## 1. 决策树基本知识

### 1.1 决策树简介

在我们的日常生活中，决策无处不在，从选择早餐吃什么，到决定是否接受一份工作，每一个选择都像是在人生的道路上进行一次小小的“分岔”。   

<img src="https://imgbed.momodel.cn/hv/202501201520422.jpg" alt="示例图片" width="60%" height="auto" style="display: block; margin: 0 auto;" >

想象一下，你站在超市的零食区，面前摆满了各种各样的巧克力、薯片和坚果。你想要挑选一种既美味又健康的零食，这时你会怎么做呢？  
你可能会先考虑自己的口味偏好，比如是喜欢甜的还是咸的；然后可能会看看包装上的营养成分表，关注热量、脂肪含量等信息；再者，你或许还会考虑价格因素，权衡性价比。  

在这个过程中，你的大脑其实就像在构建一棵“决策树”，通过层层筛选和判断，最终做出选择。

#### 1.1.1 什么是决策树
<img src="https://imgbed.momodel.cn/hv/202501201524707.png" alt="示例图片" width="60%" height="auto" style="display: block; margin: 0 auto;" >

决策树，又称判断树，它是一种以树形数据结构来展示决策规则和分类结果的模型，作为一种归纳学习算法，其重点是将看似无序、杂乱的已知实例，通过某种技术手段将它们转化成可以预测未知实例的树状模型，每一条从根结点到叶子结点的路径都代表一条决策的规则。

决策树模型是机器学习算法科学家 J.R. 昆兰于 1986 年提出。最早用于对未知数据进行分类和回归，后来被广泛运用于商业决策、医疗诊断、投资分析以及日常生活中的选择。

一句话概括决策树：一张根据各种条件层层判断、最终得出结论的树状线路图。

### 1.3 决策树详解
在介绍如何选择属性之前，我们需要先知道熵，信息增益，增益比的概念，以帮助我们更好地理解。

#### 1.3.1 熵及其相关概念
<img src="https://imgbed.momodel.cn/hv/202501210909283.gif" alt="示例图片" width="60%" height="auto" style="display: block; margin: 0 auto;" >
在我们的生活中，似乎存在着一种神奇的力量，它在默默地影响着世间万物的秩序与混乱。   

想象一下，你刚刚整理好的书桌，书本摆放整齐，文具归类有序，可没过几天，书桌上就又变得杂乱无章，书本东倒西歪，文具散落各处。又或者，地上掉落了一个苹果，随着时间的推移，苹果逐渐腐蚀、变质，最终失去了原有的美味与活力。  

这些现象背后，其实都隐藏着一个深刻的物理概念——熵。

**熵的定义**  
熵，即衡量系统无序程度的物理量。  

熵（Entropy）是信息论中的一个核心概念，由克劳德·香农（Claude Shannon）在 1948 年提出，用于量化信息的不确定性或随机性。

熵就像是一个量化“混乱度”的指标，熵值越大，表明系统的无序程度越高；反之，熵值越小，系统的有序性越强。  
举个例子，当书架上的书本整齐摆列时，说明它熵值很小。但经过一段时间后，书架上的书变得杂乱。它的熵值就比较高。书架上的书从整齐变得杂乱，对应的正是熵增的过程。

对于决策树的某个结点而言，它在对样本数据进行分类划分后，我们当然希望划分后的结果能使得整个样本集在各自的类别中尽可能有序，即希望某个属性被划分后，能最大程度地降低样本数据的熵。  

比如下边的两幅图中，从 图 1 到图 2 表示了熵增的过程。

**熵的计算**    


随机变量 S 的熵定义公式为：
$$
H(Y) = - \sum_{i=1}^{n} p_i \log_2 p_i
$$
H(Y)即事件Y的熵，其结果是把每个事件的概率乘以它概率的对数，然后加起来，前面再加个负号。

以 1.2 的例子可得，在表格中：
<img src="https://imgbed.momodel.cn/hv/202501151008680.png" alt="示例图片" width="25%" height="600" style="display: block; margin: 0 auto;" >
- 拖欠贷款者为“是”的比例：$p_{\text{是}} = \frac{3}{10}$
- 拖欠贷款者为“否”的比例：$p_{\text{否}} = \frac{7}{10}$

将这些值代入熵公式：

$$
H(Y) = - \left( p_{\text{是}} \log_2 p_{\text{是}} + p_{\text{否}} \log_2 p_{\text{否}} \right)
$$
解得：
$$ H(Y) = -(-0.881) = 0.881 $$
所以该表格中这十条数据的熵为 0.881



#### 1.3.2 划分选择

**条件熵**

条件熵：即在一定的条件上，事物的熵。  
可以理解为在已知某个条件的情况下（有房），目标变量（是否会拖欠贷款）的不确定性。  
我们用它来衡量一个属性对目标变量的影响。记作H(Y|X)，其数学表达式为：
$$
H(Y|X) = \sum_{i=1}^k p_i \, H(Y|X=x_i)
$$

同样的，以 1.2 的表格为例：
<img src="https://imgbed.momodel.cn/hv/202501161708893.png" alt="示例图片" width="70%" height="600" style="display: block; margin: 0 auto;" >
- 在有房者表格中，“拖欠贷款者”的分布为：
  - 否：$ p_{\text{否}} = 1 $（3/3）
  - 是：$ p_{\text{是}} = 0 $（0/3）
- 计算熵：
  $$
  H(Y|X=\text{是}) = -\left( p_{\text{否}} \cdot \log_2 p_{\text{否}} + p_{\text{是}} \cdot \log_2 p_{\text{是}} \right) = 0
  $$
- 在无房者表格中，“拖欠贷款者”的分布为：
  - 否：$ p_{\text{否}} = \frac{4}{7} $（4/7）
  - 是：$ p_{\text{是}} = \frac{3}{7} $（3/7）
- 计算熵：
  $$
  H(Y|X=\text{否}) = -\left( p_{\text{否}} \cdot \log_2 p_{\text{否}} + p_{\text{是}} \cdot \log_2 p_{\text{是}} \right) \approx 0.985
  $$
- **计算条件熵 $ H(Y|X) $：**
  - 有房比例：$ p_{\text{有房}} = \frac{3}{10} $
  - 无房比例：$ p_{\text{无房}} = \frac{7}{10} $
  - 加权求和：
    $$
    H(Y|X) = p_{\text{有房}} \cdot 0 + p_{\text{无房}} \cdot 0.985 \approx 0.690
    $$
    
因此最后在有房的条件下，目标变量是否会拖欠贷款的条件熵为 0.690。

**信息增益**  

信息增益是衡量一个属性对目标变量分类能力的一个指标。 

在判断是否会拖欠贷款时，如果没有其他信息，我们很难确定一个人是否会拖欠。   
但当我们知道一个人是否有房后，情况就变了——有房的人通常更稳定，拖欠贷款的可能性小；无房的人可能风险更高。  
这种通过“是否有房”这个属性，让我们的判断变得更清晰，不确定性减少的部分，就是信息增益。  

简单来说，信息增益高的属性，对数据的分类能力越强，对判断目标变量（比如是否会拖欠贷款）的帮助就越大。
信息增益的公式是：

$$
g(Y, X) = H(Y) - H(Y|X)
$$
H(Y) 是目标变量的初始熵。  
H(Y∣X) 是在X的条件下，目标变量的条件熵。  
g(Y∣X) 信息增益等于初始不确定性 H(D) 减去引入属性后的不确定性。H(D∣X)，衡量了这个属性对减少目标变量的混杂程度有多大的帮助。

**信息增益的计算及属性的选择**   
在计算熵的过程中，我们已知最开始的表格中十条数据的初始熵为：
$$ H(Y) = 0.881 $$
在计算条件熵的过程中，我们已知在有房的条件下，目标变量是否会拖欠贷款的条件熵为：
$$
    H(Y|X) = 0.690
    $$
因此，住房情况这个属性对目标变量的信息增益为：
$$
g(Y, X) = g(Y, 有房者) = H(Y) - H(Y|有房者)=0.191
$$

由此类推，婚姻情况这个属性对目标变量的信息增益为：
$$
g(Y, X) = g(Y, 婚姻) = H(Y) - H(Y|婚姻) = 0.105
$$
年收入情况这个属性对目标变量的信息增益为：
$$
g(Y, X) = g(Y, 年收入) = H(Y) - H(Y|年收入) = 0.09
$$
由：
$$
g(Y, 有房者)>g(Y, 婚姻)>g(Y, 年收入)
$$
我们选择信息增益最大的住房情况作为决策树模型根节点。   

当我们确定根节点后，同理，我们可以计算余下的属性的信息增益来确定内部节点。  
由：
$$
g(Y, 有房者—婚姻) > g(Y, 有房者—年收入)
$$
我们选择婚姻情况这个属性作为决策树模型的内部节点。

最后，仅剩下年收入这个属性，我们可以直接把年收入作为内部节点构建决策树模型。这样，依据信息增益的计算，我们就得到了一个属性选择为 (有房者->婚姻->年收入)这样结构顺序的最优的决策树模型。

**信息增益率**   

信息增益率是对信息增益的改进，旨在解决信息增益在属性选择时对取值较多的情况下，属性的偏向性问题。  

在判断是否会拖欠贷款的例子中，年收入是一个具有多个取值的属性，而是否有房是一个二分类属性，只有“有房”和“无房”两种情况。  
在信息增益的计算中，年收入由于取值较多，可能会获得较高的信息增益，但这并不完全是因为它对目标变量的区分能力更强，而是因为它本身的多样性更高。  

信息增益率通过将信息增益除以属性的内在熵（即属性本身的混杂程度）来校正这种偏差。
<img src="https://imgbed.momodel.cn/hv/202501171653303.png" alt="示例图片" width="50%" height="600" style="display: block; margin: 0 auto;" >
𝑔(𝐷, 𝑋) 为信息增益，𝐻x(𝐷) 为属性的熵。信息增益率 𝑔R(𝐷, 𝑋)为信息增益与特征固有信息的比值，信息增益率计算的具体步骤与信息增益类似，在此暂不过多赘述。


## 2. 决策树算法


### 2.1 ID3 和 C4.5 算法
**定义**  ：   
ID3 算法的原理以**信息增益**作为属性选择的准则构建决策树。它的核心思想是通过选择信息增益最大的属性作为节点，直到每个叶子节点内部类别唯一。最终构建出决策树模型。

C4.5 算法是 ID3 的改进版本，它们的大致原理是相同的。而 C4.5 算法引入了**增益率**作为特征选择的标准。它避免了信息增益倾向于选择取值多的特征的问题。

**方法**：  
a.从根结点开始，对结点计算所有可能的属性的信息增益，选择信息增益最大的属性作为结点的特征。 

b.由该属性的不同取值建立子节点，再对子结点重复地调用以上方法，构建决策树；直到属性的类别唯一为止。  

c.最后得到一个决策树。  

**流程**：
<img src="https://imgbed.momodel.cn/hv/202501201728060.png" alt="示例图片" width="30%" height="600" style="display: block; margin: 0 auto;" >

**对比**:  

ID3 算法：  
优点：简单易实现，生成的决策树可解释性强。   
缺点：只能处理离散特征，容易过拟合，不支持缺失值。   
适用场景：小规模、离散特征的数据集，对模型可解释性要求高。 

C4.5 算法：  
优点：支持连续特征和缺失值，通过增益率和剪枝减少过拟合。  
缺点：计算复杂度较高，生成的树可能较复杂。  
适用场景：大规模、包含连续特征和缺失值的数据集，需要更好的泛化能力。

### 2.2 拓展
决策树作为一种经典的机器学习算法，虽然简单直观，但在处理复杂数据时容易出现过拟合或欠拟合的问题。为了提升决策树的性能，研究者提出了集成学习方法，其中 Bagging 和 Boosting 是两种最重要的方向。它们通过组合多个模型，显著提升了预测精度和泛化能力。

#### 2.2.1 Bagging


**发展历史**  
Bagging 是由 Leo Breiman 在 1996 年提出的，旨在通过降低模型的方差来提高稳定性。  
随机森林（Random Forest）是 Bagging 的典型代表，由 Breiman 在 2001 年提出，进一步引入了特征随机选择，增强了模型的多样性。

**核心原理**  
<img src="https://imgbed.momodel.cn/hv/202501231739700.png" alt="示例图片" width="40%" height="" style="display: block; margin: 0 auto;"  >  
Bagging 的核心思想是 “人多力量大”。它通过训练多个模型，然后让这些模型一起投票或取平均值来做决定。

Bagging 通过对原始数据集进行有放回的随机抽样（Bootstrap 抽样），生成多个数据子集，然后在每个子集上独立训练一个模型。这些模型通常是同类型的，比如决策树。最后，通过投票（分类任务）或平均（回归任务）的方式将所有模型的预测结果结合起来，得到最终的预测。

Bagging 的主要目标是减少模型的方差（Variance），即降低模型对训练数据的过度拟合。通过在不同的数据子集上训练多个模型，Bagging 使得每个模型都具有一定的随机性，从而避免了单个模型因过度拟合而导致的高方差。  

**应用算法**  
- **随机森林（Random Forest）**：随机森林是 Bagging 方法的一个经典应用，由 Leo Breiman 在 2001 年提出。它在 Bagging 的基础上引入了特征随机性，即在构建每棵决策树时，每次分裂节点时只随机选择一部分特征，进一步增强了模型的多样性。随机森林在分类和回归任务中表现出色，广泛应用于图像识别、医疗诊断等领域。
- **Bagging 分类器/回归器**：在机器学习库（如 Scikit-learn）中，Bagging 分类器或回归器可以直接用于构建基于决策树或其他算法的集成模型。

#### 2.2.2 Boosting

**发展历史**  
Boosting的发展可以追溯到 1989 年，当时 Robert E. Schapire 首次提出了 Boosting 的概念。其核心思想是通过组合多个弱学习器（通常是简单的模型，如单层决策树）来构建一个强学习器。Boosting 的一个重要里程碑是 1995 年提出的 Adaboost 算法，它成为了 Boosting 领域的经典方法。

**核心原理**  
<img src="https://imgbed.momodel.cn/hv/202501231739657.png" alt="示例图片" width="40%" height="" style="display: block; margin: 0 auto;"  >  
Boosting 是一种迭代算法，通过串行训练一系列弱学习器 ，每个弱学习器都试图纠正前一个学习器的错误，最终将所有弱学习器的预测结果加权求和得到最终结果。  

Boosting 的核心思想是顺序训练，即每个新的模型都会基于前一个模型的错误进行修正。具体来说，Boosting 算法会赋予那些被前一个模型错误分类的样本更高的权重，使得后续模型更关注这些“难处理”的样本。通过这种方式，每个新的模型都在逐步改进整体模型的性能。

Boosting 的主要目标是减少模型的偏差（Bias），即提高模型对数据的拟合能力。与 Bagging 不同，Boosting 的模型是串行生成的，每个模型的训练都依赖于前一个模型的结果。

**应用算法**  
- **Adaboost**：Adaboost 是最经典的 Boosting 算法之一，适用于分类任务。它通过不断调整样本权重，使得后续模型更关注之前模型的错误样本，最终通过加权投票的方式得到强学习器。
- **梯度提升树（Gradient Boosting Machine, GBM）**：GBM 是Boosting 的另一种实现方式，它通过梯度下降的方式逐步减小模型的误差。在每一轮迭代中，GBM 会根据前一个模型的残差训练一个新的弱学习器（通常是决策树），并将这些弱学习器的预测结果加权求和，得到最终的预测结果。
- **XGBoost、LightGBM 和 CatBoost**：这些是 GBM 的改进版本，分别在计算效率、处理大数据和处理类别特征方面表现出色。例如，XGBoost 在机器学习竞赛中广泛使用，LightGBM 适合处理大规模数据，而 CatBoost 则在处理类别特征时更加高效。
