## 1. 神经网络简介

机器学习是一个广泛的概念，它包括了各种让计算机从数据中学习的技术和方法。
神经网络是机器学习中的一个特定领域，它使用类似于人脑神经元的节点和连接来模拟学习过程。
深度学习则是神经网络的一个子集，它利用多层神经网络结构来学习数据的深层表示，并在许多任务上取得了显著的性能提升。

简而言之，机器学习是一个大的领域，神经网络是机器学习的一个子集，而深度学习是神经网络的一个进阶方向。

### 1.2 神经元

#### 1.2.1 生物学基础与模型
对于神经元的研究由来已久，1904 年生物学家就已经知晓了神经元的组成结构。

一个神经元通常具有多个树突，主要用来接受传入信息；而轴突只有一条，轴突尾端有许多轴突末梢可以给其他多个神经元传递信息。轴突末梢跟其他神经元的树突产生连接，从而传递信号。这个连接的位置在生物学上叫做 **“突触”**。

人脑中的**神经元形状**可以用下图做简单的说明：

神经元信号传递的具体流程开始于神经元的细胞体或树突接收到刺激，引发电信号，即动作电位的产生。  
这个电信号沿着神经元的轴突传播，直至到达突触末端。    
在突触末端，电信号触发神经递质的释放，这些化学物质弥散穿过突触间隙，与下一个神经元的突触后膜上的受体结合，从而将化学信号转换回电信号，引发下一个神经元的兴奋或抑制，实现信息的传递和处理。  
这个过程涉及到电信号和化学信号的转换，是神经系统进行信息传递和神经网络功能的基础。

#### 1.2.2 神经元模型的数学表示与计算

神经元模型是一个包含**输入，输出与计算**功能的模型。**输入**可以类比为**神经元的树突**，而**输出**可以类比为**神经元的轴突**，**计算**则可以类比为**细胞核**。

下图是一个典型的神经元模型：包含有 3 个输入，1 个输出，以及 2 个计算功能。

注意中间的箭头线。这些线称为 **“连接”**。

**连接**是神经元中最重要的东西。每一个连接上都有一个权重。

一个神经网络的训练算法就是让权重的值调整到最佳，以使得整个网络的预测效果最好。

下图中我们使用 a 来表示**输入**，用 w 来表示**权值**。一个表示连接的有向箭头可以这样理解：在初端，传递的信号大小仍然是 a，端中间有加权参数 w，经过这个加权后的信号会变成 a × w，因此在连接的末端，信号的大小就变成了 a × w。

在其他绘图模型里，有向箭头可能表示的是值的不变传递。而在神经元模型里，每个有向箭头表示的是值的加权传递。

可见 z 是在输入和权值的线性加权和叠加了一个函数 g 的值。在神经元模型里，函数 g 是 **sgn 函数**，也就是**符号函数**。     
符号函数 sgn 是一种简单的激活函数。       
sgn 函数当输入大于 0 时，输出 1，否则输出 -1。

sgn的公式如下：

$$
sgn(n) = 
\begin{cases} 
1 & \text{如果 } n \geq 0 \\
-1 & \text{如果 } n < 0 
\end{cases}
$$

在神经元模型中，g 将 z 转换为 1 或 -1，从而引入非线性，使神经网络能够处理更复杂的数据。   
具体公式如下：   
$$
z=sgn(a1×w1+a2×w2+a3×w3)
$$

####  1.2.3 神经元模型的使用

神经元模型的使用可以这样理解：

我们有一个数据，称之为**样本**。样本有四个属性，其中三个属性已知，一个属性未知。我们需要做的就是通过三个已知属性预测未知属性。

具体办法就是使用神经元的公式进行计算。三个已知属性的值是 a<sub>1</sub>，a<sub>2</sub>，a<sub>3</sub>，未知属性的值是 z。z 可以通过公式计算出来。

这里，已知的属性称之为特征，未知的属性称之为目标。假设特征与目标之间确实是线性关系，并且我们已经得到表示这个关系的权值 w<sub>1</sub>，w<sub>2</sub>，w<sub>3</sub>。那么，我们就可以通过神经元模型预测新样本的目标。

<div class='insertContainerBox column'>
<div align=center><img src="https://imgbed.momodel.cn/202501020858012.png" width="1000px"/></div>
</div>





**举个例子**

假设我们有一个样本，样本有四个属性，其中三个已知属性分别为房屋的面积（a<sub>1</sub>）、房间数量（a<sub>2</sub>）、楼层数（a<sub>3</sub>），未知属性为房价（z）。已知房屋面积、房间数量、楼层数与房价之间存在线性关系，且已得到表示这个关系的权值 w1=0.5，w2=0.3，w3=0.2。

那么根据神经元模型的公式，房价 z 可以通过以下公式计算得出：


$$
z = w_1 \times a_1 + w_2 \times a_2 + w_3 \times a_3
$$




例如，有一个房屋，其面积 a<sub>1</sub> 为 120 平方米，房间数量 a<sub>2</sub> 为 3 个，楼层数 a<sub>3</sub> 为 5 层，那么该房屋的房价 z 可以计算为：


$$
z = 0.5 × 120 + 0.3 × 3 + 0.2 × 5 = 60 + 0.9 + 1 = 61.9
$$



所以，该房屋的预测房价为 61.9 万元。


#### 1.2.4 神经元模型的影响与发展

1943 年发布的**神经元模型**，虽然简单，但已经建立了神经网络大厦的地基。但是，**神经元模型**中，权重的值都是预先设置的，因此不能学习。

1949 年心理学家 Hebb 提出了 **Hebb 学习律**，认为人脑神经细胞的突触（也就是连接）上的强度上可以变化的。于是计算科学家们开始考虑用**调整权值**的方法来让机器学习。这为后面的学习算法奠定了基础。

尽管神经元模型与 Hebb 学习律都已诞生，但限于当时的计算机能力，直到接近 10 年后，第一个真正意义的神经网络才诞生。


### 1.3 神经网络架构
神经网络是深度学习的核心组成部分，它是一种模仿人脑神经元网络结构的计算模型。神经网络通过构建多个层次的节点（神经元）和连接这些节点的权重，来学习输入数据和输出结果之间的复杂映射关系。

<div class='insertContainerBox column'>
<div class='insertItem' align=center><img src="https://imgbed.momodel.cn/202501141406316.png" width="1000px"/></div>
</div>
     
神经网络其实是一个非常宽泛的称呼，它包括两类，一类是用计算机的方式去模拟人脑，这就是我们常说的ANN（人工神经网络），另一类是研究生物学上的神经网络，又叫生物神经网络。

在人工神经网络之中，又分为前馈神经网络和反馈神经网络这两种。

- 前馈神经网络（Feedforward Neural Network, FNN）
    - **结构图**：信息从输入层单向流动到输出层，没有回路。
    - **特点**：每一层的神经元只接收前一层的输出，不会将信息传递回去。
    - **图例**：
    ```
    输入层 → 隐藏层 → 输出层
    ```
信息流动是单向的，没有反馈。


- 反馈神经网络（Feedback Neural Network, FNN）
    - **结构图**：信息在网络中循环流动，存在回路。
    - **特点**：神经元可以将信息传递回之前的层或自身，形成动态系统。
    - **图例**：
    ```
    输入层 → 隐藏层 → 输出层
               ↑____________↓
    ```
信息在网络中循环流动，形成反馈。


**实际应用场景**

**前馈神经网络的使用场景**
- **特点**：适合处理静态数据，输入和输出之间没有时间依赖关系。
- **典型应用**：
    - **图像分类**：输入是一张图片，输出是类别标签（如猫、狗）。
    - **手写数字识别**：输入是手写数字图像，输出是数字类别（0-9）。
    - **推荐系统**：输入是用户特征和商品特征，输出是推荐评分。

**反馈神经网络的使用场景**
- **特点**：适合处理序列数据或时间依赖数据，能够记忆之前的状态。
- **典型应用**：
    - **自然语言处理（NLP）**：输入是单词序列，输出是翻译或文本生成（如 RNN、LSTM）。
    - **时间序列预测**：输入是历史数据（如股票价格、天气数据），输出是未来值。
    - **语音识别**：输入是音频信号序列，输出是文本。


| **特性**            | **前馈神经网络**                  | **反馈神经网络**                  |
|---------------------|----------------------------------|----------------------------------|
| **信息流动**         | 单向，无回路                     | 双向，有回路                     |
| **时间依赖**         | 无时间依赖，适合静态数据         | 有时间依赖，适合序列数据         |
| **典型结构**         | 多层感知机（MLP）、卷积神经网络（CNN） | 循环神经网络（RNN）、Hopfield网络 |
| **应用场景**         | 图像分类、手写识别               | 语音识别、时间序列预测           |

具体到前馈神经网络中，就有了本文中所分别描述的三个网络：单层神经网络，双层神经网络，以及多层神经网络。

红色的是**输入层**，绿色的是**输出层**，蓝色的是**隐藏层**（也叫**中间层**）。

简而言之，一个完成的神经网络架构：

- **输入层**：接收原始数据或特征向量。
- **隐藏层**：包含多个全连接层，每个全连接层中的神经元都通过激活函数与前一层中的所有神经元相连。隐藏层负责提取和学习数据的特征。
- **输出层**：产生神经网络的最终输出。对于分类任务，输出层通常使用 Softmax 激活函数以产生概率分布；对于回归任务，输出层可能直接输出预测值。




### 1.4 激活函数
**激活函数（Activation Function）** 在人工神经网络中起着至关重要的作用。它们的主要目的是在神经元的线性变换之后引入非线性，使得网络能够学习和执行更复杂的任务。

想象一下，你正在建造一个房子，房子有很多房间，每个房间都有不同的功能，比如厨房、卧室、客厅等。在神经网络中，这些房间就像是神经元，它们可以接收信息（比如房间接收阳光、空气等），处理信息（比如厨房烹饪食物），然后发送信息（比如卧室提供休息的地方）。

现在，假设你想要让房子的某个房间（神经元）只有在特定条件下才工作，比如只有当阳光充足时，太阳能电池板房间才开始发电。这时候，你需要一个开关来控制这个房间的工作状态。在神经网络中，激活函数就相当于这个开关。

激活函数在神经网络中就像是给每个神经元添加的一个特殊功能，它决定了这个神经元在给定的输入下是否应该被“激活”或者说是否应该对网络的输出产生贡献。

**前馈神经网络中常见的激活函数**包括：
      
1. Sigmoid：将输入压缩到 (0, 1) 范围内。
    - 主要用于输出层，特别是二分类问题
$$
\sigma(z) = \frac{1}{1 + e^{-z}}
$$

2. ReLU：对正输入提供线性输出，对负输入输出 0。
    - 默认激活函数，广泛用于隐藏层
$$
\text{ReLU}(z) = \max(0, z)
$$

## 2. 单层神经网络（感知机）

### 2.1 感知机的起源

1958 年，计算科学家 Rosenblatt 提出了由两层神经元组成的神经网络。他给它起了一个名字--**“感知机”（Perceptron）**。

感知机是当时首个可以学习的**人工神经网络**。Rosenblatt 现场演示了其学习识别简单图像的过程，在当时的社会引起了轰动。

人们认为已经发现了智能的奥秘，许多学者和科研机构纷纷投入到神经网络的研究中。美国军方大力资助了神经网络的研究，并认为神经网络比“原子弹工程”更重要。这段时间直到 1969 年才结束，这个时期可以看作神经网络的第一次高潮。

### 2.2 感知机模型
在“感知机”中，有两个层次。分别是输入层和输出层。
- 输入层里的 **“输入单元”** 只负责传输数据，不做计算。
- 输出层里的 **“输出单元”** 则需要对前面一层的输入进行计算。

我们把需要计算的层次称之为“计算层”，并把拥有一个计算层的网络称之为“单层神经网络”。   
有一些文献会按照网络拥有的层数来命名，例如把“感知机”称为两层神经网络。但在本文里，我们根据计算层的数量来命名。我们认为感知机为单层神经网络。

目前已知一个神经元的输出可以向多个神经元传递。我们改用二维的下标，用 w<sub>x,y</sub> 来表达一个权值。下标中的 x 代表后一层神经元的序号，而 y 代表前一层神经元的序号（序号的顺序从上到下）。

### 2.4 感知机模型的影响与发展　

感知机只能做简单的线性分类任务。但是当时的人们热情太过于高涨，并没有人清醒的认识到这点。于是，当人工智能领域的巨擘 Minsky 指出这点时，事态就发生了变化。

Minsky 在 1969 年出版了一本叫《Perceptron》的书，里面用详细的数学证明了感知机的弱点，尤其是感知机对 XOR（异或））这样的简单分类任务都无法解决。 

    XOR: 当两个输入值不同时，输出为真（1）；当两个输入值相同时，输出为假（0）。

Minsky 认为，如果将计算层增加到两层，计算量则过大，而且没有有效的学习算法。所以，他认为研究更深层的网络是没有价值的。  
由于 Minsky 的巨大影响力以及书中呈现的悲观态度，让很多学者和实验室纷纷放弃了神经网络的研究。神经网络的研究陷入了冰河期。这个时期又被称为“AI winter”。  

接近 10 年以后，对于两层神经网络的研究才带来神经网络的复苏。

## 3. 多层神经网络

### 3.1 两层神经网络（多层感知机）

#### 3.1.1 两层神经网络的兴起

在两层神经网络发展的同时，神经网络开始了大范围的推广与使用。

Minsky 说过单层神经网络无法解决异或问题。但是当增加一个计算层以后，两层神经网络不仅可以解决异或问题，而且具有非常好的非线性分类效果。不过两层神经网络的计算是一个问题，没有一个较好的解法。

1986 年，Rumelhr 和 Hinton 等人提出了**反向传播**（Backpropagation，BP）算法，解决了两层神经网络所需要的复杂计算量问题，从而带动了业界使用两层神经网络研究的热潮。目前，大量的教授神经网络的教材，都是重点介绍两层（带一个隐藏层）神经网络的内容。 

这时候的 Hinton 还很年轻，30 年以后，正是他重新定义了神经网络，带来了神经网络复苏的又一春。

#### 3.1.3 两层神经网络的非线性分类能力
与单层神经网络不同。理论证明，两层神经网络可以无限逼近任意连续函数。

这是什么意思呢？也就是说，面对复杂的非线性分类任务，两层（带一个隐藏层）神经网络可以分类的很好。

    - 数据点（用“+”和“□”表示）可以通过一条直线分开。
    - 在这种情况下，使用线性分类器（如线性回归或感知机）就可以有效地将两类数据分开。
    - 这类似于用一个筛子直接将按直线分布的球分开。   
    
2. 线性不可分，但非线性可分（右图）：

#### 3.1.4 两层神经网络的影响与发展

两层神经网络在多个地方的应用说明了其效用与价值。10 年前困扰神经网络界的异或问题被轻松解决。神经网络在这个时候，已经可以发力于语音识别，图像识别，自动驾驶等多个领域。

历史总是惊人的相似，神经网络的学者们再次登上了《纽约时报》的专访。人们认为神经网络可以解决许多问题。就连娱乐界都开始受到了影响，当年的《终结者》电影中的阿诺都赶时髦地说一句：我的 CPU 是一个神经网络处理器，一个会学习的计算机。

但是神经网络仍然存在若干的问题：尽管使用了 **BP 算法**（反向传播算法，在后面会有介绍到），一次神经网络的训练仍然耗时太久，而且困扰训练优化的一个问题就是局部最优解问题，这使得神经网络的优化较为困难。同时，隐藏层的节点数需要调参，这使得使用不太方便，工程和研究人员对此多有抱怨。

90 年代中期，由 Vapnik 等人发明的 SVM（Support Vector Machines，支持向量机）算法诞生，很快就在若干个方面体现出了对比神经网络的优势：无需调参；高效；全局最优解。基于以上种种理由，SVM迅速打败了神经网络算法成为主流。

神经网络的研究再次陷入了冰河期。当时，只要你的论文中包含神经网络相关的字眼，非常容易被会议和期刊拒收，研究界那时对神经网络的不待见可想而知。


### 3.2 损失函数

在 Rosenblat 提出的感知机模型中，模型中的参数可以被训练，但是使用的方法较为简单，并没有使用目前机器学习中通用的方法，这导致其扩展性与适用性非常有限。从两层神经网络开始，神经网络的研究人员开始使用机器学习相关的技术进行神经网络的训练。例如用大量的数据（1000-10000 左右），使用算法进行优化等等，从而使得模型训练可以获得性能与数据利用上的双重优势。

在机器学习中，我们想要训练一个模型，让它能够准确地预测结果。为了做到这一点，我们首先会给模型的参数（也就是模型中的数字）随机赋值。然后，我们用这些参数来预测一些数据，看看预测的结果（我们称之为 y<sup>p</sup>）和实际的结果（我们称之为 y ）之间有多大的差距。    
我们用一个叫做 **“损失”** （loss）的数来衡量这个差距，计算方法很简单：就是预测结果和实际结果之差的平方。我们的目标就是让这个损失值**尽可能的小**。    


$$
\text{loss} = (y^p - y)^2
$$



如果将先前的神经网络预测的矩阵公式带入到 y<sup>p</sup> 中（因为有 z = y<sup>p</sup>），那么我们可以把损失写为关于参数（parameter）的函数，这个函数称之为**损失函数**（loss function）。下面的问题就是求：如何优化参数，能够让损失函数的值最小。

为了让损失值变小，我们需要调整模型的参数。这就像在解一个复杂的谜题，我们需要找到最佳的参数组合，让损失值达到最小。这个过程叫做优化。      
优化的方法有很多，但最常用的是**梯度下降算法**。这个算法会告诉我们当前参数应该如何调整，才能让损失值减少。我们按照算法的指示不断调整参数，直到损失值不能再减少了，这时我们就认为找到了最好的参数。     
简单来说，**损失函数**就是**衡量模型预测有多准确**（衡量你的预测和实际结果的差距。）的一个工具，而**梯度下降算法**就是**帮助我们找到最佳模型参数**（找到调整预测的方向和幅度，让差距越来越小）的方法。    



接下来我们举一个**例子**来解释损失函数

你有一个朋友，他每天都会告诉你他感觉的温度（比如“今天很热”或“今天很冷”），但他不会直接告诉你具体的温度数值。你的任务是猜出实际的温度是多少。

- 初始猜测：你一开始随便猜了一个温度，比如 20 度。

- 对比结果：你的朋友告诉你，实际温度是 25 度。你发现你的猜测（20 度）和实际温度（25 度）之间有差距。

- 计算差距：你用一个简单的规则来衡量差距，比如用实际温度减去你的猜测，得到差距是 5 度。这个差距就是“损失”。

- 调整猜测：为了让差距变小，你决定调整你的猜测。你发现，如果你把猜测的温度提高一点，差距就会变小。于是你把猜测从 20 度调整到 23 度。

- 重复过程：你再次问朋友实际温度，发现差距变小了（现在是 2 度）。你继续调整猜测，直到你的猜测和实际温度几乎一样。

**损失函数和梯度下降的作用**

- 损失函数：在这个例子里，损失函数就是用来衡量你的猜测和实际温度之间的差距（比如 5 度或 2 度）。你的目标是让这个差距尽可能小。

- 梯度下降：梯度下降就是帮助你调整猜测的方法。它告诉你应该往哪个方向调整（比如提高温度还是降低温度），以及调整多少（比如每次调整 3 度还是 1 度）。

通过不断调整，你的猜测会越来越接近实际温度，损失也会越来越小。这就是训练模型的基本思想！


### 3.3 反向传播算法

在神经网络模型中，由于结构复杂，每次计算梯度的代价很大。因此需要使用**反向传播算法**。**反向传播算法**，简称BP（Backpropagation）算法，是利用了神经网络的结构进行的计算。    
每一次计算所有参数的梯度，而是从后往前。首先计算输出层的梯度，然后是第二个参数矩阵的梯度，接着是隐藏层的梯度，再然后是第一个参数矩阵的梯度，最后是输入层的梯度。计算结束以后，所要的两个参数矩阵的梯度就都有了。     

反向传播算法可以直观的理解为下图。梯度的计算从后往前，一层层反向传播。前缀 E 代表着**相对导数**的意思。

<div class='insertContainerBox column'>
<div align=center><img src="https://imgbed.momodel.cn/202501021020529.png" width="600px"/></div>
</div>

**举个例子**，想象一下，你有一座多层的房子，每一层都代表神经网络中的一层。现在，你想计算每层楼（也就是每一层的参数）对最终结果的影响。反向传播算法就是从最后一层开始，一层一层向前计算，直到到达第一层。这样做的好处是，我们可以更高效地计算出每一层参数对最终结果的影响（也就是梯度），而不需要一次性计算所有参数的梯度。

通过反向传播算法，神经网络能够根据损失函数的梯度信息，逐步调整每一层的权重和偏置，从而使得模型的预测结果越来越接近真实值。这种逐层反向传播的机制，不仅提高了计算效率，还为深度学习模型的训练提供了坚实的基础。

总结来说，反向传播算法是神经网络训练的核心技术之一。它通过从输出层到输入层的逐层梯度计算，高效地更新网络参数，使得神经网络能够通过大量数据学习到复杂的特征和模式。正是由于反向传播算法的引入，神经网络才得以在图像识别、自然语言处理等领域取得突破性进展。

尽管使用了BP算法，神经网络训练仍然耗时且优化困难的**原因**包括：
- 计算量大：神经网络参数众多，每次迭代更新参数需要大量计算。
- 局部最优：梯度下降法容易陷入局部最优解，难以找到全局最优。
- 调参困难：隐藏层节点数等超参数需要手动调整，费时费力。

### 3.4 多层神经网络（深度学习）

#### 3.4.1 深度学习的兴起　

在被人摒弃的 10 年中，有几个学者仍然在坚持研究。这其中就有加拿大多伦多大学的 Geoffery Hinton 教授。

2006 年，Hinton 在《Science》和相关期刊上发表了论文，首次提出了“深度信念网络”的概念。与传统的训练方式不同，“深度信念网络”有一个 **“预训练”** （pre-training）的过程，这可以方便的让神经网络中的权值找到一个接近最优解的值，之后再使用 **“微调”** (fine-tuning)技术来对整个网络进行优化训练。这两个技术的运用大幅度减少了训练多层神经网络的时间。他给多层神经网络相关的学习方法赋予了一个新名词--**“深度学习”**。

很快，深度学习在语音识别领域暂露头角。接着，2012 年，深度学习技术又在图像识别领域大展拳脚。Hinton 与他的学生在 ImageNet 竞赛（一个计算机图像识别大赛）中，用多层的卷积神经网络成功地对包含一千类别的一百万张图片进行了训练，取得了分类错误率 15% 的好成绩，这个成绩比第二名高了近 11 个百分点，充分证明了多层神经网络识别效果的优越性。

在这之后，关于深度神经网络的研究与应用不断涌现。

#### 3.4.2 多层神经网络的结构

#### 3.4.3 深度学习的特征表示
与两层神经网络不同。多层神经网络中的层数增加了很多。

增加更多的层次有什么好处？

更深入的表示特征，以及更强的函数模拟能力。
    
更深入的表示特征可以这样理解，随着网络的层数增加，每一层对于前一层次的抽象表示更深入。          
在神经网络中，每一层神经元学习到的是前一层神经元值的更抽象的表示。    
例如第一个隐藏层学习到的是“边缘”的特征，第二个隐藏层学习到的是由“边缘”组成的“形状”的特征，第三个隐藏层学习到的是由“形状”组成的“图案”的特征，最后的隐藏层学习到的是由“图案”组成的“目标”的特征。      
通过抽取更抽象的特征来对事物进行区分，从而获得更好的区分与分类能力。   

接下来**举一个例子**来说明多层神经网络如何通过不同隐藏层来学习特征并进行分类：

1. **输入层**：接收原始数据，比如一张手写数字的图片。

2. **第一隐藏层**：学习基本特征，比如图片中的边缘和线条。

3. **第二隐藏层**：组合第一层的特征，学习更复杂的形状，比如圆形或角形。

4. **第三隐藏层**：识别由形状组成的图案，比如数字“8”的两个圆圈。

5. **第四隐藏层**：从图案中识别出完整的数字，比如数字“5”的钩形。

6. **输出层**：将第四隐藏层的特征组合，预测图片中的数字是0到9中的哪一个。

<div class='insertContainerBox column'>
<div class='insertItem' align=center><img src="https://imgbed.momodel.cn/202501171448873.png" width="800px"/></div>
</div>

通过这种方式，神经网络能够像剥洋葱一样，一层一层地从简单到复杂学习识别手写数字。每一层都建立在前一层学习到的特征之上，最终实现对数字的准确分类。

通过研究发现，在参数数量一样的情况下，更深的网络往往具有比浅层的网络更好的识别效率。这点也在 ImageNet 的多次大赛中得到了证实。从 2012 年起，每年获得 ImageNet 冠军的深度神经网络的层数逐年增加，2015 年最好的方法 GoogleNet 是一个多达22层的神经网络。

在最新一届的 ImageNet 大赛上，目前拿到最好成绩的 MSRA 团队的方法使用的更是一个深达 152 层的网络！

#### 3.4.4 深度学习的影响与未来方向

目前，深度神经网络在人工智能界占据统治地位。但凡有关人工智能的产业报道，必然离不开深度学习。  
神经网络界当下的四位引领者除了前文所说的杰弗里·辛顿（Geoffrey Hinton）以外，还有 CNN 的发明人 Yann Lecun，《Deep Learning》的作者 Bengio，以及 DeepLearning.AI 创始人吴恩达。

ImageNet 竞赛也在不断召开，有更多的方法涌现出来，刷新以往的正确率。

## 4. 回顾

### 4.1 神经网络的发展历程与起伏
我们回顾一下神经网络发展的历程。神经网络的发展历史曲折荡漾，既有被人捧上天的时刻，也有摔落在街头无人问津的时段，中间经历了数次大起大落。

从单层神经网络（感知机）开始，到包含一个隐藏层的两层神经网络，再到多层的深度神经网络，一共有三次兴起过程。详见下图。

历史最大的好处是可以给现在做参考。科学的研究呈现螺旋形上升的过程，不可能一帆风顺。同时，这也给现在过分热衷深度学习与人工智能的人敲响警钟，因为这不是第一次人们因为神经网络而疯狂了。1958 年到 1969 年，以及 1985 年到 1995 年，这两个十年间人们对于神经网络以及人工智能的期待并不现在低，可结果如何大家也能看的很清楚。

因此，冷静才是对待目前深度学习热潮的最好办法。如果因为深度学习火热，或者可以有“钱景”就一窝蜂的涌入，那么最终的受害人只能是自己。神经网络界已经两次有被人们捧上天了的境况，相信也对于捧得越高，摔得越惨这句话深有体会。因此，神经网络界的学者也必须给这股热潮浇上一盆水，不要让媒体以及投资家们过分的高看这门技术。很有可能，三十年河东，三十年河西，在几年后，神经网络就再次陷入谷底。根据上图的历史曲线图，这是很有可能的。

| 激活函数名称 | 函数公式 | 函数描述 | 作用 | 应用场景 |
|--------------|----------|----------|------|----------|
| Sigmoid      | $$\sigma(x) = \frac{1}{1 + e^{-x}}$$ | 输出范围为 (0, 1)，S 形曲线 | 将输入映射到 (0, 1) 区间，用于二分类问题的输出层 | 二分类问题、逻辑回归 |
| ReLU         | $$\text{ReLU}(x) = \max(0, x)$$ | 输出为输入的正值部分，负值部分为 0 | 加速梯度下降，缓解梯度消失问题 | 卷积神经网络 (CNN) 的隐藏层 |
| Leaky ReLU   | $$\text{Leaky ReLU}(x) = \begin{cases} x & \text{if } x \geq 0 \\ \alpha x & \text{if } x < 0 \end{cases}$$（其中 $$\alpha$$ 是一个小常数） | 解决 ReLU 的“死亡”问题，负值部分有非零输出 | 避免 ReLU 的梯度为零问题 | 卷积神经网络、深度神经网络 |
| Tanh         | $$\tanh(x) = \frac{e^{x} - e^{-x}}{e^{x} + e^{-x}}$$ | 输出范围为 (-1, 1)，S 形曲线 | 将输入映射到 (-1, 1) 区间，中心化输出 | 循环神经网络 (RNN) 的隐藏层、某些深度神经网络 |
| Softmax      | $$\text{Softmax}(x_i) = \frac{e^{x_i}}{\sum_{j} e^{x_j}}$$ | 将输入向量转换为概率分布 | 用于多分类问题的输出层，输出概率分布 | 多分类问题 |
| ELU          | $$\text{ELU}(x) = \begin{cases} x & \text{if } x \geq 0 \\ \alpha(e^{x} - 1) & \text{if } x < 0 \end{cases}$$（其中 $$\alpha$$ 是一个小常数） | 负值部分平滑过渡，缓解梯度消失 | 提供负值输出，缓解梯度消失问题 | 深度神经网络 |
| Swish        | $$\text{Swish}(x) = x \cdot \sigma(x)$$ | 自门控激活函数，结合了 ReLU 和 Sigmoid 的优点 | 平滑且非单调，适用于复杂模型 | 深度学习模型，如 Transformer |
