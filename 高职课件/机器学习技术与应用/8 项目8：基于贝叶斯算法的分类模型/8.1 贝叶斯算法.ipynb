## 2. 贝叶斯决策论


### 2.1 贝叶斯决策论

#### 2.1.1 条件概率
$P(A|B)$：表示事件 $B$ 发生的前提下，事件 $A$ 发生的概率
$$P(A|B) = \frac{P(A \cap B)}{P(B)}$$

$P(B|A)$：表示事件 $A$ 发生的前提下，事件 $B$ 发生的概率
$$P(B|A) = \frac{P(A \cap B)}{P(A)}$$

那么，就有 $P(A|B) \times P(B) = P(B|A) \times P(A)$，即：
$$P(A|B) = \frac{P(B|A) \times P(A)}{P(B)}$$

#### 2.1.2 贝叶斯决策理论基本思想
贝叶斯决策理论方法是统计模型决策中的一个基本方法，其基本思想是：

- 已知类条件概率密度参数表达式和先验概率
- 利用贝叶斯公式转换成后验概率
- 根据后验概率大小进行决策分类

贝叶斯概率计算公式表达：
$$后验概率=先验概率×似然概率（即新增信息所带来的调节程度）$$

### 2.2 贝叶斯推断

#### 2.2.1 贝叶斯推断的基本概念
贝叶斯推断是一种基于贝叶斯公式进行分析的统计学方法。

为了计算 $P(A|B)$，可预先估计一个先验概率 $P(A)$；然后根据实际情况来计算 $P(B|A)/P(B)$，用其作为调节因子，基于观测到的新增信息，对先验概率 $P(A)$ 不断修正，得到更为准确的后验概率 $P(A|B)$。

有两个碗，第一个碗中装有 30 个水果糖和 10 个巧克力糖，第二个碗中装有 20 个水果糖和 20 个巧克力糖，现在随机选择一个碗，从中取出一颗糖，发现是水果糖，请求出这颗水果糖来自一号碗的概率有多大？

从碗1、碗2拿糖果的先验概率：
$$P(碗1) = P(碗2) = \frac{1}{2}$$

从碗1拿到水果糖的似然概率是：
$$P(水果糖|碗1) = \frac{30}{40} = \frac{3}{4}$$

从碗2拿到水果糖的似然概率是：
$$P(水果糖|碗2) = \frac{20}{40} = \frac{1}{2}$$

拿到水果糖的先验概率是：
$$P(水果糖) = P(水果糖|碗1) \times P(碗1) + P(水果糖|碗2) \times P(碗2) = \frac{3}{4} \times \frac{1}{2} + \frac{1}{2} \times \frac{1}{2} = \frac{5}{8}$$

拿到水果糖来自碗1的后验概率是：
$$P(碗1|水果糖) = \frac{P(水果糖|碗1) \times P(碗1)}{P(水果糖)} = \frac{\frac{3}{4} \times \frac{1}{2}}{\frac{5}{8}} = \frac{3}{5}$$

#### 2.2.2 贝叶斯推断：广告邮件自动识别
在互联网产业日益蓬勃发展的今天，很多用户可能都有过邮箱被不感兴趣的广告邮件所充斥的经历。
基于贝叶斯推断，现在我们来设计一个简单的广告邮件自动识别系统。

假设采集到了 10000 个邮件样本，其中有 4000 封邮件被认定为广告邮件、6000 封被认定为正常邮件。这 4000 封广告邮件中，出现“红包”关键词的有 1000 封 ，而在 6000 封正常邮件中仅有 6 封包含“红包”这个词。

请问这封包含了“红包”单词的邮件是不是广告邮件呢？从已知历史事实来看，它很有可能是广告邮件，如何用贝叶斯推断来计算这个可能性呢？

根据先前收集的 10000 封邮件，可以计算广告邮件和正常邮件的先验概率:

$$P(广告邮件) = \frac{4000}{10000} = \frac{2}{5}；
P(正常邮件) = \frac{6000}{10000} = \frac{3}{5}$$

广告邮件中出现 “红包” 关键词的条件概率为
$$P(出现“红包”|广告邮件) = \frac{1000}{4000} = \frac{1}{4}$$

正常邮件中出现“红包”关键词的条件概率为：
$$P(出现“红包”|正常邮件) = \frac{6}{6000} = \frac{1}{1000}$$

根据贝叶斯定理，可以得到邮件中出现“红包”关键词这一前提下，分别该邮件被认为是正常邮件或者广告邮件的后验概率是:


$$P(正常邮件|出现“红包”) = P(正常邮件) \times P(出现“红包”|正常邮件) / P(出现“红包”)
= \frac{0.6 \times 0.001}{1006 \div 10000} \approx 0.006$$

$$P(广告邮件|出现“红包”) = P(广告邮件) \times P(出现“红包”|广告邮件) / P(出现“红包”)
= \frac{0.4 \times 0.25}{1006 \div 10000} \approx 0.994$$


根据计算结果， $P(广告邮件|出现“红包”)$ 远远大于 $P(正常邮件|出现“红包”)$，故该邮件为广告邮件。

#### 2.2.3 贝叶斯推断：广告邮件自动识别的代码实现
接下来我们来进行贝叶斯推断的代码实现。假设采集到了 10000 个邮件样本，其中有 4000 封邮件被认定为广告邮件、6000 封被认定为正常邮件。这 4000 封广告邮件中，出现“红包”关键词的有 1000 封 ，而在 6000 封正常邮件中仅有 6 封包含“红包”这个词。

请问如果收到包含了“红包”单词的邮件它是不是广告邮件呢？

```python
# 广告邮件的数量 
ad_number = 4000
# 正常邮件的数量
normal_number = 6000
# 所有广告邮件中，出现 “红包” 关键词的邮件的数量
ad_hongbao_number = 1000
# 所有正常邮件中，出现 “红包” 关键词的邮件的数量
normal_hongbao_number = 6
print('广告邮件的数量:%s,正常邮件的数量:%s'%(ad_number, normal_number))
print('所有广告邮件中出现 “红包” 关键词的邮件的数量:', ad_hongbao_number)
print('所有正常邮件中出现 “红包” 关键词的邮件的数量:', normal_hongbao_number)
```

```python
# 用户收到广告邮件的先验概率为
P_ad = ad_number / (ad_number + normal_number)
print("用户收到广告邮件的先验概率为 P(广告邮件) = " + str(P_ad))
# 用户收到正常邮件的先验概率为
P_normal = normal_number / (ad_number + normal_number)
print("用户收到正常邮件的先验概率为 P(正常邮件) = " + str(P_normal))
```

```python
# 广告邮件中出现 “红包” 关键词的条件概率
P_hongbao_ad = ad_hongbao_number / ad_number
print("广告邮件中出现 “红包” 关键词的条件概率为 P(出现“红包”|广告邮件) = " + str(P_hongbao_ad))
```

```python
# 正常邮件中出现 “红包” 关键词的条件概率
P_hongbao_normal = normal_hongbao_number / normal_number
print("正常邮件中出现“红包”关键词的条件概率为 P(出现“红包”|正常邮件) = " + str(P_hongbao_normal))
```

**后验概率计算 **

```python
# 红包出现的概率
P_hongbao = (normal_hongbao_number + ad_hongbao_number) / (ad_number + normal_number)
print("邮件出现“红包”关键词的概率为 P(出现“红包”)=" + str(P_hongbao))
# 当邮件中出现 “红包” ，其为正常邮件的后验概率
P_normal_hongbao = P_normal * P_hongbao_normal / P_hongbao
print("当邮件中出现 “红包” ，其为正常邮件的后验概率为 P(正常邮件|出现“红包”) = " + str(P_normal_hongbao))
# 当邮件中出现 “红包” ，其为广告邮件的后验概率
P_ad_hongbao = P_ad * P_hongbao_ad / P_hongbao
print("当邮件中出现 “红包” ，其为广告邮件的后验概率为 P(广告邮件|出现“红包”) = " + str(P_ad_hongbao))
```

### 2.3 贝叶斯决策论的用途

#### 2.3.1 贝叶斯决策理论的优缺点

**贝叶斯决策理论的优点**     
（1）贝叶斯决策能对信息的价值或是否需要采集新的信息做出科学的判断。       
（2）它能对调查结果的可能性加以数量化的评价,而不是像一般的决策方法那样,对调查结果或者是完全相信,或者是完全不相信。     
（3）如果说任何调查结果都不可能完全准确,先验知识或主观概率也不是完全可以相信的,那么贝叶斯决策则巧妙地将这两种信息有机地结合起来了。    
（4）它可以在决策过程中根据具体情况下不断地使用,使决策逐步完善和更加科学。      

**贝叶斯决策理论的缺点**            
（1）它需要的数据多,分析计算比较复杂,特别在解决复杂问题时,这个矛盾就更为突出。     
（2）有些数据必须使用主观概率,有些人不太相信,这也妨碍了贝叶斯决策方法的推广使用。      

#### 2.3.2 贝叶斯决策理论的用处
在有限的信息下，能够帮助我们预测出概率。所有需要作出概率预测的地方都可以见到贝叶斯定理的影子。

特别地，贝叶斯是机器学习的核心方法之一。例如垃圾邮件过滤，中文分词，艾滋病检查，肝癌检查等。

## 3. 极大似然估计


### 3.1 极大似然估计基本概念

#### 3.1.1 什么是极大似然估计
极大似然估计方法 (Maximum Likelihood Estimate，MLE) 也称为最大概似估计或最大似然估计，是求估计的另一种方法，最大概似是1821年首先由德国数学家高斯 (C. F. Gauss) 提出，
但是这个方法通常被归功于英国的统计学家罗纳德·费希尔 (R. A. Fisher) 。

极大似然估计与贝叶斯推断是统计中两种对模型的参数确定的方法，两种参数估计方法使用不同的思想。
后者属于贝叶斯派，认为参数也是服从某种概率分布的，已有的数据只是在这种参数的分布下产生的；
前者来自于频率派，认为参数是固定的，需要根据已经掌握的数据来估计这个参数。



#### 3.1.2 为什么使用极大似然估计
当我们使用机器学习解决具体现实问题时，我们是无法确切知道具体的数据分布情况的。那我们可不可以用部分已知数据去预测整体的分布 呢？
极大似然估计就是一个解决这类问题的方法。

例子：有两个外形完全相同的箱子，甲箱中有99只白球，1只黑球；乙箱中有99只黑球，1只白球。一次试验取出了一球，结果取出的是黑球。你觉得黑球从哪个箱子中取出？
你可能会觉得：“此黑球最像是从乙箱中取出的”，这个推断符合人们的经验事实。“最像”就是“极大似然”之意，这种想法常称为“极大似然原理”。

### 3.2 极大似然估计原理

#### 3.2.1 极大似然估计原理
极大似然估计是建立在极大似然原理的基础上的一个统计方法，是概率论在统计学中的应用。
极大似然估计提供了一种给定观察数据来评估模型参数的方法，即：“模型已定，参数未知”。 通过若干次试验，观察其结果，利用试验结果得到某个参数值能够使样本出现的概率为最大，则称为极大似然估计

我们来看一个具体例子：
一个硬币被抛了 100 次，有 61 次正面朝上，假设硬币向上的概率猜测有三个 $  \frac{1}{3} $，$ \frac{1}{2} $，$ \frac{2}{3} $

问：以上三个哪个是最大似然估计？
以下是修改后的 Markdown 格式内容：



**方法一**

假设唯一参数为 $p$ ，因此由概率论的知识，我们可以列出以下公式：

- $ P(H=61|p=\frac{1}{3})=\binom{100}{61}\left(\frac{1}{3}\right)^{61}\left(1-\frac{1}{3}\right)^{39}\approx9.6\times10^{-9} $

- $ P(H=61|p=\frac{1}{2})=\binom{100}{61}\left(\frac{1}{2}\right)^{61}\left(1-\frac{1}{2}\right)^{39}=0.007 $

- $P(H=61|p=\frac{2}{3})=\binom{100}{61}\left(\frac{2}{3}\right)^{61}\left(1-\frac{2}{3}\right)^{39}\approx0.40 $

比较以上三个值，可以得出 $ p=\frac{2}{3} $ 是极大似然估计值。

---

**方法二**

除了以上的方法，我们还可以利用导数为 0 得到极大似然估计，主动计算极大似然估计值。

已知：

$$ P(H=61|p)=\binom{100}{61}p^{61}(1-p)^{39} $$

通过求导数，可以得到以下公式：

\
\begin{aligned}
\frac{d}{dp}\left[\binom{100}{61}p^{61}(1-p)^{39}\right] &=\binom{100}{61}\left(61p^{60}(1-p)^{39}-39p^{61}(1-p)^{38}\right) \\
&=\binom{100}{61}p^{60}(1-p)^{38}\left(61(1-p)-39p\right) \\
&=\binom{100}{61}p^{60}(1-p)^{38}(61-100p) \\
&=0
\end{aligned}


可知，当 $ p=0,\frac{61}{100},p=1 $ 时该式等于 0，因为 $ 0<p<1 $，因此 $ p=\frac{61}{100} $ 是极大似然估计值。

---

#### 3.2.2 极大似然估计值求解步骤



由于样本集中的样本都是独立同分布，可以只考虑一类样本集$D$，来估计参数向量$\theta$。记已知的样本集为:
$$D = \{x_1, x_2, \ldots, x_N\}$$

联合概率密度函数$p(D|\theta)$称为相对于$\{x_1, x_2, \ldots, x_N\}$的$\theta$的似然函数:
$$l(\theta) = p(D|\theta) = p\{x_1, x_2, \ldots, x_N|\theta\} = \prod_{i=1}^N p(x_i|\theta)$$

如果$\hat{\theta}$是参数空间中能使似然函数$l(\theta)$最大的$\theta$值，则$\hat{\theta}$应该是“最可能”的参数值，那么$\hat{\theta}$就是$\theta$的极大似然估计量。
它是样本集的函数，记作
$$\hat{\theta} = d(x_1, x_2, \ldots, x_N) = d(D)$$

$$\hat{\theta}(x_1, x_2, \ldots, x_N)称作极大似然函数估计值$$

算法的关键在于如何把参数$\theta$的极大似然估计$\hat{\theta}$求出。

**对于如何求解极大似然估计值**     
一般可以利用$\ln l(\theta)$是$\theta$的增函数，故$\ln l(\theta)$与$l(\theta)$在同一点处达到最大值，于是对似然函数$l(\theta)$取对数，利用微分学知识转化为求解对数似然方程。解此方程并对解做进一步的判断。但由最值原理，如果最值存在，此方程组求得的驻点即为所求的最值点，就可以得到参数的极大似然估计。         
极大似然估计法一般属于这种情况，所以可以直接按上述步骤求极大似然估计。

#### 3.2.3 极大似然估计的简单应用

求极大似然估计(MLE)的一般步骤：  
（1）由总体分布导出样本的联合概率函数（或联合密度）；  
（2）把样本联合概率函数（或联合密度）中自变量看成已知常数，而把参数$\theta$看作自变量，得到似然函数$l(\theta)$；  
（3）求似然函数$l(\theta)$的最大值点（常常转化为求$\ln l(\theta)$的最大值点），即$\theta$的MLE；  
（4）在最大值点的表达式中，用样本值带入就得到参数的极大似然估计。  


下面我们使用极大似然估计来解决一个数学问题：  
若随机变量$x$服从一个数学期望为$\mu$、方差为$\sigma^2$的正态分布，记为$N(\mu, \sigma^2)$，假设$\mu=30$，$\sigma=2$。

我们先引入相关库 

```python
import numpy as np
from scipy.stats import norm
import matplotlib.pyplot as plt
```

定义数学期望、方差和x，并使x服从正态分布。

```python
μ = 30  # 数学期望
σ = 2  # 方差
x = μ + σ * np.random.randn(10000)  # 正态分布
```

我们可以用`norm.fit(x)`来求出极大似然估计。

```python
plt.hist(x, bins=100)  # 直方图显示
plt.show()
print(norm.fit(x))  # 返回极大似然估计，估计出参数约为30和2
```

### 3.3 极大似然估计的特点

- 比其他估计方法更加简单；
- 收敛性：无偏或者渐近无偏，当样本数目增加时，收敛性质会更好；
- 如果假设的类条件概率模型正确，则通常能获得较好的结果。但如果假设模型出现偏差，将导致非常差的估计结果。


**极大似然估计和贝叶斯推断的区别和联系**


极大似然估计和贝叶斯推断最大区别便在于估计的**参数不同**：

- 极大似然估计要估计的参数`θ`被当作是固定形式的一个未知变量，然后结合真实数据通过最大化似然函数来求解这个固定形式的未知变量；
- 贝叶斯估计则是将参数视为是有某种已知先验分布的随机变量，即这个参数不是一个固定的未知数，而是符合一定先验分布如：随机变量θ符合正态分布等;
- 我们通过贝叶斯规则将参数的先验分布转化成后验分布进行求解。        


在先验概率能保证问题有解的情况下，最大似然估计和贝叶斯推断在训练样本趋近于无穷时得到的结果是一样的。 

但是实际的模式识别问题中，训练样本总是有限的，我们应如何选择使用哪种模型呢？

- 计算复杂度不同：最大似然估计中只需要使用简单的微分运算，而贝叶斯需要用到非常复杂的多重积分；
- 准确性不同：当采用的样本数据很有限时，贝叶斯推断比极大似然估计误差更小；

## 4. 朴素贝叶斯分类器

### 4.1 朴素贝叶斯分类器概念

#### 4.1.1 朴素贝叶斯分类器概念
朴素贝叶斯分类器是一系列假设特征之间强（朴素）独立条件下以贝叶斯定理为基础的简单概率分类器。     
该分类器模型会给问题实例分配用特征值表示的类标签，类标签取自有限集合。        
所有朴素贝叶斯分类器都假定样本每个特征与其他特征都不相关。

朴素贝叶斯的思想基础是：对于给出的待分类项，求解在此项出现的条件下各个类别出现的概率，哪个最大，就认为此待分类项属于哪个类别。

如：如果一种水果其具有红，圆，直径大概3英寸等特征，该水果可以被判定为是苹果。
朴素贝叶斯分类器 认为红，圆，直径大概3英寸这些特征在判定该水果是否为苹果的概率分布上独立的。

对于某些类型的概率模型，在监督式学习的样本集中能获取得非常好的分类效果。
在许多实际应用中，朴素贝叶斯模型参数估计使用最大似然估计方法；换而言之，在不用到贝叶斯概率或者任何贝叶斯模型的情况下，朴素贝叶斯模型也能奏效。

尽管是带着这些朴素思想和过于简单化的假设，但朴素贝叶斯分类器在很多复杂的现实情形中仍能够获取相当好的效果

#### 4.1.2 朴素贝叶斯分类器实际案例
小明同学在学校论坛上发帖征求外卖美食推荐，接下来的几天他的下单记录如下表所示。     
现在，有人向小明同学推荐一家“价位低、口味偏甜、距离远”的店铺，他会在该店铺下单吗？      

| 店铺价位 | 店铺口味 | 店铺距离 | 是否下单 |
|----------|----------|----------|----------|
| 高       | 偏甜     | 近       | 是       |
| 高       | 清淡     | 近       | 否       |
| 高       | 偏辣     | 远       | 否       |
| 高       | 偏甜     | 远       | 否       |
| 低       | 偏甜     | 近       | 是       |
| 低       | 偏甜     | 近       | 是       |
| 低       | 清淡     | 远       | 否       |
| 低       | 偏辣     | 远       | 是       |

该同学在收到8次推荐后，下单4次和没有下单4次，则其“下单”和“不下单”的概率：  
$$P(下单) = \frac{4}{8} = 0.5$$ 

$$P(不下单) = \frac{4}{8} = 0.5$$ 


该同学对“价位低、口味偏甜、距离远”这次推荐的“下单”或“不下单”的似然概率为（基本假设：店铺价位、口味、距离等特质相互独立）：  

$$P(价位=低, 口味=偏甜, 距离=远|下单) $$
$$= P(价位=低|下单) \times P(口味=偏甜|下单) \times P(距离=远|下单) $$
$$= \frac{3}{4} \times \frac{3}{4} \times \frac{1}{4} $$$$\approx 0.141$$  

$$P(价位=低, 口味=偏甜, 距离=远|不下单) $$
$$= P(价位=低|不下单) \times P(口味=偏甜|不下单) \times P(距离=远|不下单) $$
$$= \frac{1}{4} \times \frac{1}{4} \times \frac{3}{4} $$$$\approx 0.047$$  


根据贝叶斯公式，该同学在“价格低、口味偏甜、距离远”的店铺下单的后验概率为：  
$$P(下单|价位=低, 口味=偏甜, 距离=远) $$
$$= \frac{P(下单) \times P(价位=低, 口味=偏甜, 距离=远|下单)}{P(价位=低, 口味=偏甜, 距离=远)} $$
$$= 0.5 \times 0.141$$
$$\approx 0.0705$$

不下单的后验概率为：  
$$P(不下单|价位=低, 口味=偏甜, 距离=远) $$
$$= \frac{P(不下单) \times P(价位=低, 口味=偏甜, 距离=远|不下单)}{P(价位=低, 口味=偏甜, 距离=远)} $$
$$= 0.5 \times 0.047 $$
$$\approx 0.0235$$  


由此可见，该同学这次会下单的概率大于不下单的概率。  


对上述计算过程简化后，核心公式为：  

$$P(下单|价位=低, 口味=偏甜, 距离=远) = \frac{P(下单) \times P(价位=低, 口味=偏甜, 距离=远|下单)}{P(价位=低, 口味=偏甜, 距离=远)}$$  

$$P(不下单|价位=低, 口味=偏甜, 距离=远) = \frac{P(不下单) \times P(价位=低, 口味=偏甜, 距离=远|不下单)}{P(价位=低, 口味=偏甜, 距离=远)}$$  


由于两个公式的分母相同，对结果的比较无影响，因此计算中可省略分母。

### 4.2 朴素贝叶斯分类算法流程

#### 4.2.1  朴素贝叶斯分类算法流程
1. 设待分类项为 $ x = \{a_1, a_2, \ldots, a_m\} $（其中 $ a_i $ 为 $ x $ 的特征属性），类别集合为 $ C = \{y_1, y_2, \ldots, y_n\} $。  


2. 根据**贝叶斯公式**，分别计算每个类别 \( y_k \in C \) 对应的后验概率：  
$$
P(y_k \mid x) = \frac{P(x \mid y_k) \cdot P(y_k)}{P(x)}
$$  


3. 若存在某个 \( y_k \)，使得：  
    $$
    P(y_k \mid x) = \max\left\{ P(y_1 \mid x), P(y_2 \mid x), \ldots, P(y_n \mid x) \right\}
    $$  
    则判定待分类项 \( x \) 属于类别 \( y_k \)。

#### 4.2.2 朴素贝叶斯分类算法条件概率计算
根据 4.2.1 中所描述的朴素贝叶斯分类算法流程，算法的关键就是如何计算第 3 步中的各个条件概率。  
我们可以这么做:  

- 找到一个已知分类的待分类项集合，这个集合叫做训练样本集 。  
- 统计得到在各类别下各个特征属性的条件概率估计。即:  
  $$ P(a_1|y_1),\ P(a_2|y_1),\ \ldots,\ P(a_m|y_1);\ P(a_1|y_2),\ P(a_2|y_2),\ \ldots,\ P(a_m|y_2);\ \ldots;\ P(a_1|y_n),\ P(a_2|y_n),\ \ldots,\ P(a_m|y_n) $$  
- 如果各个特征属性是条件独立的，则根据贝叶斯定理有如下推导 :  
  $$ P(y_i|x) = P(x|y_i)P(y_i)/P(x) $$  


因为分母对于所有类别为常数，因为我们只要将分子最大化即可。  
又因为各特征属性是条件独立的，所以有:  

$$
\begin{align*}
P(x|y_i)P(y_i) &= P(a_1, a_2, \ldots, a_m|y_i)P(y_i) \\
&= P(a_1|y_i)P(a_2|y_i)\ldots P(a_m|y_i)P(y_i) \\
&= P(y_i) \prod_{j=1}^m P(a_j|y_i)
\end{align*}
$$
