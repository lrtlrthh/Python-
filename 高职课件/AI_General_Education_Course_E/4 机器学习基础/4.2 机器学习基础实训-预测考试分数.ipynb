# 线性回归实训：预测考试分数
本实训通过构建一个从学习时间预测考试分数的线性回归模型来展示数据的可视化。

假设我们收集了8位同学的学习时间与考试分数，如下。如何构建一个从学习时间预测考试分数的模型呢？


| 学习时间 | 2  | 3  | 4  | 6  | 7  | 5  | 8  | 1  |
|----------|----|----|----|----|----|----|----|----|
| 考试分数 | 50 | 55 | 65 | 70 | 72 | 80 | 85 | 88 |

## 1. 导入所需库

- 首先需要导入所需的库：
    - `numpy`是用来做数值计算的基础库；
    - 从`sklearn.linear_model`引入`LinearRegression`模型;
    - 从`sklearn.metrics`引入`mean_squared_error`损失函数；
    - `matplotlib`是一个基于`Python`的开源绘图库。


```python
#导入所需库
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
import matplotlib.pyplot as plt
```

## 2. 准备数据集
- 假设我们收集了 8 名学生的学习时间和考试分数，并进行数据集的划分。
- 为了简化操作流程，我们在此省略验证集，仅将标注好的数据集划分为训练集和测试集。

    创建了两个 NumPy 数组：`X`表示学习时间（特征），`y`表示考试分数（目标变量）

    将数据分为训练集（前 6 个样本）和测试集（后 2 个样本）

    特征`X`需要是二维数组（即使只有一个特征），因为 scikit-learn 要求特征矩阵是二维的

```python
#数据集：学习时间（小时）和考试分数
X=np.array([[1],[2],[3],[4],[5],[6],[7],[8]])#学习时间
    
y=np.array([50,55,65,70,72,80,85,88])#考试分数
    
    
#划分数据集：训练数据和测试数据
X_train=X[:6]#前6个数据用于训练
y_train=y[:6]

X_test=X[6:]#后2个数据用于测试
y_test=y[6:]
```

数据集划分是机器学习中的重要步骤：

- 训练集：用于训练模型参数

- 测试集：用于评估模型性能

- 常见划分比例：70-80%训练，20-30%测试

- 本例采用简单的前后划分，实际中应随机划分

## 3. 构建并训练模型

1. 创建 LinearRegression 对象

2. 使用 fit()方法在训练数据上训练模型

3. 训练过程会计算最佳拟合直线的参数（斜率和截距）

```python
#创建线性回归模型并训练
model=LinearRegression()
model.fit(X_train,y_train)
```

## 4. 查看模型参数
- model.coef_：斜率（θ₁），表示学习时间每增加 1 小时，考试分数平均增加多少分

- model.intercept_：截距（θ₀），表示学习时间为 0 小时的预测分数

根据输出结果，回归方程为：
$$ 考试分数=44.73+5.89×学习时间$$

```python
#查看模型参数
print("Slope (theta_1):",model.coef_[0])
print("Intercept (theta_O):",model.intercept_)
```

## 5. 评估模型性能
使用训练好的模型在测试集上进行预测      
计算预测值(y_pred)与实际值(y_test)之间的均方误差(MSE)         
MSE 是回归模型常用的评估指标    
  
- MSE（均方误差）：$\text{MSE} = \frac{1}{n}\sum_{i=1}^{n}(y_i - \hat{y}_i)^2$
    - 衡量预测值与实际值之间的平均平方差
    - 值越小表示模型越好



```python

#用测试数据进行预测
y_pred=model.predict(X_test)
#计算均方误差（MSE）
mse = mean_squared_error(y_test,y_pred)
print("Mean Squared Error （MSE):",mse)


```

## 6. 可视化模型与数据点
- 使用散点图展示训练数据（圆圈）和测试数据（三角形）
- 生成 0-9 小时范围内的 100 个点
- 对这些点进行预测，绘制回归直线
- 添加坐标轴标签和图例

```python
# 绘制训练数据和测试数据的散点图
plt.scatter(X_train, y_train, marker='o', label='训练数据')
plt.scatter(X_test, y_test, marker='^', label='测试数据')

# 绘制回归直线
X_line = np.linspace(0, 9, 100).reshape(-1, 1)
y_line = model.predict(X_line)
plt.plot(X_line, y_line, color='green', label='回归直线')

# 添加标签和图例
plt.xlabel('学习时间(小时)')
plt.ylabel('考试分数')
plt.title('学习时间与考试成绩关系')
plt.legend()

# 添加网格线
plt.grid(True, linestyle='--', alpha=0.7)

# 在图中显示回归方程
equation = f'y = {model.intercept_:.2f} + {model.coef_[0]:.2f}x'
plt.text(0.5, 45, equation, fontsize=12, bbox=dict(facecolor='white', alpha=0.8))

plt.show()
```

## 7. 预测新学生的考试分数

- 创建新的输入数据（学习 5 小时）

- 使用训练好的模型进行预测

- 输出预测分数（保留两位小数）

```python
#预测新学生的考试分数
new_data=np.array([[5]])
predicted_score=model.predict(new_data)
print(f"Predicted exam score(after 5hours of study):{predicted_score[0]:.2f}")
```

本实训完整展示了线性回归的工作流程：

- 数据准备：收集数据并划分为训练集/测试集
- 模型训练：使用训练数据拟合回归直线
- 模型评估：计算 MSE 评估预测准确性
