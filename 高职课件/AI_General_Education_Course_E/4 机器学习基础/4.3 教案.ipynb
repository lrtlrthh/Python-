# 机器学习概述教案

## 1. 课时安排

**4 学时** 

## 2. 教学课型

理论课 [√]   
实验课 [√]    
习题课 [ ]    
实习课 [ ]    
其它    [ ]


## 3. 教学目标  

1. **知识掌握**：  
    - 说出机器学习与早期“规则灌输”方法的本质差异，理解经验知识 vs 数据知识  
    - 区分监督、无监督、强化学习三者的输入、输出与核心任务  
    - 复述监督学习完整流程：训练集-验证集-测试集、损失函数、过拟合  
    - 列举回归与分类常用损失函数各一种并写出公式  
    - 描述 K-means 的目标及算法步骤  
    - 列举强化学习五大元素：Agent、Environment、State、Action、Reward 

2. **能力提升**：  
    - 针对给定场景判断应使用哪类学习范式并说明理由  
    - 用伪代码写出 K-means 的一次迭代  
    - 用 NFL 定理批判“万能算法”幻想  
    - 估算一个简单监督任务所需的标注成本   

3. **素养培育**：  
    - 树立“数据即知识”的数据驱动思维  
    - 认识模型泛化失败可能带来的伦理风险（偏见、资源浪费）  
    - 培养“先审视数据质量，再谈模型优劣”的工程素养

## 4. 教学重点、难点  
重点：监督学习流程与损失函数、无监督聚类思想、强化学习元素  
难点：损失与泛化的关系、NFL 定理的启示、聚类评估的主观性

## 5. 教学方法

**教学方式**：讲授、讨论
    
**教学手段**：多媒体教学、案例分析



## 6. 预期学习成果

**预期学习成果**

- 能正确区分监督学习、无监督学习和强化学习三类学习范式，并理解其各自的核心任务和应用场景。
- 能独立写出均方误差（MSE）或交叉熵损失函数的公式，并解释其在模型训练中的作用。
- 能完成一次 K-means 聚类的手工迭代，并解释聚类结果的意义。
- 理解并掌握机器学习的基本概念，包括从数据中学习知识、经验知识表示、基于学习的人工智能等。
- 能够分析机器学习在实际应用中的优势与局限，并根据具体任务选择合适的学习范式。



## 7. 教学过程


**模块 1：机器学习基本概念 (25 分钟)**

- 基于知识的人工智能
- 经验知识表示
- 基于学习的人工智能
- 机器学习：从数据中学习知识
- 机器学习的概念与分类


**模块 2：监督学习 (25 分钟)**

- 监督学习是什么
- 监督学习的重要元素
- 监督学习的任务类型
- 监督学习的优势与局限


**模块 3：无监督学习 (25 分钟)**

- 无监督学习是什么
- 聚类
- 无监督学习的优势与局限


**模块 4：强化学习 (15 分钟)**

- 强化学习元素
- 示例
- 强化学习的优势与局限

**模块 5：实训-预测考试分数 ( 分钟)**

本实训通过构建一个从学习时间预测考试分数的线性回归模型来展示数据的可视化




## 8. 思考
  1. 简述人工智能、机器学习和深度学习之间的关系。
  2. 解释监督学习、无监督学习和强化学习的区别。
  3. 举例说明机器学习在实际生活中的应用。
