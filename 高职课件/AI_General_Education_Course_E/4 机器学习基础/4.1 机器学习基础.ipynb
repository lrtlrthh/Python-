# 机器学习基础



## 1. 机器学习基本概念

### 1.1 基于知识的人工智能

早期的人工智能学者倾向于将人的知识和思维方式“灌输”给机器，从而让机器拥有思维能力。这好像老师把知识总结成知识点，硬性地灌输给学生。这种方法在定理证明、专家系统等任务上取得了巨大成功。

然而，这种知识灌输方法存在很大局限性。首先人为总结各种知识非常繁琐，耗时耗力；其次机器只能在人类设计好的知识框架里活动，限制了能力的发挥。



### 1.2 经验知识表示

经验知识一般可用“如果…那么”这样的形式表示，例如：“如果天阴且湿度大，那么会下雨” 。这种知识表示方式称为产生式规则。

### 1.3  基于学习的人工智能



1959年，IBM 公司的工程师亚瑟•塞缪尔正式提出“机器学习”（machine learning）的概念，将其定义为“让机器具有不需要明确编程而具有的一种学习能力（the ability to learn without being explicitly programmed）”的研究，其目标是构造一种学习机器（learning machine），使之像人一样具有自我学习能力， 而非按部就班完成预设任务。

塞缪尔在西洋跳棋游戏中进行了验证。在实验中，只是告诉机器游戏的规则，走棋的大概思路，以及一些控制走棋方式的参数。塞缪尔发现，通过调整这些参数，机器用 8-10 个小时即可学习到超过程序设计者的棋艺。

亚瑟•塞缪尔（1901-1990）美国计算机学家，机器学习的奠基人。1959年，他在《IBM Journal of Research and Development》杂志上发表文章，介绍了通过学习使计算机学会下跳棋的方法。这篇文章被认为是机器学习的开端。

### 1.5 机器学习的概念与分类

机器学习通过对数据的优化学习，建立能够刻画数据中所蕴含语义概念或分布结构等信息的模型。在模型学习过程中，采用合适手段来利用有标签数据或无标签数据， 对模型参数不断进行优化，从而提升模型性能。

从数据利用的角度，可将机器学习划分为**监督学习（supervised learning）、无监督学习（unsupervised learning）及强化学习（reinforcement learning）**。

### 2.1 监督学习是什么
首先看一个监督学习的示例

> “监督学习是一种机器学习方法，其中模型通过学习带有标签的数据来预测未知数据的标签。”  
> —— 周志华，《机器学习》（西瓜书）

在监督学习中，我们有**输入特征**和**对应的输出标签**，目标是让模型学会从输入到输出的映射关系。

**监督学习的基本形式：**

- 输入：一个包含多个特征的数据点 $ x $
- 输出：一个对应的标签 $ y $
- 模型的目标：学习一个函数 $ f(x) \approx y $

我们可以将监督学习理解为一个“老师教学生”的过程，老师（训练数据）告诉学生（模型）每一个问题（输入）的正确答案（标签），学生不断调整自己的思维方式（模型参数），直到能够独立解决新问题。

### 2.2 监督学习的重要元素
- 标注数据:标识了类别信息的数据`学什么`
- 学习模型:如何学习得到映射模型`如何学`
- 损失函数:如何对学习结果进行度量`学到否`


      没有免费午餐定理（NFL）:Wolpert 和Macerday1997年在最优化理论中提出，指出“任何机器学习模型在所有问题上的性能都是相同的，其总误差和模型本身是没有关系的。一种算法（算法A）在特定数据集上的表现优于另一种算法（算法B）的同时，一定伴随着算法A在另外某一个特定的数据集上有着不如算法B的表现”

       Wolpert, D.H., Macready, W.G., No Free Lunch Theorems for Optimization, IEEE Transactions on Evolutionary Computation, 1997



 

#### 2.2.1 训练集 (Training Set)

训练集是监督学习的基础，它由一组**带标签的数据样本**组成。每个样本包括：
- **特征向量** $ x = (x_1, x_2, ..., x_n) $
- **标签** $ y $

例如，在房价预测任务中，训练集可能如下所示：

| 面积（平方米） | 卧室数量 | 地理位置评分 | 房价（万元） |
|----------------|----------|----------------|----------------|
| 80             | 2        | 7              | 450            |
| 100            | 3        | 9              | 600            |
| 70             | 1        | 5              | 380            |

训练集的作用是教会模型如何从输入特征中预测输出标签。

#### 2.2.2  损失函数

损失函数用来衡量模型预测值与真实标签之间的差距。常见的损失函数包括：

**（1）均方误差（Mean Squared Error, MSE）**

适用于回归任务：

$$
\text{MSE} = \frac{1}{N} \sum_{i=1}^{N}(y_i - \hat{y}_i)^2
$$

其中：
- $ y_i $ 是第 $ i $ 个样本的真实标签
- $ \hat{y}_i $ 是模型的预测值
- $ N $ 是样本总数

**（2）交叉熵损失（Cross Entropy Loss）**

适用于分类任务：

$$
\text{CE} = -\frac{1}{N} \sum_{i=1}^{N} [y_i \log(\hat{y}_i) + (1 - y_i)\log(1 - \hat{y}_i)]
$$

损失函数越小，说明模型的预测越准确。


将映射函数记为 $f$，第 $i$ 个训练数据记为 $(x_i, y_i)$，且 $f$ 对 $x_i$ 的预测结果记为 $\hat y_i$（即 $\hat y_i = f(x_i)$），可定义损失函数 $\text{Loss}(f(x_i), y_i)$ 来估量预测值 $\hat y_i$ 与真实值 $y_i$ 之间的差异。  
显然，在训练过程中希望映射函数在训练集上的累积差异最小，即  
$$\min \sum_{i=1}^{n} \text{Loss}(f(x_i), y_i)$$

#### 2.2.3 验证集 (Validation Set)

验证集是从训练集中划分出来的一小部分数据，用于：
- 调整模型超参数（如学习率、正则化系数等）
- 评估模型在未见过的数据上的表现
- 防止过拟合

通常我们会采用 **K折交叉验证（K-Fold Cross Validation）** 来更有效地利用数据。


#### 2.2.4 测试集 (Test Set)

测试集是完全不参与模型训练和调参的数据集，用于：
- 最终评估模型的泛化能力
- 判断模型是否可以部署到实际环境中

测试集的结果是最客观地反映模型性能的指标。

一旦在训练集上完成了模型参数优化后，需要在测试数据集上对模型性能进行测试。为了在训练优化过程中挑选更好的模型参数，一般可将训练集中一部分数据作为验证集（validation set）。在训练集上训练模型的同时会在验证集上对模型进行评估，以便得到最佳参数，最后在测试集上进行测试，将测试结果作为模型性能最终结果。         
要注意的是，训练集、验证集和测试集所包含数据之间没有任何交叉。可以说，测试集用于模型训练（好比学生的练习册）、验证集用于评估模型以调整相应参数（好比学生的模拟考卷或小测验）、测试集用于得到模型的优劣水平（好比真正考试）。

### 2.3 监督学习的任务类型

监督学习主要包括两类任务，回归任务和分类任务。

#### 2.3.1 回归任务（Regression Task）

在现实生活中，往往需要分析若干变量之间的关系，如碳排放量与气候变暖之间的关系、某一商品广告投入量与该商品销售量之间的关系等，这种分析不同变量之间存在关系的研究叫回归分析，刻画不同变量之间关系的模型被称为回归模型。如果这个模型是线性的，则称为线性回归模型。

一旦确定了回归模型，就可以进行预测等分析工作，如从碳排放量预测气候变化程度、从广告投入量预测商品销售量等。

<div class='insertItem' align=center><img src="https://imgbed.momodel.cn/hv/20250726134200567.png" width="600px"/></div>
</div>

父母平均身高每增加一个单位, 其成年子女平均身高只增加0.516个单位，它反映了这种“衰退(regression)”效应（“回归”到正常人平均身高）。
虽然x和y之间并不总是具有“衰退”（回归）关系，但是“线性回归”这一名称就保留了下来了。 



#### 2.3.2 分类任务（Classification Task）

基于样本的特征预测样本的类别。如基于大小和价格序测房子是别墅还是普通住宅，基于颜色和形状预测植物种类，基于形状预测数字，基于面部表情预测情绪。

- 若为桔子，则 $y_i = 0$。

### 2.4 监督学习的优势与局限
| 特性         | 优势                                       | 局限                                       |
|--------------|--------------------------------------------|--------------------------------------------|
| 数据需求     | 需要大量标注数据                           | 标注成本高                                 |
| 可解释性     | 某些模型（如线性回归、决策树）可解释性强   | 复杂模型（如神经网络）解释性差             |
| 泛化能力     | 在训练数据上表现良好                       | 容易过拟合，对新数据适应能力有限           |
| 应用场景     | 广泛应用于图像识别、自然语言处理等领域     | 不适用于无标签数据或动态环境               |


## 3. 无监督学习

### 3.1 无监督学习是什么
下图是无监督学习的一个例子，学习的目标是将水果进行归类。      
收集一些水果的图片，但并不知道图片中的水果的名字，因此不包含监督信息。
用这些图片训练一个模型，这一模型可以将相似的水果聚成一堆。这一过程称为“聚类”。       
将一幅没见过苹果图片送入模型，模型将它归入苹果一类。

> “无监督学习是一种机器学习方法，模型在没有标签的数据中寻找隐藏的结构或模式。”  
> —— 周志华，《机器学习》（西瓜书）

与监督学习不同，**无监督学习**不依赖于带标签的数据。它试图从原始数据中发现内在的结构、分布或关系。

**无监督学习的基本形式：**

- 输入：仅包含输入特征 $ x $
- 输出：模型自动发现数据中的结构，例如将数据分组、降维或检测异常

我们可以将无监督学习理解为“自己摸索”的过程：模型像一个观察者一样，尝试理解数据之间的相似性、差异性和潜在规律。

#### 3.2.1 K 均值聚类 (K-means 聚类)
- 物以类聚，人以群分(《战国策·齐策三》)
- 输入：n个数据（无任何标注信息）
- 输出：k个聚类结果
- 目的：将n个数据聚类到k个集合（也称为类簇）

- 若干定义：  
    - 给定 $n$ 个 $m$ 维数据 $\{x_1,x_2,\dots,x_n\}$，其中 $x_i\in\mathbb R^m\;(1\le i\le n)$。  
    - 两个 $m$ 维数据之间的欧氏距离定义为  
    $$
    d(x_i,x_j)=\sqrt{(x_{i1}-x_{j1})^2+(x_{i2}-x_{j2})^2+\dots+(x_{im}-x_{jm})^2}.
    $$  
    $d(x_i,x_j)$ 越小，表示 $x_i$ 与 $x_j$ 越相似；反之越不相似。  
    - 给定聚类数目 $k$
    
问题：  
如何将这 $n$ 个数据依据相似度大小分别聚类到 $k$ 个集合，使每个数据仅属于一个聚类集合？

### 3.3 无监督学习的优势与局限

| 特性         | 优势                                       | 局限                                       |
|--------------|--------------------------------------------|--------------------------------------------|
| 数据需求     | 不需要标注数据                             | 缺乏明确的目标指导                         |

## 4. 强化学习
我们想训练一只小狗，当说“苹果”或“桔子”时，它可以把正确的水果送过来。但是小狗听不懂我们说话。怎么办呢？

一种做法是用奖励信号代替监督信号来引导它主动学习。方法如下：如果它拿对了，就给它一块骨头做奖励；如果拿错了，就没有奖励。久而久之，它就可以听懂主人的命令了。
这是一种有别于监督学习和无监督学习的另一种学习方法，称为“强化学习”。

强化学习旨在让**智能体**（如机器人、游戏角色、下棋机器人等）通过与**环境**的互动学习，以达到最大化收益或最小化损失的目的。强化学习的核心思想是通过智能体与环境之间的互动，智能体可以学习出**最佳的行为策略**。
强化学习与其他机器学习方法（如监督学习、无监督学习等）的区别在于，强化学习通过智能体与环境的互动来学习，而其他方法通过训练数据来学习。




### 4.1 强化学习元素
- 智能体（agent）：智能体是强化学习算法的主体，它能够根据经验做出主观判断并执行动作，是整个智能系统的核心。
- 环境（environment）：智能体以外的一切统称为环境，环境在与智能体的交互中，能被智能体所采取的动作影响，同时环境也能向智能体反馈状态和奖励。
- 状态（state）：状态可以理解为智能体对环境的一种理解和编码，通常包含了对智能体所采取决策产生影响的信息。
- 动作（action）：动作是智能体对环境产生影响的方式。
- 策略（policy）：策略是智能体在所处状态下去执行动作的规则，即给定一个状态，智能体可根据一个策略来选择应该采取的动作。
- 奖励（reward）：奖励是智能体序贯式采取一系列动作后从环境获得的收益。

### 4.2 示例
强化学习是一种间接的监督学习，通过奖励让机器知道应该如何行动。强化学习广泛应用在棋牌游戏、机器人动作等任务。
在这些任务中，即时的监督信号很难获得，但可以通过行为产生的后果对模型进行引导，使得行为产生的整体收益最大。

### 4.3 强化学习的优势与局限

| 优势                                       | 局限                                       |
|--------------------------------------------|--------------------------------------------|
| 能处理复杂、动态的环境                     | 训练时间长，收敛困难                       |
| 可自动学习最优策略                         | 对超参数敏感                               |
| 适用于连续动作空间                         | 奖励设计复杂，容易出现稀疏奖励问题         |
| 可模拟人类试错学习的过程                   | 安全性和稳定性难以保障                     |
