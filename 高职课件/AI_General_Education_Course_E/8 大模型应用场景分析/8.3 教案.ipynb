# 大语言应用案例教案

## 1. 课时安排

**2 学时** 

## 2. 教学课型

理论课 [√]   
实验课  [√]    
习题课 [ ]    
实习课 [ ]    
其它    [ ]


## 3. 题目

大模型应用案例 - 课程设计与实践

## 4. 教学重点、难点

**教学重点**：大模型各个领域的应用、大模型的使用、利用大模型生成 PPT、利用大模型进行数据分析

**教学难点**：大模型工具的选择、如何灵活运用大模型工具进行实际案例分析

## 5. 教学方法

**教学方式**：理论讲授、案例分析、互动讨论

**教学手段**：互联网环境下的上机操作


## 6. 预期学习成果

**预期学习成果**

- 了解大模型的应用场景
- 能够进行简单的数据分析
- 理解大模型在实际项目中的作用




## 7. 教学过程

**模块 1：大模型应用概述 8分钟**

- 回顾旧课程内容
- 大模型在生活与工作中的实际应用概述

**模块 2：大模型应用中的主要考虑因素 7 分钟**

- 使用大模型的关键注意事项
- 大模型工具的分类及其特点分析


**模块 3：大模型在 PPT 制作中的应用 15 分钟**

- 生成 PPT 的大模型工具介绍与对比分析
- 使用讯飞智文进行 PPT 制作的实操演示

**模块 4：大模型应用场景实战——数据分析 25 分钟**

- 大模型工具在数据分析中的应用
- 数据分析的大模型工具介绍
- 使用小浣熊进行数据分析

**模块 5：总结回顾 5 分钟**

- 总结学习内容及应用场景



## 8. 作业

**思考题**：

- 使用大模型工具做数据分析要注意什么？
- 给出新的数据集，请应用大模型工具进行分析，并得出分析结论。（可以选择除小浣熊以外的其他分析工具，并总结出不同工具的优劣势）
