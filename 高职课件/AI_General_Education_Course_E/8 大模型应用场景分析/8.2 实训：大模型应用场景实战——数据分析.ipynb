# 大模型应用场景实战

## 1. 实验描述

### 1.1 数据分析的背景与重要性
1. 定义

    数据分析是通过对数据进行收集、清洗、处理和建模，从中提取有价值的信息，帮助决策者做出科学决策的过程。

2. 应用场景

    *   商业领域：通过分析销售数据，优化产品定价和营销策略。
    *   金融领域：通过分析市场数据，预测股票走势和风险评估。
    *   医疗领域：通过分析患者数据，辅助疾病诊断和治疗方案制定。
    *   教育领域：通过分析学生成绩数据，优化教学方法和课程设计。

3. 价值

    *   帮助企业降低成本、提高效率。
    *   发现潜在的市场机会和业务增长点。
    *   支持科学决策，减少主观判断的误差。

### 1.2 大模型在数据分析中的应用

  大模型是指基于深度学习技术构建的、具有海量参数的人工智能模型，如 GPT、BERT 等。这些模型能够处理复杂的非结构化数据（如文本、图像、语音等），并从中提取深层次的信息。

1. 大模型的优势：

    *   自动化：能够自动完成数据清洗、特征提取和模型训练等任务。
    *   高效性：能够快速处理大规模数据集，节省人力和时间成本。
    *   智能化：能够从数据中发现复杂的模式和规律，提供更精准的分析结果。

2. 大模型在数据分析中的典型应用

    *   文本分析：通过自然语言处理（NLP）技术，分析客户评论、社交媒体数据等。
    *   图像分析：通过计算机视觉技术，分析医学影像、工业检测图像等。
    *   预测分析：通过时间序列分析，预测未来的销售趋势、市场需求等。

下面我们用 python 代码来验证生成的图表是否正确：

```python
# 数据已经成功读取，接下来需要对数据进行处理，提取出订单创建时间、总金额和退款金额，并按日期汇总。
# 将订单创建时间列转换为日期格式，方便后续按天汇总
data['订单创建时间'] = pd.to_datetime(data['订单创建时间'], errors='coerce')

# 提取2月份的数据
data_feb = data[data['订单创建时间'].dt.month == 2]

# 打印处理后的数据，检查是否正确
data_feb[['订单创建时间', '总金额', '退款金额']].head()

# 先将总金额和退款金额列转换为数值类型
data_feb['总金额'] = pd.to_numeric(data_feb['总金额'], errors='coerce')
data_feb['退款金额'] = pd.to_numeric(data_feb['退款金额'], errors='coerce')

# 按天汇总总金额和退款金额
daily_sales_refunds = data_feb.groupby(data_feb['订单创建时间'].dt.date)[['总金额', '退款金额']].sum()

# 打印汇总结果
daily_sales_refunds.head()
```

```python
import matplotlib.pyplot as plt

# 创建一个图形
plt.figure(figsize=(10, 6))

# 绘制总金额的折线图
plt.plot(daily_sales_refunds.index, daily_sales_refunds['总金额'], label='总金额', marker='o')

# 绘制退款金额的折线图
plt.plot(daily_sales_refunds.index, daily_sales_refunds['退款金额'], label='退款金额', marker='o')

# 添加标题和标签
plt.title('2月份每天的总金额和退款金额')
plt.xlabel('日期')
plt.ylabel('金额（元）')

# 显示图例
plt.legend()

# 旋转x轴标签以便更好地显示
plt.xticks(rotation=45)

# 显示图表
plt.tight_layout()
plt.show()
```

可以发现执行出来的结果与数据分析小伙伴的结果一致。

```python
# 发现“订单付款时间”列名后面有多余的空格，修正列名后再继续处理
data_feb.columns = data_feb.columns.str.strip()

# 重新处理订单付款时间列，将其转换为日期格式
data_feb['订单付款时间'] = pd.to_datetime(data_feb['订单付款时间'], errors='coerce')

# 计算每天创建订单的人数（按订单创建时间分组，统计唯一的订单编号）
daily_order_creators = data_feb.groupby(data_feb['订单创建时间'].dt.date)['订单编号'].nunique()

# 计算每天订单付款的人数（按订单付款时间分组，统计唯一的订单编号）
daily_order_payers = data_feb.groupby(data_feb['订单付款时间'].dt.date)['订单编号'].nunique()

# 计算创建订单人数与付款人数的差值
daily_difference = daily_order_creators - daily_order_payers.reindex(daily_order_creators.index, fill_value=0)

# 计算利润 = 买家实际支付金额 - 退款金额
daily_profit = data_feb.groupby(data_feb['订单创建时间'].dt.date).apply(
    lambda x: (x['买家实际支付金额'] - x['退款金额']).sum()
)

# 合并所有计算结果到一个DataFrame
daily_summary = pd.DataFrame({
    '创建订单人数': daily_order_creators,
    '订单付款人数': daily_order_payers.reindex(daily_order_creators.index, fill_value=0),
    '差值': daily_difference,
    '利润': daily_profit
})

daily_summary
```

1.   使用大模型工具做数据分析要注意什么？
    - 提示词技巧：需要精确的说出需求，展现形式、计算方法等
    - 需要注意数据安全！重要数据不能上传给大模型，避免数据泄露，可以做数据脱敏
    - 脱敏技巧：
      - 替换敏感信息：将姓名、地址、电话号码等敏感信息替换为非敏感的标识符，如使用“用户 1 ”等代替真实姓名。
      - 数据随机化：使用随机生成的数据替代真实数据，但保持原有的数据分布特性。
