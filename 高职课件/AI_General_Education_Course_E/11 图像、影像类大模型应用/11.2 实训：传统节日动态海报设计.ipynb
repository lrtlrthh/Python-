# 图像、影像类大模型应用 - 传统节日动态海报设计

## 教学目标
- 体验图像、影像类大模型的实际应用场景
- 掌握与大模型的交互方式，以及如何将大模型应用于实际的图像生成，特定视频生成。

## 教学方法
- **实践操作**：通过亲身使用图像、影像类大模型，思考在与大模型交互过程中需要注意的地方。

- **讨论与反思**：根据大模型生成效果，思考当下图像、影像类大模型还存在什么局限。

## 教学内容

### 1. 项目背景


随着 DALL·E 、MidJourney 、即梦 AI 等图像类大模型的崛起，这些技术正以前所未有的速度重塑现代设计创作的方式。从降低创作门槛到激发艺术灵感，它们极大地拓展了视觉表达与设计的边界。

本次实训将以即梦 AI 为核心，结合其动态视频生成功能，结合 DeepSeek 的提示词设计能力，打造"传统节日 - 春节 - 数字焕新计划"。学生将通过"文生图→图生视频"的完整工作流，掌握节日文化视觉设计的现代化表达技巧。

借助即梦 AI 强大的图像生成能力，学生们将在实训过程中获得以下收获：

- 结合文字描述生成高质量海报，帮助学生掌握从概念到视觉呈现的全流程设计方法。

- 根据用户的个性化需求调整生成参数，确保作品风格与主题的高度契合。

- 提供多种优化方案，对生成的图像进行评估，指出其中的不足，并给出切实可行的改进建议。

- 通过持续交互优化，逐步提升学生的视觉表达能力，帮助学生突破创作瓶颈，提升作品的整体质量。

- 掌握“图生视频”核心技术，将 AI 生成的静态节日海报转化为动态短片（如灯笼飘动、烟花绽放等特效）。

- 在实训过程中灵活使用多类型的大模型，图像大模型与文本类大模型结合，实现创意与技术的完美结合。

通过这些实训，学生不仅能快速掌握节日视觉设计技能，还能深入理解 AI 工具在创意领域的实际应用，为未来的艺术创作和职业发展提供强大助力。

### 2. 实训步骤

#### 2.1 文生图


##### 2.1.1 思路设计

问题分析：针对海报设计思路，需重点考虑以下三方面：

1. 核心要素：确定海报需要包含的关键元素

2. 视觉风格：统一字体、配色及背景设计风格

3. 版式布局：优化各元素位置与比例，提升整体视觉效果

    Prompt：
    现在请你作为一个资深平面设计师，为 2026 春节主题的商业海报分析以下内容：1. 海报的核心要素有哪些2. 海报的视觉风格是怎样的3. 海报的版式布局是怎样的 ？伴随返还你的设计思路。
    我希望的风格 ：体现阖家欢乐的主题，要有过年的气息，背景可以是农村雪天，要有马（年）元素。

#### 2.2 图生视频

##### 2.2.1 思路设计

首先我们选择一个刚才生成的比较满意的图片。

<div class='insertContainerBox column'>

问题分析 ：现在我们要做的是将当前这张图片进行动态化处理，使其能够在一定时间内进行动态的变化，从而达到更好的视觉效果，那我们便需要考虑以下方面：
1. 动作设计 ： 我们需要设计合适的动作来控制对象的变化，例如奔跑，跳跃等。我们需要根据具体的需求来选择合适的动作。
2. 动画效果 ： 我们需要设计合适的动画效果来增强对象的视觉效果，例如闪烁，缩放等。我们需要根据具体的需求来选择合适的动画效果。
3. 动作的合理化时间分配 ： 我们需要合理地分配动作的时间，使得对象的变化能够在一定时间内完成。我们需要根据具体的需求来选择合适的时间分配。

    Prompt：
    我现在有这样一张图片，我准备对他进行动态化处理（生成一个五秒的视频），请你按照以下两个方面进行设计：1.动作设计 2.动画效果，并返还你的设计思路。
