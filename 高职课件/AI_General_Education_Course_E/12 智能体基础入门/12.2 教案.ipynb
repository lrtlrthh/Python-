## 3. 题目

智能体基础入门

## 4. 教学重点、难点

**教学重点**：智能体的定义、组成、核心能力及基于大模型的架构与应用。

**教学难点**：将复杂技术原理通俗化，结合实际案例帮助理解。



## 5. 教学方法

**教学方式**：讲授、讨论
    
**教学手段**：多媒体教学、案例分析



## 6. 预期学习成果

**预期学习成果**


- 理解智能体的定义和组成，包括感知器、决策系统、执行器等核心组件。
- 掌握智能体的核心能力，如大模型（LLM）、思考（Brain）、感知（Perception）、行动（Action）。
- 熟悉基于大模型的智能体架构，包括用户请求、智能体/大脑、规划、记忆等模块。
- 能够将智能体的知识应用到实际场景中，理解其在不同领域的应用潜力。




## 7. 教学过程

**模块 1：智能体的定义（30 分钟）**

- 智能体初认知
- 技术定义
- 智能体的研究示例


**模块 2：智能体的核心能力（30 分钟）**

- 大模型
- 思考
- 感知
- 行动

**模块 3：多层神经网络（30 分钟）**

- LLM Agent 结构
- 智能体的大脑
- 智能体的规划
- 智能体的定义

## 8. 作业

**思考题**：

- 智能体的哪些核心能力使其能够在复杂环境中有效运作？请结合实例说明。
