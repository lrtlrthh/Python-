# 数字图像基本概念


## 1. 图片的本质：数字的集合

我们每天用手机拍照、看电脑屏幕，这些彩色的、栩栩如生的图片，在计算机里究竟是什么样子呢？它们不是真的“画”，而是一串串“数字”！
*   **像素 (Pixel)**：
    *   **概念**：图片就像由无数个小方块拼成的马赛克。每一个这样的小方块，
        我们称之为“像素”(Pixel)，是图像的最小单位。
    *   **可视化**：当图片被放大很多倍时，你就会看到这些小方块，每个小方块都只有一种纯色。

## 2. 像素矩阵：图片如何排列

这些小像素点并不是随意排列的，它们像棋盘一样，整齐地排列成一个“矩阵”（或者叫网格）。每一行、每一列都有对应的位置，就像坐标系一样。

 **思考**：图片的分辨率（比如1920x1080）代表了什么？

分辨率是指数字图像中**像素点的数量**，通常用“宽度 × 高度”来表示。例如，一张1920x1080像素的图片意味着它在水平方向上有1920个像素，垂直方向上有1080个像素。分辨率越高，图像包含的像素点越多，细节就越丰富，图像也就越清晰。

## 3. 颜色模式：像素的色彩密码 (以 RGB 为例)

一个像素为什么能显示出各种各样的颜色呢？
*   **RGB 模式**：
    *   **概念**：最常见的颜色模式是 RGB，它代表红（Red）、绿（Green）、蓝（Blue）三原色。
        就像我们小时候学过的三原色光一样，通过不同比例的红、绿、蓝光混合，就可以创造出
        几乎所有的颜色。
    *   **数值表示**：每个像素的颜色，就是由这三种颜色的亮度值组成的。
        每个颜色通道（R、G、B）通常用0到255的数字表示，0表示该颜色最暗（没有），
        255表示该颜色最亮（最大）。
    *   **举例**：
        *   (255, 0, 0) 表示纯红色。
        *   (0, 255, 0) 表示纯绿色。
        *   (0, 0, 255) 表示纯蓝色。
        *   (0, 0, 0) 表示黑色（没有光）。
        *   (255, 255, 255) 表示白色（所有光混合）。
        *   (128, 0, 128) 表示紫色（红色和蓝色中等亮度混合）。
    *   **可视化**：


接下来我们使用 Python 代码来理解一下 RGB。


### 3.1 加载并显示一张图像
我们将使用示例图片为一张苹果照片，他已经放在了我们的文件夹中，直接加载就行。


````python
from PIL import Image
import os
from IPython.display import display
import warnings                                                                                                     
warnings.filterwarnings("ignore") # 忽略一些不必要的警告                                                            
                                                          

image_path = './1.jpg' 

# 检查文件是否存在
if not os.path.exists(image_path):
    print(f"错误：图片文件 '{image_path}' 不存在。请检查路径或使用您自己的图片路径。")
else:
    # 加载图像
    img = Image.open(image_path)

    # 显示图像
    print("原始图片：")
    display(img) 

    print(f"\n图片成功加载并显示。")
````

### 3.2 像素点操作：获取特定像素的 RGB 值

我们将获取图片左上角 (0,0) 位置的像素颜色值。

````python
width, height = img.size
# 获取 (0,0) 坐标的像素值
# 注意：图像坐标系通常以左上角为 (0,0)，x轴向右，y轴向下
pixel_rgb = img.getpixel((0, 0))
print(f"\n图片左上角 (0,0) 像素的 RGB 值: {pixel_rgb}")

# 如果图片模式不是RGB（例如是灰度图'L'），getpixel会返回一个整数
if isinstance(pixel_rgb, int):
    print(f"该图片是灰度图，像素值为亮度值。")
else:
    # 进一步解析 RGB 值
    r, g, b = pixel_rgb
    print(f"R (红色) 分量: {r}")
    print(f"G (绿色) 分量: {g}")
    print(f"B (蓝色) 分量: {b}")

# 尝试获取图片中心点像素的 RGB 值
center_x = width // 2
center_y = height // 2
center_pixel_rgb = img.getpixel((center_x, center_y))
print(f"\n图片中心 ({center_x},{center_y}) 像素的 RGB 值: {center_pixel_rgb}")
````

### 3.3 颜色通道分离与可视化

我们将把一张 RGB 图像分离成红、绿、蓝三个通道的灰度图，帮助理解每个通道对最终颜色的贡献。




````python
from PIL import Image
import matplotlib.pyplot as plt

# 确保图片是RGB模式
if img.mode != 'RGB':
    img = img.convert('RGB')

# 分离通道并转换为灰度图
r_channel = img.split()[0].convert('L')
g_channel = img.split()[1].convert('L')
b_channel = img.split()[2].convert('L')

# 使用matplotlib展示图像
plt.figure(figsize=(15, 5))

plt.subplot(1, 4, 1)
plt.imshow(r_channel, cmap='gray')
plt.title('Red Channel (Grayscale)')
plt.axis('off')

plt.subplot(1, 4, 2)
plt.imshow(g_channel, cmap='gray')
plt.title('Green Channel (Grayscale)')
plt.axis('off')

plt.subplot(1, 4, 3)
plt.imshow(b_channel, cmap='gray')
plt.title('Blue Channel (Grayscale)')
plt.axis('off')

plt.subplot(1, 4, 4)
plt.imshow(img.convert('L'), cmap='gray')
plt.title('Original Image (Grayscale)')
plt.axis('off')

plt.show()
````

通过观察，你可以看到在红色通道图中，红色区域会更亮；在蓝色通道图中，蓝色区域会更亮；在绿色通道图中，绿色区域会更亮。
