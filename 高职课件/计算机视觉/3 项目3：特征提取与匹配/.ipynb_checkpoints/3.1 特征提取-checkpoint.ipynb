# 特征提取



在计算机视觉中，**特征**（Feature）是图像中具有代表性和可区分性的点、线或区域。它们就像是图像的“指纹”或“关键信息点”。
通过对图像特征的提取，可以将图像转换为可处理的数字形式，从而使计算机能够理解和处理图像。

**为什么需要特征？**

当你要识别一张照片中的物体。你不会去看照片中的每一个像素，而是会关注一些特别的地方，比如物体的轮廓、角点、特殊的纹理等。计算机也是一样，通过提取特征，我们可以：

1.  **简化数据**：将一张巨大的图像（几百万个像素）简化为几百个特征点。
2.  **实现鲁棒性**：好的特征在图像发生旋转、缩放、光照变化时，仍然能够被识别出来。
3.  **应用广泛**：特征是实现图像识别、目标跟踪、图像拼接、三维重建等高级应用的基础。

**好的特征应具备的特点：**

| 特点 | 描述 |
| :--- | :--- |
| **可重复性** (Repeatability) | 即使图像发生变化（如视角、光照），该特征点仍能被检测到。 |
| **可区分性** (Distinctiveness) | 每个特征点周围的区域都应该足够独特，以便与其他特征点区分开。 |
| **局部性** (Locality) | 特征只依赖于其周围的小区域，这使得它对图像遮挡不敏感。 |
| **数量适中** (Quantity) | 数量不能太少（信息不足），也不能太多（计算量过大）。 |

## 1. 边缘检测与角点检测

在计算机视觉中，边缘和角点是图像中具有重要意义的特征。

关于边缘检测的内容在上一节中我们已经讲解过，本节课的重点我们将放在角点检测上。

### 1.1 边缘检测
边缘是图像中灰度值明显变化的地方，可以用来表示图像中物体的边界信息。
在计算机视觉中，常用的边缘检测算法包括 Sobel 算法、Canny 算法等。

### 1.2 角点检测
**角点**（Corner）是比边缘更稳定的特征。它通常是两条或多条边缘的交点，在图像中具有很高的局部独特性。

**角点的特点：**

*   **平坦区域**：在任何方向上移动小窗口，灰度变化都很小。
*   **边缘区域**：沿着边缘方向移动小窗口，灰度变化很小；垂直于边缘方向移动，灰度变化很大。
*   **角点区域**：在**任何方向**上移动小窗口，灰度变化都很大。

#### 1.2.1 Harris 角点检测原理

Harris 角点检测是最早、最著名的角点检测算法之一。它的核心思想是：**利用一个滑动窗口，计算窗口内像素灰度值在各个方向上的变化量。**

1.  **计算灰度变化**：通过计算图像的梯度（$I_x$ 和 $I_y$），来衡量窗口在 $x$ 和 $y$ 方向上移动时的灰度变化。
2.  **构建结构张量**：利用梯度信息构建一个 $2 \times 2$ 的矩阵 $M$（也叫结构张量）。
3.  **计算响应函数 $R$**：通过矩阵 $M$ 的特征值 $\lambda_1$ 和 $\lambda_2$ 来计算一个角点响应值 $R$。
    $$R = \det(M) - k \cdot (\text{Trace}(M))^2$$
    其中 $\det(M) = \lambda_1 \lambda_2$，$\text{Trace}(M) = \lambda_1 + \lambda_2$。
    *   如果 $R$ 值很大且为正，说明 $\lambda_1$ 和 $\lambda_2$ 都很大，是**角点**。
    *   如果 $R$ 值很小（接近于零），说明 $\lambda_1$ 或 $\lambda_2$ 至少有一个很小，是**平坦区域**或**边缘**。
4.  **阈值化**：设置一个阈值，将 $R$ 值大于阈值的点标记为角点。



Harris 角点检测算法基于图像的灰度值变化来判断是否为角点。该算法计算每个像素点周围邻域的自相关矩阵，并用矩阵的特征值来判断该点是否为角点。

如果特征值都比较大，则该点为角点；如果只有一个特征值比较大，则该点为边缘点；如果两个特征值都比较小，则该点为平滑区域。

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# 1. 读取图像
image_path = "1.jpg"  # 替换为你的图片路径
img = cv2.imread(image_path)

# 确保图像存在
if img is None:
    print("无法加载图像，请检查路径是否正确！")
    exit()

# 2. 转换为灰度图像（Harris 角点检测需要灰度图像）
gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# 3. 转换为浮点型（Harris 需要）
img_float = np.float32(gray_img)

# 4. Harris 角点检测参数
block_size = 2  # 邻域大小
ksize = 3       # Sobel 导数的光圈参数
k = 0.04        # Harris 参数

# 5. 应用 Harris 角点检测
dst = cv2.cornerHarris(img_float, block_size, ksize, k)

# 6. 对检测结果进行阈值处理，突出显示角点
# 提高阈值，减少角点数量
threshold = 0.1 * dst.max()  

# 获取角点的坐标
coords = np.argwhere(dst > threshold)

# 在原图上绘制较大的圆形标记角点
for (y, x) in coords:
    cv2.circle(img, (x, y), radius=5, color=(0, 0, 255), thickness=-1)  # 半径为5的红色圆点

# 7. 显示结果
# 使用 Matplotlib 显示图像
plt.figure(figsize=(8, 6))
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))  # 将 BGR 转换为 RGB
plt.title("Harris Corner Detection")
plt.axis("off")  # 关闭坐标轴
plt.show()
```

上面这个图我们可以看到一些红点，这些红点就是使用 Harris 检测出的角点。

#### 1.2.2 Shi-Tomasi 角点检测算法

Shi-Tomasi 角点检测算法是对 Harris 角点检测算法的改进，它计算每个像素点周围邻域的最小特征值，并用最小特征值来判断该点是否为角点。该算法比 Harris 算法更加稳健，能够更好地处理图像中的噪声和图像变换。

Shi-Tomasi 算法通过计算每个像素点的自相关矩阵（也称为协方差矩阵）来评估该点是否为角点。

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# 1. 读取图像
image_path = "1.jpg"  # 替换为你的图片路径
img = cv2.imread(image_path)

# 确保图像存在
if img is None:
    print("无法加载图像，请检查路径是否正确！")
    exit()

# 2. 转换为灰度图像（Shi-Tomasi 角点检测需要灰度图像）
gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# 3. Shi-Tomasi 角点检测参数
max_corners = 100  # 最大角点数量
quality_level = 0.1  # 提高角点检测的质量水平（阈值）
min_distance = 10  # 角点之间的最小欧几里得距离
block_size = 3  # 邻域大小
use_harris = False  # 是否使用 Harris 角点检测（False 表示使用 Shi-Tomasi）
k = 0.04  # Harris 参数（如果 use_harris=True）

# 4. 应用 Shi-Tomasi 角点检测
corners = cv2.goodFeaturesToTrack(gray_img, maxCorners=max_corners, qualityLevel=quality_level,
                                  minDistance=min_distance, blockSize=block_size, useHarrisDetector=use_harris, k=k)

# 5. 绘制角点
if corners is not None:
    corners = np.int32(corners)  # 转换为整数坐标
    for corner in corners:
        x, y = corner.ravel()  # 获取角点坐标
        cv2.circle(img, (x, y), radius=5, color=(0, 255, 0), thickness=-1)  # 用绿色圆圈标记角点

# 6. 显示结果
# 使用 Matplotlib 显示图像
plt.figure(figsize=(8, 6))
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))  # 将 BGR 转换为 RGB
plt.title("Shi-Tomasi Corner Detection")
plt.axis("off")  # 关闭坐标轴
plt.show()
```

同样的，上面图片中的绿点就是检测出的角点。

#### 1.2.3 FAST 角点检测算法
FAST 全称 Features from accelerated segment test,一种用于角点检测的算法，该算法的原理是取图像中检测点，以该点为圆心的周围的16个像素点判断检测点是否为角点。通俗的讲就是中心的的像素值比大部分周围的像素值要亮一个阈值或者暗一个阈值则为角点。


```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# 1. 读取图像
image_path = "1.jpg"  # 替换为你的图片路径
img = cv2.imread(image_path)

# 确保图像存在
if img is None:
    print("无法加载图像，请检查路径是否正确！")
    exit()

# 2. 转换为灰度图像（FAST 角点检测需要灰度图像）
gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# 3. 初始化 FAST 检测器
fast = cv2.FastFeatureDetector_create()

# 4. 设置 FAST 检测器的参数
fast.setNonmaxSuppression(True)  # 启用非极大值抑制
fast.setThreshold(50)  # 提高亮度差阈值，减少角点数量

# 5. 检测角点
keypoints = fast.detect(gray_img, None)

# 6. 绘制角点
img_with_keypoints = cv2.drawKeypoints(img, keypoints, None, color=(0, 255, 0))

# 7. 显示结果
plt.figure(figsize=(8, 6))
plt.imshow(cv2.cvtColor(img_with_keypoints, cv2.COLOR_BGR2RGB))  # 将 BGR 转换为 RGB
plt.title("FAST Corner Detection")
plt.axis("off")  # 关闭坐标轴
plt.show()
```

上方图片中的空心绿圈，就是检测出的角点。

#### 1.2.4 Harris 角点检测、Shi-Tomasi 角点检测和 FAST 角点检测的比较
| 特性        | Harris 角点检测                | Shi-Tomasi 角点检测                                                       | FAST 角点检测                            |
| --------- | -------------------------- | --------------------------------------------------------------------- | ------------------------------------ |
| **算法原理**  | 基于自相关矩阵的特征值，检测局部变化最大的点。    | 基于最小特征值，选择局部变化最大的方向上的角点。                                              | 检测一个像素周围的圆环区域，判断是否为角点。               |
| **角点质量**  | 对角点的定义较为严格，适合检测强角点。        | 更注重角点的稳定性和可靠性，适合检测弱角点。                                                | 检测速度快，但角点质量可能不如 Harris 和 Shi-Tomasi。 |
| **参数调整**  | 需要调整阈值和邻域大小（`block_size`）。 | 需要调整质量水平（`quality_level`）、最大角点数量（`max_corners`）和最小距离（`min_distance`）。 | 主要参数是亮度差阈值（`threshold`）和是否启用非极大值抑制。  |
| **角点数量**  | 阈值较高时，角点数量较少；降低阈值可增加角点数量。  | 通过调整 `quality_level` 和 `max_corners` 控制角点数量。                          | 提高阈值减少角点数量，降低阈值增加角点数量。               |
| **计算复杂度** | 计算较为复杂，需要计算自相关矩阵的特征值。      | 计算相对简单，直接基于最小特征值。                                                     | 计算非常快速，适合实时应用。                       |
| **适用场景**  | 适用于对角点质量要求较高的场景，如特征匹配。     | 适用于需要检测大量角点的场景，如光流跟踪。                                                 | 适用于需要快速检测角点的实时应用，如视频处理。              |
| **优点**    | 角点质量高，适合复杂图像。              | 稳定性强，适合检测弱角点。                                                         | 速度快，适合实时应用。                          |
| **缺点**    | 计算复杂，速度较慢。                 | 参数较多，调整复杂。                                                            | 角点质量可能较低，依赖于阈值。                      |


## 2. 特征描述符 (Descriptor)

仅仅找到特征点（如角点）是不够的，我们还需要一个“身份证”来描述这个特征点周围的区域，这就是**特征描述符**（Feature Descriptor）。

**特征描述符的作用：**

*   它是一个**向量**，用于量化描述特征点周围的局部图像信息（如梯度方向、像素分布等）。
*   它必须具有**独特性**，以便在另一张图像中找到与之匹配的特征点。
*   它必须具有**不变性**，即图像发生旋转、缩放、光照变化时，描述符的值应该保持不变或变化很小。


### 2.1 ORB 描述符

ORB（Oriented FAST and Rotated BRIEF）是一种高效且性能优异的特征提取和描述算法。它结合了：

*   **FAST**：用于快速检测关键点（Keypoint）。
*   **BRIEF**：用于生成紧凑的二值描述符（Descriptor）。

ORB 描述符的特点是：**计算速度快**，并且生成的描述符是**二值**的（只包含 0 和 1），这使得后续的匹配速度也非常快。

**主要流程是这样的：**             
1. 关键点检测：ORB 算法先用 FAST 算法在图像里找出一些重要的点，这些点就是关键点，就像在图片中找到一些特别显眼的位置。
2. 特征描述：找到关键点后，用 BRIEF 算法给每个关键点“画个像”，生成一个特征描述子，方便后续比较这些点。
3. 提高鲁棒性：为了让算法在图像旋转后也能认出关键点，ORB 给每个关键点定了一个方向。在描述特征时，把描述子按照这个方向调整一下，这样不管图像怎么旋转，描述子都能保持一致，这就是旋转不变性。

总结一下，ORB 算法就是先找关键点，再给关键点“画像”，并且通过方向调整让这个“画像”在旋转后也能被认出来，这样算法就更稳定了。

```python
import cv2
import matplotlib.pyplot as plt

# 读取图像
image = cv2.imread('1.jpg', cv2.IMREAD_GRAYSCALE)  # 替换为你的图像路径
if image is None:
    print("无法加载图像，请检查路径是否正确！")
    exit()

# 初始化 ORB 检测器，提高阈值
# 默认阈值为20，提高阈值可以减少检测到的关键点数量
orb = cv2.ORB_create(nfeatures=500, edgeThreshold=15, patchSize=31, fastThreshold=30)

# 检测关键点并计算描述子
keypoints, descriptors = orb.detectAndCompute(image, None)

# 绘制关键点及其方向箭头
image_with_keypoints = cv2.drawKeypoints(image, keypoints, None, color=(0, 255, 0), flags=cv2.DRAW_MATCHES_FLAGS_DRAW_RICH_KEYPOINTS)

# 使用 matplotlib 显示图像
plt.figure(figsize=(10, 6))
plt.imshow(image_with_keypoints, cmap='gray')  # 使用灰度图显示
plt.title('ORB Key Points with Direction')
plt.axis('off')  # 关闭坐标轴
plt.show()
```

**结果解释：**
- 绿色圆圈：表示检测到的关键点位置，数量减少。
- 方向箭头：表示关键点的方向，箭头方向与图像局部结构相关。
- 箭头长度：表示关键点的尺度大小，较长的箭头表示关键点周围的特征区域较大。

**ORB 算法的关键点和描述子**
- 关键点检测（FAST）：ORB 算法使用 FAST 算法来检测关键点。FAST 算法通过检测图像中的角点来快速找到关键点，这些角点通常是图像中变化最显著的地方。
- 描述子提取（BRIEF）：检测到关键点后，ORB 算法使用 BRIEF 算法来提取每个关键点的描述子。BRIEF 是一种非常高效的描述子提取方法，它通过比较关键点周围像素对的亮度差异来生成描述子。

### 2.2 SIFT 特征检测
SIFT（Scale-Invariant Feature Transform，尺度不变特征变换）是一种非常经典的特征检测和描述算法，广泛应用于计算机视觉领域，如图像匹配、目标识别和三维重建等。

SIFT 算法能够检测到图像中的关键点，并为每个关键点生成一个描述子，这些描述子具有尺度不变性和旋转不变性。

**尺度不变性**
- 尺度不变性是指特征检测和描述算法能够在不同尺度（大小）的图像中检测到相同的特征点，并生成一致的描述子。这意味着即使图像被放大或缩小，算法仍能识别出相同的特征点，从而保证特征匹配的准确性。  

**旋转不变性**
- 旋转不变性是指特征检测和描述算法能够在旋转后的图像中检测到相同的特征点，并生成一致的描述子。这意味着无论图像如何旋转，算法都能识别出相同的特征点，从而保证特征匹配的准确性。

```python
import cv2
import matplotlib.pyplot as plt

# 读取图像
image = cv2.imread('1.jpg', cv2.IMREAD_GRAYSCALE)  # 替换为你的图像路径
if image is None:
    print("无法加载图像，请检查路径是否正确！")
    exit()

# 初始化 SIFT 检测器
sift = cv2.SIFT_create()

# 检测关键点并计算描述子
keypoints, descriptors = sift.detectAndCompute(image, None)

# 绘制关键点及其方向箭头
image_with_keypoints = cv2.drawKeypoints(image, keypoints, None, color=(0, 255, 0), flags=cv2.DRAW_MATCHES_FLAGS_DRAW_RICH_KEYPOINTS)

# 使用 matplotlib 显示图像
plt.figure(figsize=(10, 6))
plt.imshow(cv2.cvtColor(image_with_keypoints, cv2.COLOR_BGR2RGB))  # 转换为 RGB 格式以正确显示颜色
plt.title('SIFT Key Points with Direction')
plt.axis('off')  # 关闭坐标轴
plt.show()
```

通过运行 SIFT 特征检测代码，结果展示了图像中的关键点及其方向箭头。这些关键点具有以下特点：
- 绿色圆圈：表示关键点的位置和尺度，较大的圆圈表示更显著的特征区域。
- 方向箭头：表示关键点的方向，帮助实现旋转不变性。
- 分布稀疏：集中在图像中最显著的特征区域。
- 描述子：虽然不可见，但每个关键点都有一个 128 维的描述子，用于后续的匹配和识别。

## 3. 其他检测算法

除了 Harris 角点检测、Shi-Tomasi、FAST 、ORB 算法和 SIDT 算法外，还有一些其他常用的特征点检测和描述算法，例如：KAZE、AKAZE、BRISK、FREAK 等。这些算法各具特点，应根据实际应用场景选择合适的算法。

| **算法**          | **检测方法**            | **描述方法**            | **特点**                      | **适用场景**                       |
| --------------- | ------------------- | ------------------- | --------------------------- | ------------------------------ |
| **Harris 角点检测** | 基于自相关矩阵的角点检测        | 无（通常仅用于检测）          | 简单高效，对光照变化敏感，无尺度不变性。        | 图像配准、目标检测（简单场景）。               |
| **Shi-Tomasi**  | 基于 Harris 改进的角点检测   | 无（通常仅用于检测）          | 更稳健，对角点的定义更严格，无尺度不变性。       | 目标跟踪、光流计算、图像配准。                |
| **FAST**        | 基于环形检测的角点检测         | 无（通常仅用于检测）          | 非常快速，适合实时应用，无尺度不变性。         | 实时目标检测、嵌入式系统。                  |
| **ORB**         | FAST 检测 + BRIEF 描述子 | BRIEF 描述子           | 快速、旋转不变性，适合嵌入式设备。           | 实时目标识别、增强现实（AR）、SLAM（同时定位与建图）。 |
| **SIFT**        | 高斯差分（DoG）检测         | 128 维 SIFT 描述子      | 尺度不变性、旋转不变性，计算复杂度较高。        | 图像匹配、目标识别、三维重建。                |
| **KAZE**        | 非线性尺度空间分析           | KAZE 描述子（非线性）       | 尺度不变性、旋转不变性，适用于复杂图像和高分辨率图像。 | 三维重建、目标识别、图像匹配，尤其是对图像细节要求高的场景。 |
| **AKAZE**       | 加速 KAZE（使用金字塔）      | AKAZE 描述子           | 快速、尺度不变性、旋转不变性，适合实时应用。      | 实时目标识别、增强现实（AR）、快速图像匹配。        |
| **BRISK**       | FAST 检测 + 阈值分割      | BRIEF 描述子变体         | 快速、旋转不变性，适合嵌入式设备和实时应用。      | 实时视觉系统、移动设备应用、快速特征匹配。          |
| **FREAK**       | FAST 检测             | 基于方向梯度的描述子          | 高效、旋转不变性，描述子长度可调。           | 实时特征匹配、目标识别、嵌入式系统。             |
| **SURF**        | Hessian 矩阵          | SURF 描述子            | 快速、尺度不变性、旋转不变性，适合复杂图像。      | 图像匹配、目标识别、三维重建，对计算资源要求较高。      |
| **GLOH**        | Hessian-Laplace 检测  | Hessian-Laplace 描述子 | 尺度不变性、旋转不变性，描述子具有高区分度。      | 图像匹配、目标识别，对特征区分度要求较高的场景。       |
| **STAR**        | 星形特征检测              | BRIEF 描述子           | 适合检测星形结构的特征点，旋转不变性。         | 星形结构特征检测、医学图像分析。               |
| **MSER**        | 最大稳定极值区域            | 无固定描述子（可结合其他描述子）    | 对光照变化和背景噪声鲁棒，适合检测稳定区域。      | 目标识别、图像分割、文字检测。                |
