## 3.2.1 为什么中文分词很重要？

### 中英文差异

**英文**：单词之间有天然的空格
```
I love Python
```

**中文**：字与字之间没有分隔
```
我喜欢Python
```

**分词的作用**：将连续的汉字切分成有意义的词语，是中文NLP的基础。

## 3.2.2 jieba分词的三种模式

### jieba.lcut() 函数说明

**函数作用**：对中文文本进行分词，返回词语列表

**基本用法**：
```python
import jieba
words = jieba.lcut(text, cut_all=False)
```

**参数说明**：
- `text`：要分词的文本（字符串）
- `cut_all`：是否使用全模式
  - `False`：精确模式（默认）
  - `True`：全模式

**返回结果**：返回一个列表，包含分词后的所有词语

**示例**：
```python
text = "我喜欢Python"
words = jieba.lcut(text)  # ['我', '喜欢', 'Python']
```

````python
import jieba

# 测试文本
text = "小明硕士毕业于中国科学院计算技术研究所"

print(f"原始文本：{text}\n")
print("="*60)
````

### 模式1：精确模式

**特点**：最精确地切分，适合文本分析

````python
# 精确模式
words = jieba.lcut(text, cut_all=False)

print("精确模式：")
print(" / ".join(words))
print(f"词数：{len(words)}")
````

### 模式2：全模式

**特点**：扫描所有可能的词，速度快但有冗余

````python
# 全模式
words = jieba.lcut(text, cut_all=True)

print("全模式：")
print(" / ".join(words))
print(f"词数：{len(words)}")
````

### 模式3：搜索引擎模式

**特点**：在精确模式基础上，对长词再次切分

#### jieba.lcut_for_search() 函数说明

**函数作用**：搜索引擎模式分词，在精确模式基础上对长词进行细分，提高搜索召回率

**基本用法**：
```python
words = jieba.lcut_for_search(text)
```

**与精确模式的区别**：
- 精确模式：`"中国科学院计算技术研究所"` → 保持完整
- 搜索引擎模式：`"中国科学院计算技术研究所"` → 拆分为 `"中国"、"科学"、"学院"、"科学院"、"研究所"、"中国科学院计算技术研究所"`

**使用场景**：
- 搜索引擎：用户可能搜索"中国"或"科学院"，都能匹配到完整词
- 信息检索：提高查找的覆盖率

**优点**：兼顾精确性和召回率

````python
# 搜索引擎模式
words = jieba.lcut_for_search(text)

print("搜索引擎模式：")
print(" / ".join(words))
print(f"词数：{len(words)}")
````

## 3.2.3 自定义词典

### 问题：专业术语被拆分

````python
text = "我在学习自然语言处理技术"

print(f"原文：{text}")
print("默认分词：", " / ".join(jieba.lcut(text)))
print("\n问题：'自然语言处理'被拆分成了多个词！")
````

### 解决方案：添加自定义词

#### jieba.add_word() 函数说明

**函数作用**：向jieba词典中添加自定义词语，让分词器能识别专业术语或新词

**基本用法**：
```python
jieba.add_word("自定义词语")
```

**使用场景**：
- 专业术语（如"自然语言处理"、"机器学习"）
- 新词、网络用语（如"性价比"、"物流快"）
- 人名、地名、品牌名

**注意**：添加后立即生效，影响后续所有分词操作

````python
# 添加自定义词
jieba.add_word("自然语言处理")

print("添加后：", " / ".join(jieba.lcut(text)))
print("\n✅ '自然语言处理'被正确识别为一个词了！")
````

## 3.2.4 应用实例

### 电商评论分词

````python
# 添加电商常用词
jieba.add_word("性价比")
jieba.add_word("物流快")

review = "这个手机性价比很高，物流快，值得购买"

print("评论：", review)
print("分词：", " / ".join(jieba.lcut(review)))
````
