## 3.3.1 什么是停用词？

**停用词**（Stop Words）是指在文本中频繁出现，但对文本分析帮助不大的词语。

### 常见的停用词

**中文停用词：**
- 语气词：的、了、吗、呢、吧
- 代词：我、你、他、这、那
- 介词：在、于、对、和、与
- 连词：但是、因此、所以

**为什么要去除？**
- 这些词出现频率高，但信息量低
- 去除后可以突出关键词
- 减少数据量，提高处理效率

## 3.3.2 实践：去除停用词



### 步骤1：分词

````python
import jieba

text = "我认为这个产品的质量很好，但是价格有点贵"

print(f"原文：{text}")
print("="*50)

# 分词
words = jieba.lcut(text)
print("\n分词结果：")
print(" / ".join(words))
````

### 步骤2：定义停用词表

````python
# 定义常见停用词（注意：根据应用场景选择是否包含某些词）
stopwords = ['的', '了', '在', '是', '我', '有', '和', '就', '不', '人', '都', '一', '一个', '上', '也', '很', '到', '说', '要', '去', '你', '会', '着', '没有', '看', '自己', '这', '但是', '有点']

print(f"停用词表（共{len(stopwords)}个）：")
print(stopwords[:10], "...")  # 只显示前10个
````

### 步骤3：过滤停用词

#### 列表推导式说明

**作用**：快速创建新列表，同时进行过滤

**基本语法**：
```python
new_list = [item for item in old_list if condition]
```

**理解方式**：
- `for item in old_list`：遍历原列表的每个元素
- `if condition`：只保留满足条件的元素
- 结果：生成一个新列表

**示例**：
```python
# 过滤停用词
filtered = [w for w in words if w not in stopwords]
# 意思是：遍历words中的每个词w，如果w不在停用词表中，就保留它
```

````python
# 去除停用词
filtered_words = [w for w in words if w not in stopwords]

print("去除停用词前：")
print(" / ".join(words))
print(f"词数：{len(words)}")

print("\n去除停用词后：")
print(" / ".join(filtered_words))
print(f"词数：{len(filtered_words)}")
````

**保留了关键词**：认为、这个、产品、质量、好、价格、贵

**去除了停用词**：我、的、很、但是、有点

## 3.3.3 使用停用词文件

实际项目中，通常使用更完整的停用词文件。

### 创建停用词文件

````python
# 创建一个简单的停用词文件
stopwords_content = """的
了
在
是
我
有
和
就
不
人
都
一
一个
很
到
这
但是
有点
"""

with open('/tmp/stopwords.txt', 'w', encoding='utf-8') as f:
    f.write(stopwords_content)

print("✅ 停用词文件创建完成：/tmp/stopwords.txt")
````

### 加载停用词文件

#### set() 函数说明

**函数作用**：将列表转换为集合（set），提高查找效率

**基本用法**：
```python
stopwords_set = set(stopwords_list)
```

**为什么使用set？**
- 列表查找：需要逐个比对，速度慢（O(n)）
- 集合查找：使用哈希表，速度快（O(1)）
- 当停用词表很大时，使用set可以显著提升性能

**示例**：
```python
# 列表方式（慢）
if word not in stopwords_list:  # 需要遍历整个列表

# 集合方式（快）
if word not in stopwords_set:   # 直接查找，速度快
```

````python
# 从文件加载停用词
def load_stopwords(file_path):
    with open(file_path, 'r', encoding='utf-8') as f:
        stopwords = [line.strip() for line in f]
    return set(stopwords)  # 用set提高查找效率

stopwords = load_stopwords('/tmp/stopwords.txt')
print(f"✅ 加载了 {len(stopwords)} 个停用词")
````

### 使用停用词文件过滤

````python
text = "我觉得这款手机的性能很强大，拍照效果也不错"

# 分词
words = jieba.lcut(text)

# 去除停用词
filtered = [w for w in words if w not in stopwords]

print(f"原文：{text}")
print(f"\n去停用词前：{' / '.join(words)}")
print(f"去停用词后：{' / '.join(filtered)}")
print(f"\n✅ 从 {len(words)} 个词减少到 {len(filtered)} 个词")
````

## 3.3.4 构建通用函数

````python
def remove_stopwords(text, stopwords_file='/tmp/stopwords.txt'):
    """
    去除停用词
    
    参数：
        text: 原始文本
        stopwords_file: 停用词文件路径
    返回：
        过滤后的词列表
    """
    # 加载停用词
    with open(stopwords_file, 'r', encoding='utf-8') as f:
        stopwords = set([line.strip() for line in f])
    
    # 分词
    words = jieba.lcut(text)
    
    # 过滤停用词和标点符号
    filtered = [w for w in words if w not in stopwords and w.strip() and w not in ['，', '。', '！', '？', '、', '；', '：']]
    
    return filtered
````

去停用词函数定义完成

### 测试函数

````python
texts = [
    "这个产品质量很好",
    "我认为价格有点贵",
    "但是性能不错"
]

print("批量处理测试：\n")
for text in texts:
    result = remove_stopwords(text)
    print(f"原文：{text}")
    print(f"结果：{' / '.join(result)}")
    print()
````

## 3.3.5 应用场景

### 场景1：关键词提取

去除停用词后，剩下的都是关键信息。

````python
from collections import Counter

# 案例：用户评论分析
text = "我觉得这款手机很好，性能好，拍照好，价格也好，总之很好"

words = jieba.lcut(text)

# 注意：这里不使用remove_stopwords函数，因为它会过滤单字
# 手动过滤停用词，但保留"好"来演示效果
manual_stopwords = ['的', '了', '在', '是', '我', '有', '和', '就', '不', '人', '都', '一', '一个', '很', '到', '这', '但是', '有点']
filtered = [w for w in words if w not in manual_stopwords and w.strip()]

print("原始分词词频统计：")
for word, count in Counter(words).most_common(5):
    print(f"  '{word}': {count}次")

print("\n去停用词后词频统计：")
for word, count in Counter(filtered).most_common(5):
    print(f"  '{word}': {count}次")
````

分析：去除'很'、'我'等停用词后，关键词'好'、'手机'等更突出

### 场景2：文本分类

文本分类时，去除停用词可以提高准确率。

````python
# 科技新闻
tech_news = "人工智能技术在各个领域都有很好的应用"

# 体育新闻  
sport_news = "中国队在比赛中表现很好取得了胜利"

print("科技新闻关键词：")
print(' / '.join(remove_stopwords(tech_news)))

print("\n体育新闻关键词：")
print(' / '.join(remove_stopwords(sport_news)))
````

关键词更能体现新闻类别
