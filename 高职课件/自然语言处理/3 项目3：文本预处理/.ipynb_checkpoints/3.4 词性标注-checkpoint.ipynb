## 3.4.1 什么是词性标注？

**词性标注**（Part-of-Speech Tagging）是给每个词标注其语法属性的过程。

### 常见词性

- **n**：名词（noun）- 手机、电脑、学生
- **v**：动词（verb）- 学习、工作、吃饭
- **a**：形容词（adjective）- 好的、漂亮、聪明
- **d**：副词（adverb）- 很、非常、特别
- **m**：数词（numeral）- 一、二、三
- **q**：量词（quantifier）- 个、台、本

### 为什么需要词性标注？

- 理解词语的语法功能
- 提取特定类型的词（如只要名词）
- 提高文本分析的准确性

## 3.4.2 使用jieba进行词性标注

### 基础用法

```python
import jieba.posseg as pseg

text = "我喜欢学习自然语言处理技术"

print(f"原文：{text}\n")
print("词性标注结果：")
print("="*50)

# 词性标注
words = pseg.cut(text)

for word, flag in words:
    print(f"{word:8s} {flag}")
```

### 词性含义说明

```python
# 词性对照表
pos_dict = {
    'n': '名词',
    'v': '动词', 
    'a': '形容词',
    'd': '副词',
    'r': '代词',
    'm': '数词',
    'q': '量词',
    'p': '介词',
    'c': '连词'
}

print("常用词性对照表：\n")
for pos, name in pos_dict.items():
    print(f"{pos:5s} - {name}")
```

## 3.4.3 根据词性筛选词语

### 提取名词

```python
text = "这款智能手机的屏幕很大，电池续航能力强"

print(f"原文：{text}\n")

# 提取所有名词
words = pseg.cut(text)
nouns = [word for word, flag in words if flag.startswith('n')]

print("提取的名词：")
print(" / ".join(nouns))
print("\n✅ 名词通常是关键信息的载体")
```

### 提取动词

```python
text = "用户可以下载、安装和使用这个软件"

print(f"原文：{text}\n")

# 提取所有动词
words = pseg.cut(text)
verbs = [word for word, flag in words if flag.startswith('v')]

print("提取的动词：")
print(" / ".join(verbs))
print("\n✅ 动词表示动作和行为")
```

### 提取形容词

```python
text = "这个产品质量好，价格便宜，性能强大"

print(f"原文：{text}\n")

# 提取所有形容词
words = pseg.cut(text)
adjs = [word for word, flag in words if flag.startswith('a')]

print("提取的形容词：")
print(" / ".join(adjs))
print("\n✅ 形容词常用于情感分析")
```

## 3.4.4 构建词性过滤函数

```python
def extract_by_pos(text, pos_list=['n', 'v', 'a']):
    """
    根据词性提取词语
    
    参数：
        text: 原始文本
        pos_list: 要提取的词性列表
    返回：
        符合词性的词列表
    """
    words = pseg.cut(text)
    result = []
    
    for word, flag in words:
        # 检查词性是否在列表中
        for pos in pos_list:
            if flag.startswith(pos):
                result.append(word)
                break
    
    return result

print("✅ 词性过滤函数定义完成")
```

### 测试函数

```python
text = "我非常喜欢这个智能手机，它的拍照功能很强大"

print(f"原文：{text}\n")

# 只提取名词
nouns = extract_by_pos(text, ['n'])
print(f"名词：{' / '.join(nouns)}")

# 提取名词和形容词
result = extract_by_pos(text, ['n', 'a'])
print(f"名词+形容词：{' / '.join(result)}")

# 提取名词、动词和形容词
result = extract_by_pos(text, ['n', 'v', 'a'])
print(f"名词+动词+形容词：{' / '.join(result)}")
```

## 3.4.5 应用场景

### 场景1：关键词提取

通常只提取名词和形容词作为关键词。

```python
text = "这款新推出的智能手表功能强大，设计精美，价格合理"

print(f"原文：{text}\n")

# 提取关键词（名词+形容词）
keywords = extract_by_pos(text, ['n', 'a'])

print("关键词：")
print(' / '.join(keywords))
print("\n✅ 这些词最能代表文本的主要内容")
```

### 场景2：情感分析

形容词和副词常表达情感倾向。

```python
text = "这部电影非常精彩，演员表演特别出色，强烈推荐"

print(f"原文：{text}\n")

# 提取情感词（形容词+副词）
sentiment_words = extract_by_pos(text, ['a', 'd'])

print("情感词：")
print(' / '.join(sentiment_words))
print("\n✅ 这些词表达了正面情感")
```
