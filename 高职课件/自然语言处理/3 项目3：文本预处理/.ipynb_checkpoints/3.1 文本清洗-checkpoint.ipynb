## 3.1.1 为什么需要文本清洗？

### 真实场景

想象你正在开发一个商品评论分析系统，从网站上爬取了大量用户评论。

这些数据存在很多问题：
- 多余的空格
- 重复的标点符号
- 表情符号
- HTML标签

**文本清洗就是要把这些脏数据变成干净、规范的数据，方便后续分析。**

## 3.1.2 文本清洗的常见任务

### 主要清洗内容

1. **去除空格**：删除多余的空格和换行符
2. **统一大小写**：英文文本的大小写规范化
3. **去除标点符号**：保留或删除标点
4. **去除特殊字符**：表情、HTML标签、URL等
5. **去除数字**：根据需求决定是否保留
6. **繁简转换**：统一为简体或繁体中文

## 3.1.3 实践：逐步清洗文本

让我们用真实的用户评论数据来演示文本清洗的各个步骤。

### 步骤1：去除首尾空格

用户输入的文本经常会有多余的空格。

```python
# 原始评论数据
comment = "   这个手机拍照效果真不错！  "

print(f"原始文本：'{comment}'")
print(f"文本长度：{len(comment)} 个字符")

# 清洗：去除首尾空格
cleaned_comment = comment.strip()

print(f"\n清洗后文本：'{cleaned_comment}'")
print(f"清洗后长度：{len(cleaned_comment)} 个字符")
print(f"节省了 {len(comment) - len(cleaned_comment)} 个字符")
```

### 步骤2：去除多余的空格和换行

文本中间可能有多个连续空格或换行符。

```python
import re

# 包含多余空格的文本
text = "这个产品    质量很好，   物流也很快！\n\n值得购买"

print("原始文本：")
print(text)
print("="*50)

# 方法1：将多个空格替换为单个空格
cleaned1 = re.sub(r'\s+', ' ', text)
print("\n方法1 - 统一为单个空格：")
print(cleaned1)

# 方法2：完全去除所有空格和换行
cleaned2 = re.sub(r'\s+', '', text)
print("\n方法2 - 完全去除空格：")
print(cleaned2)
```

### 步骤3：去除或规范化标点符号

```python
import re

# 包含重复标点的文本
text = "太好了！！！真的很棒。。。强烈推荐？？？"

print(f"原始文本：{text}")
print("="*50)

# 方法1：将连续标点替换为单个
cleaned1 = re.sub(r'([！。？])+', r'\1', text)
print(f"\n方法1 - 规范化标点：{cleaned1}")

# 方法2：完全删除所有标点
import string
chinese_punctuation = '！？。，'
all_punctuation = string.punctuation + chinese_punctuation

cleaned2 = text
for p in all_punctuation:
    cleaned2 = cleaned2.replace(p, '')
    
print(f"方法2 - 删除所有标点：{cleaned2}")
```

### 步骤4：构建综合清洗函数

```python
import re
import string

def clean_text(text, remove_punctuation=False, remove_digits=False):
    # 去除首尾空格
    text = text.strip()
    
    # 去除数字
    if remove_digits:
        text = re.sub(r'\d+', '', text)
    
    # 去除标点符号
    if remove_punctuation:
        chinese_punctuation = '！？。，、；：''【】《》'
        all_punctuation = string.punctuation + chinese_punctuation
        for p in all_punctuation:
            text = text.replace(p, '')
    
    # 去除多余空格
    text = re.sub(r'\s+', ' ', text)
    text = text.strip()
    
    return text

print("文本清洗函数定义完成！")
```

### 测试清洗函数

```python
# 测试数据
test_texts = [
    "   这个手机真不错！！！  ",
    "我买了2个，花了3000元",
    "太棒了。。。。。强烈推荐？？？"
]

print("文本清洗测试结果：")
print("="*70)

for i, text in enumerate(test_texts, 1):
    cleaned = clean_text(text, remove_punctuation=False, remove_digits=False)
    print(f"\n测试 {i}:")
    print(f"  原文：{text}")
    print(f"  清洗：{cleaned}")
```
