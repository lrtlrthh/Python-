## 3.1.1 为什么需要文本清洗？

### 真实场景

想象你正在开发一个商品评论分析系统，从网站上爬取了大量用户评论：

```
"   这个手机真不错！！！  "
"质量很好😊😊😊推荐购买"
"价格有点贵.....但是值得"
```

这些数据存在很多问题：
- 多余的空格
- 重复的标点符号
- 表情符号

**文本清洗就是要把这些脏数据变成干净、规范的数据，方便后续分析。**

## 3.1.2 文本清洗的常见任务

### 主要清洗内容

1. **去除空格**：删除多余的空格和换行符
2. **去除标点符号**：保留或删除标点
3. **去除数字**：根据需求决定是否保留
4. **去除特殊字符**：表情、HTML标签、URL等

## 3.1.3 实践：逐步清洗文本

让我们用真实的用户评论数据来演示文本清洗的各个步骤。

### 步骤1：去除首尾空格

用户输入的文本经常会有多余的空格。

#### strip() 函数说明

**函数作用**：去除字符串开头和结尾的空白字符（空格、换行、制表符等）

**基本用法**：
```python
cleaned_text = text.strip()
```

**注意**：只去除首尾空格，不影响中间的空格

````python
# 原始评论数据
comment = "   这个手机拍照效果真不错！  "

print(f"原始文本：'{comment}'")
print(f"文本长度：{len(comment)} 个字符")

# 使用strip()去除首尾空格
cleaned_comment = comment.strip()

print(f"\n清洗后：'{cleaned_comment}'")
print(f"清洗后长度：{len(cleaned_comment)} 个字符")
print(f"节省了 {len(comment) - len(cleaned_comment)} 个字符")
````

### 步骤2：去除多余的空格和换行

文本中间可能有多个连续空格或换行符。我们有两种处理方式。

#### re.sub() 函数说明

**函数作用**：使用正则表达式查找并替换文本中的内容

**基本用法**：
```python
import re
cleaned_text = re.sub(pattern, replacement, text)
```

**参数说明**：
- `pattern`：要查找的模式（正则表达式）
- `replacement`：替换成什么内容
- `text`：原始文本

**常用正则表达式**：
- `\s`：匹配任何空白字符（空格、换行、制表符）
- `\s+`：匹配一个或多个连续的空白字符
- `\d`：匹配数字

#### 方法1：将多个空格替换为单个空格

适用于需要保留单词间距的场景。

````python
import re

text = "这个产品    质量很好，   物流也很快！\n\n值得购买"

print("原始文本：")
print(text)
print("="*50)

# 将多个空格/换行替换为单个空格
cleaned = re.sub(r'\s+', ' ', text)

print("\n清洗后：")
print(cleaned)
````

#### 方法2：完全去除所有空格和换行

适用于中文文本，因为中文不需要空格分隔。

````python
import re

text = "这个产品    质量很好，   物流也很快！\n\n值得购买"

print("原始文本：")
print(text)
print("="*50)

# 完全去除所有空格和换行
cleaned = re.sub(r'\s+', '', text)

print("\n清洗后：")
print(cleaned)
````

### 步骤3：处理标点符号

用户经常使用重复的标点符号来表达情绪。我们也有两种处理方式。

#### 方法1：将重复标点替换为单个

保留标点，但规范化数量。适合情感分析等需要保留标点信息的场景。

````python
import re

text = "太好了！！！真的很棒。。。强烈推荐？？？"

print(f"原始文本：{text}")
print("="*50)

# 将连续的标点替换为单个
cleaned = re.sub(r'([！。？])+', r'\1', text)

print(f"\n清洗后：{cleaned}")
print("\n✅ 保留了标点，但去除了重复")
````

#### 方法2：完全删除所有标点

适合文本分类等只需要关键词的场景。

````python
import string

text = "太好了！！！真的很棒。。。强烈推荐？？？"

print(f"原始文本：{text}")
print("="*50)

# 定义要删除的标点
chinese_punctuation = '！？。，、；：'
all_punctuation = string.punctuation + chinese_punctuation

# 删除所有标点
cleaned = text
for p in all_punctuation:
    cleaned = cleaned.replace(p, '')

print(f"\n清洗后：{cleaned}")
print("\n✅ 只保留了纯文本")
````

### 步骤4：构建综合清洗函数

现在我们把前面学到的所有清洗方法整合成一个通用函数。

**函数参数：**
- `text`：要清洗的文本
- `remove_punctuation`：是否删除标点（默认False）
- `remove_digits`：是否删除数字（默认False）
- `normalize_punctuation`：是否规范化重复标点（默认True）

**清洗步骤：**
1. 去除首尾空格
2. 规范化重复标点
3. 可选删除数字
4. 可选删除标点
5. 去除多余空格

````python
import re
import string

def clean_text(text, remove_punctuation=False, remove_digits=False, normalize_punctuation=True):
    """文本清洗函数"""
    
    # 1. 去除首尾空格
    text = text.strip()
    
    # 2. 规范化重复标点
    if normalize_punctuation and not remove_punctuation:
        text = re.sub(r'([！？。，、；：])+', r'\1', text)
        text = re.sub(r'([!?.,;:])+', r'\1', text)
    
    # 3. 去除数字
    if remove_digits:
        text = re.sub(r'\d+', '', text)
    
    # 4. 去除标点符号
    if remove_punctuation:
        chinese_punctuation = '！？。，、；：''""【】《》'
        all_punctuation = string.punctuation + chinese_punctuation
        for p in all_punctuation:
            text = text.replace(p, '')
    
    # 5. 去除多余空格
    text = re.sub(r'\s+', ' ', text)
    text = text.strip()
    
    return text

print("✅ 文本清洗函数定义完成！")
````

## 3.1.4 测试清洗函数

我们用几个测试案例来验证函数的效果。

````python
# 测试数据
test_texts = [
    "   这个手机真不错！！！  ",
    "我买了2个，花了3000元",
    "太棒了。。。。。强烈推荐？？？"
]

print("文本清洗测试：")
print("="*70)
````

### 测试1：基础清洗（规范化标点，保留数字）

````python
print("\n【测试1：规范化标点，保留数字】\n")

for i, text in enumerate(test_texts, 1):
    cleaned = clean_text(text, 
                        remove_punctuation=False,
                        remove_digits=False)
    print(f"{i}. 原文：{text}")
    print(f"   清洗：{cleaned}\n")
````

### 测试2：彻底清洗（删除标点和数字）

````python
print("\n【测试2：删除标点和数字】\n")

for i, text in enumerate(test_texts, 1):
    cleaned = clean_text(text,
                        remove_punctuation=True,
                        remove_digits=True)
    print(f"{i}. 原文：{text}")
    print(f"   清洗：{cleaned}\n")
````

## 3.1.5 应用场景

不同的NLP任务需要不同的清洗策略。

### 场景1：情感分析

在情感分析中，标点符号（特别是感叹号）是情感强度的重要标志，应该保留。

````python
review = "这个产品太好了！！！强烈推荐！！！"

cleaned = clean_text(review, 
                    remove_punctuation=False,
                    normalize_punctuation=True)

print(f"原文：{review}")
print(f"清洗：{cleaned}")
print("\n✅ 保留了标点'！'，表达强烈正面情感")
````

### 场景2：文本分类

在文本分类中，我们只需要关键词，可以删除标点和数字。

````python
news = "科技公司发布新产品，预计销售额达5000万元！！"

cleaned = clean_text(news,
                    remove_punctuation=True,
                    remove_digits=True)

print(f"原文：{news}")
print(f"清洗：{cleaned}")
print("\n✅ 只保留关键词，用于分类")
````

### 场景3：关键词提取

提取关键词时，删除标点但可能保留数字（如果数字有意义）。

````python
text = "   人工智能、大数据、云计算...这些技术正在改变世界！！！  "

cleaned = clean_text(text,
                    remove_punctuation=True,
                    remove_digits=False)

print(f"原文：{text}")
print(f"清洗：{cleaned}")
print("\n✅ 提取出纯文本，方便后续分词")
````
