## 1.4 自然语言处理的核心概念

### 为什么要学习这些概念？

**职场现实：**
- 面试时经常被问到基本概念
- 工作和技术同事交流需要专业术语
- 看技术文档需要理解这些词汇

### 1.4.1 五个核心概念

1. 文本（Text）

**通俗理解：** 就是要处理的文字内容   
**生活例子：**
- 一条朋友圈："今天天气真好，心情不错！"
- 一条淘宝评论："质量很好，物流快，推荐购买"
- 一封邮件："关于明天会议安排的通知"


2. 语料库（Corpus）

**通俗理解：** 大量文本的集合，用来"训练"AI    
**比喻：**
- 单篇文章 = 一本书
- 语料库 = 整个图书馆

**实际案例：**
- **新闻语料库：** 人民日报10年的所有文章
- **微博语料库：** 1亿条微博内容
- **客服语料库：** 淘宝所有客服对话记录

**重要性：** "垃圾进，垃圾出"
- 好的语料库 = 好的AI模型
- 坏的语料库 = 垃圾AI模型

3. 词汇表（Vocabulary）

**通俗理解：** AI能认识的全部词语   
**比喻：** 个人的"词典"

**简单例子：**
```
词汇表示例：
["我", "爱", "中国", "自然语言", "处理", "技术"]
```

4. 特征（Features） 

**通俗理解：** 把文字转换成数字，让电脑能计算    
**比喻：** 体检报告不写"你很健康"，而是写各项指标

**简单特征提取：**
```
文本："这个手机很好用"
特征提取：
- 字符数：7个字
- 词数：4个词
- 正面词：2个("好", "很")
- 负面词：0个
```

5. 模型（Model）

**通俗理解：** 经过训练的程序，能完成特定任务    
**比喻：**
- 训练前 = 新手学生
- 训练后 = 专业人士

**常见模型类型：**
- **分类模型：** 判断好评/差评
- **生成模型：** 写文章、回答问题
- **匹配模型：** 搜索引擎、推荐系统

### 1.4.2 实战演示：从文本到模型的完整流程

接下来，我们通过实际代码来理解这5个核心概念。我们会用两条商品评论作为例子，一步步演示整个NLP处理流程。

#### 1.4.2.1 第一步：文本（Text）

**文本**是我们要处理的原始数据。让我们先准备两条商品评论：

````python
# 准备两条商品评论文本
text1 = "这个手机拍照很清晰，推荐购买！"
text2 = "电池太不耐用了，用了一天就没电，不推荐"

print("文本1:", text1)
print("文本2:", text2)
````

#### 1.4.2.2 第二步：语料库（Corpus）

**语料库**就是把多条文本收集在一起，形成一个集合。在实际项目中，语料库可能有几万甚至几百万条数据。

````python
# 把文本放到一个列表里，这就是语料库
corpus = [text1, text2]

print("语料库包含", len(corpus), "条文本")
print("这是一个简化的例子，实际的语料库会包含成千上万条数据")
````

#### 1.4.2.3 第三步：词汇表（Vocabulary）

**词汇表**是从所有文本中提取出来的不重复词语列表。这就像 AI 的"字典"。

````python
import jieba
# 从所有文本中提取词汇
all_words = set()  # 用集合自动去重

for text in corpus:
    # 使用jieba进行中文分词
    words = jieba.lcut(text)
    all_words.update(words)

# 排序后转成列表（去除标点符号）
vocabulary = sorted([word for word in all_words if word.strip() and word not in ['，', '！', '。', '、']])

print("词汇表包含", len(vocabulary), "个不同的词：")
print(vocabulary)
````

#### 1.4.2.4 第四步：特征提取（Features）

**特征**是把文字转换成数字。电脑只能处理数字，所以我们要把文本"翻译"成数字形式。

这里我们用一个简单的方法：统计正面词和负面词的数量。

````python
# 定义情感词典
positive_words = ["好", "清晰", "推荐", "购买", "耐用"]  # 正面词
negative_words = ["不", "没", "太", "差"]              # 负面词

# 对每条评论提取特征
print("特征提取结果：\n")

for i, text in enumerate(corpus, 1):
    # 使用jieba进行中文分词
    words = jieba.lcut(text)
    
    # 统计正负面词数量
    pos_count = sum(1 for word in words if word in positive_words)
    neg_count = sum(1 for word in words if word in negative_words)
    
    print(f"文本{i}: {text}")
    print(f"  分词结果: {words}")
    print(f"  正面词数量: {pos_count}")
    print(f"  负面词数量: {neg_count}")
    
    # 额外特征：文本长度和词语数量
    print(f"  字符数: {len(text)}")
    print(f"  词语数: {len(words)}")
    print()
````

#### 1.4.2.5 第五步：模型预测（Model）

**模型**就是根据特征做出判断的"规则"。这里我们的规则很简单：正面词多就是好评，负面词多就是差评。

````python
# 简单的情感分析模型
print("模型预测结果：\n")

for i, text in enumerate(corpus, 1):
    # 使用jieba进行分词
    words = jieba.lcut(text)
    
    # 提取特征
    pos_count = sum(1 for word in words if word in positive_words)
    neg_count = sum(1 for word in words if word in negative_words)
    
    # 模型判断
    if pos_count > neg_count:
        prediction = "好评 ⭐⭐⭐⭐⭐"
    elif pos_count < neg_count:
        prediction = "差评 ⭐"
    else:
        prediction = "中性 ⭐⭐⭐"
    
    print(f"文本{i}: {text}")
    print(f"  正面词: {pos_count} 个, 负面词: {neg_count} 个")
    print(f"  预测结果: {prediction}")
    print()
````

#### 1.4.2.6 流程总结

**完整的NLP处理流程：**

```
文本 → 语料库 → 词汇表 → 特征提取 → 模型预测
```

**每一步的作用：**
1. **文本**：原始数据
2. **语料库**：数据集合
3. **词汇表**：AI的"字典"
4. **特征**：文字转数字
5. **模型**：做出判断

**记住**：这只是一个超级简化的例子！真实的NLP系统会复杂得多，但基本思路是一样的。
