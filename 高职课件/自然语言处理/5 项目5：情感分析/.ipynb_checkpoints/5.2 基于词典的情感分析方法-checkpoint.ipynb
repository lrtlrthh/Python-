## 2. 什么是情感词典？

### 2.1 生活中的例子

想象你要教一个外国朋友理解中文的情感表达：

**正面词汇：**
- 好、棒、优秀、完美、满意、喜欢

**负面词汇：**
- 差、烂、糟糕、失望、讨厌、生气

你会给TA一个列表，告诉TA哪些词是正面的，哪些词是负面的。

**情感词典就是这样一个列表，记录了词语的情感倾向。**

### 2.2 情感词典的定义

**情感词典（Sentiment Lexicon）**：一个包含情感词汇及其情感倾向的列表或数据库。

**基本结构：**

| 词语 | 情感倾向 | 情感分数 |
|------|---------|----------|
| 好 | 正面 | +1 |
| 优秀 | 正面 | +2 |
| 差 | 负面 | -1 |
| 糟糕 | 负面 | -2 |
| 一般 | 中性 | 0 |

### 2.3 情感词典的类型

**1. 基础情感词典**
- 包含基本的正面词和负面词
- 例如：好、坏、美、丑

**2. 程度副词词典**
- 表示程度的词语
- 例如：非常、很、稍微、有点
- 用来增强或减弱情感强度

**3. 否定词词典**
- 表示否定的词语
- 例如：不、没、无、未
- 用来反转情感倾向

**4. 领域专用词典**
- 针对特定领域的情感词
- 例如：电影领域（精彩、烂片）、餐饮领域（美味、难吃）

- 标注详细

**适用场景：**

## 4. 基于词典的情感分析原理

### 4.1 基本思路

**核心思想**：统计文本中正面词和负面词的数量，计算情感得分。

**基本公式：**

```
情感得分 = 正面词数量 - 负面词数量

如果得分 > 0  → 正面情感
如果得分 < 0  → 负面情感
如果得分 = 0  → 中性情感
```

### 4.2 简单示例

**文本：** "这个手机很好用，屏幕清晰，性能强大"

**分析过程：**

1. 分词：这个 / 手机 / 很 / 好用 / 屏幕 / 清晰 / 性能 / 强大
2. 匹配词典：
   - 正面词：好用、清晰、强大（共3个）
   - 负面词：无（0个）
3. 计算得分：3 - 0 = 3
4. 判断情感：得分>0，所以是**正面情感**

## 5. 实现简单的词典情感分析

让我们从零开始，实现一个简单的基于词典的情感分析系统。

### 5.1 准备情感词典

首先，我们需要准备正面词和负面词列表。

````python
# 准备简单的情感词典
positive_words = [
    '好', '棒', '优秀', '完美', '满意', '喜欢', '开心', '高兴',
    '不错', '赞', '精彩', '漂亮', '舒服', '方便', '快', '清晰',
    '强大', '实用', '值得', '推荐', '超值', '划算', '便宜', '实惠',
    '好用', '给力', '靠谱', '专业', '优质', '良好'
]

negative_words = [
    '差', '烂', '糟糕', '失望', '讨厌', '生气', '难受', '不满',
    '垃圾', '坑', '骗人', '后悔', '问题', '故障', '慢', '卡',
    '贵', '坑爹', '难用', '不好', '不行', '太差', '质量差', '服务差',
    '浪费', '上当', '欺骗', '虚假', '劣质'
]

print(f"正面词数量: {len(positive_words)}")
print(f"负面词数量: {len(negative_words)}")
print(f"\n正面词示例: {positive_words[:5]}")
print(f"负面词示例: {negative_words[:5]}")
````

### 5.2 实现基础情感分析函数

实现一个简单的函数，统计正负面词的数量。

````python
import jieba

def simple_sentiment_score(text):
    """
    简单的情感分析函数
    返回情感得分和情感类别
    """
    # 分词
    words = jieba.lcut(text)
    
    # 统计正负面词
    pos_count = sum(1 for word in words if word in positive_words)
    neg_count = sum(1 for word in words if word in negative_words)
    
    # 计算得分
    score = pos_count - neg_count
    
    # 判断情感类别
    if score > 0:
        sentiment = '正面'
    elif score < 0:
        sentiment = '负面'
    else:
        sentiment = '中性'
    
    return {
        'score': score,
        'sentiment': sentiment,
        'pos_count': pos_count,
        'neg_count': neg_count
    }

print("✅ 情感分析函数定义完成！")
````

### 5.3 测试情感分析函数

````python
# 准备测试文本
test_texts = [
    "这个手机很好用，屏幕清晰，性能强大，非常满意",
    "质量太差了，用了一天就坏了，非常失望",
    "还可以吧，价格便宜，质量一般",
    "物流很快，包装不错，推荐购买",
    "垃圾产品，浪费钱，坑爹"
]

print("情感分析测试：")
print("="*80)

for text in test_texts:
    result = simple_sentiment_score(text)
    print(f"\n文本: {text}")
    print(f"正面词数: {result['pos_count']}, 负面词数: {result['neg_count']}")
    print(f"情感得分: {result['score']}")
    print(f"情感类别: {result['sentiment']}")
    print("-"*80)
````

## 6. 处理程度副词

程度副词会影响情感的强度，我们需要在分析中考虑它们。

### 6.1 程度副词词典

不同的程度副词有不同的强度。

````python
# 程度副词及其权重
degree_words = {
    # 极强程度（权重2.0）
    '极其': 2.0, '极': 2.0, '超级': 2.0, '超': 2.0, '特别': 2.0,
    '十分': 2.0, '非常': 2.0, '相当': 2.0, '格外': 2.0,
    
    # 较强程度（权重1.5）
    '很': 1.5, '挺': 1.5, '蛮': 1.5, '颇': 1.5, '最': 1.5,
    '更': 1.5, '太': 1.5, '尤其': 1.5,
    
    # 中等程度（权重1.2）
    '比较': 1.2, '还': 1.2, '较': 1.2,
    
    # 较弱程度（权重0.8）
    '稍': 0.8, '稍微': 0.8, '略': 0.8, '略微': 0.8,
    '有点': 0.8, '有些': 0.8, '一点': 0.8
}

print("程度副词词典：")
print(f"总数: {len(degree_words)}个")
print("\n示例：")
for word, weight in list(degree_words.items())[:8]:
    print(f"  {word}: {weight}")
````

### 6.2 改进的情感分析函数

现在让我们改进函数，考虑程度副词的影响。

````python
def improved_sentiment_score(text):
    """
    改进的情感分析函数，考虑程度副词
    """
    # 分词
    words = jieba.lcut(text)
    
    # 初始化得分
    score = 0.0
    pos_count = 0
    neg_count = 0
    
    # 遍历每个词
    for i, word in enumerate(words):
        # 检查是否是情感词
        if word in positive_words:
            # 默认权重为1
            weight = 1.0
            
            # 检查前一个词是否是程度副词
            if i > 0 and words[i-1] in degree_words:
                weight = degree_words[words[i-1]]
            
            score += weight
            pos_count += 1
            
        elif word in negative_words:
            # 默认权重为1
            weight = 1.0
            
            # 检查前一个词是否是程度副词
            if i > 0 and words[i-1] in degree_words:
                weight = degree_words[words[i-1]]
            
            score -= weight
            neg_count += 1
    
    # 判断情感类别
    if score > 0:
        sentiment = '正面'
    elif score < 0:
        sentiment = '负面'
    else:
        sentiment = '中性'
    
    return {
        'score': round(score, 2),
        'sentiment': sentiment,
        'pos_count': pos_count,
        'neg_count': neg_count
    }

print("✅ 改进的情感分析函数定义完成！")
````

### 6.3 对比测试

````python
# 测试程度副词的影响
test_texts_degree = [
    "这个产品好",
    "这个产品很好",
    "这个产品非常好",
    "这个产品超级好",
    "这个产品有点好"
]

print("程度副词影响测试：")
print("="*80)

for text in test_texts_degree:
    result_simple = simple_sentiment_score(text)
    result_improved = improved_sentiment_score(text)
    
    print(f"\n文本: {text}")
    print(f"简单方法得分: {result_simple['score']}")
    print(f"改进方法得分: {result_improved['score']}")
    print("-"*80)
````

## 7. 处理否定词

否定词会反转情感倾向，这是情感分析中的重要问题。

### 7.1 否定词词典

````python
# 否定词列表
negation_words = [
    '不', '没', '无', '非', '未', '否', '别', '莫',
    '不是', '不能', '不会', '没有', '不要', '不用',
    '不太', '不够', '不怎么'
]

print(f"否定词数量: {len(negation_words)}")
print(f"否定词列表: {negation_words}")
````

### 7.2 完整的情感分析函数

现在让我们实现一个完整的函数，同时考虑程度副词和否定词。

````python
def complete_sentiment_score(text):
    """
    完整的情感分析函数，考虑程度副词和否定词
    """
    # 分词
    words = jieba.lcut(text)
    
    # 初始化得分
    score = 0.0
    pos_count = 0
    neg_count = 0
    
    # 遍历每个词
    i = 0
    while i < len(words):
        word = words[i]
        
        # 检查是否是情感词
        if word in positive_words or word in negative_words:
            # 默认权重和符号
            weight = 1.0
            sign = 1 if word in positive_words else -1
            
            # 检查前面的程度副词（向前看1-2个词）
            if i > 0 and words[i-1] in degree_words:
                weight = degree_words[words[i-1]]
            
            # 检查前面的否定词（向前看1-3个词）
            negation_count = 0
            for j in range(max(0, i-3), i):
                if words[j] in negation_words:
                    negation_count += 1
            
            # 奇数个否定词反转情感
            if negation_count % 2 == 1:
                sign *= -1
            
            # 累加得分
            score += sign * weight
            
            # 统计情感词数量
            if word in positive_words:
                pos_count += 1
            else:
                neg_count += 1
        
        i += 1
    
    # 判断情感类别
    if score > 0:
        sentiment = '正面'
    elif score < 0:
        sentiment = '负面'
    else:
        sentiment = '中性'
    
    return {
        'score': round(score, 2),
        'sentiment': sentiment,
        'pos_count': pos_count,
        'neg_count': neg_count
    }

print("✅ 完整的情感分析函数定义完成！")
````

### 7.3 测试否定词处理

````python
# 测试否定词的影响
test_texts_negation = [
    "这个产品很好",
    "这个产品不好",
    "这个产品不是很好",
    "这个产品质量差",
    "这个产品质量不差",
    "这个产品质量不是不好"
]

print("否定词影响测试：")
print("="*80)

for text in test_texts_negation:
    result = complete_sentiment_score(text)
    
    print(f"\n文本: {text}")
    print(f"情感得分: {result['score']}")
    print(f"情感类别: {result['sentiment']}")
    print("-"*80)
````

## 8. 综合测试

````python
# 综合测试
test_texts_final = [
    "这个手机非常好用，屏幕清晰，性能强大，非常满意！",
    "质量太差了，用了一天就坏了，非常失望，不推荐",
    "价格有点贵，但是质量不错，物流也很快",
    "本来很期待，但是收到货后很失望，质量不好",
    "不是特别好，但也不差，还可以吧",
    "客服态度很好，但是产品质量一般般",
    "超级棒！强烈推荐！物超所值！",
    "垃圾产品，浪费钱，根本不值这个价"
]

print("综合情感分析测试：")
print("="*80)

for text in test_texts_final:
    result = complete_sentiment_score(text)
    
    print(f"\n文本: {text}")
    print(f"正面词数: {result['pos_count']}, 负面词数: {result['neg_count']}")
    print(f"情感得分: {result['score']}")
    print(f"情感类别: {result['sentiment']}")
    print("-"*80)
````

## 9. 方法的优缺点分析

### 9.1 优点

**1. 不需要训练数据**
- 只需要情感词典就可以使用
- 适合快速开发原型

**2. 实现简单**
- 代码逻辑清晰
- 容易理解和调试

**3. 可解释性强**
- 可以明确看到哪些词影响了结果
- 便于分析和改进

**4. 计算速度快**
- 只需要简单的字符串匹配
- 适合处理大量数据

**5. 领域适应性**
- 可以针对不同领域定制词典
- 灵活性高

### 9.2 缺点

**1. 依赖词典质量**
- 词典不完整会影响效果
- 需要持续维护和更新

**2. 难以处理复杂语言现象**
- 讽刺、反语难以识别
- 隐含情感难以捕捉

**3. 上下文理解能力弱**
- 无法理解词语在特定语境下的含义
- 例如："这部电影太沉重了"（可能是褒义）

**4. 准确率相对较低**
- 通常低于机器学习方法
- 对于复杂文本效果较差

**5. 新词识别困难**
- 网络流行语更新快
- 需要频繁更新词典

### 9.3 适用场景

**适合使用词典方法的场景：**

- 快速原型开发
- 没有训练数据
- 需要可解释的结果
- 领域词汇相对固定
- 文本表达相对直接

**不适合使用词典方法的场景：**

- 文本包含大量讽刺、反语
- 追求高准确率
- 文本表达复杂隐晦
- 需要细粒度情感分析
