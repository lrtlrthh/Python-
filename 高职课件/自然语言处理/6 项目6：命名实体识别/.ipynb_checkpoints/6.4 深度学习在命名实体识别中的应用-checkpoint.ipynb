## 2. 为什么需要深度学习？

### 2.1 传统方法的局限性

在前面的课程中，我们学习了两种方法：

**规则方法（词典+正则表达式）**
- ❌ 覆盖范围有限
- ❌ 无法处理新词
- ❌ 需要大量人工编写规则

**机器学习方法（CRF）**
- ❌ 需要人工设计特征
- ❌ 特征设计需要专业知识
- ❌ 难以捕捉复杂的语言模式

**示例：**

```
文本1："苹果公司发布了iPhone"
→ "苹果"是机构名（需要根据上下文判断）

文本2："我买了一个苹果"
→ "苹果"不是实体（需要理解语义）

文本3："库克是苹果的CEO"
→ 需要理解"库克"和"苹果"的关系
```

传统方法难以很好地处理这些复杂情况。

### 2.2 深度学习的优势

**深度学习方法**：使用神经网络自动学习特征和模式

**优势：**

✅ **自动特征学习**
```
不需要人工设计特征
神经网络自动学习什么特征重要
```

✅ **更强的语义理解**
```
可以理解词语的语义
可以捕捉长距离的依赖关系
```

✅ **更好的泛化能力**
```
对未见过的词和句子也能较好识别
迁移学习能力强
```

✅ **更高的准确率**
```
在各种NER任务上都取得了最好的效果
```

## 3. 深度学习基础概念

### 3.1 什么是神经网络？

**神经网络（Neural Network）**：模拟人脑神经元工作方式的计算模型

**简单理解：**

就像人脑处理信息的过程：
- **输入层**：接收信息（看到一句话）
- **隐藏层**：处理信息（理解句子的意思）
- **输出层**：得出结论（判断每个词是什么实体）

**示例：**

```
输入："马云在杭州创立阿里巴巴"
  ↓
隐藏层1：识别字符特征
隐藏层2：识别词语特征
隐藏层3：理解语义和上下文
  ↓
输出：B-PER I-PER O B-LOC I-LOC O O B-ORG I-ORG I-ORG I-ORG
```

### 3.2 什么是循环神经网络（RNN）？

**循环神经网络（Recurrent Neural Network，RNN）**：专门处理序列数据的神经网络

**特点：**
- 有"记忆"功能
- 处理当前输入时，会考虑之前的信息

**类比：**

就像看电影：
- 你不会只看当前这一帧画面
- 你会记住之前的剧情
- 结合前后情节理解当前发生的事

**在NER中的应用：**

```
处理"马云在杭州"：

看到"马" → 记住：可能是人名开始
看到"云" → 结合前面的"马"，确认：这是人名
看到"在" → 记住：人名已结束
看到"杭" → 记住：可能是地名开始
看到"州" → 结合前面的"杭"，确认：这是地名
```

### 3.3 什么是LSTM？

**长短期记忆网络（Long Short-Term Memory，LSTM）**：改进的RNN，记忆能力更强

**问题：**

普通RNN的记忆有限：
```
"马云出生于1964年，后来在杭州创立了阿里巴巴"
                              ↑
                        处理到这里时，
                        可能已经忘记前面说过"马云"
```

**LSTM的改进：**
- 可以记住长距离的信息
- 可以选择性地遗忘不重要的信息
- 可以选择性地记住重要的信息

**类比：**

就像笔记本：
- RNN：只能记住最近几句话
- LSTM：可以记录重点，随时查看，长期记忆

### 3.4 什么是BiLSTM？

**双向长短期记忆网络（Bidirectional LSTM，BiLSTM）**：同时从前往后和从后往前处理序列

**为什么需要双向？**

有时候后面的信息也很重要：

```
"苹果很好吃" → "苹果"不是实体（后面是"好吃"）
"苹果发布会" → "苹果"是机构名（后面是"发布会"）
```

**BiLSTM的工作方式：**

```
从左往右：马 → 云 → 在 → 杭 → 州
从右往左：州 → 杭 → 在 → 云 → 马

处理"云"时：
- 知道前面是"马"（从左往右）
- 知道后面是"在杭州"（从右往左）
- 综合判断："马云"是人名
```

## 4. BiLSTM-CRF模型

### 4.1 模型结构

**BiLSTM-CRF**：结合BiLSTM和CRF的优势

**模型架构：**

```
输入层（文本）
    ↓
词嵌入层（将字转换为向量）
    ↓
BiLSTM层（学习上下文特征）
    ↓
CRF层（保证标签序列合理）
    ↓
输出层（BIO标签）
```

**各层的作用：**

**1. 词嵌入层（Embedding Layer）**
```
将每个字转换为数值向量
"马" → [0.2, -0.5, 0.8, ...]
"云" → [0.1, 0.3, -0.2, ...]
语义相近的字，向量也相近
```

**2. BiLSTM层**
```
自动学习字符的上下文特征
不需要人工设计特征
可以捕捉长距离依赖
```

**3. CRF层**
```
确保标签序列合理
避免出现不合法的标签转移
如：B-PER后面不能直接跟I-LOC
```

### 4.2 BiLSTM-CRF的优势

**相比传统CRF：**

| 特点 | 传统CRF | BiLSTM-CRF |
|------|---------|------------|
| 特征提取 | 人工设计 | 自动学习 |
| 上下文理解 | 窗口有限 | 更长的上下文 |
| 语义理解 | 较弱 | 较强 |
| 准确率 | 中等 | 更高 |
| 训练时间 | 快 | 较慢 |
| 数据需求 | 较少 | 较多 |

**实际效果对比：**

```
在CoNLL 2003英文NER任务上：
- 传统CRF：F1约84%
- BiLSTM-CRF：F1约91%

在人民日报中文NER任务上：
- 传统CRF：F1约87%
- BiLSTM-CRF：F1约94%
```

### 4.3 简单示例代码（概念性）

下面是BiLSTM-CRF模型的简化代码框架，帮助理解其结构。

**注意**：这是概念性代码，实际运行需要更多配置和计算资源。

```python
# 使用PyTorch框架的BiLSTM-CRF模型示例（概念性）

import torch
import torch.nn as nn

class BiLSTM_CRF(nn.Module):
    def __init__(self, vocab_size, tag_size, embedding_dim, hidden_dim):
        super(BiLSTM_CRF, self).__init__()
        
        # 词嵌入层：将字转换为向量
        self.embedding = nn.Embedding(vocab_size, embedding_dim)
        
        # BiLSTM层：学习上下文特征
        self.lstm = nn.LSTM(
            embedding_dim,      # 输入维度
            hidden_dim // 2,    # 隐藏层维度（双向所以除以2）
            bidirectional=True  # 双向
        )
        
        # 输出层：将LSTM输出转换为标签分数
        self.hidden2tag = nn.Linear(hidden_dim, tag_size)
        
        # CRF层：保证标签序列合理
        # （这里简化省略，实际需要实现CRF的forward和viterbi解码）
    
    def forward(self, sentence):
        # 1. 词嵌入
        embeds = self.embedding(sentence)
        
        # 2. BiLSTM处理
        lstm_out, _ = self.lstm(embeds)
        
        # 3. 转换为标签分数
        tag_scores = self.hidden2tag(lstm_out)
        
        # 4. CRF解码（得到最优标签序列）
        # ...
        
        return tag_scores

# 模型参数说明：
# vocab_size: 词表大小（有多少个不同的字）
# tag_size: 标签数量（如B-PER, I-PER, B-LOC, I-LOC, O等）
# embedding_dim: 词向量维度（如100、200等）
# hidden_dim: LSTM隐藏层维度（如256、512等）
```

## 5. 预训练语言模型

### 5.1 什么是预训练模型？

**预训练模型（Pre-trained Model）**：在大规模文本上预先训练好的模型

**类比：**

就像学习一门外语：
- **从零开始学**：效率低，需要大量时间和数据
- **有语言基础再学**：效率高，迁移能力强

**预训练模型的思想：**

```
步骤1：预训练（Pre-training）
→ 在海量文本上训练（如维基百科、新闻等）
→ 学习通用的语言知识

步骤2：微调（Fine-tuning）
→ 在具体任务的数据上继续训练
→ 适应特定任务（如NER）
```

### 5.2 BERT模型

**BERT（Bidirectional Encoder Representations from Transformers）**：最著名的预训练模型之一

**特点：**

✅ **双向理解**
```
同时看前面和后面的上下文
更全面地理解词语的意思
```

✅ **强大的语义理解**
```
"银行"在不同语境的理解：
"去银行取钱" → 金融机构
"河岸边" → 地理位置
BERT可以根据上下文正确理解
```

✅ **迁移能力强**
```
在多种NLP任务上都表现优秀
只需少量标注数据就能获得好效果
```

**在NER中的应用：**

```
1. 使用预训练的BERT模型
2. 在BERT输出层加一个分类器
3. 在NER数据上微调
4. 获得比BiLSTM-CRF更好的效果
```

### 5.3 其他预训练模型

**中文预训练模型：**

**1. BERT-wwm（Whole Word Masking）**
- 针对中文优化的BERT
- 在中文NER任务上效果更好

**2. RoBERTa**
- BERT的改进版
- 训练策略优化
- 效果更好

**3. ERNIE（百度）**
- 融入知识图谱信息
- 更好地理解实体

**4. GPT系列**
- 生成式预训练模型
- 可以用于零样本或少样本NER

**效果对比：**

```
在中文NER任务上（人民日报数据集）：
- 传统CRF：F1约87%
- BiLSTM-CRF：F1约94%
- BERT-CRF：F1约96%
- BERT-wwm：F1约97%
```

### 5.4 使用预训练模型的优势和挑战

**优势：**

✅ **效果好**
- 在各种NER任务上都达到最佳效果

✅ **数据需求少**
- 有预训练知识，少量数据也能训练出好模型

✅ **迁移能力强**
- 可以快速适应不同领域

**挑战：**

❌ **计算资源要求高**
```
BERT模型很大（110M-340M参数）
训练和推理需要GPU
```

❌ **训练时间长**
```
即使微调也需要较长时间
预训练从零开始需要几天到几周
```

❌ **部署成本高**
```
模型大，占用内存多
推理速度较慢
不适合对延迟要求高的场景
```

## 6. 方法选择指南

### 6.1 不同方法对比

| 方法 | 优点 | 缺点 | 适用场景 |
|------|------|------|----------|
| **规则方法** | 简单快速、无需数据 | 覆盖有限、维护困难 | 格式固定的实体、快速原型 |
| **CRF** | 效果较好、可解释 | 需要特征工程 | 中等规模数据、领域适应 |
| **BiLSTM-CRF** | 自动特征、效果好 | 需要GPU、数据较多 | 充足数据和计算资源 |
| **BERT等预训练模型** | 效果最好、数据需求少 | 计算成本高、部署难 | 追求最高效果、有GPU资源 |

### 6.2 选择建议

**根据数据量选择：**

```
无标注数据 → 规则方法
少量数据（<1000句） → 规则方法 + 预训练模型微调
中等数据（1000-10000句） → CRF 或 BiLSTM-CRF
大量数据（>10000句） → BiLSTM-CRF 或 BERT
```

**根据计算资源选择：**

```
无GPU → 规则方法 或 CRF
有GPU但有限 → BiLSTM-CRF
GPU充足 → BERT等预训练模型
```

**根据应用场景选择：**

```
离线批处理 → 任何方法
实时在线服务 → 规则方法 或 CRF
对延迟要求低 → 深度学习方法
需要可解释性 → 规则方法 或 CRF
```

**混合方法：**

实际应用中，常常结合多种方法：

```
1. 规则方法识别格式固定的实体（电话、邮箱等）
2. BERT识别复杂实体（人名、地名、机构名等）
3. 后处理：使用词典和规则纠正错误
```

### 6.3 实践建议

**开发流程：**

```
步骤1：快速原型
→ 使用规则方法快速验证可行性
→ 评估任务难度

步骤2：数据标注
→ 标注少量数据（100-500句）
→ 训练CRF模型
→ 评估效果

步骤3：模型优化
→ 如果效果不够，增加数据
→ 尝试BiLSTM-CRF
→ 或使用预训练模型微调

步骤4：生产部署
→ 权衡效果和成本
→ 优化模型大小和速度
→ 持续监控和更新
```
