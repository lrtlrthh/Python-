## 2. 为什么需要特征提取？

### 2.1 计算机不懂文字

想象一下，你要教一个从未见过文字的外星人理解中文：

- 你说："这是一篇科技新闻"
- 外星人："什么是'科技'？什么是'新闻'？"

**计算机也是一样，它只认识数字，不认识文字。**

所以我们需要把文字转换成数字，这个过程就叫**特征提取**。

### 2.2 特征提取的目标

**把文本转换成数字向量，同时保留文本的重要信息。**

**示例**：

```
文本："华为发布新款手机"
      ↓ 特征提取
向量：[0.2, 0.5, 0.0, 0.8, 0.3, ...]
```

这个数字向量要能够代表文本的含义，让计算机能够区分不同类别的文本。

## 3. 词袋模型（Bag of Words）

### 3.1 什么是词袋模型？

**词袋模型**：把文本看作一个装满词语的袋子，只关心每个词出现了几次，不关心词的顺序。

**生活类比**：

想象你有一个购物袋，里面装着：
- 苹果 × 3个
- 香蕉 × 2个
- 橙子 × 1个

你不关心这些水果放进袋子的顺序，只关心每种水果有几个。

### 3.2 词袋模型的原理

词袋模型通过三个步骤将文本转换为数字向量：

**步骤1：建立词汇表**

收集所有文本中出现的词语，建立一个词汇表。

**步骤2：统计词频**

对每个文本，统计词汇表中每个词出现的次数。

**步骤3：生成向量**

把词频转换成数字向量。

### 3.3 词袋模型示例

让我们通过一个简单的例子来理解词袋模型。

**代码说明：**

- **enumerate()函数**：在遍历列表时同时获取"序号"和"内容"
  - 语法：`enumerate(列表, 起始序号)`
  - 例如：`for i, text in enumerate(texts, 1)` 表示从1开始编号
  - 返回：每次循环得到两个值（序号i 和 元素text）
  - 如果不指定起始序号，默认从0开始

````python
# 准备三个简单的文本
texts = [
    "我 喜欢 机器学习",
    "我 喜欢 深度学习",
    "机器学习 很 有趣"
]

print("原始文本：")
for i, text in enumerate(texts, 1):
    print(f"{i}. {text}")
````

#### 步骤1：建立词汇表

从所有文本中提取不重复的词语。

**代码说明：**

- **set()集合**：自动去除重复元素的数据结构
- **split()方法**：将字符串按空格分割成列表
- **update()方法**：批量添加元素到集合，自动去重
- **sorted()函数**：对集合排序，方便查看

````python
# 建立词汇表
vocabulary = set()
for text in texts:
    words = text.split()
    vocabulary.update(words)

vocabulary = sorted(vocabulary)  # 排序便于查看

print("词汇表：")
for i, word in enumerate(vocabulary):
    print(f"{i}: {word}")
````

#### 步骤2：统计词频并生成向量

对每个文本，统计词汇表中每个词出现的次数，然后转换成数字向量。

**什么是统计词频？**

词频 = 某个词在文本中出现的次数

例如："我 喜欢 机器学习" 中，"我"出现1次，"喜欢"出现1次，"机器学习"出现1次

**为什么要统计词频？**

因为计算机需要数字，不认识文字。我们要把每个文本变成一串数字（向量），这样计算机才能处理。

**如何生成向量？（重点理解）**

假设词汇表是：["喜欢", "机器学习", "我", "很", "有趣", "深度学习"]

现在要把文本"我 喜欢 机器学习"转换成向量，步骤如下：

**第一步：准备一个空向量，长度和词汇表一样**
```
向量 = [   ?   ,    ?    ,  ?  , ? ,   ?  ,    ?     ]
位置     第0位   第1位   第2位 第3位 第4位   第5位
对应词   喜欢   机器学习   我   很   有趣   深度学习
```

**第二步：填入每个词在文本中出现的次数**

看文本"我 喜欢 机器学习"中：
- 第0位对应"喜欢"，文本中"喜欢"出现了**1次** → 第0位填入数字1
- 第1位对应"机器学习"，文本中"机器学习"出现了**1次** → 第1位填入数字1  
- 第2位对应"我"，文本中"我"出现了**1次** → 第2位填入数字1
- 第3位对应"很"，文本中"很"出现了**0次** → 第3位填入数字0
- 第4位对应"有趣"，文本中"有趣"出现了**0次** → 第4位填入数字0
- 第5位对应"深度学习"，文本中"深度学习"出现了**0次** → 第5位填入数字0

**最终向量：**
```
向量 = [ 1 ,  1  , 1 , 0 , 0 ,  0  ]
       ↑    ↑     ↑    ↑   ↑     ↑
    喜欢出  机器  我   很  有   深度
    现1次  学习   出   出  趣   学习
          出现   现   现  出   出现
          1次   1次  0次  现   0次
                         0次
```

**关键理解：**
- 向量的**位置**（第几位）= 对应词汇表中的第几个词
- 向量的**值**（填什么数字）= 这个词在文本中出现了几次

"我"是词汇表第2个词（第2位），文本中"我"出现1次，所以第2位填入1

**代码说明：**

- **count()方法**：统计元素在列表中出现的次数
- **append()方法**：在列表末尾添加元素

````python
# 为每个文本生成词袋向量
def text_to_bow(text, vocabulary):
    """将文本转换为词袋向量"""
    words = text.split()  # 把文本分割成词语列表
    vector = []  # 创建空列表，用来存放向量
    
    # 遍历词汇表中的每个词
    for vocab_word in vocabulary:
        # 统计这个词在当前文本中出现了几次
        count = words.count(vocab_word)
        # 把次数添加到向量中
        vector.append(count)
    
    return vector  # 返回生成的向量

print("词袋向量表示：\n")
print(f"词汇表: {vocabulary}\n")

# 为每个文本生成向量
for i, text in enumerate(texts, 1):
    vector = text_to_bow(text, vocabulary)
    print(f"文本{i}: {text}")
    print(f"向量{i}: {vector}")
    print(f"  解释: 词汇表中每个词在文本{i}中出现的次数\n")
````

### 3.4 使用sklearn实现词袋模型

在实际应用中，我们使用sklearn库来实现词袋模型，更加方便高效。

**重要提示：**

sklearn的CountVectorizer会自动过滤掉**单个字符**的词（比如"我"、"很"），只保留2个字符及以上的词。这是为了过滤掉一些无意义的字符。

如果你想保留单字词，需要设置参数：`CountVectorizer(token_pattern=r'(?u)\b\w+\b')`

**代码说明：**

- **sklearn库**：Python最流行的机器学习库，安装：`pip install scikit-learn`
- **CountVectorizer类**：sklearn中的词袋模型实现
- **fit_transform()方法**：一次完成"学习词汇表"和"转换为向量"两个步骤
- **get_feature_names_out()方法**：获取学到的词汇表
- **toarray()方法**：将稀疏矩阵转为普通数组，方便查看

````python
from sklearn.feature_extraction.text import CountVectorizer

# 创建CountVectorizer对象
vectorizer = CountVectorizer()

# 拟合并转换文本
bow_matrix = vectorizer.fit_transform(texts)

# 查看词汇表
print("sklearn的词汇表：")
print(vectorizer.get_feature_names_out())

# 查看词袋矩阵
print("\nsklearn的词袋矩阵：")
print(bow_matrix.toarray())
````

**对比说明：**

- 手动实现的词汇表有6个词：['喜欢', '很', '我', '有趣', '机器学习', '深度学习']
- sklearn实现的词汇表只有4个词：['喜欢', '有趣', '机器学习', '深度学习']
- 原因：sklearn自动过滤掉了单字词'我'和'很'

### 3.5 词袋模型的优缺点

**优点：**
- 简单易懂
- 实现方便
- 计算速度快

**缺点：**
- 丢失了词序信息（"我喜欢你" 和 "你喜欢我" 的向量相同）
- 忽略了词的重要性（"的"、"了" 等常见词权重过高）
- 向量维度高（词汇表越大，向量越长）

## 4. TF-IDF

### 4.1 TF-IDF是什么？

**TF-IDF（Term Frequency-Inverse Document Frequency）**：词频-逆文档频率

TF-IDF是对词袋模型的改进，它不仅考虑词频，还考虑词的重要性。

### 4.2 TF-IDF的核心思想

**一个词越重要，需要同时满足两个条件：**

1. **在当前文档中出现频率高**（TF高）
2. **在其他文档中出现频率低**（IDF高）

**生活类比：**

假设你要写一篇关于"人工智能"的文章：
- "人工智能" 在你的文章中出现很多次 → TF高
- "人工智能" 在其他文章中很少出现 → IDF高
- 所以"人工智能"是你文章的关键词

但是：
- "的"、"了" 在你的文章中也出现很多次 → TF高
- 但"的"、"了" 在所有文章中都很常见 → IDF低
- 所以"的"、"了"不是关键词

### 4.3 TF-IDF的计算公式

**TF（词频）**：

```
TF = 某个词在文档中出现的次数 / 文档的总词数
```

**IDF（逆文档频率）**：

```
IDF = log(文档总数 / 包含该词的文档数)
```

**TF-IDF**：

```
TF-IDF = TF × IDF
```

### 4.4 TF-IDF手动计算示例

让我们通过一个例子来理解TF-IDF的计算过程。

**代码说明：**

- **math模块**：Python的数学计算模块
- **math.log()函数**：计算对数，用于IDF公式

````python
import math

# 准备文档（增加了更多词语，更接近真实文本）
documents = [
    "机器学习 是 人工智能 的 重要 分支 很 有趣 也 很 实用 值得 深入 学习",
    "深度学习 使用 神经网络 模型 很 强大 可以 处理 图像 和 文本 应用 广泛",
    "机器学习 和 深度学习 都 是 人工智能 领域 很 重要 的 技术 发展 很 快"
]

print("文档集合：")
for i, doc in enumerate(documents, 1):
    print(f"文档{i}: {doc}")
    print(f"       (共{len(doc.split())}个词)\n")
````

#### 计算文档1中"机器学习"的TF-IDF

**代码说明：**

- **生成器表达式**：`sum(1 for doc in documents if word in doc)` 是一种简洁的循环写法，统计包含某个词的文档数

````python
# 计算TF
doc1_words = documents[0].split()
word = "机器学习"

tf = doc1_words.count(word) / len(doc1_words)
print(f"文档1中'{word}'的TF:")
print(f"  出现次数: {doc1_words.count(word)}")
print(f"  总词数: {len(doc1_words)}")
print(f"  TF = {doc1_words.count(word)} / {len(doc1_words)} = {tf:.4f}")

# 计算IDF
total_docs = len(documents)
docs_with_word = sum(1 for doc in documents if word in doc)

idf = math.log(total_docs / docs_with_word)
print(f"\n'{word}'的IDF:")
print(f"  文档总数: {total_docs}")
print(f"  包含该词的文档数: {docs_with_word}")
print(f"  IDF = log({total_docs} / {docs_with_word}) = {idf:.4f}")

# 计算TF-IDF
tfidf = tf * idf
print(f"\n'{word}'的TF-IDF:")
print(f"  TF-IDF = {tf:.4f} × {idf:.4f} = {tfidf:.4f}")
````

#### 对比常见词"很"的TF-IDF

观察常见词的TF-IDF值如何变化。

````python
# 计算"很"的TF-IDF
word = "很"

tf = doc1_words.count(word) / len(doc1_words)
docs_with_word = sum(1 for doc in documents if word in doc)
idf = math.log(total_docs / docs_with_word)
tfidf = tf * idf

print(f"文档1中'{word}'的TF-IDF:")
print(f"  TF = {tf:.4f}")
print(f"  IDF = {idf:.4f} (出现在{docs_with_word}个文档中)")
print(f"  TF-IDF = {tfidf:.4f}")
````

'很'的TF-IDF很低，因为它在所有文档中都出现了

### 4.5 使用sklearn实现TF-IDF

sklearn提供了TfidfVectorizer，可以直接计算TF-IDF。

**代码说明：**

- **TfidfVectorizer类**：sklearn的TF-IDF实现，和CountVectorizer类似
- **注意**：sklearn的计算结果可能与手动计算略有不同，因为sklearn使用了标准化等优化技术

````python
from sklearn.feature_extraction.text import TfidfVectorizer

# 创建TfidfVectorizer对象
tfidf_vectorizer = TfidfVectorizer()

# 拟合并转换文本
tfidf_matrix = tfidf_vectorizer.fit_transform(documents)

# 查看词汇表
print("词汇表：")
feature_names = tfidf_vectorizer.get_feature_names_out()
print(feature_names)

# 查看TF-IDF矩阵
print("\nTF-IDF矩阵：")
print(tfidf_matrix.toarray())
````

### 4.6 可视化TF-IDF结果

让我们用表格的形式展示每个文档中每个词的TF-IDF值。

**代码说明：**

- **pandas库**：Python最常用的数据分析库，安装：`pip install pandas`
- **DataFrame**：pandas的表格数据结构，以表格形式展示数据
- **round()方法**：保留小数位数

````python
import pandas as pd

# 创建DataFrame展示结果
tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=feature_names,
    index=[f"文档{i+1}" for i in range(len(documents))]
)

print("TF-IDF值表格：\n")
print(tfidf_df.round(4))
````

观察哪些词的TF-IDF值较低、哪些词的TF-IDF值较高


### 4.7 TF-IDF的优缺点

**优点：**
- 考虑了词的重要性
- 降低了常见词的权重
- 提高了关键词的权重
- 在文本分类中效果好

**缺点：**
- 仍然丢失了词序信息
- 无法捕捉词的语义关系
- 向量维度仍然很高

## 5. 词向量简介

### 5.1 什么是词向量？

**词向量（Word Embedding）**：用低维稠密向量表示词语，能够捕捉词的语义信息。

**与词袋模型、TF-IDF的区别：**

| 方法 | 向量维度 | 语义信息 | 示例 |
|------|---------|---------|------|
| 词袋模型 | 高维稀疏 | 无 | [0, 1, 0, 0, 1, 0, ...] |
| TF-IDF | 高维稀疏 | 无 | [0, 0.5, 0, 0, 0.8, 0, ...] |
| 词向量 | 低维稠密 | 有 | [0.2, -0.5, 0.8, 0.1, -0.3] |

#### 什么是"向量维度"？

**向量维度 = 向量中有多少个数字**

就像：
- 2维向量：`[3, 5]` → 有2个数字
- 3维向量：`[1, 2, 3]` → 有3个数字  
- 100维向量：`[0.1, 0.2, ..., 0.9]` → 有100个数字

---

#### 高维稀疏 vs 低维稠密

**1. 词袋模型和TF-IDF：高维稀疏**

- **高维**：向量很长，可能有几千甚至上万个数字
  - 假设词汇表有10000个词
  - 那么每个文本的向量就有10000个数字

- **稀疏**：大部分数字都是0
  - 例如：`[0, 1, 0, 0, 1, 0, 0, 0, 0, ...]`
  - 一个文本只用到词汇表中的少数词，所以大部分位置是0

**示例：**
```
词汇表有10000个词
文本："我 喜欢 机器学习"（只有3个词）

向量：[0, 0, 0, ..., 1, 0, 0, ..., 1, 0, ..., 1, 0, 0, ...]
      ↑ 10000个数字，但只有3个是1，其余9997个都是0
```

**2. 词向量：低维稠密**

- **低维**：向量很短，通常只有100-300个数字
  - 不管词汇表有多大
  - 每个词都用固定长度的向量表示（比如100维）

- **稠密**：大部分数字都不是0
  - 例如：`[0.2, -0.5, 0.8, 0.1, -0.3, ...]`
  - 每个位置都有有意义的数值

**示例：**
```
词汇表有10000个词
但每个词只用100维向量表示

"机器学习" → [0.2, -0.5, 0.8, 0.1, -0.3, ..., 0.7]
              ↑ 只有100个数字，每个都有意义
```

---

#### 为什么维度很重要？

| 特点 | 高维稀疏（词袋/TF-IDF） | 低维稠密（词向量） |
|------|----------------------|------------------|
| 存储空间 | 大（虽然大部分是0） | 小 |
| 计算速度 | 慢 | 快 |
| 语义信息 | 无 | 有 |
| 适用场景 | 小数据集 | 大数据集 |

---

#### 直观理解

想象你要描述一个人：

**高维稀疏方式**（词袋模型）：
- 列出世界上所有可能的特征（10000个）
- 这个人有的特征标记为1，没有的标记为0
- `[0, 0, 1, 0, 0, 0, 1, 0, 0, ..., 0]`（很长，大部分是0）

**低维稠密方式**（词向量）：
- 只用几个关键维度描述（比如5个）
- 每个维度都有具体数值
- `[身高:1.75, 体重:70, 年龄:25, 智商:120, 情商:85]`（很短，每个都有意义）

### 5.2 词向量的特点

**1. 语义相似的词，向量也相似**

```
"国王" 和 "皇帝" 的向量很接近
"苹果" 和 "香蕉" 的向量很接近
```

**2. 可以进行向量运算**

```
"国王" - "男人" + "女人" ≈ "女王"
"北京" - "中国" + "日本" ≈ "东京"
```

**3. 维度低，计算效率高**

```
词袋模型：10000维（词汇表大小）
词向量：100-300维
```

### 5.3 常见的词向量模型

**1. Word2Vec**
- Google开发
- 最流行的词向量模型
- 训练速度快

**2. GloVe**
- 斯坦福大学开发
- 基于全局词频统计

**3. FastText**
- Facebook开发
- 考虑了词的内部结构
- 能处理未登录词

### 5.4 词向量的局限性

**注意：**

词向量是一个高级话题，需要大量数据和计算资源来训练。在本课程中，我们主要使用词袋模型和TF-IDF，它们更简单、更适合初学者。

**词向量的挑战：**
- 需要大量训练数据
- 训练时间长
- 需要较强的计算能力
- 对于小数据集，效果可能不如TF-IDF

## 6. 如何选择特征提取方法？

不同的任务和场景适合不同的特征提取方法。

### 6.1 选择建议

**使用词袋模型的场景：**
- 数据量小
- 需要快速实现
- 词频信息很重要
- 示例：垃圾邮件分类

**使用TF-IDF的场景：**
- 文本长度差异大
- 需要突出关键词
- 有很多常见词干扰
- 示例：新闻分类、文档检索

**使用词向量的场景：**
- 数据量大（至少几万条）
- 需要捕捉语义信息
- 有足够的计算资源
- 示例：情感分析、问答系统

### 6.2 实践建议

**对于初学者和小项目：**

1. **先尝试TF-IDF**
   - 效果通常不错
   - 实现简单
   - 计算快速

2. **如果效果不好，再尝试词袋模型**
   - 有时候简单的方法反而更好

3. **有经验后再考虑词向量**
   - 需要更多的调参经验
   - 需要更多的计算资源

## 7. 课程总结

### 7.1 本节课我们学习了：

**1. 特征提取的必要性**
- 计算机只认识数字，需要把文本转换成数字向量

**2. 词袋模型**
- 统计每个词出现的次数
- 简单但丢失了词序信息

**3. TF-IDF**
- 考虑词的重要性
- 降低常见词权重，提高关键词权重
- 在文本分类中效果好

**4. 词向量**
- 低维稠密向量
- 能捕捉语义信息
- 需要大量数据和计算资源

**5. 方法选择**
- 初学者推荐使用TF-IDF
- 根据任务特点选择合适的方法

### 7.2 关键要点

- 特征提取是文本分类的核心步骤
- TF-IDF是最常用的特征提取方法
- 没有最好的方法，只有最合适的方法
- 实践中要多尝试、多对比
