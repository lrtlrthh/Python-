## 2. 什么是分类算法？

### 2.1 生活中的分类

想象你是一个老师，需要教幼儿园小朋友如何识别水果：

- **苹果**：圆形、红色、表面光滑
- **橙子**：圆形、橙色、表面粗糙
- **香蕉**：长条形、黄色、表面光滑

小朋友通过学习这些特征，就能识别新的水果。

**分类算法就是让计算机学习这些规律，自动给新数据打标签。**

### 2.2 分类算法的工作原理

**训练阶段：**

```
输入：带标签的训练数据
  文本1 + 标签1
  文本2 + 标签2
  ...
  ↓
学习：找出特征和标签之间的关系
  ↓
输出：训练好的模型
```

**预测阶段：**

```
输入：新文本（没有标签）
  ↓
模型：根据学到的规律判断
  ↓
输出：预测的标签
```

## 3. 朴素贝叶斯算法

### 3.1 为什么叫朴素贝叶斯？

#### 3.1.1 什么是贝叶斯？

贝叶斯不是特指某一个算法，而是一套**核心逻辑思想**—— 简单说就是用结果反推原因的概率。

生活里到处都能用到这个思路：

比如看到衣服湿了（这是结果），你会猜原因：是下雨淋到了，还是不小心打翻了水？

贝叶斯的作用，就是帮你计算每种原因发生的概率，让你知道哪种原因更可能。

它的核心逻辑可以总结为：先有一个初步判断（比如平时下雨的概率），再根据新出现的结果（衣服湿了），调整对原因概率的判断，最终得到更准确的结论。

#### 3.1.2 什么是朴素？

朴素（Naive）是朴素贝叶斯独有的特点，不是笨，而是简化假设—— 让复杂的贝叶斯思路变得好算、好用。

这个简化假设就是：**导致结果的各个线索（专业叫特征）之间，互相独立、互不影响**。

举个通俗例子：

要判断一个水果是不是草莓（结果），我们有两个线索：① 颜色是红色 ② 形状是心形。

朴素的假设就是：颜色是红色和形状是心形这两件事没关系 —— 不管水果是不是红色，它是心形的概率不变；反之也一样。

为啥要做这个假设？因为如果不假设独立，线索之间可能有复杂关联（比如红色的水果更可能是心形），计算量会变得极大，普通人根本算不动。朴素就是帮我们化繁为简，让算法能快速出结果。

### 3.2 什么是朴素贝叶斯？

朴素贝叶斯 = 贝叶斯核心逻辑 + 线索互相独立的简化假设。

它是贝叶斯思路的**具体落地算法**—— 完全遵循用结果反推原因的逻辑，但因为加了朴素的假设，计算难度大幅降低，能直接用到实际场景中。

用判断水果是不是草莓举例，看它怎么工作：

#### 步骤 1：先积累已知经验（收集数据）

统计 100 个水果的情况，得到关键信息：

- 总水果中，草莓占 30 个（初步概率 30%，没看任何线索时，随便拿一个是草莓的概率）
- 草莓中：90% 是红色，80% 是心形
- 非草莓（比如苹果、橙子）中：20% 是红色，10% 是心形

#### 步骤 2：用新线索反推概率

现在来了一个水果：① 红色 ② 心形（两个线索），判断它是不是草莓。

根据线索独立的假设，计算超简单：

1. 算如果是草莓，同时满足红色+心形的概率：30%（草莓基础概率）× 90%（红色）× 80%（心形）= 21.6%
2. 算如果不是草莓，同时满足红色+心形的概率：70%（非草莓基础概率）× 20%（红色）× 10%（心形）= 1.4%
3. 比大小：21.6% 远大于 1.4%，所以判断这个水果是草莓

### 3.3 朴素贝叶斯的优点

**1. 简单高效**
- 训练速度快
- 预测速度快
- 代码实现简单

**2. 对小数据集效果好**
- 不需要大量训练数据
- 适合快速原型开发

**3. 对文本分类效果好**
- 特别适合垃圾邮件过滤
- 适合新闻分类
- 适合情感分析

### 3.4 使用sklearn实现朴素贝叶斯

下面我们通过一个简单的**新闻分类例子**（科技新闻 vs 体育新闻）来学习如何使用朴素贝叶斯分类器。

#### 准备阶段：了解需要用到的工具包

在开始之前，我们需要了解两个重要的工具：

**1. MultinomialNB（多项式朴素贝叶斯分类器）**
- 作用：实现朴素贝叶斯算法的分类器
- 为什么叫多项式？因为它特别适合处理文本中的词频数据
- 就像一个可以训练的判断专家

**2. TfidfVectorizer（TF-IDF向量化器）**
- 作用：把文本转换成数字向量
- 为什么需要？因为计算机只认识数字，不认识文字
- 例如：华为 发布 新款 手机 → [0.5, 0.3, 0.8, 0.6, ...]

现在让我们开始实现：

````python
# 导入需要的工具包
from sklearn.naive_bayes import MultinomialNB
from sklearn.feature_extraction.text import TfidfVectorizer

print("工具包导入成功！")
````

#### 步骤1：准备训练数据

首先，我们需要准备一些带标签的训练数据。

**什么是训练数据？**
- 就是一些已经知道类别的文本
- 每条文本都有对应的标签（比如科技或体育）
- 计算机通过学习这些数据，找出哪些词通常出现在哪个类别

**我们的例子：**
- 科技新闻：通常包含手机、发布、芯片、技术等词
- 体育新闻：通常包含比赛、进球、球员、赛季等词

````python
# 准备训练文本（已分词）
train_texts = [
    # 科技类新闻（8条）
    "华为 发布 新款 旗舰 手机 搭载 最新 芯片",
    "小米 推出 新 技术 提升 充电 速度",
    "苹果 公司 发布 最新 操作系统 更新",
    "人工智能 技术 在 医疗 领域 取得 突破",
    "新款 笔记本 电脑 性能 大幅 提升",
    "5G 网络 覆盖 范围 持续 扩大",
    "芯片 制造 工艺 实现 新 突破",
    "智能 手表 新 功能 受到 用户 欢迎",
    
    # 体育类新闻（8条）
    "中国 足球队 在 比赛 中 获得 胜利",
    "篮球 运动员 创造 单场 得分 新 纪录",
    "世界杯 足球赛 精彩 进球 集锦",
    "羽毛球 选手 夺得 冠军 奖牌",
    "NBA 球员 转会 消息 引发 关注",
    "乒乓球 运动员 在 比赛 中 表现 出色",
    "马拉松 比赛 吸引 众多 跑步 爱好者",
    "游泳 选手 打破 世界 记录"
]

# 对应的标签
train_labels = [
    # 前8条是科技类
    "科技", "科技", "科技", "科技", "科技", "科技", "科技", "科技",
    # 后8条是体育类
    "体育", "体育", "体育", "体育", "体育", "体育", "体育", "体育"
]

# 查看数据
print("训练数据准备完成！")
print(f"共有 {len(train_texts)} 条训练数据")
print(f"其中科技类 {train_labels.count('科技')} 条，体育类 {train_labels.count('体育')} 条")
print("\n示例数据：")
print(f"文本: {train_texts[0]}")
print(f"标签: {train_labels[0]}")
````

#### 步骤2：文本向量化（转换成数字）

计算机不认识文字，只认识数字。所以我们需要把文本转换成数字向量。

**TfidfVectorizer 的作用：**
- 把每个文本转换成一个数字列表
- 数字的大小代表这个词在文本中的重要程度

**主要方法：**
- `fit_transform(texts)`：学习所有词汇，并转换文本（用于训练数据）
- `transform(texts)`：使用已学习的词汇转换文本（用于测试数据）

**简单理解：**
```
输入: ["华为 发布 手机", "足球 比赛 进球"]
输出: 一个数字矩阵，每行代表一个文本
```

````python
# 创建向量化器
vectorizer = TfidfVectorizer()

# 将训练文本转换为数字向量
X_train = vectorizer.fit_transform(train_texts)

# 查看转换结果
print("文本向量化完成！")
print(f"向量矩阵的形状: {X_train.shape}")
print(f"  - {X_train.shape[0]} 行表示有 {X_train.shape[0]} 条文本")
print(f"  - {X_train.shape[1]} 列表示提取了 {X_train.shape[1]} 个不同的词")

# 查看提取的部分词汇
feature_names = vectorizer.get_feature_names_out()
print(f"\n提取的词汇示例（前10个）: {list(feature_names[:10])}")
````

#### 步骤3：训练朴素贝叶斯模型

现在我们有了数字向量和对应的标签，可以开始训练模型了。

**训练过程：**
- 输入：数字向量（X_train）+ 标签（train_labels）
- 过程：计算每个词在不同类别中出现的概率
- 输出：训练好的分类器

**MultinomialNB 的方法：**
- `fit(X, y)`：训练模型
  - X：特征矩阵（数字向量）
  - y：标签列表

**简单理解：**
```
模型学习：
  手机这个词经常出现在科技类 → 记住这个规律
  进球这个词经常出现在体育类 → 记住这个规律
```

````python
# 创建朴素贝叶斯分类器
classifier = MultinomialNB()

# 训练模型
classifier.fit(X_train, train_labels)

print("模型训练完成！")
print(f"模型已学会识别 {len(classifier.classes_)} 个类别: {list(classifier.classes_)}")
````

#### 步骤4：预测新文本的类别

现在我们的模型已经训练好了，可以用来预测新文本的类别。

**预测流程：**
```
新文本 → vectorizer.transform() 转成数字 
       → classifier.predict() 预测类别 
       → 得到预测结果
```

**重要提示：**
- 新文本要用 `transform()` 而不是 `fit_transform()`
- 为什么？因为要使用训练时学到的词汇表，不能重新学习

**使用的方法：**
- `vectorizer.transform(texts)`：将新文本转换成数字向量
- `classifier.predict(X)`：预测类别

````python
# 准备测试数据（新的、没见过的文本）
test_texts = [
    "小米 公司 发布 最新 5G 手机",
    "中国队 在 足球 比赛 中 进球 得分",
    "人工智能 芯片 技术 获得 突破",
    "篮球 运动员 在 比赛 中 夺得 冠军"
]

# 第一步：将测试文本转换为数字向量
X_test = vectorizer.transform(test_texts)

# 第二步：使用模型进行预测
predictions = classifier.predict(X_test)

# 显示预测结果
print("预测结果：")
print("=" * 50)
for i, (text, pred) in enumerate(zip(test_texts, predictions)):
    print(f"\n文本{i+1}: {text}")
    print(f"预测类别: {pred}")
````

#### 步骤5：查看预测概率

朴素贝叶斯不仅能告诉我们预测的类别，还能告诉我们每个类别的概率，这样我们可以知道模型有多确定。

**使用的方法：**
- `classifier.predict_proba(X)`：预测每个类别的概率
  - 返回一个矩阵，每行是一个文本，每列是一个类别的概率
  - 所有概率加起来等于 1（即100%）
  
- `classifier.classes_`：获取所有类别名称

**概率的含义：**
```
如果显示：
  科技: 85.32%  ← 概率高，模型很确定
  体育: 14.68%
  
说明模型认为这个文本有 85.32% 的可能是科技类新闻
```

````python
# 获取预测概率
probabilities = classifier.predict_proba(X_test)

# 获取类别名称
classes = classifier.classes_

# 显示详细的预测概率
print("预测概率详情：")
print("=" * 50)

for i, text in enumerate(test_texts):
    print(f"\n文本{i+1}: {text}")
    print(f"预测结果: {predictions[i]}")
    print("各类别概率:")
    
    for j, class_name in enumerate(classes):
        prob = probabilities[i][j]
        # 用方块表示概率大小（可视化）
        bar = "█" * int(prob * 20)
        print(f"  {class_name}: {prob:6.2%}  {bar}")
````

#### 小结：朴素贝叶斯实现的完整流程

让我们回顾一下刚才学习的完整流程：

```
1. 导入工具包
   ↓
2. 准备训练数据（文本 + 标签）
   ↓
3. 文本向量化（TfidfVectorizer）
   ↓
4. 训练模型（MultinomialNB.fit()）
   ↓
5. 预测新文本（transform + predict）
   ↓
6. 查看预测概率（predict_proba）
```

**关键要点：**
- 训练数据要用 `fit_transform()`
- 测试数据要用 `transform()`（不能用 fit_transform）
- 可以通过概率值判断模型的置信度

## 4. 其他常用分类算法

除了朴素贝叶斯，还有很多其他分类算法可以用于文本分类。

### 4.1 逻辑回归（Logistic Regression）

**原理：**
- 通过线性组合特征，计算属于每个类别的概率
- 虽然名字叫回归，但实际是分类算法

**优点：**
- 训练速度快
- 效果稳定
- 可解释性强

**适用场景：**
- 二分类问题（如垃圾邮件识别）
- 特征维度高的问题

````python
from sklearn.linear_model import LogisticRegression

# 创建逻辑回归分类器
lr_classifier = LogisticRegression(max_iter=1000)

# 训练模型
lr_classifier.fit(X_train, train_labels)

# 预测
lr_predictions = lr_classifier.predict(X_test)

print("逻辑回归预测结果：")
print("=" * 50)
for text, pred in zip(test_texts, lr_predictions):
    print(f"文本: {text}")
    print(f"预测类别: {pred}\n")
````

### 4.2 支持向量机（SVM）

**原理：**
- 找到一个最优的分界线（或超平面）
- 使不同类别之间的间隔最大化

**优点：**
- 在高维空间表现好
- 对小样本数据效果好
- 泛化能力强

**缺点：**
- 训练时间较长
- 对大数据集不友好

**适用场景：**
- 文本分类
- 图像分类
- 数据量中等的问题

````python
from sklearn.svm import SVC

# 创建SVM分类器
svm_classifier = SVC(kernel='linear', probability=True)

# 训练模型
svm_classifier.fit(X_train, train_labels)

# 预测
svm_predictions = svm_classifier.predict(X_test)

print("SVM预测结果：")
print("=" * 50)
for text, pred in zip(test_texts, svm_predictions):
    print(f"文本: {text}")
    print(f"预测类别: {pred}\n")
````

### 4.3 随机森林（Random Forest）

**原理：**
- 构建多个决策树
- 每棵树独立投票
- 最终结果由多数投票决定

**优点：**
- 准确率高
- 不容易过拟合
- 能处理高维数据

**缺点：**
- 模型较大
- 预测速度较慢

**适用场景：**
- 特征较多的问题
- 需要高准确率的场景

````python
from sklearn.ensemble import RandomForestClassifier

# 创建随机森林分类器
rf_classifier = RandomForestClassifier(n_estimators=100, random_state=42)

# 训练模型
rf_classifier.fit(X_train, train_labels)

# 预测
rf_predictions = rf_classifier.predict(X_test)

print("随机森林预测结果：")
print("=" * 50)
for text, pred in zip(test_texts, rf_predictions):
    print(f"文本: {text}")
    print(f"预测类别: {pred}\n")
````

## 5. 算法对比

让我们对比一下不同算法在同一数据上的表现。

````python
# 对比所有算法的预测结果
print("算法对比：")
print("=" * 60)

for i, text in enumerate(test_texts):
    print(f"\n文本{i+1}: {text}")
    print("-" * 60)
    print(f"朴素贝叶斯: {predictions[i]}")
    print(f"逻辑回归:   {lr_predictions[i]}")
    print(f"SVM:       {svm_predictions[i]}")
    print(f"随机森林:   {rf_predictions[i]}")
````

## 6. 如何选择分类算法？

不同的场景适合不同的算法，没有绝对最好的算法。

### 6.1 选择建议

| 场景 | 推荐算法 | 原因 |
|------|---------|------|
| 数据量小（<1000条） | 朴素贝叶斯 | 训练快，不易过拟合 |
| 数据量中等（1000-10000条） | 逻辑回归、SVM | 效果稳定，速度适中 |
| 数据量大（>10000条） | 逻辑回归、随机森林 | 训练速度可接受 |
| 需要快速原型 | 朴素贝叶斯 | 实现简单，训练快 |
| 追求高准确率 | 随机森林、SVM | 模型复杂，效果好 |
| 需要可解释性 | 逻辑回归 | 可以看到特征权重 |

### 6.2 实践建议

**对于初学者：**

1. **先用朴素贝叶斯**
   - 最简单
   - 效果通常不错
   - 适合快速验证想法

2. **再试逻辑回归**
   - 效果往往更好
   - 训练速度也很快

3. **有时间再试其他算法**
   - SVM、随机森林
   - 对比效果，选择最好的

**记住：**
- 没有最好的算法，只有最合适的算法
- 实践中要多尝试、多对比
- 数据质量比算法选择更重要
