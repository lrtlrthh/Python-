## 2.1 NumPy - 数值计算库

NumPy是Python科学计算的基础库，提供高效的数组操作

```python
import numpy as np

# 创建数组 - 存储新闻点击量
clicks = np.array([120, 350, 280, 500, 190])
print("新闻点击量:", clicks)
```

```python
# NumPy的基本统计功能
clicks = np.array([120, 350, 280, 500, 190])
print(f"平均点击量: {clicks.mean():.1f}")
print(f"最高点击量: {clicks.max()}")
print(f"最低点击量: {clicks.min()}")
print(f"总点击量: {clicks.sum()}")
```

```python
# 数组运算 - 计算增长率
last_week = np.array([100, 300, 250, 450, 180])
this_week = np.array([120, 350, 280, 500, 190])

growth_rate = ((this_week - last_week) / last_week) * 100
print("每条新闻的点击增长率(%):", growth_rate.round(2))
```

## 2.2 Pandas - 数据分析库

Pandas是Python最强大的数据分析库，提供DataFrame数据结构

```python
import pandas as pd

# 创建新闻数据表
news_data = {
    "标题": ["AI技术突破", "体育赛事报道", "经济形势分析", "娱乐圈新闻"],
    "类别": ["科技", "体育", "财经", "娱乐"],
    "字数": [800, 600, 1200, 500],
    "点击量": [5000, 3000, 4500, 8000]
}

df = pd.DataFrame(news_data)
print("新闻数据表:")
print(df)
```

```python
# 查看数据基本信息
print("数据表形状:", df.shape)
print("列名:", df.columns.tolist())
print("数据类型:")
print(df.dtypes)
```

```python
# 数据筛选 - 找出点击量超过4000的新闻
popular_news = df[df["点击量"] > 4000]
print("热门新闻（点击量>4000）:")
print(popular_news)
```

```python
# 数据统计
print("数值列统计信息:")
print(df[["字数", "点击量"]].describe())
```

```python
# 按类别分组统计
category_stats = df.groupby("类别")["点击量"].mean()
print("各类别平均点击量:")
print(category_stats)
```

```python
# 添加新列 - 计算每字点击率
df["每字点击率"] = (df["点击量"] / df["字数"]).round(2)
print("添加每字点击率后:")
print(df)
```

## 2.3 Jieba - 中文分词库

Jieba是最流行的中文分词库，在NLP任务中必不可少

```python
# !pip install jieba -q
```

```python
import jieba

# 中文分词示例
text = "自然语言处理是人工智能的重要分支"
words = jieba.cut(text)

print("原文:", text)
print("分词结果:", " / ".join(words))
```

```python
# 精确模式 vs 全模式
text = "我来到北京清华大学"

words_precise = jieba.cut(text, cut_all=False)
print("精确模式:", " / ".join(words_precise))

words_full = jieba.cut(text, cut_all=True)
print("全模式:", " / ".join(words_full))
```

```python
# 词频统计
from collections import Counter

text = "人工智能技术正在改变世界，人工智能的应用越来越广泛，技术创新推动社会进步"
words = jieba.cut(text)

words_filtered = [w for w in words if len(w) > 1]
word_counts = Counter(words_filtered)

print("词频统计（前5）:")
for word, count in word_counts.most_common(5):
    print(f"- {word}: {count}次")
```
