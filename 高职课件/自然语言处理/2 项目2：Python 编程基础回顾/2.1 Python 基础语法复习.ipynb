# 1. Python基础语法复习



## 1.1 变量与数据类型

Python中最常用的基本数据类型包括：

- **字符串（str）**：文本数据
- **整数（int）**：整数数字
- **浮点数（float）**：小数
- **列表（list）**：有序的可变集合
- **字典（dict）**：键值对集合

```python
# 字符串示例
news_title = "人工智能改变世界"
print(f"新闻标题: {news_title}")
print(f"标题长度: {len(news_title)} 个字符")
```

```python
# 数字类型示例
word_count = 500
reading_speed = 250.5
reading_time = word_count / reading_speed

print(f"文章共 {word_count} 字")
print(f"预计阅读时间: {reading_time:.2f} 分钟")
```

```python
# 列表示例 - 存储多篇新闻标题
news_list = [
    "科技公司发布新产品",
    "体育赛事精彩纷呈",
    "经济数据持续向好"
]

print("新闻列表:")
for i, news in enumerate(news_list, 1):
    print(f"{i}. {news}")
```

```python
# 字典示例 - 存储新闻详细信息
news_info = {
    "标题": "人工智能技术突破",
    "类别": "科技",
    "作者": "张三",
    "字数": 1200
}

print("新闻详情:")
for key, value in news_info.items():
    print(f"{key}: {value}")
```

## 1.2 字符串操作

在文本处理中，字符串操作非常重要。常用操作包括：

- **拼接**：使用 + 或 f"" 格式化
- **切片**：获取子字符串
- **分割**：使用 .split()
- **替换**：使用 .replace()
- **查找**：使用 .find() 或 in

```python
# 字符串拼接
title = "Python编程"
subtitle = "数据处理基础"
full_title = title + " - " + subtitle
print(full_title)
```

```python
# 字符串切片
text = "自然语言处理技术应用广泛"
print(f"前5个字: {text[:5]}")
print(f"后5个字: {text[-5:]}")
print(f"中间部分: {text[4:8]}")
```

```python
# 字符串分割 - 处理句子
sentence = "Python是一种简单、易学、功能强大的编程语言"
words = sentence.split("、")
print("分割后的词组:")
for word in words:
    print(f"- {word}")
```

```python
# 字符串替换和查找
news = "今天天气很好，适合出门游玩。"

new_news = news.replace("今天", "明天")
print(f"替换后: {new_news}")

if "天气" in news:
    print("这是一条天气相关的新闻")
```

## 1.3 控制流程



### 1.3.1 条件判断（if-elif-else）

根据不同条件执行不同的代码

```python
# 根据新闻字数分类
word_count = 800

if word_count < 500:
    news_type = "短新闻"
elif word_count < 1000:
    news_type = "中等新闻"
else:
    news_type = "长新闻"

print(f"这是一篇{news_type}（{word_count}字）")
```

### 1.3.2 循环（for 和 while）

重复执行某段代码

```python
# for循环示例 - 处理多条新闻
news_categories = ["科技", "体育", "财经", "娱乐", "社会"]

print("新闻分类:")
for index, category in enumerate(news_categories, 1):
    print(f"{index}. {category}类新闻")
```

```python
# while循环示例 - 统计字数
text = "人工智能是未来科技发展的重要方向"
count = 0
index = 0

while index < len(text):
    if text[index] == "是":
        count += 1
    index += 1

print(f"'是'字出现了 {count} 次")
```

## 1.4 函数定义

函数可以让代码更加模块化、易于重用

```python
# 定义一个简单的文本处理函数
def clean_text(text):
    """清理文本：去除首尾空格"""
    cleaned = text.strip()
    return cleaned

# 测试函数
raw_text = "  Python编程很有趣  "
result = clean_text(raw_text)
print(f"原文本: '{raw_text}'")
print(f"清理后: '{result}'")
```

```python
# 带参数和返回值的函数
def analyze_news(title, content):
    """分析新闻基本信息"""
    info = {
        "标题长度": len(title),
        "内容长度": len(content),
        "总长度": len(title) + len(content)
    }
    return info

# 使用函数
news_title = "科技改变生活"
news_content = "随着科技的发展，人们的生活方式发生了巨大变化。"

result = analyze_news(news_title, news_content)
print("新闻分析结果:")
for key, value in result.items():
    print(f"- {key}: {value}")
```

## 1.5 列表推导式

列表推导式是Python的高效特性，可以快速创建列表

```python
# 传统方式：筛选长标题
titles = ["AI应用", "深度学习技术发展", "NLP", "自然语言处理技术应用"]

long_titles = []
for title in titles:
    if len(title) > 5:
        long_titles.append(title)

print("传统方式结果:", long_titles)
```

```python
# 列表推导式：更简洁
titles = ["AI应用", "深度学习技术发展", "NLP", "自然语言处理技术应用"]
long_titles = [title for title in titles if len(title) > 5]

print("列表推导式结果:", long_titles)
```
