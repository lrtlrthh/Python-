# <span class="motutor-highlight motutor-id_ennncoo-id_3i59qa3"><i></i>使用 DeepSeek 实现 RAG 应用</span>

随着人工智能技术的不断发展，大规模预训练语言模型（LLM）在自然语言处理（NLP）领域取得了巨大突破。

然而，在面对海量数据和特定领域知识时，单一生成模型存在一定局限。

为了更好地结合检索系统与生成模型，RAG 应运而生。

RAG 技术通过引入外部检索模块，能够将领域相关的知识检索出来，并与生成模型进行融合，从而提高回答的准确性和覆盖率。

DeepSeek 大模型作为一种新兴技术方案，其在高效知识检索与文本生成方面具有独到优势，因此成为了构建 RAG 应用的理想选择。

## <span class="motutor-highlight motutor-id_4fgr69n-id_f7mi0v5"><i></i>1. DeepSeek 大模型简介</span>

### 1.1 模型架构与核心设计
DeepSeek 大模型是近年来针对知识密集型任务开发的一款新型大规模预训练模型，其在 Transformer 架构的基础上进行了一系列优化与改进，主要体现在以下几个方面：

- **深度与宽度的优化**：模型通过增加层数和参数量，提升了对复杂语义信息的捕捉能力。

- **多任务预训练策略**：在预训练阶段，DeepSeek 同时学习了生成、问答、信息检索等多种任务，使其在跨领域应用中具备更强的适应性。

- **高效推理与响应机制**：针对实时应用场景，模型在推理阶段进行诸如量化、剪枝等优化措施，确保在保证质量的同时达到低延迟响应。

### <span class="motutor-highlight motutor-id_xbmktcx-id_qyotxl1"><i></i>1.2 生成与检索能力的协同</span>
DeepSeek 模型不仅在自然语言生成上表现出色，同时还具备优秀的信息检索能力。具体来说：

- **语义向量表示**：利用深层语义编码器将输入文本转化为高维向量，这种向量化表示能够捕捉到文本中深层的语义特征，为后续检索提供基础。

- **动态信息融合**：通过与检索模块无缝对接，DeepSeek 可以在生成过程中动态整合外部知识，从而弥补训练数据中的知识盲点。

- **上下文增强生成**：生成模型在接收查询时，不仅仅依赖内部固有知识，而是结合检索到的相关文档，通过上下文增强机制生成更为准确和细致的回答。

### <span class="motutor-highlight motutor-id_pszohk1-id_00uttpe"><i></i>1.3 RAG 应用中 DeepSeek 的角色</span>
- **动态知识更新：**
传统生成模型存在知识更新滞后的问题，而在 RAG 应用中，DeepSeek 通过与实时检索模块相结合，可以获取最新的外部信息。这一机制确保系统在面对实时性要求较高的场景（如在线新闻、社交媒体分析等）时，依然能够提供准确有效的答案。

- **提高生成精度：**
通过引入外部检索结果，DeepSeek 在生成阶段能够获得更加丰富的上下文信息，从而减少因数据不足或领域偏差引起的生成错误。外部信息为生成模型提供了明确的事实支撑，使得生成内容不仅语言流畅，同时具备较高的准确性和可信度。

- **系统模块化设计：**
在 RAG 系统中，DeepSeek 模型通常作为生成引擎存在，而检索模块则独立运作。这种模块化设计具有明显优势：

    - **可扩展性**：检索模块和生成模块可以分别独立升级或更换，便于系统的持续迭代与优化。

    - **灵活部署**：根据不同的业务需求，可以定制检索策略和生成策略，实现跨领域的定制化应用。

    - **高效协同**：各模块之间通过标准化接口通信，既保证了系统整体的灵活性，又确保了数据在各环节中的高效传递。

## <span class="motutor-highlight motutor-id_l4ixucj-id_kksox52"><i></i>2. RAG 的基石：（数据准备与向量化)</span>

现在，我们要为 RAG 系统准备“检索”环节所需的“燃料”——也就是我们的外部知识库，并学习如何将其转化为机器能够理解和检索的格式。

### <span class="motutor-highlight motutor-id_ap49827-id_va2n3b4"><i></i>2.1 准备你的知识库</span>

RAG 的威力来自于它所连接的知识库。这个知识库可以包含任何你希望 LLM 能够参考的信息。来源可以多种多样：

本地文件：
- 纯文本文件 (.txt)
- PDF 文件 (.pdf)
- Markdown 文件 (.md)
- CSV 或 Excel 文件 (.csv, .xlsx)
- JSON 或 JSONL 文件 (.json, .jsonl)
- Word 文档 (.docx) - 可能需要额外库处理
- 数据库： SQL 数据库、NoSQL 数据库中的文本内容。
- 网页内容： 通过爬虫抓取的特定网站信息。
- API 数据： 从其他服务 API 获取的文本数据。

选择知识库的关键考量：

- **相关性**： 知识库内容应与你期望 RAG 系统回答的问题领域高度相关。

- **质量**： 信息的准确性、完整性和时效性直接影响最终答案的质量。尽量使用可靠来源的数据。

- **格式**： 选择你易于处理和解析的格式。对于非结构化文本（如 .txt, .pdf, .md），处理相对直接。结构化数据（如 .csv, .json）可能需要提取出关键文本字段。

### <span class="motutor-highlight motutor-id_ggnju3x-id_ps8jofw"><i></i>2.2 文本加载与分割（Loading & Chunking）策略</span>

原始文档通常太长，无法直接、有效地用于检索或放入 LLM 的上下文窗口（Context Window）。

LLM 的上下文窗口是指模型一次能处理的文本长度（以 token 计），超过这个长度的信息会被忽略。

同时，检索的目标是找到与问题最相关的 具体 信息片段，而不是整个冗长的文档。

因此，我们需要两个步骤：

- **加载（Loading）**： 从源（文件、数据库等）读取文本内容到程序中。

- **分割（Chunking / Splitting）**： 将加载的长文本分割成更小的、有意义的、相互之间可能略有重叠（Overlap）的块（Chunks）。

**加载（Loading）**：

对于简单的 .txt 文件，Python 的内置文件操作即可完成，对于 PDF、Word 等复杂格式，通常需要借助第三方库，如 pypdf (用于 PDF), python-docx (用于 Word)。

LangChain 或 LlamaIndex 等框架提供了更方便的文档加载器（Document Loaders）来处理多种格式。

**分割（Chunking）**：

将文本分割成块是 RAG 中的关键一步。好的分块策略能显著提升检索效果。常见策略包括：

- **固定大小分块（Fixed Size Chunking）**： 按固定字符数（或 token 数）分割，可以设置块之间的重叠（Overlap）。

    - 优点： 实现简单。
    - 缺点： 可能在句子或段落中间断开，破坏语义完整性。
    
    
- **重叠（Overlap）**： 设置重叠（如 50 个字符）有助于保留块边界处的上下文连续性。

- **递归字符分割（Recursive Character Text Splitting）**： 尝试按一系列分隔符（如 \n\n 段落, \n 换行, . 句子,  空格）递归地分割文本，直到块大小满足要求。这是 LangChain 等框架常用的策略，通常效果较好。

- **语义分块（Semantic Chunking）**：（较高级）利用 Embedding 模型来判断文本的语义边界进行分割，力求让每个块包含相对独立的语义单元。

### <span class="motutor-highlight motutor-id_ecud5q6-id_simuw7i"><i></i>2.3 文本向量化（Embeddings）基础</span>

机器无法直接理解文本的“含义”。

为了让计算机能够比较文本之间的相似性（例如，判断哪个文档块与用户的问题最相关），我们需要将文本转换成向量（Vectors），这是一种数学表示形式（通常是包含几百到几千个数字的列表/数组）。

这个转换过程叫做文本向量化或生成嵌入（Generating Embeddings）。

如下图将 man、king、woman 和 queen 这几个词映射到的向量空间：
![](https://imgbed.momodel.cn/20250411110735997.png)

核心思想：通过一个 Embedding 模型，将具有相似语义的文本映射到向量空间中相近的位置。

- **Embedding 模型**：这是一个预训练好的神经网络模型，专门用于理解文本并将其转换为向量。常见的有 Google 的 Universal Sentence Encoder (USE), Sentence-BERT (以及其各种变体，如 all-MiniLM-L6-v2), OpenAI 的 text-embedding-ada-002，以及现在很多公司/机构提供的模型。

- **向量（Vector）**： 一个由数字组成的列表（例如 [0.12, -0.45, 0.88, ..., -0.05]）。向量的维度（列表中的数字个数）由所使用的 Embedding 模型决定（例如，all-MiniLM-L6-v2 生成 384 维向量）。

- **语义相似性**： 如果两个文本片段在含义上相近（即使措辞不同），它们的向量在向量空间中的距离也会很近（通常用 余弦相似度 Cosine Similarity 来衡量）。

## <span class="motutor-highlight motutor-id_6hpuohj-id_e8b7g6h"><i></i>3. 信息检索：找到相关的知识片段</span>

有了向量化的知识块，我们需要一种方法来存储它们，并能根据用户问题的向量快速找到最相似的知识块向量。

这就是信息检索环节，通常借助向量数据库（Vector Database）或向量索引（Vector Index）来实现。

### <span class="motutor-highlight motutor-id_yz0pvsl-id_zm6ww9m"><i></i>3.1 向量数据库（Vector Database）简介</span>

向量数据库是专门设计用来存储、管理和高效查询高维向量数据的数据库系统。

想象一下，你有数百万甚至数十亿个几百维的向量（对应你的知识库文档块），当用户提出问题时，你需要：

将用户问题也转换成相同维度的向量（使用同一个 Embedding 模型）。
在数据库中快速找到与用户问题向量“最相似”（距离最近）的 K 个知识块向量。

传统的关系型数据库或文档数据库不擅长处理这种高维向量的相似度搜索。向量数据库则使用特殊的索引结构（如 HNSW, IVF-Flat, LSH 等）和优化算法来实现快速、近似的最近邻搜索（Approximate Nearest Neighbor, ANN）。

### <span class="motutor-highlight motutor-id_tllomi0-id_sttvr2d"><i></i>3.2 常见的向量数据库选型</span>

市面上有多种向量数据库/索引库可供选择，各有特点：

**内存型（In-Memory）/ 轻量级库**：

- **FAISS (Facebook AI Similarity Search)**: 由 Facebook 开发的高性能向量相似度搜索库。功能强大，支持多种索引类型，可以在内存中运行，也可以与持久化存储结合。需要 C++ 基础或通过 Python 包装使用。

- **ScaNN (Scalable Nearest Neighbors)**: Google 开发的向量搜索库，注重性能。

- **Annoy (Approximate Nearest Neighbors Oh Yeah)**: Spotify 开发的库，侧重于内存效率和简单的 API。

**持久化/托管型向量数据库**：

- **ChromaDB**: 开源，易于使用，支持本地运行和客户端/服务器模式，API 友好。

- **Milvus**: 开源，云原生，功能丰富，为大规模生产环境设计。

- **Pinecone**: 商业化托管服务，易于扩展，提供简单 API。

- **Weaviate**: 开源，支持 GraphQL 接口，可以存储对象和向量。

- **Qdrant**: 开源，用 Rust 编写，注重性能和可靠性。

- **Elasticsearch**: 虽然是搜索引擎，但也增加了对向量相似度搜索（ANN）的支持。

**选择建议**：

**学习/原型阶段**： 从简单的 NumPy/SciPy 或轻量级库如 FAISS (CPU 版) 或 ChromaDB 开始。

**生产/大规模应用**： 考虑使用持久化、可扩展的向量数据库如 Milvus, Pinecone, Weaviate, Qdrant 等。

### <span class="motutor-highlight motutor-id_zi53v3q-id_m72mbal"><i></i>3.3 实现相似度搜索（Similarity Search）</span>

最常用的衡量向量相似度的指标是 余弦相似度（Cosine Similarity）。
它衡量两个向量之间夹角的余弦值。

- 值域范围：[-1, 1]
- 值越接近 1，表示两个向量方向越一致，语义越相似。
- 值越接近 -1，表示方向相反。
- 值接近 0，表示向量（几乎）正交，相关性不高。

**计算公式**：
Cosine Similarity (A, B) = (A · B) / (||A|| * ||B||)
其中 A · B 是向量 A 和 B 的点积，||A|| 和 ||B|| 分别是向量 A 和 B 的 L2 范数（长度）。

## <span class="motutor-highlight motutor-id_jzlgdn5-id_73e71jg"><i></i>4. 整合：构建 RAG 应用流程</span>

我们已经分别掌握了 RAG 的“检索”部分（加载数据、分块、向量化、搜索）和“生成”部分（调用 DeepSeek API）。

现在，是时候将它们串联起来，形成一个完整的 RAG 应用流程了。

![](https://imgbed.momodel.cn/20250411110344931.png)

一个典型的 RAG 应用处理用户问题的流程如下：

- **接收用户问题 (Query)**: 获取用户输入的自然语言问题。

- **向量化查询 (Embed Query)**: 使用与文档向量化 相同 的 Embedding 模型，将用户问题转换成向量。

- **搜索相关文档 (Search)**: 使用查询向量，在预先构建好的向量索引（或数据库）中执行相似度搜索，找到 Top-K 个最相关的文档块。

- **检索文档内容 (Retrieve)**: 获取这 Top-K 个文档块的原始文本内容。

- **构建增强 Prompt (Augment Prompt)**: 将检索到的文档块内容（作为上下文 Context）与用户的原始问题结合起来，按照特定模板构建一个新的、信息更丰富的 Prompt。

- **调用 LLM 生成答案 (Generate)**: 将这个增强后的 Prompt 发送给大语言模型（在这里是 DeepSeek API）。

- **返回结果 (Return Result)**: 将 LLM 生成的答案呈现给用户。
