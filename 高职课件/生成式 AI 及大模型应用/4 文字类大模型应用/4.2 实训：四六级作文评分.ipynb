# 文字类大模型应用-DeepSeek：四六级英语作文评分

## 教学目标
- 体验文本类大模型的实际应用场景
- 掌握与大模型的交互方式，以及如何将大模型应用于个性化生产

## 教学方法
- **实践操作**：通过亲身使用文本类大模型，思考在与大模型交互过程中需要注意的地方。

- **讨论与反思**：根据大模型生成效果，思考当下文本类大模型还存在什么局限。

## 教学内容

### 1. 项目背景介绍
随着 GPT、DeepSeek、文心一言等文字类大模型的崛起，这些技术正以前所未有的速度重塑现代学生的学习方式。从提供个性化的学习建议到激发创新思维，它们极大地拓展了知识获取与实践的边界。

本次实训将以当下广受关注的 DeepSeek 文字类模型为基础，结合学生们普遍关注的英语四六级作文，作为切入点。我们将通过实战演练，帮助大家学会如何使用 DeepSeek 解决四六级复习中的难点——命题作文。借助 DeepSeek 强大的信息库，学生们将在实训过程中获得以下收获：

*   结合大数据进行四六级作文预测，帮助学生掌握考试趋势和作文命题规律。
*   根据用户的个性化需求生成作文命题，确保针对性地提升写作能力。
*   评分并提供反馈，对用户提交的作文进行评估，指出其中的不足，并给出切实可行的改进方案。
*   通过持续交互优化，逐步提升学生的写作水平，帮助学生突破写作瓶颈，提升作文的整体质量。
*   针对所给出的命题给出范文作为参考，给予学生一些参考案例，可以应用于自己的作文当中。

通过这些实训，学生们不仅能更高效地准备四六级考试，还能深刻理解大语言模型在实际学习中的应用，为未来的学习和创新提供有力支持。

### 2. 实训步骤

#### 2.1 作文题目预测
问题分析：
针对于我们所要达到的输出：大模型给予预测我们可以将问题拆分成以下部分：

 1. 角色设定：我们需要给大模型设定一个身份，以便其完成我们想要的功能。
 2. 提出需求：我们需要大模型为我们提供几个其预测的题目。
 3. 更多要求：我们可以针对性的要求大模型完成要求，比如：“请结合2015年到2024年的四级作文规律为我提供预测”。

        提示词：“假设你是一位拥有10年大学英语四级命题经验的专家，请结合2015-2024年真题库规律和当前社会发展，预测2025年英语四级作文可能出现的3-5个主题。”

#### 2.4 大模型个性化出题
在刚才的实践过程中，我们已经尝试了让大模型去预测四级作文题目、返还范文并对范文进行讲解。那么在我们背诵了大模型所整理出的好词好句后，将他投入实战中吧，接下来让我们尝试让大模型为我们生成一个四级作文题目。

        提示词：“我已经看完了你所提供的范文，并且掌握了其中的一些写作技巧和单词，现在请你结合2015-2024年的四级作文为我出一道命题作文，注意：命题作文的要求应该严格符合四级考试过程中的要求。”

### 3. 实战演练

相信各位同学已经掌握了基本的文本类大模型的使用流程，在刚才的实训过程中我们完成了从作文命题的预测、范文生成并解析、到最后生成命题并评分。

现在请各位同学尝试使用文本类大模型去预测一下六级的作文命题，范文解析并且生成一份六级的命题作文完成后交给大模型进行评分吧。
