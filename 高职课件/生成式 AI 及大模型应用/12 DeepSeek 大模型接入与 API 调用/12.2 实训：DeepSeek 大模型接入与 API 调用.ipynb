# 实训：DeepSeek 大模型接入与 API 调用



## 1. 实验描述
本实验旨在通过接入和调试 DeepSeek API，使学习者掌握大语言模型 API 的使用方法和流程。实验将基于 DeepSeek 平台进行开发，涵盖 API Key 的创建、API 调用、接口调试等关键环节。通过这些步骤，学习者将能够开发出一个能够与 DeepSeek AI 模型进行交互的应用。实验重点在于 API Key 的管理、API 调用的方法、接口调试的技巧，以及应用的测试，旨在帮助学习者全面实践大语言模型 API 应用开发的全过程。

## 2. 实验准备

- 培养接口调试的能力。

## 3. 实验步骤


#### 3.1 创建 API key
要能调用 API 首先要学会创建属于自己的 DeepSeek API key。     
[deepseek api](https://platform.deepseek.com/api_keys)
点击链接进入 deepseek api网站。

首次注册登录的用户， Deepseek 会赠送 10 块钱的 tokens 额度（先看看有没有，也有可能不送了，导致你的 API_KEY 不能用，没有额度的话得先充钱）。

在 Deepseek 开放平台的左侧边栏找到 API Keys，点击它并在此创建一个 API Key，创建完成之后，将这个 API Key 复制下来，并保存在一个只有你知道的地方。

<div class='insertContainerBox column'>

#### 3.2 调用 API
当 API Key 有了之后，我们就可以调用它了，Deepseek 的官方 API 文档在这里：https://api-docs.deepseek.com/zh-cn/。

接下来我们使用一种一款工具来调用 deepseek 的 api key。

下文我们通过 [Apifox](https://apifox.com) 来调试 API，并将 AI 模型给我们返回的消息进行流式输出。

Apifox 是一个集成了 API 文档、API 调试、API 设计、API 测试、API Mock、自动化测试的 API 一体化协作平台。

1. 设置 API Key     
登录 Apifox 并在里面新建一个 HTTP 项目后，在项目右上角的 “环境管理” 中，点击“正式环境”，在服务前置 URL 那里填写 `https://api.deepseek.com` ，然后添加一个环境变量，可以将其命名为 API_KEY，这个环境变量的值就是上面创建的 Deepseek API 的 API Key，添加完成后保存即可。

<div class='insertContainerBox column'>
<div class='insertItem' align=center><img src="https://imgbed.momodel.cn/hv/20250627104750111.png" width="800px"/></div>
</div>

2. 新建接口     
API Key 设置完成后，在项目里新建一个接口，然后将 Deepseek API 文档里的对话 API 的`curl`复制下来。

<div class='insertContainerBox column'>
<div class='insertItem' align=center><img src="https://imgbed.momodel.cn/hv/20250627104750112.png" width="800px"/></div>
</div>

你也可以复制这里的：

    curl https://api.deepseek.com/chat/completions \
      -H "Content-Type: application/json" \
      -H "Authorization: Bearer {{API_KEY}}" \
      -d '{
            "model": "deepseek-chat",
            "messages": [
              {"role": "system", "content": "你是一个全能助手"},
              {"role": "user", "content": "广东的省会是哪里？"}
            ],
            "stream": false
          }'

- role:系统提示词，是预先设定的指令，用于指导 AI 的行为和输出，通常在对话开始时由系统提供，以确保 AI 按照特定的规则或目标进行交互。
- user:用户提示词，用户输入的文本，用于向AI提出问题或请求，引导AI生成相应的回答或执行特定的操作。

回到 Apifox，将这个`curl`直接按"Ctrl + V"粘贴到接口路径那里即可，Apifox 会自动解析这里的`curl`。

<div class='insertContainerBox column'>
<div class='insertItem' align=center><img src="https://imgbed.momodel.cn/hv/20250627104750113.png" width="800px"/></div>
</div>

在解析出来的接口中，点击 Headers，然后更改 Authorization 的参数值，将其更改为`Bearer {{API_KEY}}`，这样在发送请求的时候才会携带存储在环境变量中的 API_KEY。

<div class='insertContainerBox column'>
<div class='insertItem' align=center><img src="https://imgbed.momodel.cn/hv/20250627104950121.png" width="800px"/></div>
</div>

>注意：要用`Bearer {{API_KEY}}`，而不是`Bearer <API_KEY>`，并且中间要有一个空格。

你也可以把密钥直接写死在这里，比如`Bearer sk-9017………………`。

<div class='insertContainerBox column'>

</div>
