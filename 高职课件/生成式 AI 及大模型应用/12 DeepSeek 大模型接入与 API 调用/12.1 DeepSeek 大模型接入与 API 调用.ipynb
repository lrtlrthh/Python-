### 1.1 什么是 DeepSeek 大模型

DeepSeek 大模型是由中国人工智能公司 杭州深度求索人工智能基础技术研究有限公司（简称 DeepSeek）开发的一系列开源大语言模型（LLMs）。

该公司成立于 2023 年 7 月，总部位于浙江杭州，由对冲基金 High-Flyer 投资并由其联合创始人梁文锋领导。

DeepSeek 的目标是通过开源和开放科学推进人工智能技术的发展，其模型因性能强大、训练成本低廉以及开放性而在全球 AI 社区中引起广泛关注。

2025 年 1 月发布的 DeepSeek-R1 模型是其技术突破的集大成者。该模型基于 671B 参数的专家混合架构（MoE），每个令牌激活 37 亿参数，专注于数学、代码与自然语言推理任务。

与传统依赖人工标注数据的模型不同，DeepSeek-R1 在后训练阶段大规模应用强化学习（RL）技术，仅需极少标注数据即可显著提升推理能力，技术路径与 OpenAI 的闭源模式形成鲜明对比。

如下图对比，基准覆盖了数学（AIME 2024、MATH-500）、编程（Codeforces、SWE-bench Verified）、科学推理（GPQA Diamond）和通用知识（MMLU），全面评估模型的多维能力。

### 1.2 <span class="motutor-highlight motutor-id_3jwuidt-id_7yvw2bt"><i></i>DeepSeek 大模型特点</span>

1. **开源性质**：
    - DeepSeek 模型（如 DeepSeek-V3 和 DeepSeek-R1）以 MIT 许可证发布，允许用户免费使用、修改和商用。
    
    这与其他闭源模型（如 OpenAI 的 GPT 系列）形成对比，促进了学术和商业领域的广泛应用。
    - 模型权重公开，但训练数据集和具体训练代码通常未完全披露，属于“开放权重”而非完全“开源”。
    
    
2. **高性能**：
    - DeepSeek-V3：拥有 6710 亿个参数，采用混合专家（Mixture-of-Experts, MoE）架构，每 token 仅激活 370 亿参数。
    
    其性能在多项基准测试中媲美甚至超越闭源模型（如 GPT-4o 和 Claude 3.5 Sonnet）。
    - DeepSeek-R1：专注于推理能力，通过大规模强化学习（RL）训练，能处理数学、编程和逻辑推理等任务，性能可与 OpenAI 的 o1 模型匹敌。
    
    
3. **高效与低成本**：
    - DeepSeek 声称其 V3 模型训练成本仅为 600 万美元，远低于 GPT-4 的 1 亿美元，且使用的计算资源（H800 GPU 小时）比竞争对手少近十倍。
    - 采用 FP8 混合精度训练和优化的 MoE 架构，大幅降低训练和推理的计算需求。
    
    
4. **创新技术**：
    - MoE 架构：通过细粒度专家分割和负载均衡策略，提高效率并减少冗余。
    - 强化学习与冷启动：DeepSeek-R1 使用“冷启动”技术（少量监督微调数据启动）结合强化学习，优化推理能力和输出可读性。
    - 多模态扩展：如 Janus 模型，支持图像和文本的统一理解与生成。

通过发送 HTTP 请求到 DeepSeek 提供的 API 接口，传递输入参数（如用户提问、配置选项），并接收模型返回的响应结果（如生成的文本）。

- **灵活扩展**：根据业务需求选择不同模型（如通用对话模型 deepseek-chat 或代码专用模型 deepseek-coder）。

### 2.4 <span class="motutor-highlight motutor-id_4c65856-id_jdc568h"><i></i>DeepSeek 大模型 API 调用</span>

#### 2.4.1 API 概念

**API** 是 `Application Programming Interface` 的缩写，中文通常翻译为 应用程序编程接口。

简单来说，**API 就像不同软件程序之间沟通和交流的桥梁或协议**。 

它定义了一套规则、约定和工具，使得一个软件应用程序可以请求另一个软件应用程序提供的服务或数据，而无需了解其内部的具体实现细节。

#### 2.4.2 什么是 API 调用

API 调用，就是你的软件程序（客户端）向另一个软件程序或服务（服务器端）发出请求，要求它执行某个操作或提供某些信息的过程。

![](https://imgbed.momodel.cn/20250317155841905.png)

你可以把 API 调用想象成你在餐厅点餐：

```
你 (客户端程序): 你是想获取信息或执行操作的一方。

服务员 (API): API 就像是餐厅的服务员，它理解你的需求，并将你的请求传递给厨房。

厨房 (服务器端程序/服务): 厨房是实际执行操作或存储数据的地方。

菜单 (API 文档): 菜单列出了餐厅提供的所有菜品（功能或数据），以及如何点餐（API 的使用规则）。

你点餐 (API 调用/请求): 你告诉服务员你想吃什么（你向 API 发出请求，指定要执行的操作或获取的数据）。

服务员把你的订单给厨房 (API 将你的请求传递给服务器): API 将你的请求按照服务器能理解的方式传递过去。

厨房准备你的菜 (服务器处理你的请求): 服务器端程序或服务执行你请求的操作或准备你请求的数据。

服务员把你的菜端给你 (API 将服务器的响应返回给你): API 将服务器处理后的结果（数据或操作结果）返回给你的程序。

```

### 2.6 使用 HTTP 请求调用 DeepSeek API

#### 2.6.1 <span class="motutor-highlight motutor-id_ueuk8h7-id_99p7lzx"><i></i>HTTP 基本概念</span>

HTTP 的全称是（Hypertext Transfer Protocol  超文本传输协议），它是互联网上应用最广泛的应用层协议。

用于在客户端（如浏览器）和服务器之间传输超文本（如网页）数据。

HTTP 协议包含由浏览器发送数据到服务器需要遵循的请求协议与服务器发送数据到浏览器需要遵循的响应协议。
 
用于 HTTP 协议交互的信息被为 HTTP 报文。
 
请求端(客户端)的 HTTP 报文做请求报文，响应端(服务器端)的做响应报文。
 
HTTP 报文本身是由多行数据构成的字文本。

![](https://imgbed.momodel.cn/20250318160911984.png)

 
常用的请求方式: get 与 post 请求

**GET**:
- 目的: 从服务器获取资源。
- 特点:
    - 请求参数通常附加在 URL 后面，以 ? 分隔，多个参数之间用 & 连接（例如：/users?id=123&name=John）。
    
    - 请求体通常为空。
    
    - GET 请求是幂等的（Idempotent），即多次发送相同的 GET 请求，结果应该相同，不会对服务器状态产生副作用。
    
    如：Get 请求用户信息，第一次请求，与第 n 次请求返回的数据不会改变（不直接修改数据库中的情况下）
    
    - 浏览器会对 GET 请求进行缓存，以提高性能。
    
    - GET 请求的 URL 长度通常有限制。
    
- 适用场景: 获取网页内容、查询数据等。

**POST**:

- 目的: 向服务器提交数据，通常用于创建新的资源或更新现有资源。

- 特点:
    - 请求参数通常放在请求体 (Request Body) 中，可以是多种格式（例如：application/x-www-form-urlencoded、multipart/form-data、application/json 等）。
    
    - POST 请求不是幂等的，多次发送相同的 POST 请求可能会产生不同的结果，例如重复创建数据。
    
    如：Post 请求创建订单，第一次创建订单，服务端返回：order-number-001，第二次相同请求，后端未做除重出路，返回：order-number-002。
    
    - 浏览器通常不会对 POST 请求进行缓存。
    
    - POST 请求没有 URL 长度限制。
    
- 适用场景: 提交表单数据、上传文件、创建用户等。

#### 2.6.2 <span class="motutor-highlight motutor-id_yydci03-id_7nwghin"><i></i>JSON 数据格式化 (JavaScript Object Notation)</span>

JSON 是一种轻量级的数据交换格式，易于人阅读和编写，同时也易于机器解析和生成。

它基于 JavaScript 语言的子集，但独立于语言。

这意味着，尽管 JSON 的语法与 JavaScript 相似，但它并不依赖于 JavaScript，可以在多种编程语言中使用，如 Python、Java、C++ 等。

**JSON 的基本语法规则**:
- 键值对 (Key-Value Pairs): 数据以键值对的形式存在，键和值之间用冒号 : 分隔。键必须是字符串（通常用双引号 " 包裹），值可以是以下类型：

    - 字符串 (String): 用双引号 " 包裹。
    
    - 数字 (Number): 可以是整数或浮点数。
    
    - 布尔值 (Boolean): true 或 false。
    
    - 数组 (Array): 用方括号 `` 包裹，元素之间用逗号 , 分隔。数组中的元素可以是任意合法的 JSON 值。
    
    - 对象 (Object): 用花括号 {} 包裹，包含多个键值对，键值对之间用逗号 , 分隔。
    - 空值 (Null): null。
    
    - 对象 (Object): 由花括号 {} 包裹，包含零个或多个键值对。
    
    - 数组 (Array): 由方括号 `` 包裹，包含零个或多个 JSON 值。

````python
import json

# JSON 数据
json_data = '''{
  "name": "John Doe",
  "age": 30,
  "isStudent": false,
  "address": {
    "street": "123 Main St",
    "city": "Anytown",
    "zipCode": "12345"
  },
  "courses": ["Math", "Science", "History"],
  "grades": [90, 85, 92],
  "spouse": null
}'''

# 解析 JSON
data = json.loads(json_data)

# 提取信息
name = data["name"]
age = data["age"]
is_student = data["isStudent"]
address = data["address"]
courses = data["courses"]
grades = data["grades"]
spouse = data["spouse"]

# 打印解析后的信息
print(f"姓名: {name}")
print(f"年龄: {age}")
print(f"是否为学生: {is_student}")
print(f"地址: {address['street']}, {address['city']}, {address['zipCode']}")
print(f"课程列表: {', '.join(courses)}")
print(f"成绩: {grades}")
print(f"平均成绩: {sum(grades) / len(grades):.2f}")
print(f"配偶: {spouse}")

````

#### 2.6.3 H<span class="motutor-highlight motutor-id_oo9gw6i-id_7fd5x2b"><i></i>TTP 调用基本流程</span>

- **参数设置**：
    确定 API 的 URL、请求头（包含 API Key 或 Token 等认证信息）以及请求体（如 prompt、max_tokens、temperature 等参数）。
    
- **请求发送**：
使用 HTTP POST/GET 方法调用 DeepSeek API，确保请求能正确触达服务端。

- **返回结果解析**
通常返回 JSON 格式的数据，需要检查状态码，并解析出生成文本、概率分布等核心信息。

- **错误处理与异常捕获**
    - 常见错误：
    包括网络超时、连接失败、权限问题以及参数错误等。
    - 处理策略：
    通过 try/except 机制捕获异常，记录详细错误日志，并采用重试机制（如指数退避算法）以应对临时网络波动。

````python
import requests

# 配置 API 相关参数
api_url = "https://api.deepseek.com/chat/completions"
api_key = "sk-xxxxxxxx"  # 请替换为实际 API Key

headers = {
    "Authorization": f"Bearer {api_key}",
    "Content-Type": "application/json"
}

# 请求数据示例
payload = {
    "model": "deepseek-chat",  
    "messages": [
        {"role": "system", "content": "你是经验丰富的内容生成助理"},
        {"role": "user", "content": "用一句话解释人工智能"},
    ],  
    "max_tokens": 150,
    "temperature": 0.7
}

# 发送请求
try:
    response = requests.post(api_url, headers=headers, json=payload, timeout=10)
    response.raise_for_status()  # 检查 HTTP 状态码是否正常
    result = response.json()['choices'][0]['message']['content']
    print("生成结果：", result)
except requests.exceptions.HTTPError as http_err:
    print("HTTP 错误:", http_err)
except requests.exceptions.RequestException as req_err:
    print("请求异常:", req_err)
````
