## 2. 大模型应用中的主要考虑因素

### 2.1 大模型应用时的注意事项

1. **设定合理的期望**：理解 AI 的<font color="#2861FC">局限性</font>，不要期望 AI 能够解决所有问题或完全替代人类的判断和创造力，适当放低期待。

**AI 幻觉现象**

    AI 无法准确反映事实，可能根据错误或不完整的信息生成完全不同的结论。 

    这种现象在写作工具中普遍存在，可能会导致用户对生成内容的信任度降低。

2. **选择适合的工具**：根据具体任务的需求，选择最合适的 AI 工具，而不是盲目追求最流行或最复杂的技术。


-   大模型逐渐<font color="#2861FC">改变产品形态</font>：网页端、app 端、桌面端

-   大模型领域<font color="#2861FC">从通用到垂类</font>，以计算机专业学生需要优化代码为例：
    - 通用类工具（如 ChatGPT）：可以生成代码片段、解释算法原理，甚至提供伪代码，但可能缺乏对特定编程语言的深度支持。
    
    - 垂类工具（如 通义灵码）： 专门为开发者设计，提供更精细更实时的功能，如代码补全、错误提示、API 文档查询等。


3. **数据隐私和安全**：在使用 AI 处理<font color="#2861FC">敏感数据</font>时，要确保遵守相关的数据保护法规，保护个人隐私。

4. **持续学习和适应**：需要持续学习最新的 AI 工具和 AI 提问技巧（提示词等）。

5. **人机协作**：将 AI 视为辅助工具，协作提高工作效率，但不应过度依赖，它只是辅助工具，无法代替你的思考。

### 2.2 大模型工具的两种模式

| 类型       | 效果         | 优势                                                         | 劣势                                                         | 举例工具                           |
|------------|--------------|--------------------------------------------------------------|--------------------------------------------------------------|------------------------------------|
| AI导师型   | AI说我来做   | 相比于去互联网信息搜索而言，可直接获取如何操作的步骤，信息清晰明确 | 1. 依然需要自己按照步骤操作<br>2. 太复杂的数据处理，给出的步骤可能是错的 | Kimi、豆包、文心一言|
| AI秘书型   | 我说AI来做   | 给到目标和对应的数据文件，通过提示词可获得结果，对非数据分析专业人士非常友好 | 1. 需要将数据上传到网站，涉及数据安全<br>3. 太复杂的数据处理可能会报错<br>4. 文件上传类型和大小有限制 | 小浣熊、讯飞智文、通义灵码   |



AI 是一个强大的工具，但它并不是万能的。  
我们需要合理地使用它，同时<font color="#2861FC">保持我们的批判性思维和创造力</font>。  
通过不断学习和适应，我们可以更好地利用 AI 来提高工作效率和学习效果。
