# 本地部署 Deepseek 

## 1. 实验描述

本项目旨在指导开发者完成 DeepSeek 大语言模型在本地环境的完整部署流程，构建一个功能完善的 AI 开发环境。通过本教程，您将掌握：

1. 核心能力：
    - Ollama 框架的安装与配置
    - 模型文件的下载与管理
    - 本地 API 服务的搭建
    - 可视化交互界面的集成

2. 技术特色：
    - 支持多种主流开源模型
    - 提供命令行和图形化两种交互方式
    - 可自定义模型存储路径
    - 完整的开发测试工具链

### 3.2 Ollama 配置
安装完成后，会在电脑右下角出现 Ollama 图标，可以不用先启动 Ollama，先退出做一些必要路径的配置！

右键图标，会出现退出的按钮「Quit Ollama」，注意：一定要退出Ollama，否则下边的环境配置无法生效！

<div class='insertContainerBox column'>
<div class='insertItem' align=center><img src="https://imgbed.momodel.cn/hv/202505060913693.png" width="400px"/></div>
</div>


                        

1. - 关闭开机自启动（可选）：

Ollama 默认会随 Windows 自动启动，可以在「文件资源管理器」的地址栏中访问以下路径，删除其中的Ollama.lnk快捷方式文件，阻止它自动启动。
 
 将下面这段地址直接复制到地址栏中即可打开。       
`%APPDATA%\Microsoft\Windows\Start Menu\Programs\Startup`


<div class='insertContainerBox column'>
<div class='insertItem' align=center><img src="https://imgbed.momodel.cn/hv/202505060916130.png" width="800px"/></div>
</div>

2. 配置环境变量（必须）：

Ollama 的默认模型存储路径如下：`C:\Users\%username%\.ollama\models`，无论 C 盘空间大小，需要安装多少模型，都建议换一个存放路径到其它盘，否则会影响电脑运行速度。

右键此电脑->高级系统设置->环境变量

打开「系统环境变量」，新建一个系统变量`OLLAMA_MODELS` ，然后设置 ollama 模型的存储路径。

- 变量名：OLLAMA_MODELS

- 变量值（路径）：E:\ollama_models

运行到这步我们已经可以使用我们部署的模型了。

Ollama 命令行指令：
| 命令类别       | 命令格式                          | 功能描述                                                                 | 示例                                                                 |
|----------------|-----------------------------------|--------------------------------------------------------------------------|----------------------------------------------------------------------|
| ​**模型管理**   | `ollama pull <model-name>`        | 从仓库下载指定模型                                                       | `ollama pull llama2:7b`                             |
|                | `ollama list`                     | 列出本地已下载的模型                                                     | `ollama list`                                       |
|                | `ollama rm <model-name>`          | 删除本地模型                                                             | `ollama rm llama2:7b`                               |
|                | `ollama show <model-name>`        | 查看模型详细信息（版本、大小等）                                         | `ollama show llama2`                                |
|                | `ollama cp <src> <dest>`          | 复制/重命名模型                                                          | `ollama cp llama2 my-llama2-copy`                   |
| ​**运行交互**   | `ollama run <model-name>`         | 启动模型交互式对话                                                       | `ollama run llama2`                                 |
|                | `ollama run -p "prompt"`          | 直接传入提示词非交互运行                                                 | `ollama run llama2 -p "写 Python 冒泡排序"`           |
|                | `ollama run --verbose`            | 显示详细推理日志                                                         | `ollama run llama2 --verbose`                       |
| ​**自定义模型** | `ollama create -f <Modelfile>`    | 根据 Modelfile 创建自定义模型                                            | `ollama create mymodel -f ./Modelfile`              |
|                | `ollama push <model-name>`        | 推送自定义模型到私有仓库                                                 | `ollama push my-org/mymodel:latest`                 |
| ​**服务管理**   | `ollama serve`                    | 启动本地 API 服务（默认端口 11434）                                      | `ollama serve`                                      |
|                | `ollama stop`                     | 停止运行的服务或模型实例                                                 | `ollama stop`                                       |
|                | `ollama ps`                       | 查看正在运行的模型实例                                                   | `ollama ps`                                         |
| ​**辅助命令**   | `ollama help`                     | 显示帮助信息                                                             | `ollama help`                                       |
|                | `ollama version`                  | 查看版本号                                                               | `ollama version`                                    |
|                | `ollama logs`                     | 查看运行日志（调试用）                                                   | `ollama logs`                                       |
|                | `ollama prune`                    | 清理未使用的模型缓存                                                     | `ollama prune`                                      |
