# 大语言模型技术——提示词框架实战应用

## 1. 实验描述

 本实验旨在通过一系列精心设计的实践任务，使学习者深入理解并掌握大语言模型提示词设计的核心原理与实战技巧。   
 实验围绕 ICIO 框架、CRISPE 框架以及一个综合框架 CO-STAR 展开，通过模拟真实应用场景，引导学习者如何高效构建和优化提示词，从而提升大语言模型在特定任务上的表现。



## 2. 实验准备


### 2.1 实验目的

1. 理论掌握：深入理解大语言模型提示词设计的基本原理，包括 ICIO 框架、CRISPE 框架及自定义 CO-STAR 框架的核心概念和应用逻辑。
2. 实战技能：通过实际操作，掌握在不同应用场景下如何运用上述框架设计高效提示词，提升大语言模型的输出质量和任务完成效率。
3. 分析评估：学会对生成的文本内容进行分析评估，识别提示词设计中的优点与不足，并能根据评估结果进行迭代优化。
4. 创新思维：鼓励学习者在实验过程中发挥创新思维，探索框架间的融合与创新应用，为提示词设计提供更多可能性。
5. 问题解决：培养学习者在面对复杂任务时，能够灵活运用所学知识，设计有效提示词解决问题的能力。

### 2.2 实验内容

本实验将完成的练习有：

- 熟悉提示词框架原理：了解 ICIO、CRISPE 和 CO-STAR 框架的基本概念、组成要素和设计原则。
- 掌握框架应用方法：通过具体案例，掌握在不同场景下如何运用这些框架设计提示词。
- 实战应用与效果评估：设计并实施实验，验证提示词框架的有效性，并对生成内容进行评估和优化。

### 2.3 实验目标

1. 通过三个框架的综合实践，全面掌握大语言模型提示词设计的核心原理和实战技巧，能够灵活应用不同框架应对多样化任务需求。
2. 培养学习者在提示词设计过程中的创新思维和问题解决能力，鼓励探索新的设计思路和技术方法，推动大语言模型技术的创新应用。

### 3.2 场景设计

**场景一：新闻摘要生成**   

- 背景：在信息爆炸的时代，快速获取新闻要点变得尤为重要。本场景要求大语言模型根据提供的新闻报道，生成简洁明了的摘要。   
- 目标：使用 ICIO 框架设计一个提示词，确保模型能够准确理解新闻报道的内容，并提取出关键信息。   
- 应用框架：ICIO 框架。   
- 预期输出：一篇简短且包含新闻报道主要信息的摘要。   

**场景二：创意写作**

- 背景：创意写作是激发想象力和创造力的过程。本场景要求大语言模型根据给定的主题或情境，创作一篇具有吸引力的短篇小说或故事。    
- 目标：利用 CRISPE 框架设计一个提示词，使模型能够生成既相关又具体，同时富有吸引力和个性化的内容。   
- 应用框架：CRISPE 框架。   
- 预期输出：一篇围绕给定主题或情境的创意短篇小说或故事。   

**场景三：产品描述优化**   
- 背景：在电商平台上，产品描述的吸引力直接影响消费者的购买决策。本场景要求大语言模型根据提供的产品信息，生成一段吸引目标受众的详细描述。   
- 目标：采用 CO-STAR 框架设计一个提示词，确保模型能够生成清晰、客观、具体且针对目标受众的描述。   
- 应用框架：CO-STAR 框架。     
- 预期输出：一段优化后的产品描述，能够吸引并说服目标受众。

再回顾一下 ICIO 框架包含四个部分：

*   Instruction(必须)：指令，即你希望模型执行的具体任务。
*   Context(选填)：背景信息，或者说是上下文信息，这可以引导模型做出更好的反应。
*   Input Data(选填)：输入数据，告知模型需要处理的数据。
*   Output Indicator(选填)：输出指示器，告知模型我们要输出的类型或格式。

#### 3.3.2 创意写作

我们使用 CRISPE 框架设计一个提示词，让大模型作为一个作家，帮我们写一个小故事，故事主题为小乌龟去看心理医生。

接下来按照 CRISPE 框架的格式编写提示词。

*   CR:Capacity and Role(能力与角色)：你希望大模型扮演怎样的角色。
*   I:Insight(背景)：背景信息和上下文。
*   S:Statement(指令)：你希望大模型做什么。
*   P:Personality(格式)：你希望大模型以什么风格或方式回答你。
*   E:Experiment(尝试)：要求大模型为你提供多个答案。

```
Capacity and Role（角色与定位）：
你是一位创意无限的故事讲述者，擅长将平凡的故事元素编织成引人入胜的叙事篇章。今天，你的任务是围绕“小乌龟去看心理医生”这一主题，创作一个温馨、鼓舞人心且富有教育意义的小故事。

Insight（背景与情境）：
在一个充满生机与和谐的动物王国里，小乌龟托比因为自己的行动缓慢而感到与众不同，这逐渐演变成了他内心的负担。在朋友的鼓励下，托比鼓起勇气，踏上了前往森林心理诊所的旅程，希望在那里找到释放内心压力的方法。

Statement（指令与要求）：
请创作一个以小乌龟托比为主角的小故事，故事中应包含托比与心理医生的互动、托比内心的转变过程以及最终他如何克服自卑、重拾自信的关键时刻。确保故事情节紧凑、情感真挚，能够触动读者的心灵。

Personality（风格与调性）：
故事应采用温暖、积极的叙事风格，穿插幽默与智慧的对话，让读者在轻松愉快的氛围中感受到成长的力量。同时，故事应传递出接纳自我、勇于面对挑战的正面信息。

Experiment（创新与多样性）：
在创作过程中，不妨尝试融入一些独特的元素或创意转折，如心理医生的特别疗法、托比意外发现的自我价值等，以使故事更加新颖且引人入胜。同时，考虑不同读者群体的喜好，确保故事既适合儿童阅读，也能引起成年读者的共鸣。

```

#### 3.3.4 三种框架的优势和缺点

**ICIO 框架：明确与灵活**

明确性：指令部分确保任务清晰，易于 AI 理解。   
灵活性：支持多种输入数据，适应不同需求。

**CRISPE 框架：定制与创新**  
个性化：风格部分允许定制AI输出，增强互动体验。      
创新性：实验部分鼓励尝试多种输出，激发创造力。    
 
**CO-STAR 框架：全面与细致**   
全面性：覆盖背景、目标、风格、语气、受众和响应，确保内容精准。     
细致性：详细描述各部分，便于理解和应用。  

总结：
- ICIO 框架：以明确的任务导向和灵活的输入多样性为特点，适用于需要明确指令和背景信息的场景，例如：数据处理与分析、信息检索。
- CRISPE 框架：通过角色定制和个性化风格增强了互动的趣味性和创新性，适用于需要 AI 模型模拟特定身份或提供多样化输出的场景，例如：创意写作、角色扮演。
- CO-STAR 框架：以其全面性和细致性见长，适用于需要构建全面且结构良好的提示词的场景，例如：学术研究与报告撰写、客户服务与沟通。
