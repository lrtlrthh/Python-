# 平台应用开发 - 实训

## 1. 实验描述

本实验旨在通过构建一个海报生成应用，帮助学习者掌握大语言模型应用开发的基本流程和技能。实验将使用扣子平台（Coze）作为开发环境，通过设计工作流、配置组件、设置界面等步骤，创建一个能够生成个性化海报的智能应用。

实验重点包括：
1. 工作流设计：学习如何设计有效的工作流程，使应用能够准确执行任务
2. 组件集成：掌握如何集成和使用各种组件，如图像处理、文本输入等
3. 界面设计：学习如何设计美观的用户界面，包括输入框、按钮等
4. 发布部署：了解如何将开发完成的应用发布到实际应用平台


## 2. 实验准备


### 2.1 实验目的

1. 掌握大语言模型应用开发的基本流程和方法
2. 学习工作流设计的核心原理和实践技巧

4. 培养应用设计思维和用户体验意识

- 优化体验：设置界面布局和交互方式
- 测试发布：进行功能测试并发布到应用平台

### 2.3 实验目标

1. 完成一个功能完整的海报生成应用

### 3.2 设计应用
在开发之前，我们需要设计一下自己的应用。

因为是基于`海报生成`，我们可以通过应用设计属于自己的海报，该模板中包括内容元素、主标题、副标题、logo、二维码、产品图这些元素。这些元素我们并不都需要使用，可以根据我们的需求修改变量名或者删除增加变量。   


开发个人个性海报，我们需要输入的变量有网名（或者真实姓名）、个性签名、个人 logo、微信交友二维码。同时我们还需要一张海报的背景图，这张背景图可以使用图像生成大模型生成我们想要的样子，所以我们还需要输入参数`image_prompt`用于生成背景图。    最后使用画板融合就可以得到我们想要的海报。

依照以上修改我们开始工作流的创建编辑。

模板原本用于海报生成各个节点的作用：
- 开始节点：设置进入工作流所需要的变量。
- 图像处理阶段
    - 图像生成：生成海报背景图，即内容元素。
    - 智能扩图：扩张海报背景图成海报所需要的大小。
    - 智能抠图：保留 logo 前景图，输出透明背景。
    - 智能抠图：保留产品图前景图，输出透明背景。
- 画板：将所有元素整合到一起。
- 结束节点：输出图片。


首先我们把开始节点中的每一个变量对应起来，然后修改变量。

| 原变量名         | 新变量名       | 原作用描述               | 新作用描述                 | 修改说明                     |
|------------------|----------------|--------------------------|--------------------------|-----------------------------|
| `BOT_USER_INPUT` | `BOT_USER_INPUT` | 用户输入内容             | 用户输入内容               | 保持不变                     |
| `image_prompt`   | `image_prompt` | 生成图片的内容元素       | 生成背景图片的内容元素      | 功能明确为生成背景图片        |
| `primary_title`  | `name`         | 主标题文字               | 网名文字                  | 更符合实际用途                |
| `second_title`   | `self_signature` | 副标题文字             | 个性签名文字              | 更符合实际用途                |
| `logo`           | `logo`         | 上传/显示logo图片        | 上传/显示logo图片          | 保持不变                     |
| `evm`            | `evm`          | 上传/显示二维码图片       | 上传/显示二维码图片        | 保持不变                     |
| `product`        | -              | 上传/显示产品图片        | -                        | 未使用，删除该变量           |
