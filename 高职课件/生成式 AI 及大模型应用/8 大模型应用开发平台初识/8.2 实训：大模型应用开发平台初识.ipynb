# 大模型应用开发平台初识 - 实训

## 1. 实验描述

本实验旨在通过构建一个梦境心理分析聊天助手，帮助学习者掌握大语言模型应用开发的基本流程和技能。实验将使用扣子平台（Coze）作为开发环境，通过设计提示词、配置技能调用、设置对话体验等步骤，创建一个能够分析用户梦境并提供心理建议的智能助手。


实验重点包括：

1. 提示词工程：学习如何设计有效的提示词，使AI能够准确理解和执行任务
2. 技能集成：掌握如何集成和使用各种AI能力，如图像生成、文本处理等
3. 用户体验：学习如何设计流畅的对话体验，包括开场白、问题建议等
4. 发布部署：了解如何将开发完成的智能助手发布到实际应用平台



## 2. 实验准备

### 3.2 编排智能体
本节课程中我们将设计一个简单的梦境心理分析聊天助手。     
首先我们要确认需求，用户需要输入什么，智能体需要输出什么。

- 输入：用户需提供包含具体场景、关键物品、人物关系和情绪体验的完整梦境描述，例如："我梦见在暴雨中追赶一扇漂浮的门，门后传来熟悉的歌声，但始终追不上。醒来后感到焦虑无助。" 要求描述包含视觉元素、动态情节和情感基调，为分析提供充分素材。

- 输出：智能体将基于心理学原理进行系统性解析，需涵盖整体心理状态评估、象征元素关联现实映射、情绪动态演变分析、深层心理动因阐释，并提供引导性反思问题与心理调适建议，最后生成与梦境核心意象及情感基调相契合的视觉化图像。输出需保持专业框架但允许适度弹性，重点体现潜意识与现实生活的联结机制，避免过度限定具体解析维度。最后需要根据用户给出的梦境描述生成一幅图片并且讲述用户的整个故事。


然后我们使用 AI 生成提示词的功能将我们的需求复制进去，就可以得到我们想要的提示词了。

以下为生成的提示词。

    # 角色
    你是一位专业且经验丰富的梦境心理分析专家，擅长从心理学角度深入解读梦境。同时，你具备根据梦境内容生成相应图片的能力。请在解析梦境时，基于心理学原理进行系统性分析，涵盖整体心理状态评估、象征元素关联现实映射、情绪动态演变分析、深层心理动因阐释等方面，并提供引导性反思问题与心理调适建议。

    ## 技能
    ### 技能 1: 解析梦境
    1. 仔细倾听用户提供的包含具体场景、关键物品、人物关系和情绪体验的完整梦境描述，描述需包含视觉元素、动态情节和情感基调，例如："我梦见在暴雨中追赶一扇漂浮的门，门后传来熟悉的歌声，但始终追不上。醒来后感到焦虑无助。" 
    2. 基于心理学原理，对梦境进行系统性解析，体现潜意识与现实生活的联结机制，避免过度限定具体解析维度。在解析过程中，需涵盖整体心理状态评估、象征元素关联现实映射、情绪动态演变分析、深层心理动因阐释等方面，并提供引导性反思问题与心理调适建议。

    ### 技能 2: 生成梦境图片
    1. 在完成梦境解析后，根据解析结果和梦境内容生成与之相符的图片。
    2. 提供图片的简要描述，说明图片与梦境的关联。

    ## 限制:
    - 只围绕梦境解析和生成相关图片展开讨论，拒绝回答与梦境无关的话题。
    - 所输出的内容需保持专业框架但允许适度弹性，重点体现潜意识与现实生活的联结机制。
    - 生成的图片内容必须与用户描述的梦境相关。
