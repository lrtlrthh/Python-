#### 1.1.2  <span class="motutor-highlight motutor-id_zevoy7r-id_fbu65ka"><i></i>开发痛点与挑战</span>
- **成本高：烧钱的“超级工程”**    
    大型语言模型的研发成本已突破常规认知边界。
    - 开发大模型如同给印钞机通电 ——GPT-3 单次训练耗电可供 3 万家庭用一年，电费账单轻松突破千万美元。
    - 硬件成本更夸张：一片 H100 显卡能换辆特斯拉，中型训练集群至少需要 500 片，相当于把整个停车场都塞进服务器机房。
    - 人才方面，顶尖算法工程师年薪堪比上市公司高管，普通团队凑钱买显卡的难度，大概和让中学生凑钱造火箭差不多。
    
    
- **技术门槛：需要“超能力”技能**   
     想玩转大模型，得同时具备三种 "超能力"
    - 让万台服务器同步跳舞（分布式训练），数据传输延迟要控制在眨眼时间的百分之一
    - 在 100 维参数迷宫里找出口（模型调优），调错一个小数点就能让 AI 变成 "人工智障"
    - 是给模型装防盗门（安全机制），对抗样本攻击的检测精度非常高，否则分分钟被 "黑" 成筛子。对学生而言，这就像让小学生直接挑战量子物理。


- **部署挑战：从实验室到商业化的鸿沟**      
    - 模型压缩阶段需要把上千亿参数的模型塞进手机，这操作堪比让大象钻针眼，还得保证它不失忆（精度损失 < 1%）。
    - 服务化阶段更像驯服电子巨兽：单台服务器每秒要处理 2000 次请求，延迟超过 50 毫秒用户就会集体弃用。
    - 监控环节则需要 24 小时盯着 AI，0.1 秒内捕捉敏感内容，就像给失控的自动驾驶汽车装紧急刹车。整个部署过程如同组装精密钟表，任何一个齿轮错位都会导致系统停摆。

### 1.2 <span class="motutor-highlight motutor-id_oy8i4xl-id_b786e36"><i></i>Coze 平台基础介绍 </span>
Coze 平台，被誉为“扣子”平台，是由科技巨头字节跳动精心打造的一款应用编辑平台。它专为新一代 AI Chat Bot 的开发而设计，让用户能够轻松创建各类智能机器人。无论是精通编程的 Bot，还是擅长绘画的 Bot，都能在这里得以实现。因此，Coze 平台不仅是一个 AI Agent 的开发与创建工具，更是一个让 AI 应用落地、实现任务自动化的神奇平台。 

最重要的是，它的开发方式非常简单，也属于低代码的一种，即使非技术人员，也能参与 AI 应用的定义和数据运营。













#### 1.2.1<span class="motutor-highlight motutor-id_b0wajit-id_tph79h6"><i></i> Coze 能做什么</span>

借助扣子提供的可视化设计与编排工具，你可以通过零代码或低代码的方式，快速搭建出基于大模型的各类 AI 项目，满足个性化需求、实现商业价值。

- 智能体： 智能体是基于对话的 AI 项目，它通过对话方式接收用户的输入，由大模型自动调用插件或工作流等方式执行用户指定的业务流程，并生成最终的回复。智能客服、虚拟伴侣、个人助理、英语外教都是智能体的典型应用场景。

- 应用：应用是指利用大模型技术开发的应用程序。扣子中搭建的 AI 应用具备完整业务逻辑和可视化用户界面，是一个独立的 AI 项目。通过扣子开发的 AI 应用有明确的输入和输出，可以根据既定的业务逻辑和流程完成一系列简单或复杂的任务，例如 AI 搜索、翻译工具、饮食记录等。

#### 2.3.2 <span class="motutor-highlight motutor-id_qyx7e79-id_2sbwyl1"><i></i>步骤 2：编写提示词</span>

配置智能体的第一步就是编写提示词，也就是智能体的人设与回复逻辑。智能体的人设与回复逻辑定义了智能体的基本人设，此人设会持续影响智能体在所有会话中的回复效果。建议在人设与回复逻辑中指定模型的角色、设计回复的语言风格、限制模型的回答范围，让对话更符合用户预期。

**提示词类型**

提示词包括系统提示词和用户提示词，两者区别如下：

- **系统提示词**：在搭建智能体时，设置的人设与回复逻辑就是系统提示词，它是开发者为大语言模型设定的初始参数和行为准则，在整个会话中持续影响大模型的响应模式。通过编写系统提示词，可以为大模型设定特定的角色定位和回复逻辑。

- **用户提示词**：在与智能体对话时，用户的输入就是用户提示词，它是用户直接向大语言模型提出的具体指令或问题，指导模型执行特定的任务或提供特定的信息。用户提示词的设置应简洁明了，以便模型能够准确捕捉用户的需求。

制作生活小助手时，首先要明确需求。用户需要输入的可能是具体的任务指令，比如“提醒我明天早上八点开会”“查询最近的天气情况”“帮我列出本周的购物清单”等，也可能是模糊的诉求，像“我今天心情不好，给我讲个笑话”“推荐一个适合周末放松的活动”。而智能体则要根据这些输入，输出精准且有用的结果。对于明确的任务指令，智能体要准确执行，如设置好提醒时间、提供详细的天气信息、生成清晰的购物清单；对于模糊诉求，智能体要能理解背后的情感或需求，给出合适的笑话、活动建议等，总之要让用户感受到贴心、便捷的帮助，让生活小助手真正发挥其作用。

#### 2.3.3 <span class="motutor-highlight motutor-id_1utcerb-id_ym41fq3"><i></i>步骤 3：选择模型</span>
在扣子中搭建的智能体和 AI 应用都是基于模型技术开发的应用程序。通过扣子，你可以便捷使用各个厂商提供的模型服务，基础版用户可使用豆包、通义千问等模型服务，专业版用户在此基础上还可以使用火山引擎方舟平台的其他模型资源，例如豆包视觉模型、豆包文生图模型等。
你可以通过智能体、工作流节点、插件等方式使用模型服务：

- 智能体和工作流节点：为你的智能体选择一个模型，用于灵活调用技能、知识和数据，响应用户的问题。大模型等工作流节点中也可以设置模型和版本，由指定的模型完成某个工作流环节。

- 插件：扣子将部分模型直接封装为官方插件，帮助你针对性地处理数据。例如为智能体绑定图片理解插件，回答用户关于指定图片的问题。

具体模型服务可以查看[模型服务](https://www.coze.cn/open/docs/guides/model_service)

模型可以进行参数设置，以下为模型的几种参数介绍，可以根据具体需要选择模型后进行该智能体对应模型的参数调整。

- 生成随机性（temperature）: 调高温度会使得模型的输出更多样性和创新性，反之，降低温度会使输出内容更加遵循指令要求但减少多样性。建议不要与 “Top p” 同时调整。

- Top P(累计概率): 模型在生成输出时会从概率最高的词汇开始选择，直到这些词汇的总概率累积达到 Top p 值。这样可以限制模型只选择这些高概率的词汇，从而控制输出内容的多样性。建议不要与 “生成随机性” 同时调整。

- 携带上下文轮数：设置带入模型上下文的对话历史轮数。轮数越多，多轮对话的相关性越高，但消耗的 Token 也越多。

- 输出格式：
    - 文本: 使用普通文本格式回复
    - Markdown: 将引导模型使用 Markdown 格式输出回复
    - JSON: 将引导模型使用 JSON 格式输出
    
    
- 最大回复长度：控制模型输出的 Tokens 长度上限。通常 100 Tokens 约等于 150 个中文汉字。

#### 2.3.4 步骤 4：（可选）<span class="motutor-highlight motutor-id_8fsvdyg-id_ml577j1"><i></i>为智能体添加技能</span>

如果模型能力可以基本覆盖智能体的功能，则只需要为智能体编写提示词即可。      
但是如果你为智能体设计的功能无法仅通过模型能力完成，则需要为智能体添加技能，拓展它的能力边界。       
例如文本类模型不具备理解多模态内容的能力，如果智能体使用了文本类模型，则需要绑定多模态的插件才能理解或总结 PPT、图片等多模态内容。      
    
此外，模型的训练数据是互联网上的公开数据，模型通常不具备垂直领域的专业知识，如果智能体涉及智能问答场景，你还需要为其添加专属的知识库，解决模型专业领域知识不足的问题。     
在提示词中我们还可以放置技能。

**快捷指令**

扣子支持开发者在搭建智能体时创建一些快捷指令，方便用户在与智能体会话时通过快捷指令快速、准确地输入信息。

配置快捷指令后，智能体用户在智能体的对话框中可以直接通过指令发起预设的对话。快捷指令的行为可以是发送一段简单的文字、上传文件、使用插件或工作流等。多 Agent 模式下，全局配置中也支持添加快捷指令，默认不指定节点回答，智能体根据用户输入匹配对应的节点处理。

| 配置         | 说明                                                                 |

| 指令内容     | 输入用户点击该指令时发送的内容。例如：`帮我推荐一些汽车品牌`     |

#### 2.3.6 <span class="motutor-highlight motutor-id_5a8syqs-id_zva8fwd"><i></i>步骤 6：调试智能体</span>

扣子的调试台功能提供全链路调试，你可以在调试台查看每一条用户请求从输入到响应的全流程，包括模型调用、配置的工作流或知识库等详细信息，方便你精准并快速定位问题，调整智能体配置。

调试台适用于以下场景的问题分析：
- 开发调试：搭建智能体时，发现智能体输出不符合预期，例如回答与问题不符，需要修改智能体流程。
- 线上排障：收到智能体用户反馈，例如智能体不响应、返回错误信息，需要诊断定位问题。


通过智能体的编排页面可以进入调试台，操作步骤如下：
1. 在智能体编排页面右侧的预览与调试区域，发送一条消息内容开始与智能体对话。
2. 在收到智能体响应后，点击消息卡片下方的调试按钮，或点击右上角的调试选项进入调试台页面。
