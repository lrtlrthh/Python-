## 3. 题目

大模型应用开发平台初识

## 4. 教学重点、难点

**教学重点**： 
- 大模型技术的最新进展、应用趋势以及面临的开发挑战。
- 低代码平台的工作原理、优势以及其在简化大模型应用开发中的关键作用。
- Coze 平台的核心功能模块和智能体搭建的基本流程。

**教学难点**：    
- 将复杂的低代码开发流程和技术概念（如模型调用、技能配置、多模态能力）简化为易于理解的内容。
- 帮助学生快速上手 Coze 平台，掌握智能体的创建、配置、调试和发布过程。
- 引导学生理解如何将低代码平台应用于实际业务场景，解决实际问题。


## 5. 教学方法

**教学方式**：讲授、实验、讨论

**教学手段**：互联网环境下的上机操作


## 6. 预期学习成果

- 学生能够概述大模型技术的发展趋势、应用领域及开发难点。
- 理解低代码平台的优势与工作原理，尤其是在简化开发流程、降低开发成本方面的价值。
- 熟练掌握 Coze 平台的基本操作，能独立创建、配置、调试和发布智能体，以满足不同业务场景的需求。
- 具备将所学知识应用于实际项目的能力，能够运用 Coze 平台开发简单的 AI 应用。



## 7. 教学过程

**模块 1：大模型应用开发平台介绍（24 分钟）**

- 大模型应用开发趋势与挑战
- Coze 平台基础介绍
- Coze 与其他开发工具的对比


**模块 2：平台使用入门（16 分钟）**

- 智能体效果
- 搭建步骤

**模块 3：实训-搭建聊天助手（45 分钟）**

**模块 4：总结（5 分钟）**



## 8. 作业

**思考题**：

- 请结合大模型开发的痛点，思考低代码平台如何降低开发成本并提高开发效率？
