## 3. 教学目标  

1. **知识掌握**：  
   - 理解大语言模型的核心局限（时效性、上下文记忆、专业深度、幻觉问题等）及其技术原理。  
   - 掌握优化大模型的三种路径（微调、检索增强生成、提示词工程）及其适用场景。  
   - 熟悉提示词工程的核心概念、作用及常见框架（ICIO、CRISPE、CO-STAR）。  

2. **能力提升**：  
   - 能通过案例分析，诊断大语言模型在实际应用中的问题（如数学错误、内容幻觉）。  
   - 能设计有效的提示词，结合不同框架（如CO-STAR）优化模型输出。  
   - 能在实验环节中实践提示词技巧，解决文本生成、任务指导等实际问题。  

3. **素养培育**：  
   - 培养对AI技术局限性的理性认知，理解模型能力与边界的平衡。  
   - 建立工程化思维，掌握通过系统方法（如提示词工程）提升AI应用效果的实践能力。  
   - 发展批判性思维，探讨大语言模型技术对社会、伦理的影响及应对策略。

## 4. 教学重点、难点

**教学重点**：大语言模型的局限、大语言模型的优化方式、提示词工程的核心概念、提示词的框架

**教学难点**：大语言模型局限性的深入理解与优化策略的实施、提示词工程的概念、提示词框架的实践应用



## 5. 教学方法

**教学方式**：讲授、实验、讨论

**教学手段**：互联网环境下的上机操作


## 6. 预期学习成果

**预期学习成果**

- 深入理解大语言模型
- 掌握大语言模型的局限与优化策略
- 精通提示词工程的设计与实施
- 熟练运用多种提示词技巧解决实际问题




## 7. 教学过程

**模块 1：大语言模型的局限 15 分钟**

- 大语言模型的能力
- 大语言模型的六个局限

**模块 2：优化大模型的三个路径 10 分钟**

- 微调、检索增强生成和提示词工程的概念
- 微调、检索增强生成和提示词工程的作用
- 一般优化流程

**模块 3：提示词工程 15 分钟**

- 提示词概念介绍
- 提示词的常见作用
- 提示词工程的概念和重要性

**模块 4：提示词技巧 10 分钟**

- 提示词工程的使用模式
- 提示词的三个框架

**模块 5：提示词框架实战练习 40 分钟**

- 通过三个框架进行实战

**模块 6：总结回顾 5 分钟**
