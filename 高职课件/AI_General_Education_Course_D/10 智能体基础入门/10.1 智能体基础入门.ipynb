### 1.1 智能体初认知


    故事背景：

    想象一下，李华有一个特别能干的虚拟助手叫小明。小明不是普通人，他是一个智能体，就像一个超级版的 Siri 或者小爱同学，但比他们更聪明、更有能力。

有一天，李华工作特别忙，需要准备一个报告，还要处理一堆邮件，更要命的是，晚上李华还有个重要的约会，需要挑选合适的衣服和餐厅。这时候，小明出现了。

**小明利用他的大语言模型理解了李华的需求，然后，他开始规划行动。**

**（写报告）** 小明知道李华需要一个突出的报告来打动老板，并且记得他上次报告的风格，同时结合他的工作数据，迅速生成了一个精彩的报告草稿。**（处理邮件）** 小明用邮件管理系统，根据邮件的紧急程度和内容，帮李华一一回复，甚至还能帮李华筛选出哪些邮件需要他亲自回复。**（准备约会）** 小明根据李华的喜好和场合，推荐了几套衣服和几个餐厅，还帮他预定了位置，为他的约会做了完美的安排。

就在李华准备离开办公室的时候，小明提醒他，根据天气预报，晚上可能会下雨，建议他带上伞。李华惊讶于小明的细心，同时也感激他的提醒。

晚上，李华穿着小明推荐的衣服，带着伞，准时到达了餐厅。约会非常成功，李华感到非常轻松和愉快。这一切都要感谢小明，这个智能体，我的数字小助手。

    1. 理解任务：李华告诉小明他的需求，小明通过自己的大语言模型（LLM）理解了李华的需求。

    2. 规划行动：小明开始规划，他知道李华需要一个突出的报告来打动老板，也需要回复那些紧急邮件，还要为约会做好准备。

    3. 记忆能力：小明记得李华上次报告的风格，也记得李华上次约会时喜欢的衣服和餐厅。

    4. 使用工具：小明开始工作了，他用他的工具——一个先进的文本生成器来帮你写报告，一个邮件管理系统来帮李华快速回复邮件，还有一个时尚顾问 AI 来帮李华挑选衣服和餐厅。

通过这个故事，可以看到智能体不仅仅是一个工具，更像是一个能够理解你、帮助你、甚至预测你需求的伙伴。他们通过大模型来理解复杂的任务，通过规划和记忆来高效完成任务，通过使用各种工具来实现目标。

**这就是智能体，一个能够让你的生活更轻松、工作更高效的神奇存在。**

#### 1.2.1 基础定义

在人工智能（Artificial Intelligence, AI）领域，**智能体（Agent）** 是一个能够感知环境并采取行动以实现某些目标的自主系统。智能体是 AI 研究的核心概念之一，它可以被看作一个**输入-处理-输出**的系统：智能体感知周围环境的信息（输入），通过内置的决策机制处理信息，并根据结果采取相应的行动（输出）。其目标可能是任务导向型的，如解决特定问题，也可能是优化某些**性能指标**。

一个简单的例子是智能家庭中的语音助手，它通过麦克风感知语音指令（感知），分析语义内容（决策），并通过扬声器回应（行动）。更复杂的例子包括自动驾驶汽车，它们需要实时感知路况，决策驾驶策略，并控制汽车实现导航。


#### 1.2.2 基本组成

1. **感知器（Perception）**   
感知器负责采集环境中的信息。它可以是物理传感器（如摄像头、麦克风、温度传感器）或数字输入接口（如 API、数据库）。感知器将采集的信息转化为系统能够理解的形式。

2. **决策系统（Decision-making）**  
决策系统是智能体的“大脑”，负责处理感知输入并做出相应的决策。决策可以是基于规则的（如条件逻辑）或基于模型的（如机器学习模型）。一些高级智能体还支持实时优化决策。

3. **执行器（Action）**  
执行器负责实施决策系统的指令，对环境进行物理或虚拟的操作。例如，机械臂是工业机器人智能体的执行器，而语音助手的执行器可能是合成语音模块。

4. **环境（Environment）**    
智能体所处的环境是其感知和行动的对象。环境可以是静态的（如棋盘游戏）或动态的（如真实世界）。环境与智能体之间的交互决定了智能体的实际表现。

5. **通信模块（Communication）（可选）**  
在多智能体系统（MAS）中，智能体可能需要与其他智能体或用户进行通信，这部分由专门的通信模块负责。


#### 1.2.3 分类
根据功能、结构和复杂性，智能体可以分为以下几类：

1. **反应式智能体（Reactive Agent）**
    - 特性：基于当前环境状态直接采取行动，无需历史数据或记忆。
    - 优点：实现简单，响应迅速。
    - 缺点：缺乏长期规划能力。
    - 实例：自动门、温控器。
    
    自动门可以看作是智能体的一种简单形式，它通过传感器感知周围环境（如人的接近），然后自动执行开门或关门的动作，具备一定的自主性和自动化能力，是智能体在日常生活中的一个基础应用实例。

2. **基于目标的智能体（Goal-based Agent）**
    - 特性：拥有明确的目标，通过推理和规划寻找实现目标的最佳路径。
    - 优点：能够适应更复杂的任务。
    - 缺点：需要更复杂的算法支持。
    - 实例：导航系统、物流机器人。
    导航系统以帮助用户到达目的地为目标，它会根据地图数据和实时路况，规划出从出发地到目的地的最佳路线，并在行驶过程中根据实际情况调整路线。

    - 特性：能够通过经验学习，不断优化决策。常用技术包括监督学习、强化学习等。
    - 优点：具备自我改进能力，适应动态环境。

    - 实例：智能推荐系统、围棋 AI（如 AlphaGo）。

#### 1.2.4 关键特性
1. **自主性（Autonomy）**    
智能体能够在很少或完全没有外部干预的情况下自主运行。
  
2. **适应性（Adaptability）**     
智能体能够根据环境的变化调整自身行为。例如，气候控制系统会根据房间温度自动调整暖气。

3. **互动性（Interactivity）**     
智能体与环境或其他智能体交互。多智能体系统中的互动是其关键特征之一。

4. **学习能力（Learning Ability）**    
一些智能体具备学习能力，能够通过历史数据或实时反馈优化自身表现。

5. **目标导向（Goal-orientedness）**     
智能体的所有行为都以实现目标为导向，这在规划型智能体中尤为明显。

 ## 2. 智能体的核心能力
 智能体结构可以拆分为四个部分：**大模型（LLM）**、**思考（Brain）**、**感知（Perception）**、**行动（Action）**。
 


### 2.1 大模型（LLM）
你真的了解什么是大型语言模型（LLM）吗？

LLM 即大型语言模型（Large Language Model），是一种人工智能技术，它通过训练大量的文本数据来学习语言的模式和结构。这些模型通常非常庞大，包含了数十亿甚至数千亿个参数，能够理解和生成自然语言文本。


    如果有一个超级学霸，它读过了互联网上几乎所有的书籍、文章、对话等等，然后你问它问题或者让它写作文，它就能根据它学到的知识给出答案或者写出文章。这个超级学霸就有点像 LLM，只不过它不是人，而是一个由代码和算法构成的虚拟存在。



| 对比维度   | 有大模型之前             | 有大模型之后             |
|:-----------|:-------------------------|:-------------------------|
| **交互模式** | 以文本交互为主           | 多模态交互，如文本、语音、图像等 |
| **信息获取** | 依赖预训练数据，更新不及时 | 能实时检索外部数据，获取最新信息 |
| **任务处理** | 多为单一任务处理         | 具备多任务处理和任务编排能力   |
| **用户体验** | 交互机械，需适应系统规则 | 能理解自然语言，提供个性化服务 |
| **应用场景** | 应用场景相对有限        | 广泛应用于多行业多领域        |

**从时间维度来看（技术视角）**

1. 专家系统 IBM 深蓝（1950s-1980s）

3. 深度学习 (2010s-2020s)
    - AlphaGo 
    - 卷积神经网络 - 图像识别 
5. 大模型 ChatGpt 智能助手（2020s-至今）

####  2.2.1 记忆（Memory）

记忆是指智能体在与用户交互或执行任务过程中动态积累和存储的信息。它可以是短期的，比如记住用户刚刚输入的指令；也可以是长期的，例如记住用户的偏好和历史交互记录。



    在一个阳光明媚的下午，小李在准备他的周末旅行。他记得上次去野餐时忘记带防晒霜，结果皮肤被晒伤了。这次，他决定提前列出所有必需品，包括防晒霜、帽子和足够的水。

    小李还回忆起上次旅行时，他因为没有提前规划路线，结果迷路了。所以这次，他提前在网上查找了地图，标记了想去的地方，并下载了离线地图以防万一。

    通过这些记忆，小李制定了一个完美的旅行计划，确保了旅途的顺利和愉快。

记忆在帮助我们避免过去的错误和做出更明智决策中起到关键作用。




#### 2.2.2  知识（Knowledge）

知识是思考和规划的基石，它为我们提供了必要的信息和洞察力，使我们能够更有效地处理信息、做出决策，并在复杂世界中导航。



    在一个小镇上，有一个叫做“智慧屋”的神奇地方。一天，居民小张想知道哪里的咖啡最好喝，他走进智慧屋进行求助。

    智慧屋的主人首先在自己的“知识库”里找，就像翻家里的老相册，找到了一些咖啡店的信息。但这些信息可能有点旧了。

    于是，他又用了一个叫“检索”的特殊技能。他先问自己：“最近哪家咖啡店最受欢迎？”然后，像侦探一样，从网上和小镇居民的聊天中找到了最新的线索。

    最后，他把新旧信息结合起来，告诉小张：“根据最新的信息，‘阳光咖啡店’的咖啡最好喝。”小张去喝了，果然很棒！

内部知识和外部知识结合，能帮我们找到最新、最好的答案。



#### 2.2.3 决策（Decision）

大模型的决策能力、推理和规划是其在复杂任务中表现的关键因素。

- 推理能力（Reasoning）

- 计划制定（Plan Formulation）

- 计划反思（Plan Reflection）



    在一个晴朗的周末，小明决定去郊外的农场体验生活。他首先确定了目标：体验农活，亲近自然。然后，他开始规划路线，准备野餐用品，并检查了天气预报，发现可能会下雨，于是带上了雨伞。

    在出发前，小明通过朋友的建议选择了一个评价很高的有机农场。到达农场后，他按照计划参与了种植和收割活动，了解了农作物的生长过程。午餐时，他享用了自己准备的食物，感受着田园的宁静。

    下午，天空突然下起了小雨，但小明并不担心，因为他已经准备了雨伞。雨中的农场别有一番景致，小明在雨中漫步，享受着清新的空气和雨后的彩虹。

良好的决策和规划对于实现目标至关重要。

如果把智能体比作一个人，感知端就如同人的五官，能够敏锐地捕捉到周围环境的变化和信息。没有高效准确的感知端，智能体就无法及时了解外界情况，难以做出明智的决策和有效的行动。

**什么是多模态 LLM**

多模态 LLM（Multimodal Large Language Models）是一种结合了文本和其他多种模态数据（如图像、音频等）的语言模型。多模态模型专注于从多种感知通道（如视觉、听觉、触觉等）中提取信息，并进行综合理解，该方向致力于通过多种模态输入生成新的内容，例如图像与文本结合生成描述性文字或视频，能够处理并理解多种不同模态输入的统一模型，从而提高模型的泛化能力。

人类的多模态感知为智能体感知端带来了极大的启发，人类通过视觉、听觉、触觉等多种感官协同工作，全方位地感知世界。所以智能体的感知端也应朝着多模态的方向发展，不仅能处理文本信息，还能融合视觉、听觉等多种模态的数据。

最常见的感知输入就是文本输入（提示词），文本中本身存在着各种语法结构、语义关系和上下文信息，智能体可以从大量的文字数据中提取关键信息，并进行深入的分析和理解。

智能体的行动端（Action）是整个智能体系统中至关重要的组成部分。它直接决定了智能体如何与外部环境进行交互，以及如何通过一系列的动作来实现其设定的目标。Action 不仅是智能体对外输出的表现形式，更是其适应和改变环境、解决问题、实现自身价值的关键手段。

1. 文本输出

    在目前的互联网产品中（尤其是助手类产品），文本输出是最常见的一种方式。目前的智能体平台、智能体框架大多都是文本输出（markdown 也算）。

2. 工具使用

    尽管智能体底层的 LLM 拥有丰富的知识储备和专业能力，但在处理具体问题时，仍可能面临鲁棒性问题、产生幻觉等一系列挑战。而工具的引入，能够有效地弥补这些不足，工具可以在专业性、事实性、可解释性等多个方面为智能体提供有力的支持。

    例如：在处理数学问题时，可以借助计算器进行精确计算；在获取实时信息时，利用搜索引擎能够快速获取最新的资讯。

    工具不仅能够解决特定问题，还能极大地扩展智能代理的行动空间。通过调用语音生成、图像生成等专家模型，智能体能够获得多模态的行动方式，从而更全面、灵活地应对各种任务和场景。

3. 具身智能

    具身（Embodyment）是指代理在与环境交互的过程中，理解、改造环境并更新自身状态的能力。具身行动（Embodied Action）则被视为虚拟智能与物理现实的互通桥梁，它使得智能体能够像人类一样，通过感知、行动来与真实世界进行深度互动。

    目前具身行动的研究仍主要集中于游戏平台《我的世界》等虚拟沙盒环境中。

    通过故事再理解一下

        在一个繁忙的办公室里，小智是一个虚拟助手，它的任务是帮助员工提高工作效率，起初，小智只能通过文本回复来提供帮助，但很快它发现这种方式在处理复杂问题时存在局限。

        有一天，一个员工需要解决一个紧急的数据分析问题，小智尝试用文本解释，但员工看起来一头雾水。这时，小智决定采取行动，它调用了一个数据分析工具，将复杂的数据转化为直观的图表和图形，员工立刻明白了问题所在，并迅速做出了决策。

        通过这次经历，小智意识到不同行动的重要性，它开始学习使用更多的工具，比如实时信息搜索和语音交互，甚至尝试与办公室的智能设备连接，以实现更直接的环境交互。

行动让智能体能够更有效地理解和解决问题，提供更丰富的交互方式，从而更好地服务于用户。


### 2.5 回顾

大模型（LLM）是智能体的基座，提供最基本的知识和推理等能力。

思考（Brain）是智能体的灵智，可以结合记忆和知识做出规划和决策。

感知（Perception）是智能体的五官，能够接收外部的各种信息。

行动（Action）是智能体的手脚，能够通过使用工具产生更多和外界交互的行为。

## 3. 基于大模型的智能体
在科技发展的历程中，人类对智能化的追求从未停止。从电影中的幻想角色，如《机器人总动员》的瓦力、《终结者》的 T-800，到现实中的智能助手，人们一直在努力打造能够自主完成任务的智能体（AI Agents）。如今，随着大型语言模型（LLM）的飞速发展，基于这些模型构建的智能体正逐渐成为现实。这些智能体通过整合LLM与规划、记忆等技术模块，能够执行复杂的任务，为我们的生活和工作带来便利。

### 3.1 LLM Agent 架构
基于LLM的智能体框架通常包括以下几个核心组件：

- **用户请求**：用户的提问或需求。
    - 用户通过自然语言向智能体提出问题或请求。例如，用户可能问：“当前欧洲最受欢迎的电动汽车品牌是什么？”
    
    
- **智能体/大脑**：作为协调者的智能体核心，负责处理和响应用户请求。
    - 智能体的核心部分，LLM，负责理解和处理用户请求。它通过分析用户的输入，确定任务的目标和需求。如果LLM有足够的知识储备，它可以直接回答问题。如果需要外部信息，它会调用相应的工具或API来获取信息。
    
    
- **规划**：帮助智能体分解任务，制定解决方案。
    - 规划模块帮助智能体分解复杂任务，制定解决方案。对于简单的问题，LLM 可以直接回答。但对于复杂的问题，如“过去十年里，欧洲电动汽车市场的增长趋势如何，这对环境政策有何影响？”，智能体需要将问题分解成多个子任务，并通过规划模块逐步解决。
    
    
- **记忆**：管理智能体过去的行为和信息。
    - 记忆模块帮助智能体存储和管理过去的行为和信息。这包括短期记忆（如当前对话的上下文）和长期记忆（如历史数据和知识库）。记忆模块使智能体能够更好地理解和处理用户的请求，提高任务执行的效率和准确性。

### 3.2 智能体的大脑
在构建以大型语言模型（LLM）为核心的智能体系统中，LLM 是至关重要的，充当系统的主脑和多任务协调的核心。这种智能体通过解析和执行基于提示模板的指令，这些模板不仅指导 LLM 具体操作，也详细定义了智能体的角色和人格，包括背景、性格、社会环境及人口统计信息等。这种人格化的描述使得智能体能更精准地理解和执行任务。 为了优化这一过程，系统设计需要综合考虑几个关键方面：

- 首先，系统需具备丰富的上下文理解和持续学习能力，不仅处理和记忆大量交互信息，还需不断优化执行策略和预测模型。
- 其次，引入多模态交互，融合文本、图像、声音等多种输入输出形式，让系统更自然有效地处理复杂任务和环境。此外，智能体的动态角色适应和个性化反馈也是提升用户体验和执行效率的关键。
- 最后，加强安全性和可靠性，确保系统稳定运行，赢得用户信任。 

整合这些元素，基于 LLM 的智能体系统能够在处理特定任务时展现出更高的效率和准确性，同时，在用户交互和系统长期发展方面展现出更强的适应性和可持续性。这种系统不仅仅是执行命令的工具，更是能够理解复杂指令、适应不同场景并持续优化自身行为的智能合作伙伴。

### 3.4 智能体的记忆
Agent 的记忆指的是它存储信息的能力，这可以是临时的或持久的。在人工智能和计算机科学中，记忆通常指的是数据存储。

其中临时数据主要指的是依靠LLM的短期记忆，持久化存储指的是依靠模型外置数据存储的长期记忆，长期当前重要的实现形式是依靠RAG。









#### 3.4.1 短期记忆
大模型的短期记忆，通常指的是LLM在处理数据时，能够临时存储和访问与当前任务相关的信息的能力。这种记忆不是永久性的，而是在模型进行计算和生成输出时临时存在的。

**特点**

1. 上下文理解：短期记忆使模型能够理解当前输入的上下文，包括之前的句子、段落或对话轮次。
2. 信息保持时间有限：与人类的短期记忆类似，大模型的短期记忆通常只保持在当前任务或会话期间。回话过期之后继续进行问答无法获取之前的信息。
3. 动态更新：当模型接收新的输入时，短期记忆中的信息会动态更新，以反映最新的上下文。是一个记忆窗口的概念，模型会不断记忆最新内容，忘记之前的内容，因此叫短期记忆。
4. 依赖于模型架构：大模型的短期记忆能力依赖于其架构设计，如 Transformer 模型中的自注意力机制。同时记忆窗口的大小，同样与模型有关，不同的模型记忆能力也不同。
5. 无物理存储：与计算机的 RAM 不同，大模型的短期记忆不是存储在物理介质上，而是通过计算过程中的状态来体现。

在这个例子中，模型需要记住用户提出的关于天气的问题，以及模型的回答，以便在用户提出去公园野餐的建议时，能够生成一个连贯的回答。这种能力是通过模型的短期记忆实现的，它允许模型在当前对话轮次中保持上下文的连贯性。

#### 3.4.2 长期记忆
大型语言模型（LLMs）的长期记忆是指模型能够存储、检索和利用在训练过程中学到的知识的能力。这种记忆不是临时的，而是模型在多次训练和学习后形成的，可以用于各种任务和决策。

1. 持久性：长期记忆的信息不是临时的，而是在模型训练后持久存储的。
2. 泛化能力：长期记忆使得模型能够泛化到新的、未见过的情境中，即使这些情境在训练数据中没有直接出现过。
3. 知识丰富：长期记忆包含了模型在训练过程中学到的广泛知识，如语言模式、语法规则、世界知识等。
4. 细粒度信息：长期记忆可以包含非常细粒度的信息，如特定领域的术语、罕见词汇的含义等。
5. 遗忘与更新：尽管长期记忆相对稳定，但在持续学习过程中，某些信息可能会被遗忘或更新。
6. 上下文无关性：长期记忆通常不依赖于特定的上下文或会话，它可以在多种不同的情境下被调用。


大型语言模型的长期记忆是其智能行为的基础，使得模型能够展现出类似于人类长期记忆的复杂认知能力。这种记忆能力是通过模型的训练过程逐渐建立起来的，并且可以通过持续学习不断优化和扩展。
