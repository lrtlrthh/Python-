## 3. 教学目标

1. 知识掌握  
- 能准确说出符号主义、联结主义、行为主义三大学派的核心主张：  
  -  符号主义：智能=符号操作+逻辑规则（课件1.1.1物理符号系统假设）  
  -  联结主义：智能=神经网络权重对数据的分布式表征（课件1.2.1Hebb规则与反向传播）  
  -  行为主义：智能=环境-动作-奖励的闭环优化（课件1.3.2强化学习框架图）  
- 对应课件案例，列举每个学派两项关键技术与应用场景：  
  -  符号主义：情境演算（灯开关案例）、专家系统 MYCIN（医疗诊断）  
  -  联结主义：CNN（AlexNet图像识别）、LSTM（Seq2seq翻译）  
  -  行为主义：Q-learning（AlphaGo Zero 自我对弈）、具身智能 Atlas 机器人搬运

2. 能力提升  
-  对照课件“自动驾驶”案例（模块 7），用“学派标签法”解析Waymo系统：指出感知模块（联结主义-CNN）、决策模块（行为主义-强化学习）、规则校验（符号主义-交通法规知识图谱）。  
-  针对课件“动物分类”规则（课件1.1.1），现场增补一条新规则并画出推理链，体验符号主义知识获取瓶颈。  
-   基于课件 AlphaGo Zero 训练曲线（模块 6），解释“灾难性干扰”并给出缓解策略（经验回放/渐进网络）。

3. 素养培育 
- 技术伦理：引用课件 Atlas 机器人强化学习片段，讨论“机器人训练中的安全试错”与“真实世界风险”之间的伦理平衡点。  
- 跨学科视角：用课件 1.3.3 控制论“反馈-通信-控制”三要素，解释行为主义与具身智能的哲学一致性。  
- 终身学习：建立“三学派技术追踪卡”——未来三个月，每月记录一个三学派融合的新案例（如神经符号系统、离线强化学习+知识图谱、具身智能+大模型）。



## 4. 教学重点、难点

**教学重点**：人工智能三大主流技术学派的核心原理、主要技术应用场景。

**教学难点**：深入理解每种学派的技术细节和实际应用中的挑战。


## 5. 教学方法

**教学方式**：讲授、讨论
    
**教学手段**：多媒体教学、案例分析



## 6. 预期学习成果

**预期学习成果**

- 理解并区分人工智能的三大技术学派：符号主义、联结主义和行为主义。
- 掌握每种学派的主要技术和应用场景。
- 通过讨论和案例分析，深入理解每种学派的技术细节和实际应用中的挑战。



## 7. 教学过程



**模块 1：AI 三大主流技术学派 (3分钟)**

- 引入和概述三大技术学派。

**模块 2：符号主义简介（10分钟）**

- 符号主义的基本原理和核心观点
- 符号主义的历史背景和发展

**模块 3：联结主义简介（10分钟）**

- 联结主义的基本原理和核心观点
- 联结主义的历史背景和发展

**模块 4：行为主义简介（10分钟）**

- 行为主义的基本原理和核心观点
- 行为主义的历史背景和发展

**模块 5：符号主义简介的主要技术和应用场景（18分钟）**

- 符号主义的主要技术和应用场景
- 符号主义的优缺点分析

**模块 6：联结主义的主要技术和应用场景（18分钟）**

- 联结主义的主要技术和应用场景
- 联结主义的优缺点分析

**模块 7：行为主义的主要技术和应用场景（18分钟）**

- 行为主义的主要技术和应用场景
- 行为主义的优缺点分析

**模块 8：总结回顾（3分钟）**
