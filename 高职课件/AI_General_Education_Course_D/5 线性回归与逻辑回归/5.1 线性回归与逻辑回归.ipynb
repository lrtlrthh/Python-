# 线性回归与逻辑回归



## 1. 线性回归

### 1.1 线性回归简介
想象一下，你正在寻找梦想中的房子。你浏览了无数房源，发现房价似乎与房屋面积密切相关。面积越大，价格越高，这似乎是一个显而易见的规律。但你是否想过，我们能否利用这个规律，更准确地预测房价呢？这就是线性回归的魅力所在！

线性回归，就像一位经验丰富的房地产专家，能够从海量数据中提炼出关键规律，帮助我们预测未来趋势。无论是预测房价、股票走势，还是分析广告投入与销售额的关系，线性回归都能大显身手。它不仅是机器学习领域的基石，更是我们理解世界、做出决策的利器。

#### 1.1.1 线性与非线性 

首先，我们来弄清楚什么是线性，什么是非线性
*   线性：两个变量之间的关系是一次函数关系的——图像是直线，叫做线性。
*   非线性：两个变量之间的关系不是一次函数关系的——图像不是直线，叫做非线性。

如上图所示，绿色图像 $y=0.5x+1$ 是线性的，而蓝色图像 $y = x^2$ 是非线性的

#### 1.1.2 回归  

看到“回归”一词，我们会不禁认为它是“回来”，“归来”的含义，但实际不然。它是一个更专业的术语。

现在假设我们有一个苹果，想要测量它的准确重量。由于测量工具（如电子秤）可能存在误差，每次测量得到的重量可能略有不同。  
例如，第一次测量得到150克，第二次测量得到148克，第三次测量得到152克……这些测量值都不是苹果的真实重量，而是带有误差的“测量值”。  

那怎样才能测到真实值呢，有这样一种方法：  
我们可以不断地重复测量苹果的重量，假设我们测量了10次，得到以下数据：150克、148克、152克、151克、149克、150克、153克、151克、149克、150克。  

最后我们通过统计方法（如计算平均值），我们可以将这10次测量值进行综合分析。计算平均值为：  
    $$
   \text{平均重量} = \frac{150 + 148 + 152 + 151 + 149 + 150 + 153 + 151 + 149 + 150}{10} = 150.3 \text{克}
   $$
这个平均值150.3克可以被视为对苹果真实重量的一个更接近的估计。

**回归**：人们在测量事物的时候因为客观条件所限，求得的都是测量值，而不是事物真实的值，为了能够得到真实值，无限次的进行测量，最后通过这些测量数据计算回归到真实值，这就是回归的由来。  
我们不妨用其原本英文的意思理解“回归”一词，用英文直译过来即 “regression”，其中：
*   **re** 表示“再次”或“重复
*   **gress** 表示“行走”或“前进”  

结合起来，“regression” 的字面意思是“重复前进”。回归的过程可以理解为通过多次测量（re），每一次测量都是向真实值的一次逼近（gress），通过不断地测量，最终计算回归到真实值。

在上面例子中，“回归”就是通过多次测量（re），让每次测量的结果不断逼近真实重量（gress），最终通过统计算平均值回归到一个更接近真实值的结果。

#### 1.1.3 线性回归 
想象一下，你有一堆数据点，它们在坐标系上散乱地分布着。每个点包含着（x，y）的坐标信息。线性回归就是想找到一条直线，这条直线能够尽可能地“贴合”**全部整体**的数据点。
<img src="https://imgbed.momodel.cn/hv/202502131552773.gif" alt="示例图片" width="50%" height="auto" style="display: block; margin: 0 auto;" >
如图所示，红色直线向蓝色数据点集中区域移动逼近的过程，就是线性回归的过程

线性回归是一种统计学方法，用于研究变量之间的线性关系。通俗来说，它就像是一种“拟合直线”的工具，目的是找到一条直线，让这条直线尽可能地接近一组数据点。从而找到一条最佳拟合直线来描述坐标轴中变量 y 和变量 x 存在的某种线性关系

当我们通过线性回归建立自变量（x）与因变量（y）之间的线性关系后，我们对线性回归中的未知量就可以实现预测，这正是线性回归在机器学习中的核心功能。

### 1.2 一元和多元线性回归


#### 1.2.1 一元线性回归
在现实生活中，我们常常需要通过已有的数据来预测未知的结果。例如，房地产市场中，我们可能希望根据房屋的面积来预测房价。一元线性回归就是一种简单而有效的统计方法，能够帮助我们实现这一目标。下面我们来通过房价预测的例子，以一元线性回归为切入点，进一步学习线性回归

我们上中学的时候，都学过二元一次方程，我们将 y 作为因变量，x 作为自变量，得到方程：  
$$y = \beta_{0} + \beta_{1}x$$
- $\beta_{0}$ 是截距项，表示当 x=0 时 y 的值。
- $\beta_{1}$ 是斜率，表示 x 每变化一个单位时 y 的平均变化量。
- $x$  是自变量，对应的是房屋面积。
- $y$  是因变量，对应的是房屋价格。

这里，有一组数据画出来的散点图，横坐标 x 代表面积，纵坐标 y 代表房价，一元线性回归找到了一条直线来尽可能全部拟合这些点。
<img src="https://imgbed.momodel.cn/hv/202502131816565.png" alt="示例图片" width="50%" height="auto" style="display: block; margin: 0 auto;" >
这里我们得到的拟合方程是y = 0.0512x + 7.1884，此时当我们获得一个新的房屋面积时，我们就可以用这个方程预测出大概的销售量。

#### 1.2.2 多元线性回归
现实中的数据可能是比较复杂的，自变量也可能不止一个，例如，影响房屋价格也很可能不止房屋面积一个因素，可能还有是否在地铁附近，房间数，层数，建筑年代等诸多因素。不过，这些因素对房价影响的权重是不同的，因此，我们可以使用多个权重来表示多个因素与房屋价格的关系：
$$
\hat{y} = w_1 * x_1 + w_2 * x_2 + \ldots + w_n * x_n + b
$$

- x：影响因素，即特征。
- w：每个x的影响力度。
- n：特征的个数。
- $ \hat{y} $：房屋的预测价格。
- b：截距项。  

当我们选择面积和卧室数量这两个因素表示与房屋价格的关系，模型如下：

$$
\text{房价} = 50 + 3 \times \text{面积} + 10 \times \text{卧室数量} + \varepsilon
$$

- 截距项 $b$：当面积和卧室数量为 0 时，房价基准值。
- $w_1$：面积每增加 1 平米，房价平均上涨 3 万元。
- $w_2$：每多一间卧室，房价平均上涨 10 万元。

#### 1.2.3 线性回归中的过拟合
当一个模型在训练数据上表现得非常好，但是对于未见过的数据（即测试数据）泛化能力差，就发生了过拟合。这意味着模型过于复杂，学习了训练数据中的噪音和细节，而不仅仅是潜在的数据分布。

在三维空间中，过拟合的模型的拟合曲面会像一个复杂的、起伏不定的地形，每一个训练数据点都像是一个山峰或山谷，模型试图去完美地“绕过”每一个点，而不是找到一个能够反映数据整体规律的平滑曲面。
<img src="https://imgbed.momodel.cn/hv/202502181421567.webp" alt="示例图片" width="50%" height="auto" style="display: block; margin: 0 auto;" >  
产生原因：

- 模型可能过于复杂，拥有太多的参数或太高的灵活性。
- 训练数据中可能包含噪音，模型学习了这些噪音而不是潜在的数据结构。
- 模型训练时间过长，迭代次数过多。

解决办法：
- 优化数据与特征：通过增加样本数量、引入噪声、减少特征数量以及进行特征工程，提升模型的泛化能力。
- 限制模型复杂度：采用正则化（如 L1 、L2 正则化）、调整模型复杂度（如减少多项式阶数）以及使用交叉验证，防止模型过拟合。
- 组合多个模型：利用集成方法来提高模型的泛化性能。

### 1.3 求解线性回归

#### 1.3.1 损失函数
- 残差

我们先从残差说起。残差说白了就是真实值和预测值间的差值（也可以理解为差距、距离），用公式表示是：$$e_i = y_i - \hat{y}_i$$
用图像表示为：
<img src="https://imgbed.momodel.cn/hv/202502140843675.png" alt="示例图片" width="50%" height="auto" style="display: block; margin: 0 auto;" >
对于某个房屋面积 $ x_i$，我们有对应的实售价 $y_i$，和预测出来的销售量 $\hat{y}_i$（通过将 $x_i$ 代入公式 $y = \beta_0 + \beta_1 x$ 计算得到），计算 $e_i = y_i - \hat{y}_i$ 的值，就是 $ x_i$ 对应的残差

- 残差平方和

一个点的残差体现了该点真实值与预测值的差距。显然，我们要评价回归模型预测数据质量的好坏，一个点的残差是远远不够的。  
对此，我们可以引入在机器学习中回归问题中最常用的损失函数：残差平方和。  

对于这个公式，我们可以看成是每一个点的残差的平方全部加在一起。

$$Q = \sum_{i=1}^{n} (y_i - \hat{y}_i)^2 = \sum_{i=1}^{n} \left(y_i - \left(\hat{\beta}_0 + \hat{\beta}_1 x_i\right)\right)^2$$

损失函数是衡量回归模型误差的函数，也就是我们要的“直线”的评价标准。这个函数的值越小，说明直线越能拟合我们的数据。




#### 1.3.2 最小二乘估计
那么如何找出对此，由最小二乘法给出的判断标准是：残差平方和的值达到最小。
$$Q = \sum_{i=1}^{n} (y_i - \hat{y}_i)^2 = \sum_{i=1}^{n} \left(y_i - \left(\hat{\beta}_0 + \hat{\beta}_1 x_i\right)\right)^2$$
上面公式中 $\hat{\beta}_0$ 和 $\hat{\beta}_1$ 未知，有两个未知参数的二次方程，画出来是一个三维空间中的图像，类似下面：
<img src="https://imgbed.momodel.cn/hv/202502181630286.png" alt="示例图片" width="50%" height="auto" style="display: block; margin: 0 auto;" >
还记得微积分知识的话，就知道导数为0时，Q取最小值，因此我们分别对 $\hat{\beta}_0$ 和 $\hat{\beta}_1$求偏导并令其为0:

$$
\frac{\partial Q}{\partial \beta_0} = 2 \sum_1^n (y_i - \hat{\beta}_0 - \hat{\beta}_1 x_i) = 0
$$
$$
\frac{\partial Q}{\partial \beta_1} = 2 \sum_1^n (y_i - \hat{\beta}_0 - \hat{\beta}_1 x_i) x_i = 0
$$

$x_i$，$y_i$（i=1,2,...n）都是已知的，全部代入上面两个式子，就可求得 $\hat{\beta}_0$ 和 $\hat{\beta}_1$ 的值啦。这就是最小二乘法，“二乘”是平方的意思。

### 1.4 总结
<img src="https://imgbed.momodel.cn/hv/202502181700386.png" alt="示例图片" width="30%" height="auto" style="display: block; margin: 0 auto;" >
线性回归作为一种基础的统计分析方法，具有简单易懂、计算高效等优点，在回归分析领域占据重要地位。


然而，它也有一定的局限性。为了进一步拓展应用范围，我们将在后续内容中引入逻辑回归。逻辑回归主要用于处理分类问题，能够将线性关系转化为概率预测，为解决更复杂的数据分析问题提供有力工具。


## 2. 逻辑回归



### 2.1 基本概念
逻辑回归是一种广泛应用于分类问题的机器学习算法。虽然名字里有"回归"二字，但它实际上解决的是分类问题，特别是二分类问题（比如判断邮件是垃圾邮件还是正常邮件，预测客户是否会违约等）。

举个生活中的例子：医生判断病人是否患有某种疾病。医生会考虑病人的各种指标（体温、血压、化验结果等），然后综合判断"患病"或"健康"。逻辑回归做的就是类似的事情，它通过学习历史数据，找出各个特征与结果之间的关系，然后对新数据进行预测。

逻辑回归旨在解决二分类问题，即预测结果属于两个可能类别之一。其基本模型形式如下：
$$
P(y=1|X) = \sigma(w^TX + b)
$$

其中：

- $ P(y=1|X)$ 表示给定输入特征 $ X $ 的情况下，样本属于类别 1 的概率。
- $ w $ 是特征的权重向量。
- $ X $ 是输入特征向量。
- $ b $) 是偏置项。
- $ \sigma $ 是 sigmoid 函数，其定义为：$\sigma(z) = \frac{1}{1 + e^{-z}}$

逻辑回归通过 sigmoid 函数将线性组合 $ w^TX + b $ 映射到 (0,1) 区间，从而得到样本属于类别 1 的概率。

### 2.2 为什么需要逻辑回归？
你可能学过线性回归，它适合预测连续值（比如预测房价、预测销售额）。但当我们要预测的是类别（是/否，好/坏）时，线性回归就不太合适了：

1. 线性回归的输出范围是负无穷到正无穷，而概率需要在0到1之间
2. 线性回归对异常值很敏感
3. 当数据分布不均匀时，线性回归效果会变差

逻辑回归通过使用 sigmoid 函数（也叫逻辑函数）解决了这些问题，它能把任何输入值压缩到 0-1 之间，正好可以表示概率。



### 2.3 Sigmoid 函数

Sigmoid 函数是逻辑回归的核心，它将任意实数转换为 (0,1) 区间内的概率值：

$$ \sigma(z) = \frac{1}{1 + e^{-z}} $$



<div align=center><img src="https://imgbed.momodel.cn//img/softmax.png" width="400px" /> </div>

特性：

- 当 $ z $ 接近正无穷大时，$ \sigma(z) $ 接近 1，表示样本非常可能属于类别 1。
- 当 $ z $ 接近负无穷大时，$ \sigma(z) $ 接近 0，表示样本非常可能属于类别 0。
- 当 $ z $ 等于 0 时，$ \sigma(z) = 0.5 $，表示样本有相等的概率属于类别 0 或类别 1。

### 2.4 决策边界
逻辑回归通过设定一个阈值（通常是0.5）来做分类决策：
- 如果 σ(z) ≥ 0.5，预测类别 1
- 如果 σ(z) < 0.5，预测类别 0

这个 0.5 的阈值对应的是$z=0$，也就是$wᵀx + b = 0$，这就是决策边界——一个将特征空间分成两部分的超平面。

实际应用中，阈值可以根据业务需求调整。比如在医疗诊断中，我们可能更重视不要漏诊，就可以把阈值调低，增加敏感性。

### 2.5 参数估计
逻辑回归通过最大化似然函数来估计参数。简单理解就是找一组参数，使得模型预测的结果与实际观察到的数据最吻合。

具体来说：
1. 对于每个样本，计算它的预测概率
2. 对于真实类别为 1 的样本，我们希望预测概率尽可能接近 1
3. 对于真实类别为 0 的样本，我们希望预测概率尽可能接近 0
4. 通过调整参数，使所有样本的联合概率最大

这个过程通常使用梯度下降法来实现，损失函数是交叉熵（也叫对数损失），而不是线性回归用的均方误差。

### 2.6 逻辑回归的实际应用案例

### 案例 1：银行信贷风险评估

**场景**：银行需要预测客户是否会违约。他们收集了客户的年龄、收入、信用评分、历史还款记录等数据。

**实现步骤**：
1. 数据准备：收集历史客户数据，标记哪些客户违约了
2. 特征工程：可能需要对收入进行分箱处理，对类别特征进行编码
3. 模型训练：使用逻辑回归学习各特征与违约概率的关系
4. 模型评估：计算准确率、召回率等指标
5. 应用：对新客户计算违约概率，决定是否批准贷款

**效果**：相比人工审批，模型可以更快速、更一致地评估风险，减少主观偏差。

### 案例 2：医疗诊断辅助

**场景**：医院想通过患者的各项检查指标预测是否患有某种疾病。

**实现步骤**：
1. 收集历史病例数据，包括检查结果和最终诊断
2. 处理缺失值和异常值
3. 训练逻辑回归模型
4. 重点关注召回率（不要漏诊）和精确率（减少误诊）
5. 医生参考模型预测结果，结合临床经验做最终判断

**效果**：模型可以帮助医生快速筛选高风险患者，提高诊断效率。

### 案例 3：电商用户购买预测

**场景**：电商平台想预测用户点击某个商品后是否会购买。

**实现步骤**：
1. 收集用户行为数据：浏览历史、点击记录、停留时间等
2. 处理类别不平衡问题（购买的用户通常远少于不购买的）
3. 使用加权逻辑回归或调整分类阈值
4. 评估模型效果，优化推荐策略

**效果**：可以更精准地投放广告和推荐商品，提高转化率。
