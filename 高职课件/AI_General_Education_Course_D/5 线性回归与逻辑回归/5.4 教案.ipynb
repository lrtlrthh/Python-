**6 学时**

理论课 [√]   
实验课 [√]

## 3. 教学目标

1. 知识掌握  
    - 说出线性回归与逻辑回归各自适用的任务类型（回归 vs 二分类）
    - 写出最小二乘损失、对数损失公式并指出其对应场景
    - 解释 Sigmoid 如何把 (−∞,+∞) 映射到 (0,1)
    - 描述过拟合在回归与分类中的不同表现及缓解思路

2. 能力提升  
    - 能用 Excel/Python 现场跑通一元/多元线性回归并解释截距、斜率含义
    - 能用纸笔手推一次梯度下降更新步骤
    - 能对给定二分类任务选择逻辑回归并给出特征工程要点

3. 素养培育 
    - 理解“简单模型先上”的工程伦理，抑制“一上来就深度”的冲动
    - 认识数据标准化、正则化对公平性与可解释性的意义



## 4. 教学重点、难点

**教学重点**：最小二乘 vs 对数损失、Sigmoid 的物理意义、梯度下降思想

**教学难点**：过拟合的数学直觉、正则项如何“惩罚”复杂度、概率输出的解释


## 5. 教学方法

**教学方式**：讲授、实验、讨论

**教学手段**：互联网环境下的上机操作


## 6. 预期学习成果

**预期学习成果**

- 理解并区分线性回归与逻辑回归的适用场景：线性回归用于回归任务，逻辑回归用于分类任务。
- 掌握线性回归的核心概念，包括最小二乘法、损失函数、过拟合及其缓解方法。
- 掌握逻辑回归的核心概念，包括 Sigmoid 函数、对数损失函数、梯度下降优化方法。
- 通过动手演练和案例分析，深入理解线性回归与逻辑回归的模型训练过程及其在实际应用中的优势与局限。
- 能够运用所学知识解决简单的回归与分类问题，并对模型结果进行合理解释。



## 7. 教学过程



**模块 1：线性回归 (60 分钟)**

- 线性回归简介
- 一元和多元线性回归
- 求解线性回归


**模块 2：逻辑回归（60 分钟）**

- 基本概念
- Sigmoid 函数
- 损失函数
- 模型训练
- 二分类应用
- 多分类扩展
- 优点与缺点
- 总结


**模块 3：线性回归实战：预测考试分数（70 分钟）**


本实训通过构建一个从学习时间预测考试分数的线性回归模型来展示数据的可视化。

**模块 4：逻辑回归实战（70 分钟）**


本实训通过一个广泛使用的电影和电视节目相关的数据集来训练逻辑回归模型，判断某一条影评是好评还是差评。

**模块 4：总结（10 分钟）**
