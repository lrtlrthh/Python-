### 2.1 字典

字典是一种可变容器，用于存储键值对。例如：

```python
# 创建字典
account = {"account_number": "123456", "balance": 1000.0, "owner": "张三"}
# 访问值
print(account["account_number"]) # 输出 123456
# 修改值
account["balance"] = 2000.0
# 添加键值对
account["type"] = "储蓄账户"
```

### 2.2 列表

列表用于存储多个有序的元素。例如：

```python
# 创建列表
transaction_history = []
# 添加元素
transaction_history.append("存款 500 元")
transaction_history.append("取款 200 元")
# 遍历列表
for transaction in transaction_history:
    print(transaction)
```

```python
# 写入文件
with open("accounts.txt", "w") as file:
    file.write("123456,张三,1000.0\n")
    file.write("789012,李四,2000.0")

# 读取文件
with open("accounts.txt", "r") as file:
    for line in file:
        account_info = line.strip().split(",")
        print(f"账号：{account_info[0]}，户主：{account_info[1]}，余额：{account_info[2]}")
```

## 4. 银行<span class="motutor-highlight motutor-id_q1u8pe7-id_zza3a30"><i></i>系统基础版  </span>


实现银行系统的注册、登录、存款、取款、查询余额等基本功能。

```python
# 定义全局变量，模拟数据库
accounts = {}  # 存储账号信息，键为账号，值为账户字典

# 注册功能
def register():
    account_number = input("请输入账号：")
    if account_number in accounts:
        print("该账号已存在！")
        return
    password = input("请输入密码：")
    balance = float(input("请输入初始余额："))
    owner = input("请输入户主姓名：")
    accounts[account_number] = {
        "password": password,
        "balance": balance,
        "owner": owner,
        "transaction_history": []  # 交易历史
    }
    print("注册成功！")

# 登录功能
def login():
    account_number = input("请输入账号：")
    if account_number not in accounts:
        print("账号不存在！")
        return None
    password = input("请输入密码：")
    if password != accounts[account_number]["password"]:
        print("密码错误！")
        return None
    return account_number

# 存款功能
def deposit(account_number):
    amount = float(input("请输入存款金额："))
    if amount <= 0:
        print("存款金额必须大于 0！")
        return
    accounts[account_number]["balance"] += amount
    accounts[account_number]["transaction_history"].append(f"存款 {amount} 元")
    print(f"存款成功！当前余额：{accounts[account_number]['balance']} 元")

# 取款功能
def withdraw(account_number):
    amount = float(input("请输入取款金额："))
    if amount <= 0:
        print("取款金额必须大于 0！")
        return
    if amount > accounts[account_number]["balance"]:
        print("余额不足！")
        return
    accounts[account_number]["balance"] -= amount
    accounts[account_number]["transaction_history"].append(f"取款 {amount} 元")
    print(f"取款成功！当前余额：{accounts[account_number]['balance']} 元")

# 查询余额功能
def check_balance(account_number):
    account = accounts[account_number]
    print(f"账号：{account_number}")
    print(f"户主：{account['owner']}")
    print(f"余额：{account['balance']} 元")
    print("交易历史：")
    for transaction in account["transaction_history"]:
        print(transaction)

# 主程序
def main():
    while True:
        print("\n===== 欢迎使用简易银行系统 =====")
        print("1. 注册")
        print("2. 登录")
        print("3. 退出")
        choice = input("请选择操作：")
        if choice == "1":
            register()
        elif choice == "2":
            account_number = login()
            if account_number:
                while True:
                    print("\n===== 欢迎，" + accounts[account_number]["owner"] + " =====")
                    print("1. 存款")
                    print("2. 取款")
                    print("3. 查询余额")
                    print("4. 退出登录")
                    operation = input("请选择操作：")
                    if operation == "1":
                        deposit(account_number)
                    elif operation == "2":
                        withdraw(account_number)
                    elif operation == "3":
                        check_balance(account_number)
                    elif operation == "4":
                        print("已退出登录。")
                        break
                    else:
                        print("无效的操作选项！")
        elif choice == "3":
            print("感谢使用简易银行系统，再见！")
            break
        else:
            print("无效的选项！")

if __name__ == "__main__":
    main()
```

## 6. 参考代码

```python
import json

# 定义全局变量
accounts_file = "accounts.json"  # 用于存储账户数据的文件

# 加载账户数据
def load_accounts():
    try:
        with open(accounts_file, "r") as file:
            return json.load(file)
    except FileNotFoundError:
        return {}  # 如果文件不存在，返回空字典

# 保存账户数据
def save_accounts(accounts):
    with open(accounts_file, "w") as file:
        json.dump(accounts, file)

# 注册功能（拓展版）
def register(accounts):
    account_number = input("请输入账号：")
    if account_number in accounts:
        print("该账号已存在！")
        return accounts
    password = input("请输入密码：")
    balance = float(input("请输入初始余额："))
    owner = input("请输入户主姓名：")
    account_type = input("请输入账户类型（普通/金卡/VIP）：")
    accounts[account_number] = {
        "password": password,
        "balance": balance,
        "owner": owner,
        "type": account_type,
        "transaction_history": []  # 交易历史
    }
    print("注册成功！")
    return accounts

# 登录功能（拓展版）
def login(accounts):
    account_number = input("请输入账号：")
    if account_number not in accounts:
        print("账号不存在！")
        return None, accounts
    password = input("请输入密码：")
    if password != accounts[account_number]["password"]:
        print("密码错误！")
        return None, accounts
    return account_number, accounts

# 转账功能
def transfer(accounts, account_number):
    target_account = input("请输入转账目标账号：")
    if target_account not in accounts:
        print("目标账号不存在！")
        return accounts
    amount = float(input("请输入转账金额："))
    if amount <= 0:
        print("转账金额必须大于 0！")
        return accounts
    if amount > accounts[account_number]["balance"]:
        print("余额不足！")
        return accounts
    # 扣除转账金额（根据账户类型设置不同的手续费）
    account_type = accounts[account_number]["type"]
    if account_type == "普通":
        fee = 2  # 普通账户转账手续费 2 元
    elif account_type == "金卡":
        fee = 1  # 金卡账户转账手续费 1 元
    else:  # VIP 账户
        fee = 0  # VIP 账户转账无手续费
    total_amount = amount + fee
    if total_amount > accounts[account_number]["balance"]:
        print("余额不足支付转账金额和手续费！")
        return accounts
    accounts[account_number]["balance"] -= total_amount
    accounts[target_account]["balance"] += amount
    # 记录交易历史
    accounts[account_number]["transaction_history"].append(f"转账支出 {amount} 元（手续费 {fee} 元）到账号 {target_account}")
    accounts[target_account]["transaction_history"].append(f"收到转账 {amount} 元来自账号 {account_number}")
    print(f"转账成功！扣除手续费 {fee} 元，实际转账 {amount} 元。当前余额：{accounts[account_number]['balance']} 元")
    return accounts

# 主程序（拓展版）
def main():
    accounts = load_accounts()  # 加载账户数据
    while True:
        print("\n===== 欢迎使用简易银行系统 =====")
        print("1. 注册")
        print("2. 登录")
        print("3. 退出")
        choice = input("请选择操作：")
        if choice == "1":
            accounts = register(accounts)
            save_accounts(accounts)  # 保存账户数据
        elif choice == "2":
            account_number, accounts = login(accounts)
            if account_number:
                while True:
                    print("\n===== 欢迎，" + accounts[account_number]["owner"] + "（" + accounts[account_number]["type"] + " 账户） =====")
                    print("1. 存款")
                    print("2. 取款")
                    print("3. 查询余额")
                    print("4. 转账")
                    print("5. 退出登录")
                    operation = input("请选择操作：")
                    if operation == "1":
                        amount = float(input("请输入存款金额："))
                        if amount > 0:
                            accounts[account_number]["balance"] += amount
                            accounts[account_number]["transaction_history"].append(f"存款 {amount} 元")
                            print(f"存款成功！当前余额：{accounts[account_number]['balance']} 元")
                            save_accounts(accounts)  # 保存账户数据
                        else:
                            print("存款金额必须大于 0！")
                    elif operation == "2":
                        amount = float(input("请输入取款金额："))
                        if amount > 0:
                            if amount > accounts[account_number]["balance"]:
                                print("余额不足！")
                            else:
                                accounts[account_number]["balance"] -= amount
                                accounts[account_number]["transaction_history"].append(f"取款 {amount} 元")
                                print(f"取款成功！当前余额：{accounts[account_number]['balance']} 元")
                                save_accounts(accounts)  # 保存账户数据
                        else:
                            print("取款金额必须大于 0！")
                    elif operation == "3":
                        print(f"账号：{account_number}")
                        print(f"户主：{accounts[account_number]['owner']}")
                        print(f"账户类型：{accounts[account_number]['type']}")
                        print(f"余额：{accounts[account_number]['balance']} 元")
                        print("交易历史：")
                        for transaction in accounts[account_number]["transaction_history"]:
                            print(transaction)
                    elif operation == "4":
                        accounts = transfer(accounts, account_number)
                        save_accounts(accounts)  # 保存账户数据
                    elif operation == "5":
                        print("已退出登录。")
                        break
                    else:
                        print("无效的操作选项！")
        elif choice == "3":
            print("感谢使用简易银行系统，再见！")
            save_accounts(accounts)  # 退出时保存账户数据
            break
        else:
            print("无效的选项！")

if __name__ == "__main__":
    main()
```
