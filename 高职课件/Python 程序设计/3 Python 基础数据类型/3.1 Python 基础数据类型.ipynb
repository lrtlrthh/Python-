## 6. <span class="motutor-highlight motutor-id_mbxn3v9-id_ooh1m2a"><i></i>实训</span>

### 6.1 讲解实操

通过本次实训，将学习 `Python` 中不同数据类型的基本特点和操作方式。重点介绍数值类型（`int` 和 `float`）以及字符串类型（`str`），并通过实践巩固对这些数据类型的理解。

### 6.2 知识点回顾

1. 数值类型

    - 整型 (`int`)：表示整数，示例：42、-5、0。
    - 浮点型 (`float`)：表示带小数部分的数字，示例：3.14、-0.01、2.0。
    
2. 字符串类型

    - 字符串 (`str`)：表示文本内容，可以包含字母、数字和符号，示例："Hello"、'Python'、"123"。
    
3. 判断数据类型

     - 使用 `type()` 函数判断变量的数据类型。

```python
# 1.数值类型
num_int = 7
num_float = 7.7

# 2.字符串
num_str = '77'

# 3.判断数据类型

# 判断变量的数据类型
print(type(num_int))  # 输出: <class 'int'>
print(type(num_float))  # 输出: <class 'float'>
print(type(num_str))  # 输出: <class 'str'>
```

### 6.3 数据类型转换

**1. 数值类型之间的转换**

- 使用 int() 将浮点型转换为整型。
- 使用 float() 将整型转换为浮点型。

```python
# 数值类型转换
num_float = 3.99
num_int = int(num_float)  # 转换为整型，舍弃小数部分
print(num_int)  # 输出: 3

num_int = 10
num_float = float(num_int)  # 转换为浮点型
print(num_float)  # 输出: 10.0
```

**示例: 判断用户输入类型并进行处理**

编写程序，读取用户输入，判断其是数字还是字符串，并执行相应操作。

### 6.4 数据类型操作

**示例 1: 判断用户输入类型并进行处理**

编写程序，读取用户输入，判断其是数字还是字符串，并执行相应操作。

```python
user_input = input("请输入一个值：")
# 判断输入值是否是整数
if user_input.isdigit():
    # 如果是整数
    num = int(user_input)
    print(f"你输入的是整数，平方是 {num ** 2}")
else:
    print(f"你输入的是字符串，长度是 {len(user_input)}")
```

**数值与字符串之间的转换**

- 使用 str() 将数值转换为字符串。
- 使用 int() 或 float() 将字符串转换为数值（前提是字符串内容为合法的数字）。

```python
# 数值转字符串
num = 42
num_str = str(num)
print(num_str)  # 输出: "42"
print(type(num_str))  # 输出: <class 'str'>

# 字符串转数值
num_str = "3.14"
num_float = float(num_str)  # 转换为浮点型
print(num_float)  # 输出: 3.14
```

```python
name = input("请输入你的名字：")
age = input("请输入你的年龄：")
# 捕获输入值有效（try-except）
try:
    age = int(age)
    print(f"你好，{name}！10 年后你将 {age + 10} 岁。")
except ValueError:
    print("年龄输入无效，请输入整数！")
```
