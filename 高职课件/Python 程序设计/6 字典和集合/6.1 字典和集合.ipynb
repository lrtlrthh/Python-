字典是一种可变容器，且可存储任意类型对象。  
字典的每个键值 `key=>value` 对用冒号 : 分割，每个键值对之间用逗号 , 分割，整个字典包括在花括号 `{}` 中 ,格式如下所示：  
`d = {key1 : value1, key2 : value2 }`  
键一般是唯一的，如果重复最后的一个键值对会替换前面的，值不需要唯一。  
集合是一个无序的不重复元素序列。

字典和集合在 Python 中也非常常用，本节课程，我们来一起学习**字典和集合**的相关知识。

## 1.<span class="motutor-highlight motutor-id_ft9spop-id_s091fs8"><i></i> 字典</span>

字典 `dictionary` ，在一些编程语言中也称为 `hash` ， `map` ，是一种由键值对组成的数据结构。

顾名思义，我们把键想象成字典中的单词，值想象成词对应的定义，那么——

一个词可以对应一个或者多个定义，但是这些定义只能通过这个词来进行查询。

**Python** 使用`key: value`这样的结构来表示字典中的元素结构。

````python
a = {}
type(a)
````

````python
a = dict()
type(a)
````

当我们 `print` 一个字典时，**Python** 并不一定按照插入键值的先后顺序进行显示,因为字典中的键本身不一定是有序的。

````python
print (a)
````

````python
print (b)
````

因此，**Python** 中不能用支持用数字索引按顺序查看字典中的值，而且数字本身也有可能成为键值，这样会引起混淆：

````python
# 会报错
a[0]
````

**<span class="motutor-highlight motutor-id_uml0pyr-id_wbg810g"><i></i>键必须是不可变的类型</span>**

出于 hash 的目的， Python 中要求这些键值对的**键**必须是**不可变**的，而值可以是任意的 Python 对象。

一个表示近义词的字典：

````python
synonyms = {}
synonyms['mutable'] = ['changeable', 'variable', 'varying', 'fluctuating',
                       'shifting', 'inconsistent', 'unpredictable', 'inconstant',
                       'fickle', 'uneven', 'unstable', 'protean']
synonyms['immutable'] = ['fixed', 'set', 'rigid', 'inflexible', 
                         'permanent', 'established', 'carved in stone']
synonyms
````

另一个例子：

````python
# 定义四个字典
e1 = {'mag': 0.05, 'width': 20}
e2 = {'mag': 0.04, 'width': 25}
e3 = {'mag': 0.05, 'width': 80}
e4 = {'mag': 0.03, 'width': 30}
# 以字典作为值传入新的字典
events = {500: e1, 760: e2, 3001: e3, 4180: e4}
events
````

键（或者值）的数据类型可以不同：

````python
people = [
    {'first': 'Sam', 'last': 'Malone', 'name': 35},
    {'first': 'Woody', 'last': 'Boyd', 'name': 21},
    {'first': 'Norm', 'last': 'Peterson', 'name': 34},
    {'first': 'Diane', 'last': 'Chambers', 'name': 33}
]
people
````

### 1.2 <span class="motutor-highlight motutor-id_xp98o05-id_g07l6g7"><i></i>适合做键的类型</span>

在不可变类型中，整数和字符串是字典中最常用的类型；而浮点数通常不推荐用来做键，原因如下：

````python
data = {}
data[1.1 + 2.2] = 6.6
# 会报错
data[3.3]
````

事实上，观察`data`的值就会发现，这个错误是由浮点数的精度问题所引起的：

````python
data
````

有时候，也可以使用元组作为键值，例如，可以用元组做键来表示从第一个城市飞往第二个城市航班数的多少：

````python
connections = {}
connections[('New York', 'Seattle')] = 100
connections[('Austin', 'New York')] = 200
connections[('New York', 'Austin')] = 400
````

元组是有序的，因此 `('New York', 'Austin')` 和 `('Austin', 'New York')` 是两个不同的键：

````python
print (connections[('Austin', 'New York')])
print (connections[('New York', 'Austin')])
````

### 1.3 <span class="motutor-highlight motutor-id_vqibbm2-id_s41islr"><i></i>字典方法</span>

**`get`方法**

之前已经见过，用索引可以找到一个键对应的值，但是当字典中没有这个键的时候，Python会报错，这时候可以使用字典的 `get` 方法来处理这种情况，其用法如下：

    `d.get(key, default = None)`

返回字典中键 `key` 对应的值，如果没有这个键，返回 `default` 指定的值（默认是 `None` ）。

````python
a = {}
a["one"] = "this is number 1"
a["two"] = "this is number 2"
````

索引不存在的键值会报错：

````python
a["three"]
````

改用get方法：

````python
print (a.get("three"))
````

指定默认值参数：

````python
print(a.get("three","none"))
````

**`pop` <span class="motutor-highlight motutor-id_yfvx3vv-id_lghlxmo"><i></i>方法删除元素</span>**

`pop` 方法可以用来弹出字典中某个键对应的值，同时也可以指定默认参数：

    `d.pop(key, default = None)`

删除并返回字典中键 `key` 对应的值，如果没有这个键，返回 `default` 指定的值（默认是 `None` ）。

**<span class="motutor-highlight motutor-id_x9i3v49-id_ddhpdnf"><i></i>`update`方法更新字典</span>**

将字典`newd`中的内容更新到`d`中去。

## 2.<span class="motutor-highlight motutor-id_y8zk3sz-id_dilffgl"><i></i> 集合</span>

之前看到的列表和字符串都是一种有序序列，而集合 `set` 是一种无序的序列。

因为集合是无序的，所以当集合中存在两个同样的元素的时候， **Python** 只会保存其中的一个（唯一性）；同时为了确保其中不包含同样的元素，集合中放入的元素只能是不可变的对象（确定性）。

<img src="https://imgbed.momodel.cn/20201009155601.png" width=600/>

### 2.1 集合生成

可以用`set()`函数来显示的生成空集合：

````python
a = set()
type(a)
````

也可以使用一个列表来初始化一个集合：

````python
a = set([1, 2, 3, 1])
a
````

集合会自动去除重复元素 `1`。

可以看到，集合中的元素是用大括号`{}`包含起来的，这意味着可以用`{}`的形式来创建集合：

````python
a = {1, 2, 3, 1}
a
````

但是创建空集合的时候只能用`set`来创建，因为在 **Python** 中`{}`创建的是一个空的字典：

````python
s = {}
type(s)
````

### 2.2 集合操作

<span class="motutor-highlight motutor-id_onasw9e-id_e6kvev6"><i></i>假设有这样两个集合：</span>

````python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
````

**并**

两个集合的并，返回包含两个集合所有元素的集合（去除重复）。

可以用方法 `a.union(b)` 或者操作 `a | b` 实现。

````python
a.union(b)
````

````python
b.union(a)
````

````python
a | b
````

<span class="motutor-highlight motutor-id_bn6qqmc-id_v1e1inp"><i></i>**交**</span>

两个集合的交，返回包含两个集合共有元素的集合。

可以用方法 `a.intersection(b)` 或者操作 `a & b` 实现。

````python
a.intersection(b)
````

````python
b.intersection(a)
````

````python
a & b
````

````python
print(a & b)
````

注意：一般使用print打印set的结果与表示方法并不一致。

**<span class="motutor-highlight motutor-id_zciu4ua-id_ifq7ql4"><i></i>差</span>**

`a` 和 `b` 的差集，返回只在 `a` 不在 `b` 的元素组成的集合。

可以用方法 `a.difference(b)` 或者操作 `a - b` 实现。

````python
a.difference(b)
````

````python
a - b
````

注意，`a - b` 与 `b - a`并不一样，`b - a` 返回的是返回 b 不在 a 的元素组成的集合：

````python
b.difference(a)
````

````python
b - a 
````

**<span class="motutor-highlight motutor-id_xir9a6m-id_yprokaq"><i></i>对称差</span>**

`a` 和`b` 的对称差集，返回在 `a` 或在 `b` 中，但是不同时在 `a` 和 `b` 中的元素组成的集合。

可以用方法 `a.symmetric_difference(b)` 或者操作 `a ^ b` 实现（异或操作符）。

````python
a.symmetric_difference(b)
````

````python
b.symmetric_difference(a)
````

````python
a ^ b
````

**<span class="motutor-highlight motutor-id_lbopk7k-id_pkh7crs"><i></i>包含关系</span>**

假设现在有这样两个集合：

````python
a = {1, 2, 3}
b = {1, 2}
````

要判断 `b` 是不是 `a` 的子集，可以用 `b.issubset(a)` 方法，或者更简单的用操作 `b <= a` ：

````python
b.issubset(a)
````

````python
b <= a
````

与之对应，也可以用 `a.issuperset(b)` 或者 `a >= b` 来判断：

````python
a.issuperset(b)
````

````python
a >= b
````

方法只能用来测试子集，但是操作符可以用来判断真子集：

````python
a <= a
````

自己不是自己的真子集：

````python
a < a
````

### 3.2 知识回顾

**字典（`dict`）**

字典是键值对结构，可用于存储商品及其销量。

```Python
sales = {
    "苹果": 120,
    "香蕉": 80,
    "橙子": 100,
    "草莓": 150
}
```
在 sales 字典中，键（`key`）是商品名，值（`value`）是销量。

字典操作:

````python
sales = {
    "苹果": 120,
    "香蕉": 80,
    "橙子": 100,
    "草莓": 150
}

sales["苹果"] += 30  # 更新苹果的销量
sales["西瓜"] = 60  # 添加新商品
del sales["香蕉"]  # 删除商品

print(f'商品销量: {sales}')
````

**集合（`set`）**

集合用于**去重**，如统计购买某商品的唯一用户数。


````python

buyers = {
    "苹果": {"张三", "李四", "王五"},
    "香蕉": {"张三", "赵六"},
}

buyers["苹果"].add("赵六")  # 添加购买用户
````

### 3.3 实现步骤

**初始化数据**


````python
# 商品销量字典
sales = {
    "苹果": 120,
    "香蕉": 80,
    "橙子": 100,
    "草莓": 150
}

# 购买用户集合字典
buyers = {
    "苹果": {"张三", "李四", "王五"},
    "香蕉": {"张三", "赵六"},
    "橙子": {"李四", "王五"},
    "草莓": {"王五", "赵六"}
}
````

**查询某个商品的销量**
```python
def query_sales(product_name):
    """查询商品销量和购买用户"""
    if product_name in sales:
        print(f"商品 {product_name} 的销量是 {sales[product_name]} 件")
        print(f"购买该商品的唯一用户数：{len(buyers[product_name])}")
    else:
        print(f"商品 {product_name} 不在销售记录中")
```

**生成热门商品排行榜**
```Python
def top_products():
    """生成销量排行榜（按销量降序排序）"""
    sorted_sales = sorted(sales.items(), key=lambda x: x[1], reverse=True)
    print("📊 热门商品排行榜（按销量）:")
    for rank, (product, count) in enumerate(sorted_sales, start=1):
        print(f"{rank}. {product} - {count} 件")
```
