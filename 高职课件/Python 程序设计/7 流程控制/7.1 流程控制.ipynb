Python 提供了现代编程语言都支持的两种基本流程控制结构，**分支结构**和**循环结构**：

**分支结构**用于实现根据条件来选择性地执行某段代码，可以使用判断语句实现，   
**循环结构**用于实现根据循环条件重复执行某段代码，可以使用循环语句实现。

## 1. <span class="motutor-highlight motutor-id_dnzgvgg-id_bck228z"><i></i>判断语句</span>

<img src="https://imgbed.momodel.cn/20201009155314.png" width=600/>

### 1.1 基本用法 

判断，基于一定的条件，决定是否要执行特定的一段代码，例如判断一个数是不是正数：

````python
x = 0.5
if x > 0:
    print ("Hey!")
    print ("x is positive")
````

在这里，如果 `x > 0` 为 `False` ，那么程序将不会执行两条 `print` 语句。

虽然都是用 `if` 关键词定义判断，但与**C，Java**等语言不同，**Python**不使用 `{}` 将 `if` 语句控制的区域包含起来。**Python**使用的是缩进方法。同时，也不需要用 `()` 将判断条件括起来。

上面例子中的这两条语句：
```python
    print "Hey!"
    print "x is positive"
```
就叫做一个代码块，同一个代码块使用同样的缩进值，它们组成了这条 `if` 语句的主体。

不同的缩进值表示不同的代码块，例如：

`x > 0` 时：

````python
x = 0.5
if x > 0:
    print ("Hey!")
    print ("x is positive")
    print ("This is still part of the block")
print ("This isn't part of the block, and will always print.")
````

`x < 0` 时：

````python
x = -0.5
if x > 0:
    print ("Hey!")
    print ("x is positive")
    print ("This is still part of the block")
print ("This isn't part of the block, and will always print.")
````

在这两个例子中，最后一句并不是`if`语句中的内容，所以不管条件满不满足，它都会被执行。

一个完整的 `if` 结构通常如下所示（注意：条件后的 `:` 是必须要的，缩进值需要一样）：
    
    if <condition 1>:
        <statement 1>
        <statement 2>
    elif <condition 2>: 
        <statements>
    else:
        <statements>

当条件1被满足时，执行 `if` 下面的语句，当条件1不满足的时候，转到 `elif` ，看它的条件2满不满足，满足执行 `elif` 下面的语句，不满足则执行 `else` 下面的语句。

对于上面的例子进行扩展：

````python
x = 0
if x > 0:
    print ("x is positive")
elif x == 0:
    print ("x is zero")
else:
    print ("x is negative")
````

`elif` 的个数没有限制，可以是1个或者多个，也可以没有。

`else` 最多只有1个，也可以没有。

可以使用 `and` ， `or` , `not` 等关键词结合多个判断条件：

````python
x = 10
y = -5
x > 0 and y < 0
````

````python
not x > 0
````

````python
x < 0 or y < 0
````

这里使用这个简单的例子，假如想判断一个年份是不是闰年，按照闰年的定义，这里只需要判断这个年份是不是能被4整除，但是不能被100整除，或者正好<span class="motutor-highlight motutor-id_rw4vris-id_a3fbhy9"><i></i>被400整除：</span>

````python
year = 1900
if year % 400 == 0:
    print ("This is a leap year!")
# 两个条件都满足才执行
elif year % 4 == 0 and year % 100 != 0:
    print ("This is a leap year!")
else:
    print ("This is not a leap year.")
````

### <span class="motutor-highlight motutor-id_tzw5lp8-id_8zkhnhg"><i></i>1.2 值的测试</span>

**Python**不仅仅可以使用布尔型变量作为条件，它可以直接在`if`中使用任何表达式作为条件：

大部分表达式的值都会被当作`True`，但以下表达式值会被当作`False`：

- False
- None
- 0
- 空字符串，空列表，空字典，空集合

````python
mylist = [3, 1, 4, 1, 5, 9]
if mylist:
    print ("The first element is:", mylist[0])
else:
    print ("There is no first element.")
````

修改为空列表：

````python
mylist = []
if mylist:
    print ("The first element is:", mylist[0])
else:
    print ("There is no first element.")
````

## 2. <span class="motutor-highlight motutor-id_e4z1qqp-id_minnwg0"><i></i>循环</span>

循环的作用在于将一段代码重复执行多次。

<img src="https://imgbed.momodel.cn/20201009155402.png" width=600/>

### 2.1 while 循环

    while <condition>:
        <statesments>

**Python**会循环执行`<statesments>`，直到`<condition>`不满足为止。

例如，计算数字`0`到`1000000`的和：

````python
i = 0
total = 0
while i < 1000000:
    total += i
    i += 1
print (total)
````

之前提到，空容器会被当成 `False` ，因此可以用 `while` 循环来读取容器中的所有元素：

````python
plays = set(['Hamlet', 'Macbeth', 'King Lear'])
while plays:
    play = plays.pop()
    print ('Perform', play)
````

循环每次从 `plays` 中弹出一个元素，一直到 `plays` 为空为止。

### 2.2 <span class="motutor-highlight motutor-id_4v8e83g-id_sgrtkkv"><i></i>for 循环</span>

    for <variable> in <sequence>:
        <indented block of code>

`for` 循环会遍历完`<sequence>`中所有元素为止

上一个例子可以改写成如下形式：

````python
plays = set(['Hamlet', 'Macbeth', 'King Lear'])
for play in plays:
    print ('Perform', play)
````

使用 `for` 循环时，注意尽量不要改变 `plays` 的值，否则可能会产生意想不到的结果。

之前的求和也可以通过 `for` 循环来实现：

````python
total = 0
for i in range(100000):
    total += i
print (total)
````

运行时间：

````python
%timeit for i in range(1000000): i = i
````

### 2.3 <span class="motutor-highlight motutor-id_03y1gqg-id_n4j0spl"><i></i>continue 语句</span>

遇到 `continue` 的时候，程序会返回到循环的最开始重新执行。

例如在循环中忽略一些特定的值：

````python
values = [7, 6, 4, 7, 19, 2, 1]
for i in values:
    if i % 2 != 0:
        # 忽略奇数
        continue
    print (i/2)
````

### 2.4 break 语句

遇到 `break` 的时候，程序会跳出循环，不管循环条件是不是满足：

````python
command_list = ['start', 
                'process', 
                'process',
                'process', 
                'stop', 
                'start', 
                'process', 
                'stop']
while command_list:
    command = command_list.pop(0)
    if command == 'stop':
        break
    print(command)
````

在遇到第一个 `'stop'` 之后，程序跳出循环。

### 2.5 e<span class="motutor-highlight motutor-id_pppx19u-id_kz2uxhz"><i></i>lse语句</span>

与 `if` 一样， `while` 和 `for` 循环后面也可以跟着 `else` 语句，不过要和`break`一起连用。

- 当循环正常结束时，循环条件不满足， `else` 被执行；
- 当循环被 `break` 结束时，循环条件仍然满足， `else` 不执行。

不执行：

````python
values = [7, 6, 4, 7, 19, 2, 1]
for x in values:
    if x <= 10:
        print ('Found:', x)
        break
else:
    print ('All values greater than 10')
````

执行：

````python
values = [11, 12, 13, 100]
for x in values:
    if x <= 10:
        print ('Found:', x)
        break
else:
    print ('All values greater than 10')
````

### 3.2 讲解实操

在日常生活中，我们经常需要对数据进行分类，例如：

- 温度分类（低于 0°C 为冰点，0-30°C 为正常温度，高于 30°C 为高温）。

- 考试成绩分类（小于 60 为不及格，60-80 为良好，80 以上为优秀）。

本次实训，我们将通过编程实现一个简单的数字分类程序：

    1. 让用户输入一组数字，并存入列表。

    2. 遍历列表中的数字，判断其大小并分类。

    3. 如果数字大于 50，打印 “较大数值”。

    4. 如果数字小于等于 50，打印 “较小数值”。

    5. 使用元组存储（数字, 判断结果），并输出分类结果。

### 3.3 实现步骤

**第一步：获取用户输入**

- 使用 input() 函数获取用户输入的多个数字。

- 使用 split() 方法拆分输入字符串，并转换为整数列表。

代码示例：

```python
numbers = input("请输入一组数字（用空格分隔）：").split()
numbers = [int(num) for num in numbers]  # 将输入转换为整数列表
```

**第二步：进行数值判断**

- 遍历 `numbers` 列表，使用 `if-else` 语句进行判断。

根据数值大小，将结果存入元组，并加入 `results` 列表。

代码示例：
```python
results = []  # 存储分类结果

for num in numbers:
    if num > 50:
        results.append((num, "较大数值"))
    else:
        results.append((num, "较小数值"))
```

**第三步：输出分类结果**

遍历 `results` 列表，使用 `print()` 逐行输出结果。

代码示例：

```python
print("\n分类结果如下：")
for num, label in results:
    print(f"数字 {num} 被分类为：{label}")
```
