- **`random` 模块：**

    - 导入 `random` 模块：`import random` 生成随机整数;
    - `random.randint(a, b)`，生成 a 到 b 之间的随机整数（包含 a 和 b）。

```python
# 导入 random 模块
import random

random_num = random.randint(1, 10) # 生成 1 到 10 之间的随机整数（包含 1 和 10）
print(random_num)

random_float = random.random() #生成0到1之间的随机浮点数
print(random_float)
```

- **`while` 循环：**

    - 基本语法：`while` 执行条件语句:
    - 循环条件为真时，重复执行循环体内的语句。

```python
i = 0
while i < 10:
    i += 1
    if i % 2 == 0: #如果是偶数
        continue #跳过本次循环，执行下一次循环
    if i > 5: #如果i大于5
        break #退出循环
    print(i) #输出1,3,5
```

- **条件判断：**

    - `if`、`elif`、`else` 语句用于根据条件执行不同的代码块。

```python
score = 85

if score >= 90:
    print("优秀")
elif score >= 80:
    print("良好") #输出 良好
elif score >= 60:
    print("及格")
else:
    print("不及格")
```

    
- **错误处理 (`try-except`)：**

    - 工作流程：
    
        1. `try` 块： 首先执行 `try` 块中的代码。
        2. 异常发生： 如果 `try` 块中的代码引发了异常，`Python` 会立即停止执行 `try` 块中的剩余代码，并查找与该异常类型匹配的 `except` 块。
        3. `except` 块： 如果找到了匹配的 `except` 块，则执行该 `except` 块中的代码。如果没有找到匹配的 `except` 块，则异常会向上层传递，如果没有更上层的 `try-except` 块处理该异常，程序将崩溃。
        
    - 常见的异常类型：
    
        - `ValueError`: 值错误，例如将字符串转换为整数时，字符串不是有效的数字。
        - `TypeError`: 类型错误，例如对不同类型的对象进行非法操作。
        - `IndexError`: 索引错误，例如访问列表或元组中不存在的索引。
        - `FileNotFoundError`: 文件未找到错误，例如尝试打开不存在的文件。
        - `ZeroDivisionError`: 除零错误，例如除数为零。

```python
try:
    numerator = int(input("请输入分子："))
    denominator = int(input("请输入分母："))
    result = numerator / denominator
    print(f"结果是：{result}")
except ValueError:
    print("输入无效，请输入整数！")
except ZeroDivisionError:
    print("除数不能为零！")
```

## 4. <span class="motutor-highlight motutor-id_jf5g0m5-id_nyaz644"><i></i>猜谜游戏基础版</span>

- 实现猜谜游戏核心逻辑；
- 记录猜谜游戏完成时次数；

```python
#导入模块
import random
# 定义变量
secret_number = random.randint(1, 100) #随机数
guess = 0 #猜测数
attempts = 0 #猜测次数
#游戏提示
print("欢迎来到猜数字游戏！")
#猜测数与随机数不相等，循环执行下面代码
while guess != secret_number:
    #处理用户输入非数字的情况
    try:
        #将输入字符串转为数值类型
        guess = int(input("请输入你猜测的数字："))
        #每循环一次，猜测数累加 1
        attempts += 1
        #判断猜测数与随机数大小，并友好提示
        if guess > secret_number:
            print("猜大了！")
        elif guess < secret_number:
            print("猜小了！")
        else:
            print(f"恭喜你，猜对了！你用了 {attempts} 次猜中。")
    #捕获错误
    except ValueError:
        print("输入无效，请输入一个整数。")
```

## 5. 猜谜游戏可<span class="motutor-highlight motutor-id_eg3eisu-id_6qmy8tt"><i></i>扩展方向</span>

在基础版本上，可以发挥想法完善整个程序，例如添加以下功能：

- 将猜谜游戏的核心逻辑封装成函数（代码复用、模块化、提高可读性）；
- 设置游戏难度（根据数字范围和猜测次数设置不同难度）；
- 记录游戏完成时间；

**参考代码**

```python
import random
import time

def generate_secret_number(min_num, max_num):
    """生成指定范围内的随机数"""
    return random.randint(min_num, max_num)

def get_user_guess(guesses_left):
    """获取用户输入，并进行异常处理"""
    while True:
        try:
            user_guess = int(input(f"请猜一个数字（你还有{guesses_left}次机会）："))
            return user_guess
        except ValueError:
            print("输入无效，请输入整数！")
            continue

def check_guess(secret_number, user_guess):
    """检查用户猜测是否正确，并给出提示"""
    if user_guess == secret_number:
        print("恭喜你，猜对了！")
        return True
    elif user_guess < secret_number:
        print("猜小了！")
    else:
        print("猜大了！")
    return False

def play_game(min_num, max_num, max_guesses):
    """进行一轮游戏"""
    secret_number = generate_secret_number(min_num, max_num)
    guesses_left = max_guesses
    start_time = time.time() #记录开始时间
    guess = 0
    while guesses_left > 0:
        user_guess = get_user_guess(guesses_left)
        guess += 1
        guesses_left -= 1
        if check_guess(secret_number, user_guess):
            end_time = time.time()
            print(f"你用了{guess}次猜中，耗时{end_time - start_time:.2f}秒")
    else:
        print(f"你没有猜中，正确答案是 {secret_number}。")

# 配置游戏参数
min_num = 1
max_num = 100
max_guesses = 7
# 开始游戏
play_game(min_num,max_num,max_guesses)
```
