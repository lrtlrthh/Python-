### 1.1 定义函数

函数`function`，通常接收输入参数，并有返回值。

它负责完成某项特定任务，而且相较于其他代码，具备相对的独立性。

````python
def add(x, y):
    """Add two numbers"""
    a = x + y
    return a
````

函数通常有一下几个特征：
- 使用 `def` 关键词来定义一个函数。
-  `def` 后面是函数的名称，括号中是函数的参数，不同的参数用 `,` 隔开， `def foo():` 的形式是必须要有的，参数可以为空；
- 使用缩进来划分函数的内容；
-  `docstring` 用 `"""` 包含的字符串，用来解释函数的用途，可省略；
-  `return` 返回特定的值，如果省略，返回 `None` 。

### 1.2 使用函数

使用函数时，只需要将参数换成特定的值传给函数。

**Python** 并没有限定参数的类型，因此可以使用不同的参数类型：

````python
print(add(2, 3))
print(add('foo', 'bar'))
````

在这个例子中，如果传入的两个参数不可以相加，那么 **Python** 会将报错：

````python
print(add(2, "foo"))
````

如果传入的参数数目与实际不符合，也会报错：

````python
print(add(1, 2, 3))
````

传入参数时，**Python** 提供了两种选项，第一种是上面使用的按照位置传入参数，另一种则是使用关键词模式，显式<span class="motutor-highlight motutor-id_fs7p9et-id_4piln4u"><i></i>地指定参数的值：</span>

````python
print(add(x=2, y=3))
print(add(y="foo", x="bar"))
````

可以混合这两种模式：

````python
print(add(2, y=3))
````

### 1.3 <span class="motutor-highlight motutor-id_5mexvg3-id_4od2ssb"><i></i>设定参数默认值</span>

可以在函数定义的时候给参数设定默认值，例如：

````python
def quad(x, a=1, b=0, c=0):
    return a*x**2 + b*x + c
````

可以省略有默认值的参数：

````python
print(quad(2.0))
````

可以修改参数的默认值：

````python
print(quad(2.0, b=3))
````

````python
print(quad(2.0, 2, c=4))
````

这里混合了位置和指定两种参数传入方式，第二个 2 是传给 `a` 的。

注意，在使用混合语法时，要注意不能给同一个值赋值多次，否则会报错，例如：

````python
print(quad(2.0, 2, a=2))
````

### 1.4 <span class="motutor-highlight motutor-id_88t5kp0-id_950ni3l"><i></i>接收不定长参数</span>

使用如下方法，可以使函数接受不定数目的参数：

````python
def add(x, *args):
    total = x
    for arg in args:
        total += arg
    return total
````

这里，`*args` 表示参数数目不定，可以看成一个元组，把第一个参数后面的参数当作元组中的元素。

````python
print(add(1, 2, 3, 4))
print(add(1, 2))
````

这样定义的函数不能使用关键词传入参数，要使用关键词，可以这样：

````python
def add(x, **kwargs):
    total = x
    for arg, value in kwargs.items():
        print("adding %s=%s"%(arg,value))
        total += value
    return total
````

这里 `**kwargs` 表示参数数目不定，相当于一个字典，键和值对应于键值对。

````python
print(add(10, y=11, z=12, w=13))
````

再看这个例子，可以接收任意数目的位置参数和键值对参数：

````python
def foo(*args, **kwargs):
    print(args, kwargs)

foo(2, 3, x='bar', z=10)
````

不过要按顺序传入参数，先传入位置参数 `args` ，再传入关键词参数 `kwargs` 。

### 1.5 <span class="motutor-highlight motutor-id_5d2uj0r-id_d75v2sx"><i></i>返回多个值</span>

函数可以返回多个值：

````python
def divid(a, b):
    """
    除法
    :param a: number 被除数
    :param b: number 除数
    :return: 商和余数
    """
    quotient = a // b
    remainder = a % b
    return quotient, remainder

quotient, remainder = divid(7,4)
print(quotient, remainder)
````

事实上，**Python** 将返回的两个值变成了元组：

````python
print(divid(7,4))
````

因为这个元组中有两个值，所以可以使用

    quotient, remainder = divid(7,4)

给两个值赋值。

列表也有相似的功能：

````python
a, b, c = [1, 2, 3]
print(a, b, c)
````

事实上，不仅仅返回值可以用元组表示，也可以将参数用元组以这种方式传入：

````python
def divid(a, b):
    """
    除法
    :param a: number 被除数
    :param b: number 除数
    :return: 商和余数
    """
    quotient = a // b
    remainder = a % b
    return quotient, remainder

z = (7,4)
print(divid(*z))
````

这里的`*`必不可少。
因为 `divid` 函数声明了两个**独立的位置参数** `a` 和 `b`：

```python
def divid(a, b): ...
```

而 `z` 是一个**整体**的元组 `(7, 4)`。

- 如果写成 `divid(z)`，相当于把**整个元组**当成第一个参数 `a`，第二个参数 `b` 缺失，会报错。  
- 写成 `divid(*z)` 时，`*` 会把元组**拆成两个独立的位置参数** `7` 和 `4`，分别对应 `a` 和 `b`，因此调用才正确。

`*` 负责“拆包”，让元组的元素逐一对应到形参。


事实上，还可以通过字典传入参数来执行函数：

````python
def divid(a, b):
    """
    除法
    :param a: number 被除数
    :param b: number 除数
    :return: 商和余数
    """
    quotient = a // b
    remainder = a % b
    return quotient, remainder

z = {'a':7,'b':4}
print(divid(**z))
````

### 1.6 <span class="motutor-highlight motutor-id_s1ihgjw-id_3251mvr"><i></i> `map` 方法生成序列</span>

其用法为：
    
    map(aFun, aSeq)

将函数 `aFun` 应用到序列 `aSeq` 上的每一个元素上，返回一个列表，不管这个序列原来是什么类型。

事实上，根据函数参数的多少，`map` 可以接受多组序列，将其对应的元素作为参数传入函数：

````python
def divid(a, b):
    """
    除法
    :param a: number 被除数
    :param b: number 除数
    :return: 商和余数
    """
    quotient = a // b
    remainder = a % b
    return quotient, remainder

a = (10, 6, 7)
b = [2, 5, 3]
print(list(map(divid,a,b)))
````

## 2.<span class="motutor-highlight motutor-id_gyyv0f2-id_wewen1h"><i></i> 模块</span>

Python 模块 是一个 .py 文件，包含 变量、函数、类，可以被导入复用。

**Python** 会将所有 `.py` 结尾的文件认定为 **Python** 代码文件，考虑下面的脚本 `ex1.py` ：

````python
%%writefile ex1.py

PI = 3.1416

def sum(lst):
    """
    计算 lst 序列所有元素的和
    :param lst: 序列 e.g. [1,2,3]
    :return: lst 序列所有元素的总和
    """
    
    # 获取 lst序列第一个元素
    tot = lst[0]
    
    # 循环遍历 lst 序列剩余元素
    for value in lst[1:]:
        tot = tot + value
    return tot

w = [0, 1, 2, 3]
print(sum(w), PI)
````

可以执行它：

````python
%run ex1.py
````

这个脚本可以当作一个模块，可以使用`import`关键词加载并执行它（这里要求`ex1.py`在当前工作目录）：

````python
import ex1

ex1
````

在导入时，**Python** 会执行一遍模块中的所有内容。

`ex1.py` 中所有的变量都被载入了当前环境中，不过要使用`ex1.变量名`
的方法来查看或者修改这些变量：

````python
print(ex1.PI)
````

````python
ex1.PI = 3.141592653
print(ex1.PI)
````

还可以用

    ex1.函数名

调用模块里面的函数：

````python
print(ex1.sum([2, 3, 4]))
````

为了提高效率，**Python** 只会载入模块一次，已经载入的模块再次载入时，**Python** 并不会真正执行载入操作，哪怕模块的内容已经改变。

例如，这里重新导入 `ex1` 时，并不会执行 `ex1.py` 中的 `print` 语句：

````python
import ex1
````

Python 只会加载模块一次，即使模块的内容已经改变，再次导入也不会重新执行。如果需要重新加载模块，可以使用`reload`函数：

````python
from imp import reload
reload(ex1)
````

删除之前生成的文件：

````python
import os
os.remove('ex1.py')
````

### 2.1 <span class="motutor-highlight motutor-id_n02c0m3-id_x8iqie8"><i></i>`__name__` 属性</span>



`__name__` 是一个内置的模块属性，用于区分模块是被直接运行还是被导入。

- **当模块被直接运行时**，`__name__` 的值为 `'__main__'`。
- **当模块被导入时**，`__name__` 的值为模块的名称。

通过检查 `__name__` 的值，可以控制某些代码只在模块被直接运行时执行，而在被导入时不会执行。





我们写一个文件 ex2.py，内容如下：

````python
%%writefile ex2.py

PI = 3.1416

def sum(lst):
    """ Sum the values in a list
    """
    tot = 0
    for value in lst:
        tot = tot + value
    return tot

def add(x, y):
    " Add two values."
    a = x + y
    return a

def test():
    w = [0,1,2,3]
    assert(sum(w) == 6)
    print('test passed.')

if __name__ == '__main__':
    test()
````

当直接运行 ex2.py 时，__name__ 的值为 '__main__'，因此 test() 函数会被执行

````python
%run ex2.py
````

当将 ex2.py 作为模块导入时，__name__ 的值为模块的名称（ex2），因此 test() 函数不会被自动执行。

````python
import ex2
````

此时，虽然 test() 不会自动运行，但模块中的变量和函数仍然可以被访问：

````python
ex2.PI
````

为了方便使用模块中的内容，可以为模块设置一个别名：

````python
import ex2 as e2
e2.PI
````

## 3. 包

**包** 是一个包含多个模块的目录，用于组织和管理代码。包的主要目的是将相关的模块和子包组织在一起，形成一个逻辑上的单元，方便代码的管理和使用。











### 3.1 包的结构

假设我们有以下目录结构：

```
foo/
    __init__.py
    bar.py  # 定义了 func 函数
    baz.py  # 定义了 zap 函数
```

在这个结构中：
- `foo` 是一个包。
- `__init__.py` 是一个特殊的文件，用于标识 `foo` 是一个包。
- `bar.py` 和 `baz.py` 是包中的模块。



````python
#### `__init__.py` 的作用

`__init__.py` 文件可以为空，但它的存在明确地告诉 Python 解释器，一个目录应该被视为一个包。此外，`__init__.py` 文件可以包含初始化代码，例如：
- 定义包的初始化行为。
- 创建和初始化包级别的变量。
- 定义包的 `__all__` 列表，控制 `from package import *` 时导入的模块。

例如，`foo/__init__.py` 可以包含以下内容：

```python
# foo/__init__.py
PI = 3.1416

def init():
    print("Initializing foo package")

init()
```

在这种情况下，当导入 `foo` 包时，`__init__.py` 中的代码会被执行：

```python
import foo
```

输出：
```
Initializing foo package
````

### 3.2 导入包中的内容

可以使用以下方式导入包中的内容：

```python
from foo.bar import func
from foo.baz import zap
```

也可以直接导入整个包，然后通过点号访问模块和函数：

```python
import foo

foo.bar.func()
foo.baz.zap()
```


### 3.3 包的搜索路径

Python 会根据 `sys.path` 列表来查找包。`sys.path` 是一个包含目录路径的列表，Python 会依次在这些目录中查找模块和包。

可以通过以下方式查看当前的 `sys.path`：

```python
import sys
print(sys.path)
```

如果需要将自定义的目录添加到 `sys.path` 中，可以使用以下代码：

```python
import sys
sys.path.append('/path/to/directory')
```

### 3.4 子包

包可以嵌套，形成子包。例如：


- mypackage/
    - __init__.py
    - module1.py
    - subpackage/
        - __init__.py
        - module2.py


在这种情况下：
- `mypackage` 是一个包。
- `subpackage` 是 `mypackage` 的子包。

可以使用以下方式导入子包中的模块：

```python
from mypackage.subpackage import module2
```

或者：

```python
import mypackage.subpackage.module2
```


### 3.5 `__all__` 列表

在包的 `__init__.py` 文件中，可以通过定义 `__all__` 列表来控制 `from package import *` 时导入的模块。例如：

```python
# mypackage/__init__.py
__all__ = ['module1', 'subpackage']
```

这样，当使用 `from mypackage import *` 时，只会导入 `module1` 和 `subpackage`。

**创建 operations.py（计算模块）**

这个模块包含基本数学运算的函数。

```Python
# calculator/operations.py
def add(a, b):
```

**创建 main.py（主程序）**

这个文件用于接收用户输入，并调用 operations.py 进行计算。

```Python
# calculator/main.py
import operations  # 导入计算模块

def calculator():
    print("欢迎使用简易计算器！")
    num1 = float(input("请输入第一个数字: "))
    operator = input("请输入运算符 (+, -, *, /): ")
    num2 = float(input("请输入第二个数字: "))

    if operator == "+":
        result = operations.add(num1, num2)
    elif operator == "-":
        result = operations.subtract(num1, num2)
    elif operator == "*":
        result = operations.multiply(num1, num2)
    elif operator == "/":
        result = operations.divide(num1, num2)
    else:
        result = "无效的运算符！"

    print(f"计算结果: {result}")

if __name__ == "__main__":
    calculator()
```
