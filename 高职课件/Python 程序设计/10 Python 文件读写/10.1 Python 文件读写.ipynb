## 1. 文件读写

### 1.1 读文件

使用 `open` 函数来读文件，使用文件名的字符串作为输入参数：


```python
f = open('test.txt')
```

默认以读的方式打开文件，如果文件不存在会报错：

```python
f1 = open('test1.txt')
```

可以使用 `read` 方法来读入文件中的所有内容：

```python
text = f.read()
print(text)
```

也可以按照行读入内容，`readlines` 方法返回一个列表，每个元素代表文件中每一行的内容：

```python
f = open('test.txt')
lines = f.readlines()
print(lines)
```

使用完文件之后，需要将文件关闭。

```python
f.close()
```

事实上，我们可以将 `f` 放在一个循环中，得到它每一行的内容：

```python
f = open('test.txt')
for line in f:
    print(line)
f.close()
```

删除刚才创建的文件：

```python
import os
os.remove('test.txt')
```

### 1.2 写文件

我们使用 `open` 函数的写入模式来写文件：

```python
f = open('myfile.txt', 'w')
f.write('hello world!')
f.close()
```

使用 `w` 模式时，如果文件不存在会被创建，我们可以查看是否真的写入成功：

```python
print(open('myfile.txt').read())
```

如果文件已经存在， `w` 模式会覆盖之前写的所有内容：

```python
f = open('myfile.txt', 'w')
f.write('another hello world!')
f.close()
print(open('myfile.txt').read())
```

除了写入模式，还有追加模式 `a` ，追加模式不会覆盖之前已经写入的内容，而是在之后继续写入：

```python
f = open('myfile.txt', 'a')
f.write('... and more')
f.close()
print(open('myfile.txt').read())
```

写入结束之后一定要将文件关闭，否则可能出现内容没有完全写入文件中的情况。

还可以使用读写模式 `w+`：

```python
f = open('myfile.txt', 'w+')
f.write('hello world!')
f.seek(6)
print(f.read())
f.close()
```

这里 `f.seek(6)` 移动到文件的第6个字符处，然后 `f.read()` 读出剩下的内容。

删除刚才创建的文件：

```python
import os
os.remove('myfile.txt')
```

### 1.3 关闭文件

在 Python 中，文件操作是一个常见的任务，但如果不正确地关闭文件，可能会导致数据丢失或文件损坏。因此，正确关闭文件非常重要。

**自动关闭文件**

在 Python 中，如果一个打开的文件不再被任何变量引用，Python 会自动关闭这个文件。然而，依赖自动关闭文件并不是一个好的实践，因为它不能保证文件的内容已经完全写入磁盘。

**手动关闭文件**

为了确保文件内容被正确写入磁盘，应该显式地调用文件的 close() 方法。如果不关闭文件，可能会出现意想不到的结果。

示例 1：未关闭文件

```python
f = open('newfile.txt', 'w')
f.write('hello world')

g = open('newfile.txt', 'r')
print(repr(g.read()))  # 输出为空，因为文件未关闭，内容未写入磁盘
g.close()
```

虽然这里我们往文件中写了内容，但是在关闭之前，这个内容并没有被写入磁盘。

示例 2：循环写入未关闭文件

```python
import os
os.remove('newfile.txt')

f = open('newfile.txt','w')
for i in range(30):
    f.write('hello world: ' + str(i) + '\n')
    

g = open('newfile.txt', 'r')
print(repr(g.read()))
f.close()
g.close()
```

循环写入的内容一样也没有被写入磁盘。

```python
import os
os.remove('newfile.txt')
```

如果在文件操作过程中出现异常，文件可能不会被正确关闭。为了避免这种情况，可以使用 try/except/finally 块来确保文件在异常发生时也能被正确关闭。

```python
f = open('newfile.txt', 'w')
try:
    for i in range(30):
        x = 1.0 / (i - 10)  # 这里会引发 ZeroDivisionError
        f.write('hello world: ' + str(i) + '\n')
except ZeroDivisionError:
    print("something bad happened")
finally:
    f.close()  # 确保文件被关闭
```

运行上述代码后，即使发生异常，文件也会被正确关闭。可以验证文件内容是否被写入：

```python
g = open('newfile.txt', 'r')
print(g.read())  # 输出已写入的内容
g.close()
```

**使用 with 语句**

为了避免手动管理文件的打开和关闭，Python 提供了`with`语句，它会自动管理文件的上下文，确保文件在退出上下文时被正确关闭。

```python
with open('newfile.txt', 'w') as f:
    for i in range(30):
        f.write('hello world: ' + str(i) + '\n')

# 文件自动关闭，无需手动调用 close()
with open('newfile.txt', 'r') as g:
    print(g.read())  # 输出已写入的内容
```

所以，写文件时候要确保文件被正确关闭。

最后我们删除刚才创建的文件：

```python
import os
os.remove('newfile.txt')
```

## 2. CSV 文件和 csv 模块

CSV（Comma-Separated Values）文件是一种常见的文本文件格式，用于存储表格数据，如电子表格或数据库。Python 的标准库中提供了 csv 模块，用于读取和写入 CSV 文件。

### 2.1 读 csv 文件

假设我们有这样的一个文件：

```python
%%file data.csv
"alpha 1",  100, -1.443
"beat  3",   12, -0.0934
"gamma 3a", 192, -0.6621
"delta 2a",  15, -4.515
```

打开这个文件，并产生一个文件 reader：

```python
import csv

# 打开 data.csv 文件
with open("data.csv", newline='') as fp:
    # 创建一个 CSV 文件读取器
    reader = csv.reader(fp)
    # 按行迭代数据
    for row in reader:
        print(row)
```

默认情况下，csv.reader 将每一行的数据作为字符串列表返回。如果需要对数据进行进一步处理，可以手动转换数据类型：

```python
data = []

with open('data.csv', newline='') as fp:
    reader = csv.reader(fp)
    for row in reader:
        data.append([row[0], int(row[1]), float(row[2])])

print(data)
```

### 2.2 写 csv 文件

可以使用 csv.writer 来写入 CSV 文件。通常建议使用`w`模式打开文件，而不是`wb`，因为 Python 3 中的 csv 模块已经支持文本模式写入。

```python
import csv

# 准备要写入的数据
data = [('one', 1, 1.5), ('two', 2, 8.0)]

# 打开文件并写入数据
with open('out.csv', 'w', newline='') as fp:
    writer = csv.writer(fp)
    writer.writerows(data)
```

显示结果：

```python
data = []

with open('out.csv') as fp:
    r = csv.reader(fp)
    for row in r:
        data.append((row[0], int(row[1]), float(row[2])))

data
```

### 2.3 读取写入的文件
可以读取刚刚写入的文件，验证内容是否正确：

```python
data = []

with open('out.csv', newline='') as fp:
    reader = csv.reader(fp)
    for row in reader:
        data.append((row[0], int(row[1]), float(row[2])))

print(data)
```

### 2.4 处理 CSV 文件中的标题

CSV 文件通常包含标题行，可以通过 csv.DictReader 和 csv.DictWriter 来处理这些标题。
示例：读取带标题的 CSV 文件
我们设计以下带标题的 CSV 文件 data_with_headers.csv：

```python
%%file data_with_headers.csv

name,value1,value2
"alpha 1",100,-1.443
"beta  3",12,-0.0934
"gamma 3a",192,-0.6621
"delta 2a",15,-4.515
```

可以使用 csv.DictReader 来读取这个文件：

```python
import csv

data = []

with open('data_with_headers.csv', newline='') as fp:
    reader = csv.DictReader(fp)
    for row in reader:
        data.append(row)

print(data)
```

示例：写入带标题的 CSV 文件

可以使用 csv.DictWriter 来写入带标题的 CSV 文件：

```python
import csv

# 准备要写入的数据
data = [
    {'name': 'one', 'value1': 1, 'value2': 1.5},
    {'name': 'two', 'value1': 2, 'value2': 8.0}
]

# 打开文件并写入数据
with open('out_with_headers.csv', 'w', newline='') as fp:
    fieldnames = ['name', 'value1', 'value2']
    writer = csv.DictWriter(fp, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(data)
```

完成操作后，可以删除创建的文件以清理环境：

```python
import os

os.remove('data.csv')
os.remove('out.csv')
os.remove('data_with_headers.csv')
os.remove('out_with_headers.csv')
```
