## 2. 知识回顾

**列表和元组**

列表是一种可变序列，使用方括号 `[ ]` 定义，元素之间用逗号分隔。可以对列表进行增删改查等操作。

例如：

````python
# 创建列表
scores = [90, 85, 78, 92, 88]
# 访问元素
print(scores[0])  # 输出 90
# 修改元素
scores[2] = 80
# 添加元素
scores.append(75)
# 删除元素
scores.remove(80)
````

元组是一种不可变序列，使用圆括号 `( )` 定义。一旦创建，不能修改其元素。

例如：

````python
# 创建元组
subjects = ("语文", "数学", "英语")
# 访问元素
print(subjects[1])  # 输出 数学
````

**字典和集合**

字典是一种键值对的集合，使用花括号 `{}` 定义。通过键来访问对应的值。

例如：

````python
# 创建字典
student_scores = {"张三": 90, "李四": 85, "王五": 82}
# 访问元素
print(student_scores["张三"])  # 输出 90
# 添加元素
student_scores["赵六"] = 78
# 修改元素
student_scores["李四"] = 88
# 删除元素
del student_scores["王五"]
````

集合是一个无序不重复元素的集合，使用花括号 `{}` 或 `set()` 函数创建。集合支持集合运算，如交集、并集、差集等。

例如：

````python
# 创建集合
set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}
# 交集
print(set1 & set2)  # 输出 {3, 4}
# 并集
print(set1 | set2)  # 输出 {1, 2, 3, 4, 5, 6}
# 差集
print(set1 - set2)  # 输出 {1, 2}
````

**`if-elif-else` 语句**


用于根据条件执行不同的代码块。例如：

````python
score = 85
if score >= 90:
    print("优秀")
elif score >= 80:
    print("良好")
elif score >= 60:
    print("及格")
else:
    print("不及格")
````

**for 循环和 while 循环**

**for** 循环用于遍历可迭代对象，如列表、元组、字典、集合、字符串等。

例如：

````python
# 遍历列表
for score in [90, 85, 78, 92, 88]:
    print(score)
````

**while** 循环需要设定循环条件，当条件为真时，持续执行循环体。

例如：

````python
i = 0
while i < 5:
    print(i)
    i += 1
````

**Python 函数**

函数是组织好的、可重复使用的代码块，用于实现单一功能。使用 `def` 关键字定义函数。

例如：

````python
# 定义函数
def add(a, b):
    return a + b

# 调用函数
result = add(3, 5)
print(result)  # 输出 8
````

**模块和包**

模块是一个包含 `Python` 定义和语句的文件，以 `.py` 为后缀名。可以使用 `import` 关键字导入模块。  
例如：

````python
# 导入 math 模块
import math

# 使用模块中的函数
print(math.sqrt(25))  # 输出 5.0
````

包是一种包含多个模块的文件夹，用于组织模块。例如：
```python
# 导入包中的模块
import mypackage.module1

# 使用包中模块的函数
mypackage.module1.my_function()

```

## 4. 学生成绩管理系统基础版

实现学生成绩管理系统的核心功能，包括添加成绩、删除成绩、查询成绩、修改成绩和显示所有成绩。

````python
# 学生成绩管理系统基础版

# 初始化学生成绩字典
student_scores = {}

# 显示系统菜单
print("欢迎使用学生成绩管理系统！")
print("1. 添加成绩")
print("2. 删除成绩")
print("3. 查询成绩")
print("4. 修改成绩")
print("5. 显示所有成绩")
print("6. 退出系统")

# 获取用户选择
while True:
    try:
        choice = int(input("请输入功能选项（1-6）："))
        if choice == 1:
            # 添加成绩
            name = input("请输入学生姓名：")
            score = int(input("请输入学生成绩："))
            student_scores[name] = score
            print(f"成功添加 {name} 的成绩！")
        elif choice == 2:
            # 删除成绩
            name = input("请输入要删除成绩的学生姓名：")
            if name in student_scores:
                del student_scores[name]
                print(f"成功删除 {name} 的成绩！")
            else:
                print("未找到该学生！")
        elif choice == 3:
            # 查询成绩
            name = input("请输入要查询成绩的学生姓名：")
            if name in student_scores:
                print(f"{name} 的成绩是：{student_scores[name]}")
            else:
                print("未找到该学生！")
        elif choice == 4:
            # 修改成绩
            name = input("请输入要修改成绩的学生姓名：")
            if name in student_scores:
                new_score = int(input("请输入新的成绩："))
                student_scores[name] = new_score
                print(f"成功修改 {name} 的成绩！")
            else:
                print("未找到该学生！")
        elif choice == 5:
            # 显示所有成绩
            print("所有学生成绩：")
            for name, score in student_scores.items():
                print(f"{name}: {score}")
        elif choice == 6:
            # 退出系统
            print("感谢使用学生成绩管理系统，再见！")
            break
        else:
            print("无效的选项，请重新输入！")
    except ValueError:
        print("输入无效，请输入一个整数选项！")
````

## 5. 学生成绩管理系统可扩展方向

在基础版本上，可以发挥想法完善整个程序，例如添加以下功能：
- 将学生成绩管理系统的功能封装成函数，使用模块化设计
- 增加数据存储和读取功能
- 记录系统操作日志等。

**<span class="motutor-highlight motutor-id_5wep8un-id_dt22epi"><i></i>参考代码**</span>

````python
# 导入模块
import json
import os
from datetime import datetime

# 定义函数：初始化系统
def initialize_system():
    """初始化学生成绩管理系统，创建或加载学生成绩数据"""
    global student_scores
    global operation_log
    if os.path.exists("student_scores.json"):
        with open("student_scores.json", "r") as file:
            student_scores = json.load(file)
    else:
        student_scores = {}
    operation_log = []

# 定义函数：保存数据
def save_data():
    """将学生成绩数据保存到文件"""
    with open("student_scores.json", "w") as file:
        json.dump(student_scores, file)

# 定义函数：记录操作日志
def log_operation(operation, name="", score=""):
    """记录系统操作日志"""
    timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    log_entry = {
        "timestamp": timestamp,
        "operation": operation,
        "name": name,
        "score": score
    }
    operation_log.append(log_entry)

# 定义函数：显示系统菜单
def display_menu():
    """显示学生成绩管理系统的菜单"""
    print("\n欢迎使用学生成绩管理系统！")
    print("1. 添加成绩")
    print("2. 删除成绩")
    print("3. 查询成绩")
    print("4. 修改成绩")
    print("5. 显示所有成绩")
    print("6. 查看操作日志")
    print("7. 退出系统")

# 定义函数：添加成绩
def add_score():
    """添加学生成绩"""
    name = input("请输入学生姓名：")
    try:
        score = int(input("请输入学生成绩："))
        student_scores[name] = score
        log_operation("添加成绩", name, str(score))
        print(f"成功添加 {name} 的成绩！")
        save_data()
    except ValueError:
        print("输入无效，请输入一个整数成绩！")

# 定义函数：删除成绩
def delete_score():
    """删除学生成绩"""
    name = input("请输入要删除成绩的学生姓名：")
    if name in student_scores:
        log_operation("删除成绩", name, str(student_scores[name]))
        del student_scores[name]
        print(f"成功删除 {name} 的成绩！")
        save_data()
    else:
        print("未找到该学生！")

# 定义函数：查询成绩
def query_score():
    """查询学生成绩"""
    name = input("请输入要查询成绩的学生姓名：")
    if name in student_scores:
        print(f"{name} 的成绩是：{student_scores[name]}")
        log_operation("查询成绩", name, str(student_scores[name]))
    else:
        print("未找到该学生！")

# 定义函数：修改成绩
def modify_score():
    """修改学生成绩"""
    name = input("请输入要修改成绩的学生姓名：")
    if name in student_scores:
        try:
            new_score = int(input("请输入新的成绩："))
            original_score = student_scores[name]
            student_scores[name] = new_score
            log_operation("修改成绩", name, f"{original_score} -> {new_score}")
            print(f"成功修改 {name} 的成绩！")
            save_data()
        except ValueError:
            print("输入无效，请输入一个整数成绩！")
    else:
        print("未找到该学生！")

# 定义函数：显示所有成绩
def display_all_scores():
    """显示所有学生成绩"""
    if student_scores:
        print("\n所有学生成绩：")
        for name, score in student_scores.items():
            print(f"{name}: {score}")
    else:
        print("暂无学生成绩数据！")

# 定义函数：查看操作日志
def view_operation_log():
    """查看系统操作日志"""
    if operation_log:
        print("\n操作日志：")
        for log in operation_log:
            print(f"{log['timestamp']} - {log['operation']} - {log['name']} - {log['score']}")
    else:
        print("暂无操作日志！")

# 初始化系统
student_scores = {}
operation_log = []
initialize_system()

# 显示系统菜单
while True:
    display_menu()
    try:
        choice = int(input("请输入功能选项（1-7）："))
        if choice == 1:
            add_score()
        elif choice == 2:
            delete_score()
        elif choice == 3:
            query_score()
        elif choice == 4:
            modify_score()
        elif choice == 5:
            display_all_scores()
        elif choice == 6:
            view_operation_log()
        elif choice == 7:
            print("感谢使用学生成绩管理系统，再见！")
            break
        else:
            print("无效的选项，请重新输入！")
    except ValueError:
        print("输入无效，请输入一个整数选项！")
````
