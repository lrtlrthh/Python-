 **长度**：用 `len` 查看列表长度

### 1.3 <span class="motutor-highlight motutor-id_yqbcdvi-id_gb38qd5"><i></i>添加元素</span>

**append**：向列表添加单个元素  
`l.append(ob)` 将元素 `ob` 添加到列表 `l` 的最后。

````python
a = [10, 11, 12]
a.append(11)
print(a)
````

`append` 每次只添加一个元素，并不会因为这个元素是序列而将其展开：

````python
a = [10, 11, 12]
a.append(['a', 'b'])
print(a)
````

**extend**: 向列表添加序列元素    
`l.extend(lst)` 将序列 `lst` 的元素依次添加到列表 `l` 的最后，作用相当于 `l += lst`。

````python
a = [10, 11, 12]
a.extend(['a', 'b'])
print(a)
````

**insert**: 插入元素       
`l.insert(idx, ob)` 在索引 `idx` 处插入 `ob` ，之后的元素依次后移。

````python
a = [10, 11, 12, 13, 11]
# 在索引 3 插入 'a'
a.insert(3, 'a')
print(a)
````

### 1.4 <span class="motutor-highlight motutor-id_eiqny0y-id_tfz8nl6"><i></i>删除元素</span>

**del**：根据下标进行删除

````python
# 根据下标进行删除
a = [1002, 'a', 'b', 'c']
del a[0]
print(a)
````

**pop**：<span class="motutor-highlight motutor-id_qx4xffe-id_hnfkxjz"><i></i>弹出元素</span>    
`l.pop(idx)` 会将索引 `idx` 处的元素删除，并返回这个元素。未指定 `idx` 时，默认为列表最后一个元素。

````python
a = [1002, 'a', 'b', 'c']
a.pop()
print(a)
````

**remove**：<span class="motutor-highlight motutor-id_auueqje-id_bt0g5fb"><i></i>根据元素的值进行删除 </span>    
`l.remove(ob)` 会将列表中第一个出现的 `ob` 删除，如果 `ob` 不在 `l` 中会报错。

````python
a = [1002, 'a', 'b', 'c', 'b']
a.remove("b")
print(a)
````

### 1.5 测试<span class="motutor-highlight motutor-id_zizudpt-id_zc37jrx"><i></i>从属关系</span>
用 `in` 来看某个元素是否在某个序列（不仅仅是列表）中；

用`not in`来判断是否不在某个序列中。

````python
a = [10, 11, 12, 13, 11]
print(10 in a)
print(10 not in a)
````

用 `index` 查找某个元素在列表中的位置：

`l.index(ob)` 返回列表中元素 `ob` 第一次出现的索引位置，如果 `ob` 不在 `l` 中会报错。

````python
a = [10, 11, 12, 13, 11]
a.index(11)
````

`count` 查找列表中某个元素出现的次数：

````python
a = [10, 11, 12, 13, 11]
a.count(11)
````

修改元素的时候，要通过下标来确定要修改的是哪个元素，然后才能进行修改

### 1.7 <span class="motutor-highlight motutor-id_i9awir2-id_8vh29rt"><i></i>排序</span>

`sort`方法将 `list` 按特定顺序重新排列，默认为由小到大，参数 `reverse=True` 可改为倒序，由大到小

````python
# 从小到大排序
a = [10, 1, 11, 13, 11, 2]
a.sort()
print(a)
````

````python
# 从大到小排序
a = [10, 1, 11, 13, 11, 2]
a.sort(reverse=True)
print(a)
````

如果不想改变原来列表中的值，可以使用 `sorted` 函数：      
`sorted` 函数把任何可迭代对象里的元素按指定规则排好序，返回一个全新的有序列表，原对象纹丝不动。

````python
a = [10, 1, 11, 13, 11, 2]
b = sorted(a)
print("a:",a)
print("b:",b)
````

### 1.8 <span class="motutor-highlight motutor-id_3g69nma-id_2mvvwg8"><i></i>列表反向</span>

`l.reverse()` 会将列表中的元素从后向前排列。

````python
a = [10, 1, 11, 13, 11, 2]
a.reverse()
print(a)
````

如果不想改变原来列表中的值，可以使用分片：

````python
a = [10, 1, 11, 13, 11, 2]
b = a[::-1]
print("a:",a)
print("b:",b)
````

## 2. <span class="motutor-highlight motutor-id_sn4q5d0-id_5ood3rt"><i></i>可变和不可变类型</span>

### 2.1 列表是可变的

````python
a = [1,2,3,4]
a
````

通过索引改变：

````python
a = [1,2,3,4]
a[0] = 100
a
````

通过方法改变：

````python
a = [1,2,3,4]
a.insert(3, 200)
a
````

````python
a = [1,2,8,3,4]
a.sort()
a
````

### 2.2 字符串是不可变的

````python
s = "hello world"
s
````

通过索引改变会报错：

````python
s = "hello world"
s[0] = 'z'
````

字符串方法只是返回一个新字符串，并不改变原来的值：

````python
s = "hello world"
print (s.replace('world', 'Mars'))
print (s)
````

如果想改变字符串的值，可以用重新赋值的方法：

````python
s = "hello world"
s = s.replace('world', 'Mars')
print (s)
````

数据类型分类：

|可变数据类型|不可变数据类型|
|--|--|
|`list`, `dictionary`, `set`, `numpy array`, `user defined objects`|`integer`, `float`, `long`, `complex`, `string`, `tuple`, `frozenset`

### 2.3 字符串不可变的原因

其一，列表可以通过以下的方法改变，而字符串不支持这样的变化。

````python
a = [1, 2, 3, 4]
b = a
b[0] = 100
a
````

## 3. <span class="motutor-highlight motutor-id_2dsvk3s-id_3m2a8sm"><i></i>元组</span>

### 3.1 基本操作

与列表相似，元组`Tuple`也是个有序序列，但是元组是不可变的，用`()`生成。

````python
t = (10, 11, 12, 13, 14)
t
````

可以索引，切片：

````python
t = (10, 11, 12, 13, 14)
t[0]
t[1:3]
````

但是元组是不可变的：

````python
# 会报错
t = (10, 11, 12, 13, 14)
t[0] = 1
````

### 3.2 单个元素的元组生成

由于`()`在表达式中被应用，只含有单个元素的元组容易和表达式混淆，所以采用下列方式定义只有一个元素的元组：

````python
a = (10,)
print (a)
print (type(a))
````

````python
a = (10)
print (type(a))
````

将列表转换为元组：

````python
a = [10, 11, 12, 13, 14]
tuple(a)
````

### 3.3<span class="motutor-highlight motutor-id_n69v2md-id_5a91emi"><i></i> 元组方法</span>

由于元组是不可变的，所以只能有一些不可变的方法，例如计算元素个数 `count` 和元素位置 `index` ，用法与列表一样。

````python
a = [10, 11, 12, 13, 14]
a.count(10)
````

````python
a = [10, 11, 12, 13, 14]
a.index(12)
````

### 4.2 知识回顾

**圆周率（`π`）**

- 圆周率是一个数学常数，表示圆的周长与直径的比值。

- 常用值：**3.14159**。

- 在 `Python` 中，可以使用 `math` 模块中的 `math.pi` 获取更精确的值。

**圆的面积公式**

- 圆的面积公式为：`area = π x r²`

- 其中 r 是圆的半径。

**列表和元组的特点**

- 列表（`list`）

- 可变的数据结构，允许添加、删除和修改元素。

- 适合存储需要频繁修改的圆半径和对应的面积。

**元组（tuple）**

- 不可变的数据结构，适合存储固定不变的圆半径和面积对。

- 提高数据的安全性和稳定性。

### 4.3 实现步骤

**第一步：导入必要模块**

- 在 Python 中，使用 math 模块中的 math.pi 获取 π 的值。

代码示例：
```python
import math
```

**第二步：创建列表存储圆的半径**

- 通过代码定义一个包含多个半径值的列表。

代码示例：
```python
radius = [1, 2, 3, 4, 5]
```

**第三步：计算面积并存储为元组**

    - 使用 `for` 循环遍历列表，计算每个半径对应的圆面积，并将结果存储为元组。

代码示例：
```python
areas = []
for r in radius:
    area = math.pi * (r ** 2)
    areas.append((r, area))  # 将半径和面积作为元组添加到列表
```

**第四步：输出结果**

- 通过循环输出每个圆的半径和面积。

代码示例：

```python
for r, a in areas:
    print(f"半径为 {r} 的圆的面积是：{a:.2f}")
```
