### 1.4 基本操作示例

#### 1.4.1 创建一个新的 Excel 文件

```python
from openpyxl import Workbook

# 创建一个新的工作簿
wb = Workbook()

# 获取当前活跃的工作表
ws = wb.active

# 设置工作表标题
ws.title = "学生成绩表"

# 添加表头
ws.append(["姓名", "语文", "数学", "英语"])

# 添加数据
ws.append(["张三", 85, 90, 88])
ws.append(["李四", 78, 82, 80])
ws.append(["王五", 92, 88, 85])

# 保存文件
wb.save("学生成绩表.xlsx")
```

#### 1.4.2 读取 Excel 文件内容

```python
from openpyxl import load_workbook

# 加载已存在的 Excel 文件
wb = load_workbook("学生成绩表.xlsx")

# 获取指定工作表
ws = wb["学生成绩表"]

# 读取数据
for row in ws.iter_rows(values_only=True):
    print(row)
```

#### 1.4.3 修改单元格内容和样式

```python
from openpyxl.styles import Font, Alignment

# 修改单元格内容
ws["B2"] = 95

# 设置字体样式
ws["A1"].font = Font(bold=True, color="FF0000")

# 设置对齐方式
ws["A1"].alignment = Alignment(horizontal="center", vertical="center")

# 保存修改
wb.save("学生成绩表.xlsx")
```

#### 1.4.4 合并单元格

```python
# 合并单元格
ws.merge_cells("A5:C5")
ws["A5"] = "以上为成绩数据"

# 保存文件
wb.save("学生成绩表.xlsx")
```

#### 1.4.5 插入柱状图

```python
from openpyxl.chart import BarChart, Reference

# 创建柱状图
chart = BarChart()
chart.title = "学生成绩对比"
chart.y_axis.title = "分数"
chart.x_axis.title = "学生"

# 设置数据范围
data = Reference(ws, min_col=2, min_row=1, max_row=4, max_col=4)
categories = Reference(ws, min_col=1, min_row=2, max_row=4)

chart.add_data(data, titles_from_data=True)
chart.set_categories(categories)

# 添加图表到工作表
ws.add_chart(chart, "F2")

# 保存文件
wb.save("学生成绩表.xlsx")
```

### 1.5 批量生成成绩报告

假设我们需要为多个班级生成成绩报告，每个班级的数据不同，我们可以使用 `openpyxl` 批量生成多个 Excel 文件：

```python
import os
from openpyxl import Workbook

# 创建文件夹
os.makedirs("成绩报告", exist_ok=True)

# 模拟多个班级的数据
classes = {
    "一班": [("张三", 85, 90, 88), ("李四", 78, 82, 80)],
    "二班": [("王五", 92, 88, 85), ("赵六", 75, 80, 78)],
}

for class_name, students in classes.items():
    wb = Workbook()
    ws = wb.active
    ws.title = f"{class_name}成绩表"
    ws.append(["姓名", "语文", "数学", "英语"])
    for student in students:
        ws.append(student)
    wb.save(f"成绩报告/{class_name}成绩表.xlsx")

print("所有成绩报告已生成完成！")
```

- 每月各部门的绩效报表

- 各地区的销售数据汇总表

手动制作这些报表不仅耗时，而且容易出错。使用 Python 可以自动化完成这些任务，大大提高工作效率。

### 3.3 示例数据准备

假设我们有以下销售数据，需要为每个地区生成一份销售报表：

```python
import pandas as pd

# 示例数据
data = {
    "地区": ["华北", "华北", "华东", "华东", "华南", "华南"],
    "销售员": ["张三", "李四", "王五", "赵六", "孙七", "周八"],
    "销售额": [50000, 60000, 55000, 65000, 70000, 72000],
    "月份": ["1月", "2月", "1月", "2月", "1月", "2月"]
}

df = pd.DataFrame(data)
```

### 3.4 批量生成报表的实现步骤

#### 3.4.1 按地区分组

```python
# 按地区分组
grouped = df.groupby("地区")
```

#### 3.4.2 遍历每个地区，生成报表

```python
import os
from openpyxl import Workbook
from openpyxl.styles import Font, Alignment
from datetime import datetime

# 创建文件夹
os.makedirs("销售报表", exist_ok=True)

for region, group in grouped:
    wb = Workbook()
    ws = wb.active
    ws.title = f"{region}销售数据"

    # 添加标题
    ws["A1"] = f"{region}地区销售报表"
    ws["A1"].font = Font(size=16, bold=True)
    ws.merge_cells("A1:D1")
    ws["A1"].alignment = Alignment(horizontal="center")

    # 添加表头
    headers = ["销售员", "月份", "销售额", "备注"]
    ws.append(headers)
    for cell in ws[2]:
        cell.font = Font(bold=True)

    # 添加数据
    for _, row in group.iterrows():
        ws.append([row["销售员"], row["月份"], row["销售额"], ""])

    # 添加制表日期
    ws["D8"] = f"制表日期：{datetime.now().strftime('%Y-%m-%d')}"

    # 保存文件
    wb.save(f"销售报表/{region}销售报表.xlsx")

print("所有销售报表已生成完成！")
```

### 3.6 应用场景介绍
当需要为学校各年级生成期中成绩报表时，可通过该方法批量生成。每个年级的报表中都包含该年级所有班级的多科平均成绩数据，以及对应的柱状图（如语文平均成绩柱状图、数学平均成绩柱状图），教师能通过报表快速了解各年级各班的成绩情况，为教学分析和指导提供便利，且相比手动制作，大幅提高效率。

```python
import pandas as pd
import matplotlib.pyplot as plt
import os

# 示例数据
data = {
    '年级': [1, 1, 1, 2, 2, 2, 3, 3, 3],
    '班级': [1, 2, 3, 1, 2, 3, 1, 2, 3],
    '语文平均成绩': [85, 88, 90, 78, 82, 80, 92, 90, 95],
    '数学平均成绩': [90, 92, 88, 85, 80, 82, 95, 93, 94],
    '英语平均成绩': [88, 90, 85, 82, 80, 83, 90, 92, 95]
}

# 创建 DataFrame
df = pd.DataFrame(data)

# 定义图片存储路径
results_path = "/home/jovyan/work/results/png/"
os.makedirs(results_path, exist_ok=True)  # 确保路径存在

# 定义一个函数来生成每个年级的报表和柱状图
def generate_report(df):
    # 创建一个 Excel 写入器
    with pd.ExcelWriter('All_Grades_Report.xlsx') as writer:
        # 遍历每个年级生成报表
        for grade in df['年级'].unique():
            # 筛选出当前年级的数据
            grade_data = df[df['年级'] == grade]
            
            # 将数据写入 Excel
            grade_data.to_excel(writer, sheet_name=f'年级_{grade}_数据', index=False)
            
            # 生成柱状图并保存到指定路径
            for subject in ['语文平均成绩', '数学平均成绩', '英语平均成绩']:
                plt.figure(figsize=(10, 6))
                plt.bar(grade_data['班级'], grade_data[subject], color='skyblue')
                plt.xlabel('班级')
                plt.ylabel('平均成绩')
                plt.title(f'年级 {grade} {subject} 柱状图')
                plt.xticks(grade_data['班级'])
                
                # 保存图片到指定路径
                img_path = os.path.join(results_path, f'Grade_{grade}_{subject}.png')
                plt.savefig(img_path)
                plt.close()
                
                # 将图片插入到 Excel 中
                worksheet = writer.sheets[f'年级_{grade}_数据']
                worksheet.insert_image(f'A{grade_data.shape[0] + 3}', img_path)

# 生成报表
generate_report(df)

print("报表生成完成！")
```
