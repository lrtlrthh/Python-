# 基于 Numpy 的公司财务数据分析实训


## 1. 实训目标
通过使用 Numpy 库对某公司的财务数据进行分析，掌握 Numpy 在数据处理和分析中的基本应用，包括数据导入、数据清洗、基本统计分析等操作，培养利用 Python 进行数据分析的能力。

## 2. 实训数据
假设某公司 2022 - 2024 年的财务数据存储在一个名为 “financial_data.csv” 的文件中，数据包含以下字段：
- 年份（Year）：数据所属的年份
- 季度（Quarter）：数据所属的季度
- 营业收入（Revenue）：该季度的营业收入，单位为万元
- 营业成本（Cost）：该季度的营业成本，单位为万元
- 利润（Profit）：该季度的利润，单位为万元

## 3. 实训步骤

### 3.1 数据导入
1. 使用 Numpy 的 `genfromtxt` 函数导入数据




```python
import numpy as np

# 设置 NumPy 打印选项，避免科学计数法
np.set_printoptions(suppress=True, formatter={'float_kind': '{:0.0f}'.format})

# 导入数据
data = np.genfromtxt('./financial_data.csv', delimiter=',', skip_header=1)
print(data)
```

   说明：`genfromtxt` 函数可以方便地从文本文件中读取数据，`delimiter` 参数指定字段之间的分隔符，`skip_header` 参数用于跳过文件头部的标题行。


2. 查看数据的形状和数据类型

```python
print(data.shape)  # 查看数据的形状
print(data.dtype)  # 查看数据类型
```

### 3.2 数据清洗
1. 检查数据中是否存在缺失值



```python
print(np.isnan(data).any())  # 检查是否存在缺失值
```

如果存在缺失值，可以根据实际情况选择填充或删除缺失值。例如，用均值填充缺失值：


```python
mean_values = np.nanmean(data, axis=0)  # 计算每列的均值
data = np.where(np.isnan(data), mean_values, data)  # 用均值填充缺失值
```

2. 检查数据中是否存在异常值（如负数的营业收入或利润等）

```python
print(np.any(data[:, 2] < 0))  # 检查营业收入是否存在负数
print(np.any(data[:, 4] < 0))  # 检查利润是否存在负数
```

如果存在异常值，需要根据实际情况进行处理，如删除异常数据行：

```python
data = data[(data[:, 2] >= 0) & (data[:, 4] >= 0)]  # 删除营业收入和利润为负的行
```

### 3.3 基本统计分析
1. 计算每年的总营业收入、总营业成本和总利润




```python
years = np.unique(data[:, 0])  # 获取所有年份
for year in years:
    year_data = data[data[:, 0] == year]  # 提取该年的数据
    total_revenue = np.sum(year_data[:, 2])  # 计算该年的总营业收入
    total_cost = np.sum(year_data[:, 3])  # 计算该年的总营业成本
    total_profit = np.sum(year_data[:, 4])  # 计算该年的总利润
    print(f"Year {int(year)}: Total Revenue = {total_revenue}万元, Total Cost = {total_cost}万元, Total Profit = {total_profit}万元")
```

2. 计算每年的平均利润



```python
for year in years:
    year_data = data[data[:, 0] == year]
    avg_profit = np.mean(year_data[:, 4])  # 计算该年的平均利润
    print(f"Year {int(year)}: Average Profit = {avg_profit}万元")
```

3. 计算各季度的平均营业收入和平均利润

```python
quarters = np.unique(data[:, 1])  # 获取所有季度
for quarter in quarters:
    quarter_data = data[data[:, 1] == quarter]
    avg_revenue = np.mean(quarter_data[:, 2])  # 计算该季度的平均营业收入
    avg_profit = np.mean(quarter_data[:, 4])  # 计算该季度的平均利润
    print(f"Quarter {int(quarter)}: Average Revenue = {avg_revenue}万元, Average Profit = {avg_profit}万元")
```

### 3.4 数据可视化（可选）
虽然 Numpy 本身不提供可视化功能，但可以结合 Matplotlib 等库进行数据可视化。例如，绘制每年的利润趋势图：






```python
import matplotlib.pyplot as plt

years = np.unique(data[:, 0])
profits = []
for year in years:
    year_data = data[data[:, 0] == year]
    total_profit = np.sum(year_data[:, 4])
    profits.append(total_profit)

plt.plot(years, profits, marker='o')
plt.xlabel('年份')
plt.ylabel('总利润（万元）')
plt.title('年度利润趋势')
plt.grid(True)
plt.show()
```
