# 数据可视化

## 1. 常见数据可视化图表
常见数据可视化图表有：柱状图、折线图、饼图、散点图、雷达图、箱型图、气泡图、词频图、桑基图、热力图、关系图、漏斗图等。

| 图表名称   | 图表简介                                                   | 用途                                                         |
|--------|----------------------------------------------------------|------------------------------------------------------------|
| 柱状图    | 展示多个数据之间的差异，通过柱形的高度来表示数值的大小。                           | 适用于小规模数据的比较。                                           |
| 折线图    | 反映数据的变化趋势，通过数据点的连接来展示随时间或其他变量的变化。                     | 适用于二维或时间序列数据。                                         |
| 散点图    | 展示不同数据的分布，通过点的位置来表示两个变量之间的关系。                           | 适用于二维数据的类别对比。                                         |
| 雷达图    | 展示不同类别数据等级差异，通过多边形的顶点来表示各个维度的数值。                       | 适用于小规模数据，展示多维度的比较。                                   |
| 箱形图    | 反映数据的离散情况，通过箱体和须来展示数据的分布和异常值。                             | 适用于展示数据的分布和异常值。                                       |
| 饼图     | 展示各数据在整体的占比，通过扇形的大小来表示各部分的比例。                             | 适用于分类数据的占比展示。                                           |
| 气泡图    | 展示三个变量之间的关系，通过气泡的大小和位置来表示数值。                               | 适用于小规模数据，展示三个变量之间的关系。                               |
| 词频图    | 展示数据出现频率大小，通过条形的长度来表示词频。                                     | 适用于用户画像/标签。                                              |
| 桑基图    | 反映信息的变化和流动状态，通过流动的宽度来表示流量的大小。                             | 适用于路径分析与流量分析。                                           |
| 热力图    | 展示数据的分布情况，通过颜色的深浅来表示数值的大小。                                 | 适用于含位置信息的数据。                                            |
| 关系图    | 展示事物相关性和关联性，通过点和线的连接来表示关系。                                 | 适用于社交关系链/品牌传播。                                         |
| 漏斗图    | 展示业务流程中的转化情况，通过漏斗的宽度来表示各阶段的转化率。                         | 适用于用户转化分析。                                               |

词云图（Word Cloud）是一种文本数据的可视化表示，它通过不同大小和颜色的词语来展示文本中各个词汇出现的频率或重要性。

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from wordcloud import WordCloud
import networkx as nx
from matplotlib.sankey import Sankey
import seaborn as sns

# 设置中文字体
plt.rcParams['font.sans-serif'] = ['SimHei']  # 使用黑体
plt.rcParams['axes.unicode_minus'] = False    # 解决负号显示问题

# 创建6×2的子图布局（竖着的2×6）
fig = plt.figure(figsize=(12, 24))
fig.suptitle('常见图表类型示例', fontsize=20, fontweight='bold', y=0.98)

# 1. 柱状图
ax1 = plt.subplot(6, 2, 1)
categories = ['A', 'B', 'C', 'D', 'E']
values = [23, 45, 56, 12, 41]
ax1.bar(categories, values, color=['#FF6B6B', '#4ECDC4', '#45B7D1', '#F9C80E', '#FF6F61'])
ax1.set_title('柱状图\n展示多个数据之间的差异\n适用于小规模数据')
ax1.set_ylabel('数值')

# 2. 折线图
ax2 = plt.subplot(6, 2, 2)
x = np.linspace(0, 10, 100)
y = np.sin(x)
ax2.plot(x, y, color='#4ECDC4', linewidth=2)
ax2.set_title('折线图\n反映数据的变化趋势\n适用于二维或时间序列数据')
ax2.set_xlabel('时间')
ax2.set_ylabel('数值')

# 3. 散点图
ax3 = plt.subplot(6, 2, 3)
np.random.seed(42)
x = np.random.randn(50)
y = np.random.randn(50)
colors = np.random.rand(50)
sizes = 100 * np.random.rand(50)
ax3.scatter(x, y, c=colors, s=sizes, alpha=0.7, cmap='viridis')
ax3.set_title('散点图\n展示不同数据的分布\n适用于一维数据比较')
ax3.set_xlabel('X轴')
ax3.set_ylabel('Y轴')

# 4. 雷达图 - 使用极坐标
ax6 = plt.subplot(6, 2, 4)
sizes = [15, 30, 25, 10, 20]
labels = ['A', 'B', 'C', 'D', 'E']
colors = ['#FF6B6B', '#4ECDC4', '#45B7D1', '#F9C80E', '#FF6F61']
ax6.pie(sizes, labels=labels, colors=colors, autopct='%1.1f%%', startangle=90)
ax6.set_title('饼图\n展示各数据在整体的占比\n适用于分类较少的数据')

# 5. 箱形图
ax5 = plt.subplot(6, 2, 5)
data = [np.random.normal(0, std, 100) for std in range(1, 5)]
ax5.boxplot(data, patch_artist=True,
           boxprops=dict(facecolor='#F9C80E', color='black'),
           medianprops=dict(color='black'))
ax5.set_title('箱形图\n反映数据的离散情况\n最大/小值、中值（四分）')
ax5.set_xticklabels(['组1', '组2', '组3', '组4'])



# 6. 饼图
ax4 = plt.subplot(6, 2, 6, polar=True)
categories = ['速度', '力量', '技巧', '耐力', '敏捷']
N = len(categories)
values = [4, 3, 5, 2, 4]
angles = np.linspace(0, 2 * np.pi, N, endpoint=False).tolist()
values += values[:1]
angles += angles[:1]
ax4.plot(angles, values, 'o-', linewidth=2, color='#45B7D1')
ax4.fill(angles, values, alpha=0.25, color='#45B7D1')
ax4.set_thetagrids(np.degrees(angles[:-1]), categories)
ax4.set_title('雷达图\n展示不同类别数据等级差异\n适用于小规模数据', y=1.1)

# 7. 气泡图
ax7 = plt.subplot(6, 2, 7)
np.random.seed(42)
x = np.random.rand(20)
y = np.random.rand(20)
z = np.random.rand(20) * 1000
scatter = ax7.scatter(x, y, s=z, alpha=0.6, c=np.random.rand(20), cmap='Blues')
ax7.set_title('气泡图\n展示三个变量之间的关系\n适用于小规模数据')
ax7.set_xlabel('X变量')
ax7.set_ylabel('Y变量')

# 8. 词频图
ax8 = plt.subplot(6, 2, 8)
text = "Python 数据 可视化 图表 柱状图 折线图 散点图 饼图 箱形图 雷达图 气泡图 词频图 桑基图 热力图 关系图 漏斗图"
wordcloud = WordCloud(width=400, height=300, background_color='white', 
                     colormap='viridis').generate(text)
ax8.imshow(wordcloud, interpolation='bilinear')
ax8.axis('off')
ax8.set_title('词频图\n展示数据出现频率大小\n适用于用户画像/标签')

# 9. 桑基图（简化版）
ax9 = plt.subplot(6, 2, 9)
sankey = Sankey(ax=ax9, scale=0.01, offset=0.2)
sankey.add(flows=[1.0, -0.3, -0.4, -0.3],
          labels=['输入', '输出1', '输出2', '输出3'],
          orientations=[0, 1, 0, -1],
          pathlengths=[0.5, 0.5, 0.5, 0.5])
sankey.finish()
ax9.set_title('桑基图\n反映信息的变化和流动状态\n适用于路径分析与流量分析')

# 10. 热力图
ax10 = plt.subplot(6, 2, 10)
np.random.seed(42)
data = np.random.rand(10, 12)
im = ax10.imshow(data, cmap='hot', interpolation='nearest')
plt.colorbar(im, ax=ax10)
ax10.set_title('热力图\n展示数据的分布情况\n适用于含位置信息的数据')

# 11. 关系图
ax11 = plt.subplot(6, 2, 11)
G = nx.Graph()
G.add_edges_from([(1, 2), (1, 3), (2, 3), (3, 4), (4, 5), (4, 6), (5, 6)])
pos = nx.spring_layout(G)
nx.draw(G, pos, ax=ax11, with_labels=True, node_color='skyblue', 
       node_size=800, font_size=12, font_weight='bold')
ax11.set_title('关系图\n展示事物相关性和关联性\n适用于社交关系链/品牌传播')

# 12. 漏斗图
ax12 = plt.subplot(6, 2, 12)
stages = ['访问', '注册', '加购', '支付', '完成']
values = [1000, 600, 300, 150, 80]
ax12.barh(range(len(stages)), values, color=['#FF6B6B', '#4ECDC4', '#45B7D1', '#F9C80E', '#FF6F61'])
ax12.set_yticks(range(len(stages)))
ax12.set_yticklabels(stages)
ax12.invert_yaxis()
ax12.set_title('漏斗图\n展示业务流程中的转化情况\n适用于用户转化分析')
ax12.set_xlabel('用户数量')

plt.tight_layout()
plt.subplots_adjust(top=0.97, hspace=0.4)
plt.show()
```

### 2.3 Python

Python 中与可视化有关的库有很多个，这些库各有特点，适用于不同的数据可视化需求。例如，Matplotlib和Seaborn适合创建静态图表，而Plotly和Bokeh适合创建交互式图表。Pandas则因其数据处理能力而常用于快速生成图表。您可以根据项目需求选择合适的库。

| 库名称          | 描述                                                         |
|---------------|------------------------------------------------------------|
| Matplotlib    | 一个广泛使用的绘图库，提供全面的绘图功能，支持多种图表类型。               |
| Seaborn       | 基于 Matplotlib 的统计数据可视化库，提供更高级的接口绘制统计图表。         |
| Pandas        | 提供数据结构和数据分析工具，内置数据可视化功能，可以快速生成图表。         |
| Plotly        | 一个交互式绘图库，支持在网页上创建动态图表，便于探索数据。               |
| Bokeh         | 专注于现代 Web 浏览器的交互式可视化库，适合创建复杂的可视化应用。           |
| WordCloud     | 用于生成词云图的库，可以展示文本数据中词语的频率或重要性。                 |
| NetworkX      | 用于创建、操作和研究复杂网络结构、动态和功能的库。                       |
| Altair        | 一个声明式统计可视化库，基于 Vega 和 Vega-Lite，提供高层次的图表定制。       |
| HoloViews     | 一个库，用于快速创建交互式和可扩展的数据可视化，基于 Bokeh。               |
| Pygal         | 一个生成 SVG 图表的库，适合在 Web 应用中使用，生成静态图表。                 |
| CartoPy       | 一个地理空间数据分析和可视化库，可以创建地图和空间数据图表。             |
| Dash          | 一个用于构建分析 Web 应用的框架，基于 Plotly，支持交互式图表。               |
