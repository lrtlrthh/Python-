# 展示水果销量

## 1. 实训导入
在前面的学习中，我们已经掌握了 Matplotlib 的基本用法。在实际的数据分析工作中，单一图表往往无法全面展示数据的特征。本次实训将带领大家学习更多 Matplotlib 图表类型，从不同维度深入分析水果销售数据，培养多角度观察和分析数据的能力。

## 2. 实训内容

### 2.1 饼图分析 (Pie Chart)
饼图（Pie Chart）是展示各部分占总体的比例关系的有效图表。

**实训要求：** 已知某商店出售 3 种水果，当天的营业额分别为苹果：100，香蕉：200，葡萄：250。请绘制饼图，显示三种水果的占比。



```python
import matplotlib.pyplot as plt
import numpy as np
# 解决中文显示问题（如果运行环境支持）
plt.rcParams['font.sans-serif'] = ['SimHei'] 
plt.rcParams['axes.unicode_minus'] = False 

labels = ['苹果', '香蕉', '葡萄'] 
sizes = [100, 200, 250] 
explode = (0, 0.1, 0)  # 突出显示香蕉，使其从饼图中分离出来

plt.figure(figsize=(6, 6))
plt.pie(
    sizes, 
    explode=explode, 
    labels=labels, 
    autopct='%1.1f%%', # 显示百分比，保留一位小数
    shadow=True, 
    startangle=90 # 从90度位置开始绘制
)
plt.title('水果销量占比')
plt.axis('equal') # 确保饼图是正圆
plt.savefig('pie_chart_example.png')
plt.show()
```

### 2.2 柱状图分析 (Bar Chart)
柱状图通过柱子的高度直观比较不同类别数据的数值大小，适合进行精确的数值对比。

```python
import matplotlib.pyplot as plt
import numpy as np

# 解决中文显示问题
plt.rcParams['font.sans-serif'] = ['SimHei'] 
plt.rcParams['axes.unicode_minus'] = False

# 数据
fruits = ['苹果', '香蕉', '葡萄']
sales = [100, 200, 250]
colors = ['red', 'yellow', 'purple']

plt.figure(figsize=(8, 6))
bars = plt.bar(fruits, sales, color=colors, alpha=0.7)

# 添加数值标签
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2., height,
             f'{height}', ha='center', va='bottom')

plt.title('水果销量对比')
plt.xlabel('水果种类')
plt.ylabel('销量')
plt.grid(axis='y', alpha=0.3)
plt.show()
```

### 2.3 折线图分析 (Line Chart)
折线图用于展示数据随时间变化的趋势，适合分析销售数据的周期性变化。

```python
# 假设有多天的数据
days = ['周一', '周二', '周三', '周四', '周五']
apple_sales = [80, 100, 120, 90, 110]
banana_sales = [150, 200, 180, 220, 190]
grape_sales = [200, 250, 230, 280, 240]

plt.figure(figsize=(10, 6))
plt.plot(days, apple_sales, 'ro-', label='苹果', linewidth=2)
plt.plot(days, banana_sales, 'yo-', label='香蕉', linewidth=2)
plt.plot(days, grape_sales, 'mo-', label='葡萄', linewidth=2)

plt.title('水果销量趋势')
plt.xlabel('日期')
plt.ylabel('销量')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
