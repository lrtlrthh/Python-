## 2. 什么是办公自动化？
办公自动化是指利用计算机技术和软件工具，自动完成日常办公任务，如数据处理、文档生成、报表制作、邮件发送等，从而提高工作效率、减少重复劳动。
常见办公自动化任务包括：
- 批量处理 Excel 表格（如合并、拆分、格式调整）
- 自动生成 Word 报告或合同
- 提取 PDF 中的文本或表格数据
- 自动发送带附件的邮件
- 批量重命名、移动、整理文件

## 3. 为什么选择 Python？
Python 是一种简洁、易学、功能强大的编程语言，拥有丰富的第三方库，特别适合处理办公自动化任务。
Python 的优势：
- 语法简洁，适合非程序员学习
- 拥有大量办公自动化相关库（如 openpyxl、python-docx、PyPDF2 等）
- 社区活跃，文档丰富，易于查找解决方案
- 可与其他工具（如邮件系统、数据库、Web 应用）集成

## 6. 第一个办公自动化小例子批量创建 Excel 文件

```python
import os
from openpyxl import Workbook

# 创建一个文件夹用于存放生成的 Excel 文件
os.makedirs("excel_files", exist_ok=True)

# 批量创建 5 个 Excel 文件
for i in range(1, 6):
    wb = Workbook()
    ws = wb.active
    ws.title = "成绩表"
    ws.append(["姓名", "语文", "数学", "英语"])
    ws.append(["张三", 85, 90, 88])
    ws.append(["李四", 78, 82, 80])
    wb.save(f"excel_files/班级{i}_成绩表.xlsx")

print("Excel 文件已生成完成！")
```
