# Pandas 

## 1. Pandas 简介


### 1.1 什么是 Pandas？

Pandas 是一个开源的 Python 库，专门用于数据操作和分析。它提供了快速、灵活且富有表现力的数据结构，旨在使“关系型”或“带标签”的数据处理变得简单直观。你可以将其想象成一个强大的 Excel 表格工具，但拥有编程的灵活性和自动化能力。


### 1.2 为什么使用 Pandas？

*   **高效的数据结构**: 提供了`Series`和`DataFrame`两种核心数据结构，能高效处理各种类型的数据。
*   **强大的数据处理能力**: 支持数据清洗、转换、合并、筛选、分组、聚合等一系列复杂操作。
*   **与Python生态系统集成**: 能与 NumPy、Matplotlib、Scikit-learn 等其他 Python 库无缝协作。
*   **广泛应用于数据科学**: 是数据科学家、分析师和工程师进行数据探索和预处理的首选工具。


### 1.3 Pandas 与 NumPy 的关系

Pandas 是建立在 NumPy 库之上的。NumPy 提供了高性能的多维数组对象，是 Python 科学计算的基础。Pandas 利用 NumPy 的数组操作能力，在此基础上增加了标签（索引）和更高级的数据操作功能，使得处理表格数据更加便捷。



```python
# 导入Pandas库，通常约定俗成使用别名 pd
import pandas as pd

print("Pandas 库已成功导入。")
```

## 2. DataFrame 与 Series 数据结构

Pandas 的两个核心数据结构是`Series`和`DataFrame`，理解它们是掌握 Pandas 的关键。


### 2.1 Series：一维带标签数组

`Series`是一种带标签的一维数组，可以存储任何数据类型（整数、字符串、浮点数、Python对象等）。它由两部分组成：数据和索引。




1. 从列表创建 Series

```python
s = pd.Series([10, 20, 30, 40, 50])
print("从列表创建的 Series:")
print(s)
print("\nSeries 的数据类型:", type(s))
print("Series 的索引:", s.index)
print("Series 的值:", s.values)
```

```python
print("\nSeries 的数据类型:", type(s))
print("Series 的索引:", s.index)
print("Series 的值:", s.values)
```

2. 从字典创建 Series (字典的键会成为索引)

```python
s_dict = pd.Series({'a': 100, 'b': 200, 'c': 300})
print("\n从字典创建的 Series:")
print(s_dict)
```

3. 访问 Series 元素

```python
print("\n访问 Series 元素:")
print(f"第一个元素 (通过位置): {s[0]}")
print(f"索引为 'b' 的元素 (通过标签): {s_dict['b']}")
```

### 2.2 DataFrame：二维带标签表格数据

`DataFrame`是Pandas中最常用的数据结构，可以看作是一个带标签的二维表格，类似于电子表格或SQL表。它由一系列`Series`组成，这些`Series`共享同一个索引。`DataFrame`有行索引和列索引。




1. 从字典创建 DataFrame

```python
data = {
    '姓名': ['张三', '李四', '王五'],
    '年龄': [25, 30, 22],
    '城市': ['北京', '上海', '广州']
}
df_example = pd.DataFrame(data)
print("从字典创建的 DataFrame:")

df_example
```

```python
print("\nDataFrame 的数据类型:", type(df_example))
print("DataFrame 的行索引:", df_example.index)
print("DataFrame 的列索引 (列名):", df_example.columns)
```

2. 从列表的列表创建 DataFrame

```python
data_list = [
    ['张三', 25, '北京'],
    ['李四', 30, '上海'],
    ['王五', 22, '广州']
]
df_list = pd.DataFrame(data_list, columns=['姓名', '年龄', '城市'])
print("\n从列表的列表创建的 DataFrame:")

df_list
```

3. DataFrame 基本操作 - 选择列

```python
print("\n选择 '姓名' 列:")
df_example['姓名']
```

```python
print("\n选择 '姓名' 和 '城市' 列:")
df_example[['姓名', '城市']]
```

4. DataFrame 基本操作 - 选择行 (使用 .loc[] 或 .iloc[])

```python
print("\n选择第一行 (通过位置索引 .iloc):")
df_example.iloc[0]

```

```python
print("\n选择索引为 1 的行 (通过标签索引 .loc):") # 默认索引从0开始

df_example.loc[1]
```

### 3.2 筛选的逻辑

表格数据中，最重要的一个功能就是要具有可筛选的能力，选出我所需要的信息，丢弃无用的信息。

**任务一**： 我们以 "Age" 为筛选条件，显示年龄在 10 岁以下的乘客信息。

```python
df[df["Age"]<10].head(3)
```

**任务二**： 以 "Age" 为条件，将年龄在 10 岁以上和 50 岁以下的乘客信息显示出来，并将这个数据命名为 midage

```python
midage = df[(df["Age"]>10) & (df["Age"]<50)]
midage.head(3)
```
