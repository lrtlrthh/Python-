### 2.1 数据集加载
我们将使用`pandas`的`read_csv`函数加载数据集。

```python
import pandas as pd
import warnings                                                                                                     
warnings.filterwarnings("ignore") # 忽略一些不必要的警告  

# 加载数据集
df = pd.read_csv('/home/jovyan/work/datasets/68b691247991eb53424a4049-momodel/online_retail.csv', encoding='ISO-8859-1')
print("数据集加载成功！")

```

### 2.2 数据集初步概览
加载数据集后，首先要对其进行初步的概览，了解其结构、数据类型、缺失值等信息。





```python
print("--- 数据集概览 ---")

# 查看数据集前5行
print("\n数据集前5行 (df.head()):")
print(df.head())

# 查看数据集的形状 (行数, 列数)
print(f"\n数据集的形状 (df.shape): {df.shape}")

# 查看数据集的列名和数据类型
print("\n数据集的列信息 (df.info()):")
df.info()

# 查看数值型列的统计摘要
print("\n数值型列的统计摘要 (df.describe()):")
print(df.describe())

# 检查每列的缺失值数量
print("\n每列的缺失值数量 (df.isnull().sum()):")
print(df.isnull().sum())
```

**解释**:
*   `df.head()`：快速预览数据，了解其大概内容和格式。
*   `df.shape`：返回一个元组，表示DataFrame的行数和列数。
*   `df.info()`：提供每列的名称、非空值数量和数据类型，帮助我们发现潜在的数据类型问题和缺失值情况。
*   `df.describe()`：为数值型列生成描述性统计信息，如计数、均值、标准差、最小值、最大值和四分位数，有助于了解数据的分布。
*   `df.isnull().sum()`：统计每列的缺失值数量，这是数据清洗的第一步。


### 2.3 数据选择与筛选

我们经常需要从数据集中选择特定的列或根据条件筛选出行。




```python
print("--- 数据选择与筛选 ---")

# 选择单列：选择 'Description' 列
product_descriptions = df['Description']
print("\n选择 'Description' 列的前5行:")
print(product_descriptions.head())

# 选择多列：选择 'InvoiceNo', 'Description', 'Quantity' 列
selected_columns = df[['InvoiceNo', 'Description', 'Quantity']]
print("\n选择 'InvoiceNo', 'Description', 'Quantity' 列的前5行:")
print(selected_columns.head())

# 根据条件筛选行：筛选出 'Country' 为 'United Kingdom' 的交易
uk_sales = df[df['Country'] == 'United Kingdom']
print("\n筛选出 'United Kingdom' 的交易 (前5行):")
print(uk_sales.head())
print(f"在 'United Kingdom' 的交易总数: {len(uk_sales)} 条")

# 组合条件筛选：筛选出 'United Kingdom' 且 'Quantity' 大于 10 的交易
uk_large_quantity_sales = df[(df['Country'] == 'United Kingdom') & (df['Quantity'] > 10)]
print("\n筛选出 'United Kingdom' 且 'Quantity' 大于 10 的交易 (前5行):")
print(uk_large_quantity_sales.head())
print(f"符合条件的交易总数: {len(uk_large_quantity_sales)} 条")

# 使用 .loc[] 进行基于标签的行/列选择
# 选择行索引0到4的所有列
print("\n使用 .loc[] 选择前5行的所有列:")
print(df.loc[0:4, :])

# 选择行索引0到4的 'Description' 和 'Quantity' 列
print("\n使用 .loc[] 选择前5行的 'Description' 和 'Quantity' 列:")
print(df.loc[0:4, ['Description', 'Quantity']])


```

**解释**:
*   通过`df['列名']`或`df[['列名1', '列名2']]`可以方便地选择一列或多列。
*   通过布尔索引 (`df[条件]`) 可以根据条件筛选出满足特定条件的行。
*   `.loc[]` 允许我们通过行标签和列标签进行选择，格式为 `df.loc[行标签, 列标签]`。
*   `.iloc[]` 允许我们通过行位置和列位置进行选择，格式为 `df.iloc[行位置, 列位置]`。

### 2.4 缺失值处理

缺失值是指数据集中某些数据点没有记录或无法获取的情况。

数据集中经常包含缺失值，这会影响分析结果，需要进行处理。








```python
print("--- 缺失值处理 ---")

# 再次查看缺失值情况
print("\n处理前每列的缺失值数量:")
print(df.isnull().sum())

# 针对 'Description' 列的缺失值：填充为 'Unknown'
# 为了不修改原始df，我们先创建一个副本进行演示
df_filled_description = df.copy()
df_filled_description['Description'].fillna('Unknown', inplace=True)
print("\n处理 'Description' 列缺失值后:")
print(df_filled_description.isnull().sum())
# 找出被填充的行并展示前5行
print("填充 'Description' 列缺失值后的前5行 (其中Description为'Unknown'):")
print(df_filled_description[df_filled_description['Description'] == 'Unknown'].head())

# 针对 'CustomerID' 列的缺失值：直接删除含有缺失值的行
# 注意：删除操作会减少数据量，需谨慎使用，因为这可能导致信息丢失
df_dropped_customer = df.copy()
df_dropped_customer.dropna(subset=['CustomerID'], inplace=True)
print("\n处理 'CustomerID' 列缺失值后 (删除含有缺失值的行):")
print(df_dropped_customer.isnull().sum())
print(f"删除缺失值后数据集的形状: {df_dropped_customer.shape}")

# 为了后续分析的方便，我们创建一个处理过缺失值的新DataFrame
# 策略：删除 CustomerID 和 Description 缺失的行，以确保数据质量
df_cleaned = df.dropna(subset=['CustomerID', 'Description']).copy()
print(f"\n最终清理后的数据集形状: {df_cleaned.shape}")
print("最终清理后数据集的缺失值情况:")
print(df_cleaned.isnull().sum())
```

**解释**:
*   `df.isnull().sum()`：快速识别哪些列存在缺失值以及数量。
*   `df.fillna(value, inplace=True)`：用指定的值填充缺失值。`inplace=True`表示直接修改原DataFrame。
*   `df.dropna(subset=['列名'], inplace=True)`：删除指定列中含有缺失值的行。`subset`参数可以指定在哪些列中检查缺失值。

### 2.5 简单分组聚合与可视化

分组聚合是数据分析中非常常用的一种操作，用于根据某个（或多个）列的值对数据进行分组，然后对每个组执行聚合计算（如求和、均值、计数等）。

我们还将使用可视化来直观展示结果。






```python
print("--- 简单分组聚合 ---")

# 计算每笔交易的销售额 (Quantity * UnitPrice)
df_cleaned['TotalPrice'] = df_cleaned['Quantity'] * df_cleaned['UnitPrice']

# 按 'Country' 分组，并计算每个国家的 'TotalPrice' 总和
sales_by_country = df_cleaned.groupby('Country')['TotalPrice'].sum().sort_values(ascending=False)
print("\n按国家统计总销售额 (前10名):")
print(sales_by_country.head(10))

# 可视化：按国家统计总销售额 (前10名)
import matplotlib.pyplot as plt
import seaborn as sns

# 设置中文字体和图表风格，确保中文显示正常
plt.rcParams['font.sans-serif'] = ['SimHei', 'Arial Unicode MS']
plt.rcParams['axes.unicode_minus'] = False # 解决负号显示问题
sns.set_style("whitegrid") # 设置Seaborn风格，使图表更美观

# 目的: 直观展示不同国家的销售贡献，便于学生理解分组聚合结果
# 数据说明: 按国家统计的总销售额 (前10名)

# 准备数据：取前10名国家进行可视化
top_10_countries = sales_by_country.head(10)
countries = top_10_countries.index
total_sales = top_10_countries.values

# 创建图形
plt.figure(figsize=(12, 7))

# 绘制柱状图，使用Seaborn的调色板
bars = plt.bar(countries, total_sales, color=sns.color_palette("viridis", len(countries)), alpha=0.8)

# 在柱子顶部添加数值标签，方便直接读取数据
for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, yval + (max(total_sales)*0.01),
             f'{yval:,.2f}', ha='center', va='bottom', fontsize=10)

# 添加标题和标签
plt.title('电商平台按国家统计的总销售额 (前10名)', fontsize=16, fontweight='bold')
plt.xlabel('国家', fontsize=12)
plt.ylabel('总销售额 (货币单位)', fontsize=12)

# 调整Y轴范围以留出标签空间
plt.ylim(0, max(total_sales) * 1.15)

# 旋转X轴标签，防止标签重叠，提高可读性
plt.xticks(rotation=45, ha='right')

# 调整布局，确保所有元素都显示完整
plt.tight_layout()

# 显示图形
plt.show()

print("\n图表解释：")
print("此柱状图直观展示了不同国家对电商平台总销售额的贡献。最高的柱子代表销售额最高的国家，")
print("通常是主要市场。通过对比柱子高度，我们可以快速识别出核心市场和次要市场，为市场策略提供数据支持。")

# 计算每个商品的平均单价
avg_price_by_product = df_cleaned.groupby('Description')['UnitPrice'].mean().sort_values(ascending=False)
print("\n按商品描述统计平均单价 (前10名):")
print(avg_price_by_product.head(10))

# 计算每个客户的订单数量
orders_by_customer = df_cleaned.groupby('CustomerID')['InvoiceNo'].nunique().sort_values(ascending=False)
print("\n按客户ID统计订单数量 (前10名):")
print(orders_by_customer.head(10))
```

**解释**:
*   `df.groupby('列名')`：根据指定列的值对DataFrame进行分组。
*   `.sum()`, `.mean()`, `.count()`, `.nunique()`：是常用的聚合函数，分别用于求和、求均值、计数和计算唯一值数量。
*   `.sort_values(ascending=False)`：对结果进行降序排序，方便查看排名靠前的项。
*   **可视化图表**: 上方的柱状图展示了销售额最高的10个国家，让我们可以一目了然地看到哪些国家是主要的销售市场，这是数据洞察的直观体现。
