````python
import os, random
from matplotlib.pyplot import imshow
import numpy as np
from PIL import Image
%matplotlib inline

def read_random_image():
    folder=r"/home/jovyan/work/datasets/ma1ogo3ushu4ju4ji2-momodel/dogs_cats/data/test/"
    file_path = folder + random.choice(os.listdir(folder))
    pil_im = Image.open(file_path, 'r')
    return pil_im
````

接着，定一个使用模型对读取出的图片进行预测的函数。

````python
def get_predict(pil_im, model):
    # 对图片进行缩放
    pil_im = pil_im.resize((200, 200))
    # 将格式转为 numpy array 格式
    array_im = np.asarray(pil_im)
    array_im = np.expand_dims(array_im, axis=0)
    # 对图片进行预测
    result = model.predict(array_im)
    if result[0][0] > 0.5:
        print("预测结果是：狗")
    else:
        print("预测结果是：猫")

````

### 第四步：迁移学习

迁移学习是一种机器学习方法，就是把为任务 A 开发的模型作为初始点，重新使用在为任务 B 开发模型的过程中。

在计算机视觉任务中，将预训练的模型作为新模型的起点是一种常用的方法，通常这些预训练的模型在开发神经网络的时候已经消耗了巨大的时间资源和计算资源，迁移学习可以将已习得的强大技能迁移到相关的的问题上。

**基于 InceptionV3 模型做迁移学习**

````python
# 导入需要的包
import sys
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from keras.applications.inception_v3 import InceptionV3
from keras.models import Model
from keras.layers import Dense, GlobalAveragePooling2D
import tensorflow as tf
tf.compat.v1.logging.set_verbosity(tf.compat.v1.logging.ERROR)
# 把 InceptionV3 的预训练模型拷贝到 .keras 文件夹下，避免重复下载
import os
from shutil import copy2
if not os.path.exists('/home/jovyan/.keras/models/'):
    os.mkdir('/home/jovyan/.keras/models')
copy2("/home/jovyan/work/datasets/ma1ogo3ushu4ju4ji2-momodel/dogs_cats/model/inception_v3_weights_tf_dim_ordering_tf_kernels_notop.h5",
             "/home/jovyan/.keras/models")
````

定义模型

````python
def define_transfer_model():
    # 构建不带分类器的预训练模型
    base_model = InceptionV3(weights='imagenet', include_top=False)

    # 添加全局平均池化层
    x = base_model.output
    x = GlobalAveragePooling2D()(x)

    # 添加一个全连接层
    x = Dense(1024, activation='relu')(x)

    # 添加一个分类器
    prediction = Dense(1, activation='sigmoid')(x)

    # 首先，我们只训练顶部的几层（随机初始化的层）
    # 锁住所有 InceptionV3 的卷积层
    for layer in base_model.layers:
        layer.trainable = False
    model = Model(inputs=base_model.input, outputs=prediction)
    # 编译模型
    model.compile(optimizer='rmsprop',
                  loss='binary_crossentropy',
                  metrics=['accuracy'])
    return model
````

运行代码查看模型结构图

````python
from keras.utils import plot_model
model = define_transfer_model()
plot_model(model, to_file='transfer_model.png')
````

定义训练模型的函数

````python
def train_transfer_model():
    # 实例化模型
    model = define_transfer_model()
    # 创建图片生成器
    datagen = ImageDataGenerator(rescale=1.0 / 255.0)
    train_it = datagen.flow_from_directory(
        '/home/jovyan/work/datasets/ma1ogo3ushu4ju4ji2-momodel/dogs_cats/data/train/',
        class_mode='binary',
        batch_size=64,
        target_size=(224, 224))
    # 训练模型
    model.fit(train_it,
                        steps_per_epoch=len(train_it),
                        epochs=1,
                        verbose=1)
````

````python
train_transfer_model()
````
