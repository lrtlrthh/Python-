**特征介绍**

1.date: 日期格式为 dd/mm/yy

2.time: 时间格式为 hh:mm:ss

3.Global_active_power: 家庭消耗的总有功功率（千瓦），在交流电路中，电源在一个周期内发出瞬时功率的平均值(或负载电阻所消耗的功率)，称为"有功功率"

4.Global_reactive_power: 家庭消耗的总无功功率（千瓦），在具有电感或电容的电路中，在每半个周期内，把电源能量变成磁场(或电场)能量贮存起来，然后，再释放，又把贮存的磁场（或电场）能量再返回给电源，只是进行这种能量的交换，并没有真正消耗能量，我们把这个交换的功率值，称为" 无功功率"
5.voltage: 平均电压（伏特）

6.Global_intensity: 平均电流强度（安培)

7.sub_metering_1: 厨房的有功功率

8.sub_metering_2: 用于洗衣机等电器的有功功率

9.sub_metering_3: 空调热水器等电器的有功功率

因为这里我们预测 ‘Global_active_power' 不仅用到过去时间的 ‘Global_active_power' 还会用到其他的特征，比如: 'Voltage‘，这时候我们把此类问题叫做多变量时间序列预测。下面我们展示完整的转换为监督学习问题的代码。

```python

def series_to_supervised(data, n_in=1, n_out=1, dropnan=True):
    """
    将时间序列数据转换为监督学习数据
    data: 输入数据（DataFrame或numpy数组）
    n_in: 输入序列长度（过去的时间步）
    n_out: 输出序列长度（未来的时间步）
    dropnan: 是否删除含缺失值的行
    """
    n_vars = 1 if type(data) is list else data.shape[1]
    dff = pd.DataFrame(data)
    cols, names = [], []

    # 输入序列：t-n_in, ..., t-1
    for i in range(n_in, 0, -1):
        cols.append(dff.shift(i))
        names += [f'var{j+1}(t-{i})' for j in range(n_vars)]

    # 输出序列：t, t+1, ..., t+n_out-1
    for i in range(0, n_out):
        cols.append(dff.shift(-i))
        if i == 0:
            names += [f'var{j+1}(t)' for j in range(n_vars)]
        else:
            names += [f'var{j+1}(t+{i})' for j in range(n_vars)]

    # 合并所有列
    agg = pd.concat(cols, axis=1)
    agg.columns = names

    # 删除含缺失值的行
    if dropnan:
        agg.dropna(inplace=True)
    return agg
```

这里我们探索的问题是根据之前一小时的有功功率和其他特征，预测未来一小时的有功功率是多少？这种类型的模型在规划支出方面可能对家庭有所帮助。在供应方面，它也可能有助于规划特定家庭的电力需求。

#### 对特征进行归一化

数据归一化处理是数据挖掘的一项基础工作。不同指标往往具有不同的量纲，数值间的差别可能很大，不进行处理可能会影响数据分析的结果。为了消除指标间的量纲和取值范围差异的影响，需要进行标准化处理，将数据按照比例进行缩放，使之落入特定的区域，以便于进行综合分析。

同样我们必须对用户电量的各指标进行处理，这里我们用到最小最大规范化。

```python
# 特征归一化（Min-Max缩放至[0,1]）
scaler = MinMaxScaler(feature_range=(0, 1))
scaled = scaler.fit_transform(df_resample.values)  # 转换为numpy数组并归一化
```

这里，我们将前三年的数据作为训练集，后一年的样本作为测试集，并将数据改为3D格式。

```python
#对样本集拆分成训练集和测试集
values = reframed.values
n_train_time = None
train = values[:n_train_time, :]
test = values[n_train_time:, :]

train_X, train_y = train[:, :-1], train[:, -1]
test_X, test_y = test[:, :-1], test[:, -1]
# 把数据转换为3维
train_X = train_X.reshape((train_X.shape[0], 1, None))
test_X = test_X.reshape((test_X.shape[0], 1, None))
print(train_X.shape, train_y.shape, test_X.shape, test_y.shape) 
```

上面我在当前时间以小时为单位显示了7个输入变量（输入系列）和1个输出变量“Global_active_power”（取决于重新采样）。

首先，我将准备好的数据集拆分为训练集和测试集。我们将在数据的前三年训练模型，然后在接下来的1年数据中对其进行评估

模型架构

1）LSTM在隐藏层中有100个神经元

2）丢弃20％，防止过拟合

3）输出层有1个神经元，用于预测 Global_active_power

4）使用均方误差（MSE）损失函数和随机梯度下降的Adam优化器

5）该模型epoch为20，批次大小为70




```python
model = Sequential()
model.add(LSTM(100, input_shape=(train_X.shape[1], train_X.shape[2])))
model.add(Dropout(0.2))
model.add(Dense(1))
model.compile(loss='mean_squared_error', optimizer='adam')


# 训练网络
history = model.fit(train_X, train_y, epochs=20, batch_size=70, validation_data=(test_X, test_y), verbose=2, shuffle=False)

plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])
plt.title('model loss')
plt.ylabel('loss')
plt.xlabel('epoch')
plt.legend(['train', 'test'], loc='upper right')
plt.show()

# 做出预测
yhat = model.predict(test_X)
test_X = test_X.reshape((test_X.shape[0], 7))

```

我们将预测与测试数据集相结合，并反演缩放。 

以预测值和实际值为原始尺度，我们可以计算模型的误差分数。 在这种情况下，我们计算出与变量本身相同的单位产生误差的均方根误差（RMSE）。

```python

# 对预测值进行反演缩放
inv_yhat = np.concatenate((yhat, test_X[:, -6:]), axis=1)
inv_yhat = scaler.inverse_transform(inv_yhat)
inv_yhat = inv_yhat[:,0]
# 对真实值进行反演缩放
test_y = test_y.reshape((len(test_y), 1))
inv_y = np.concatenate((test_y, test_X[:, -6:]), axis=1)
inv_y = scaler.inverse_transform(inv_y)
inv_y = inv_y[:,0]
# 计算RMSE
rmse = np.sqrt(mean_squared_error(inv_y, inv_yhat))
print('Test RMSE: %.3f' % rmse)
```
